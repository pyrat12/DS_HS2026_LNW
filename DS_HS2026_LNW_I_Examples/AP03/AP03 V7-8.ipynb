# V7: Neural Networks

## Neural Networks: Aus Daten lernen

- Definition: Ein neuronales Netz lernt anhand von Beispielen, Eingaben in passende Vorhersagen umzuwandeln
- Zweck: Komplexe Muster erkennen, bspw. Spam, Bildinhalte, Sprache oder Zusammenhänge in Messdaten
- Input: Numerische Merkmale, Pixelwerte, Textrepräsentationen oder andere Zahlenarrays
- Lernen: Gewichte und Biases werden so angepasst, dass der Trainingsfehler kleiner wird
- Output: Je nach Aufgabe ein Zahlenwert, eine Klasse, Wahrscheinlichkeiten oder eine gelernte Repräsentation
- Erkenntnis: Entscheidend ist die Leistung auf neuen Beobachtungen, gute Trainingsresultate allein reichen nicht

### Warum wurden neuronale Netze erfolgreich?

| Entwicklung | Bedeutung |
|---|---|
| Perzeptron, 1957 | Lernbarer linearer Klassifikator |
| XOR-Problem | Ein einzelnes lineares Perzeptron kann nicht linear trennbare Muster wie XOR nicht darstellen |
| AI Winter | Überhöhte Erwartungen und Finanzierungskürzungen bremsen die Entwicklung |
| Backpropagation, 1986 | Effizientes Training mehrschichtiger Netze wird verbreitet |
| GPUs und CUDA | Parallele Berechnungen ermöglichen das Training grösserer Netze |
| AlexNet, 2012 | Durchbruch bei der Bildklassifikation auf ImageNet |
| AlphaGo, 2016 | Verbindet neuronale Netze mit Selbstspiel und Suche |

- Erkenntnis: Der Fortschritt entstand durch bessere Algorithmen, mehr Daten und mehr Rechenleistung zusammen

## Spam-Beispiel: Von E-Mails zu Eingabemerkmalen

- Beobachtung: Eine Zeile entspricht einer E-Mail
- Zielvariable: `0 = legitime E-Mail`, `1 = Spam`
- Eingaben: 16 manuell entwickelte Features, bspw. Textlänge, Wortanzahl, Grossbuchstabenanteil, URLs, Ausrufezeichen und Häufigkeit bestimmter Wörter
- Vorbereitung: Numerische Features anhand der Trainingsdaten standardisieren und dieselbe Transformation auf Validierung und Test anwenden
- Aufteilung: 18'000 E-Mails zum Training, 6000 zur Validierung und 6000 zum abschliessenden Test
- Erkenntnis: Der Klassifikator erhält Zahlen, die Qualität dieser Repräsentation beeinflusst, welche Muster er erkennen kann

## Neuron: Gewichtete Summe plus Aktivierung

- Definition: Ein Neuron kombiniert seine Eingaben mit Gewichten, addiert einen Bias und wendet eine Aktivierungsfunktion an
- Format: `z = w₁ × x₁ + … + wₙ × xₙ + b`, danach `a = f(z)`
- Gewichte: Bestimmen, wie stark die jeweiligen Eingaben zum Score beitragen
- Bias: Verschiebt den Score unabhängig von den aktuellen Eingabewerten
- Logit: Roher Output-Score vor der Umwandlung in eine Wahrscheinlichkeit

### Bsp.

`z = 0,2 × 0,8 + 0,5 × 1,2 + 0,4 = 1,16`

Nach Sigmoid ergibt sich ungefähr `p = 0,761`

- Interpretation: Das Modell schätzt für diese Eingaben ungefähr 76,1 % Spam-Wahrscheinlichkeit
- Achtung: Der Score 1,16 ist selbst noch keine Wahrscheinlichkeit

## MLP: Multilayer Perceptron

- Definition: Neuronales Netz mit verbundenen Schichten, verborgene Schichten lernen Zwischenrepräsentationen
- Zweck: Nichtlineare Zusammenhänge und Kombinationen von Merkmalen erfassen
- Feedforward: Informationen laufen von der Eingabe zur Ausgabe, ohne Schleifen
- Tiefe: Anzahl aufeinanderfolgender Schichten
- Breite: Anzahl Neuronen pro Schicht

| Schicht | Aufgabe |
|---|---|
| Input Layer | Eingabemerkmale übernehmen |
| Hidden Layers | Merkmale gewichten, kombinieren und transformieren |
| Output Layer | Zur Aufgabe passenden Output erzeugen |

- Bsp. Spam-Netz: `16 Eingaben → 16 ReLU → 16 ReLU → 1 Sigmoid`
- Erkenntnis: Wir wählen die Architektur, das Training lernt die Gewichte und Biases

## Aktivierungsfunktionen

| Funktion | Wirkung | Verwendung |
|---|---|---|
| Linear | `f(z) = z`, Score unverändert weitergeben | Output einer unbeschränkten Regression |
| Sigmoid | `1 / (1 + exp(−z))`, Wert zwischen 0 und 1 | Wahrscheinlichkeit einer binären Zielklasse |
| ReLU | `max(0, z)`, negative Werte werden 0 | Häufig in verborgenen Schichten |
| Softmax | Mehrere Scores in Wahrscheinlichkeiten mit Summe 1 umwandeln | Mehrklassenklassifikation |

- Zweck: Nichtlineare Aktivierungen ermöglichen komplexere Beziehungen zwischen Eingabe und Ausgabe
- Erkenntnis: Mehrere rein lineare Schichten lassen sich zu einer einzigen linearen Abbildung zusammenfassen
- Achtung: Mehr Schichten ohne nichtlineare Aktivierungen lösen dieses Problem nicht

## Modellparameter und Hyperparameter

| Modellparameter | Hyperparameter |
|---|---|
| Werden beim Training gelernt | Werden vorgegeben oder anhand der Validierung ausgewählt |
| Gewichte und Biases | Schichtanzahl, Einheiten, Aktivierungen, Lernrate, Batchgrösse, Optimizer und Regularisierung |

- Erkenntnis: Mehr Schichten, grössere Netze oder längeres Training sind nicht automatisch besser
- Auswahl: Einstellungen auf Validierungsdaten vergleichen, den Test für die abschliessende Bewertung zurückhalten

## Wahrscheinlichkeit, Entscheidung und Fehlerkosten

- Wahrscheinlichkeit: Einschätzung des Modells, bspw. 80 % Spam-Wahrscheinlichkeit
- Entscheidung: Entsteht erst durch einen Schwellenwert, bspw. Spam markieren bei `p ≥ 0,84`
- Interpretation: Dieselbe Wahrscheinlichkeit kann bei verschiedenen Schwellenwerten zu unterschiedlichen Entscheidungen führen

| Begriff | Bedeutung im Spam-Beispiel |
|---|---|
| TP | Spam korrekt markiert |
| FP | Legitime E-Mail fälschlich markiert |
| FN | Spam übersehen |
| Precision | `TP / (TP + FP)`, Anteil echten Spams unter den markierten E-Mails |
| Recall | `TP / (TP + FN)`, Anteil erkannten Spams unter allen Spam-Mails |

- Beispielkosten: `5 × FP + FN`, eine fälschlich markierte legitime Mail wird fünfmal stärker gewichtet als übersehener Spam
- Erkenntnis: Nicht nur Kennzahlen maximieren, auch Fehlertypen und geschäftliche Folgen berücksichtigen

| PyTorch | Flexible Definition und Steuerung neuronaler Netze |

## Modell verändern oder Repräsentation verändern

| Änderung | Was verändert sich? |
|---|---|
| MLP durch XGBoost ersetzen | Modell, dieselben manuell entwickelten Features |
| E-Mail mit vortrainiertem Encoder darstellen | Repräsentation, bspw. 384 Embedding-Werte statt 16 manueller Features |
| Mehr Hidden Layers ergänzen | Architektur und Kapazität des Klassifikators |

- XGBoost: Ensemble aus Bäumen, kein neuronales Netz
- Frozen Encoder: Vortrainierte Gewichte bleiben unverändert, ein neuer Klassifikator lernt auf seinen Merkmalen
- Erkenntnis: Ein Embedding-System ist nicht einfach nur ein tieferer Klassifikator
- Achtung: Unterschiede zwischen vollständigen Systemen isolieren nicht die Wirkung einer einzelnen Änderung

## Trainingsablauf: Forward Pass bis Update

| Schritt | Was geschieht? |
|---|---|
| Forward Pass | Aus Eingaben eine Vorhersage berechnen |
| Loss berechnen | Vorhersage mit bekanntem Trainingsziel vergleichen |
| Backpropagation | Gradienten vom Output rückwärts durch das Netz berechnen |
| Optimizer-Update | Gewichte und Biases anhand der Gradienten verändern |
| Wiederholen | Mit den neuen Parametern weitere Trainingsbeispiele verarbeiten |

- Erkenntnis: Beim Lernen werden die Modellparameter verändert, nicht das vorgegebene Label
- Generalisierung: Gesucht ist gute Leistung auf neuen Daten, der Trainingsfehler ist häufig zu optimistisch

## Binary Cross-Entropy: BCE

- Definition: Loss-Funktion für binäre Zielvariablen und deren vorhergesagte Wahrscheinlichkeiten
- Zweck: Wahrscheinlichkeiten bestrafen, die nicht zum beobachteten Label passen
- Format: `L = −[y × ln(p) + (1 − y) × ln(1 − p)]`
- Bei `y = 1`: Loss ist `−ln(p)`, eine hohe Wahrscheinlichkeit ist günstig
- Bei `y = 0`: Loss ist `−ln(1 − p)`, eine niedrige Wahrscheinlichkeit ist günstig
- Erkenntnis: Sichere falsche Vorhersagen werden besonders stark bestraft
- Bsp. tatsächlicher Spam mit `p = 0,646` ergibt ungefähr `L = 0,437`
- Achtung: Loss einer einzelnen Beobachtung und mittleren Loss eines Batches unterscheiden

## Gradient und Gradient Descent

- Definition: Ein Gradient enthält eine partielle Ableitung je Gewicht und Bias
- Zweck: Beschreiben, wie sich der Loss bei kleinen Parameteränderungen lokal verändert
- Positiver Gradient: Eine kleine Verringerung des Parameters kann den Loss senken
- Negativer Gradient: Eine kleine Erhöhung des Parameters kann den Loss senken
- Update-Regel: `neuer Parameter = alter Parameter − Lernrate × Gradient`

### Bsp.

`0,2000 − 0,1 × (−0,1664) ≈ 0,2166`

- Lernrate: Bestimmt die Schrittgrösse
- Zu klein: Langsame Annäherung
- Zu gross: Instabile Schritte oder Überspringen günstiger Parameterbereiche
- Sigmoid plus BCE: Ableitung nach dem Output-Logit ist `p − y`
- Achtung: `p − y` ist nicht automatisch der Gradient jedes einzelnen Gewichts

## Backpropagation

- Definition: Effiziente Berechnung der Gradienten mit der Kettenregel
- Ablauf: Am Output beginnen und die Beiträge schrittweise zu früheren Schichten zurückverfolgen
- Mehrere Pfade: Erreicht ein Neuron den Output über mehrere Wege, werden deren Gradientenbeiträge addiert
- Effizienz: Gemeinsame Zwischenergebnisse werden berechnet und wiederverwendet
- Reihenfolge: Zuerst Gradienten für den aktuellen Parameterstand berechnen, anschliessend das Update durchführen
- Merksatz: Backpropagation berechnet Gradienten, der Optimizer aktualisiert Gewichte

### Folienbeispiel

- Das Netz `16 → 16 → 16 → 1` hat 561 trainierbare Gewichte und Biases
- Wiederverwendung spart im gezeigten Rechenvergleich ungefähr Faktor 17 bei der Gradientenberechnung, inkl. Updates ungefähr Faktor 9
- Diese Zahlen zählen Rechenoperationen, sie sind kein gemessener Laufzeitvergleich
- Bei einer Spam-Mail steigt die Wahrscheinlichkeit nach einem Update von 64,6 % auf 68,6 %, der Loss sinkt von 0,437 auf 0,377
- Erkenntnis: Verbesserung bei einer Trainingsmail beweist keine Verbesserung bei neuen E-Mails

## Batch, Trainingsschritt und Epoche

| Begriff | Erklärung |
|---|---|
| Batch | Gruppe von Beobachtungen für eine gemeinsame Gradientenschätzung |
| Trainingsschritt | Ein Parameterupdate |
| Epoche | Ein vollständiger Durchlauf durch die Trainingsdaten |
| Batchgrösse | Anzahl Beispiele pro Update |
| Lernrate | Grösse des Updates |

- Bsp. 1000 E-Mails bei Batchgrösse 100 ergeben zehn Updates pro Epoche
- Bei 20 Epochen entstehen somit 200 Updates
- Erkenntnis: Batchgrösse, Epochen und Lernrate steuern unterschiedliche Dinge

## Optimizer

| Optimizer | Funktionsweise | Einordnung |
|---|---|---|
| Mini-Batch SGD | Schritt entgegen dem mittleren Gradienten des aktuellen Batches | Einfach, Lernrate wichtig |
| SGD mit Momentum | Vergangene Gradienten geglättet berücksichtigen | Kann Zickzackbewegungen reduzieren |
| Adam | Gradienten und quadrierte Gradienten zur Anpassung der Schritte verwenden | Praktischer Ausgangspunkt, weiterhin Lernrate nötig |

- Erkenntnis: Kein Optimizer ist immer überlegen
- Vergleich: Validierungsleistung bei vergleichbarem Abstimmungsaufwand beurteilen
- Startwerte aus den Folien: Bspw. ein bis zwei Hidden Layers, 16 bis 64 Einheiten, Adam mit Lernrate 0,001 und Batchgrösse 32 oder 64
- Achtung: Das sind mögliche Ausgangspunkte, keine allgemeinen Optimalwerte

## Overfitting und Early Stopping

- Overfitting: Das Netz passt sich zu stark an Trainingsdetails an und generalisiert schlechter
- Erkennbar: Trainings-Loss sinkt weiter, während Validierungs-Loss steigt
- Early Stopping: Training beenden, wenn die Validierungsleistung länger nicht besser wird
- Patience: Erlaubte Anzahl Epochen ohne Verbesserung
- Checkpoint: Gespeicherter Parameterstand eines Modells
- Bsp. bestes Modell in Epoche 20 und Patience 5, Stopp nach Epoche 25 und Wiederherstellung von Epoche 20
- Erkenntnis: Besten Validierungsstand verwenden, nicht automatisch den letzten Trainingsstand
- Achtung: Testdaten nicht für Early Stopping verwenden

## L2-Regularisierung und Dropout

| Verfahren | Wirkung | Achtung |
|---|---|---|
| L2-Regularisierung | Grosse Gewichte durch eine zusätzliche Strafe im Trainingsziel unattraktiver machen | Zu starke Strafe kann Underfitting verursachen |
| Dropout | Während des Trainings zufällig Hidden-Aktivierungen auf 0 setzen | Bei der normalen Vorhersage ausgeschaltet |

- L2-Grundidee: `Trainingsziel = mittlerer Loss + λ × Summe quadrierter Gewichte`
- Dropout-Rate: Wahrscheinlichkeit, mit der eine Aktivierung auf 0 gesetzt wird
- Neue Maske: Bei weiteren Trainingsdurchläufen werden andere Aktivierungen ausgewählt
- Erkenntnis: Dropout reduziert die Abhängigkeit von einzelnen Aktivierungen, Neuronen werden nicht dauerhaft gelöscht
- Auswahl: Regularisierungsstärke und Dropout-Rate auf Validierungsdaten abstimmen

## Bilder als Zahlen

| Bildtyp | Darstellung |
|---|---|
| Graustufenbild | Eine Matrix mit einem Intensitätswert pro Pixel |
| 8-Bit-Graustufen | 0 ist schwarz, 255 weiss |
| RGB-Bild | Drei ausgerichtete Matrizen für Rot, Grün und Blau |
| 64 × 64 RGB | `64 × 64 × 3 = 12'288` Werte |

- Vorbereitung: Division durch 255 skaliert Intensitäten auf 0 bis 1
- Flattening: Bildwerte in einem langen Vektor anordnen, selbst keine trainierbaren Parameter
- Problem dichter Netze: Jede Position erhält eigene Verbindungen, lokale Bildstruktur und Wiederverwendung verschobener Muster sind nicht eingebaut
- Bsp. 12'288 Eingaben und 128 Neuronen benötigen `12'288 × 128 + 128 = 1'572'992` Parameter
- Erkenntnis: CNNs sind für räumliche Strukturen häufig geeigneter
- Zeitliche Daten: Reihenfolge und längere Abhängigkeiten motivieren andere Architekturen, bspw. RNNs, LSTMs und Transformer

## Convolution: Lokale Muster erkennen

- Definition: Ein kleiner Filter bzw. Kernel wird über lokale Bildausschnitte geschoben
- Berechnung: Passende Werte multiplizieren und die Produkte addieren
- Feature Map: Räumliches Raster der Filterantworten
- Sobel: Manuell festgelegter Filter zur Hervorhebung gerichteter Intensitätsänderungen bzw. Kanten
- CNN: Filtergewichte und Biases werden anhand des Trainings-Loss gelernt
- Training: Backpropagation liefert Gradienten für Filter und Klassifikator, der Optimizer passt sie an
- Vorhersage: Gelernte Gewichte bleiben fest, kein Referenzlabel erforderlich

## Filter, Channels und Weight Sharing

| Begriff | Bedeutung |
|---|---|
| Input Channels | Eingangskanäle, am Anfang bspw. RGB |
| Filter | Gewichte zur Erkennung lokaler Muster |
| Output Channels | Anzahl erzeugter Feature Maps |
| Weight Sharing | Derselbe Filter wird an allen Bildpositionen verwendet |

- Ein räumlicher `3 × 3`-RGB-Filter verwendet `3 × 3 × 3` Gewichte plus einen Bias
- Ein Filter erzeugt eine Feature Map, mehrere Filter erzeugen mehrere Maps
- Spätere Channels entsprechen gelernten Merkmalen, sie sind keine neuen Farbkanäle
- Erkenntnis: Eine Feature Map enthält lokale Antworten, noch keine Wahrscheinlichkeit für eine Bildklasse

## Stride und Padding

| Einstellung | Wirkung |
|---|---|
| Stride | Schrittweite beim Verschieben des Filters |
| Grösserer Stride | Weniger ausgewertete Positionen und kleinere räumliche Ausgabe |
| Padding | Zusätzlicher Rand, bspw. mit Nullen |
| Passendes Padding | Kann die räumliche Grösse erhalten |

### Bsp. bei 64 × 64 Eingabe und 5 × 5 Filter

| Einstellung | Ausgabe pro Filter |
|---|---|
| Stride 1, kein Padding | 60 × 60 |
| Stride 1, Padding 2 | 64 × 64 |
| Stride 2, kein Padding | 30 × 30 |

- Format pro Raumrichtung: `abrunden((Eingabe + 2 × Padding − Filtergrösse) / Stride) + 1`

## Warum CNNs Parameter sparen

- Dichte Schicht: Eigene Gewichte für jede Eingabe-Ausgabe-Verbindung
- Conv-Schicht: Kleine lokale Filter, deren Gewichte über Positionen geteilt werden
- Format: `Filteranzahl × (Filterhöhe × Filterbreite × Eingangskanäle + 1)`
- Bsp. sechs RGB-Filter mit `5 × 5` benötigen `6 × (5 × 5 × 3 + 1) = 456` Parameter
- Eine dichte Schicht mit derselben Ausgabegrösse `60 × 60 × 6` hätte 265'442'400 Parameter
- Achtung: Weniger Parameter bedeuten nicht gleich wenig Rechenaufwand, jeder Filter wird an vielen Positionen angewendet
- Erkenntnis: Gleiche Ausgabeform garantiert weder gleiche Genauigkeit noch gleiche Ausdrucksfähigkeit

## Pooling

- Definition: Benachbarte Aktivierungen einer Feature Map zusammenfassen
- Max Pooling: Pro Fenster den grössten Wert übernehmen
- Bsp. `2 × 2`-Fenster mit Stride 2 machen aus `60 × 60 × 6` genau `30 × 30 × 6`
- Parameter: Max Pooling hat keine trainierbaren Gewichte
- Vorteil: Weniger Werte und geringere Empfindlichkeit gegenüber manchen kleinen Positionsänderungen
- Nachteil: Schwächere Aktivierungen und genaue Positionsinformationen gehen verloren
- Achtung: Pooling garantiert keine vollständige Verschiebungsinvarianz

## CNN-Aufbau und Softmax

### Beispielarchitektur für EuroSAT

| Schritt | Ausgabe |
|---|---|
| RGB-Bild | 64 × 64 × 3 |
| Convolution und ReLU | 60 × 60 × 6 |
| Max Pooling | 30 × 30 × 6 |
| Weitere Convolution und ReLU | 30 × 30 × 16 |
| Max Pooling | 15 × 15 × 16 |
| Global Average Pooling | Ein Mittelwert pro Map, insgesamt 16 Werte |
| Klassifikationskopf | Zehn rohe Klassenscores |
| Softmax | Zehn Wahrscheinlichkeiten mit Summe 1 |

- EuroSAT: 27'000 beschriftete RGB-Kacheln in zehn Landbedeckungsklassen, bspw. Wald und Strasse
- Softmax-Format: `pₖ = exp(zₖ) / Summe aller exp(z)`
- Interpretation: Ein Klassenscore wird relativ zu den anderen Klassenscores bewertet
- Bsp. Waldscore 2, Strassenscore 1 und acht weitere Scores 0 ergeben ungefähr 40,8 %, 15,0 % und je 5,5 %
- Binärer Zusammenhang: Zwei-Klassen-Softmax entspricht Sigmoid der Differenz beider Scores
- Achtung: Die gezeigten Trainingsverfahren berechnen den Loss direkt aus Logits, dort vorher keine zusätzliche Sigmoid- oder Softmax-Umwandlung anwenden
- Erkenntnis: Die Architekturzeichnung allein belegt keine gute Modellleistung

## CNNs bewerten und Bildantworten interpretieren

- Baseline: Einfache Vergleichsmethode, bei EuroSAT liegt die Mehrheitsbaseline laut Folien bei 11,1 %
- Bewertung: Neben Accuracy auch Macro-F1, Recall pro Klasse und Confusion Matrix betrachten
- Macro-F1: F1 pro Klasse berechnen und gleichgewichtet mitteln, seltene Klassen zählen gleich stark
- Bsp. Wald gegen übrige Klassen mit `TP = 18`, `FN = 2` und `FP = 9` ergibt Recall 90 % und Precision ungefähr 67 %
- Geografischer Holdout: Räumlich getrennte Gebiete zum Testen verwenden, benachbarte Kacheln können stark ähnlich sein
- Verschiebungsäquivarianz: Verschiebt sich das Bild, verschiebt sich die Filterantwort im passenden Fall mit
- Grenzen: Ränder, Stride und Pooling können die genaue Entsprechung verändern
- Achtung: Eine Kante ist kein Beweis für eine Strasse, auch Feldgrenzen erzeugen Kanten
- Geschäftlicher Einsatz: Unsichere Fälle prüfen lassen, Fehlerkosten berücksichtigen und auf geeigneten Testdaten bewerten

## Learned Representations und Embeddings

- Definition: Gelernte Repräsentationen wandeln Rohdaten in Merkmale um, die für ein Trainingsziel nützlich sind
- Zweck: Weniger manuelle Feature-Entwicklung, kompaktere Darstellung und Wiederverwendung für weitere Aufgaben
- Embedding: Gelernter Zahlenvektor, bspw. eine kompakte Bild- oder Textrepräsentation
- Verwendung: Klassifikation, Clustering, Anomalieerkennung oder weitere Modelle
- Achtung: Einzelne Dimensionen besitzen nicht zwingend verständliche Bedeutungen
- Erkenntnis: Eine gelernte Repräsentation ist nur dann hilfreich, wenn sie relevante Informationen für die spätere Aufgabe erhält

## Autoencoder

- Definition: Netz, das eine Eingabe über eine Zwischenrepräsentation möglichst gut rekonstruiert
- Format: `Eingabe → Encoder → Code z → Decoder → Rekonstruktion`
- Encoder: Verdichtet bzw. transformiert die Eingabe
- Code bzw. Bottleneck: Gelernte Zwischenrepräsentation
- Decoder: Erzeugt daraus eine Rekonstruktion
- Bsp. `10 Eingabewerte → 24 → 3 Codewerte → 24 → 10 rekonstruierte Werte`
- Trainingsziel: Bei einem gewöhnlichen Autoencoder ist die Eingabe selbst das Ziel
- Einordnung: Keine externen Klassenlabels erforderlich, das Ziel wird aus den Daten erzeugt
- Erkenntnis: Trainiert wird die Rekonstruktion, nicht automatisch eine geschäftlich sinnvolle Gruppierung
- Achtung: Auch ein kleiner Code kann overfitten, auf neuen Beobachtungen prüfen

## Denoising und Rekonstruktionsfehler

| Anwendung | Eingabe | Trainingsziel |
|---|---|---|
| Gewöhnlicher Autoencoder | Originaldaten | Dieselben Originaldaten rekonstruieren |
| Denoising | Verrauschte Daten | Saubere Referenz rekonstruieren |
| Anomalieerkennung | Häufig überwiegend normale Daten | Normale Struktur gut rekonstruieren |
| Generierung mit VAE | Trainingsbeispiele | Strukturierte latente Verteilung lernen |

- MSE: Quadrierte Rekonstruktionsabweichungen mitteln, grosse Fehler zählen stärker
- Alternativen: MAE für kontinuierliche Werte oder BCE für binäre Ziele
- Anomaliehinweis: Ungewöhnlich hoher Rekonstruktionsfehler kann auf eine Abweichung hindeuten
- Achtung: Hoher Fehler beweist keine Anomalie, niedriger Fehler garantiert keine fachlich richtige Repräsentation
- Informationsverlust: Erhalten zwei Eingaben denselben Code, kann der Decoder sie anhand dieses Codes nicht unterscheiden
- Erkenntnis: Rekonstruktion ist normalerweise näherungsweise, exakte Wiederherstellung ist nicht garantiert

## ResNet-18 und Residual Blocks

- Definition: CNN mit 18 gewichtstragenden Schichten und Residual-Verbindungen
- Residual Block: Lernt eine Veränderung `F(x)` und addiert die ursprüngliche Eingabe
- Format: `Output = F(x) + x`
- Shortcut: Führt die Eingabe an den lernenden Schichten vorbei
- Zweck: Gradienten erhalten einen direkteren Weg, das Training tiefer Netze wird erleichtert
- Folienmodell: Auf ImageNet zur Klassifikation vortrainiert, nicht zur Rekonstruktion

## Transfer Learning

- Definition: Wissen bzw. Merkmale eines bereits trainierten Modells für eine neue Aufgabe wiederverwenden
- Zweck: Mit vorhandenen Repräsentationen starten, statt alle Merkmale neu zu lernen

| Vorgehen | Was wird trainiert? |
|---|---|
| Von Grund auf | Gewichte des neuen Modells |
| Frozen Feature Extraction | Nur der neue Klassifikationskopf |
| Fine-Tuning | Neuer Kopf und zusätzlich Gewichte des vortrainierten Encoders |

- Bsp. ResNet liefert 512 Bildmerkmale, ein neuer Kopf ordnet diese fünf Kleidungsklassen zu
- Trainierbare Parameter des neuen Kopfs: `512 × 5 + 5 = 2565`
- Erkenntnis: Frozen bedeutet festgehaltene Gewichte, nicht ein untrainiertes Modell
- Achtung: Vortraining hilft nur, wenn die Merkmale ausreichend zur neuen Aufgabe passen

## Fashion-MNIST und Übertragbarkeit

- Datensatz: 70'000 Graustufenbilder mit `28 × 28` Pixeln und zehn Kleidungsklassen, im Beispiel fünf Klassen
- Anpassung: Graustufenkanal dreimal wiederholen, Resize, zentraler Zuschnitt und passende ImageNet-Normalisierung
- Erkenntnis: Drei identische Kanäle erzeugen keine Farbinformation, Vergrössern erzeugt keine neuen Details
- Achtung: Zuschneiden kann Randinformationen entfernen
- Ablauf: Je 5000 Bilder für Training, Validierung und Test, Checkpoints und Repräsentation auf Validierung auswählen

| Verfahren | Test-Accuracy |
|---|---|
| Immer dieselbe Klasse bei fünf gleich grossen Klassen | 20,00 % |
| Klassifikator auf Rohpixeln | 93,56 % |
| Klassifikator auf Frozen-ResNet-Merkmalen | 96,00 % |

- Interpretation: Verbesserung um 2,44 Prozentpunkte gegenüber Rohpixeln
- Achtung: Kein Fine-Tuning-Ergebnis, auch Vorverarbeitung und externes Vortraining unterscheiden sich
- Domain Shift: Echte Katalogfotos können andere Hintergründe, Farben und Produkte enthalten
- Geschäftlicher Nutzen: Repräsentative Bilder, Fehlerarten, manuelle Nacharbeit, Laufzeit und Kosten prüfen

# V8: Foundations of LLMs

## Textrepräsentationen im Überblick

- Zweck: Text in numerische Eingaben für Modelle umwandeln
- Entwicklung: Von manuell definierten Regeln und Wortzählungen zu gelernten und kontextabhängigen Repräsentationen
- Erkenntnis: Die Darstellung bestimmt, welche Informationen das nachfolgende Modell nutzen kann

| Verfahren | Was wird dargestellt? | Zentrale Grenze |
|---|---|---|
| Manuelle Features | Ausgewählte Eigenschaften wie URLs oder Textlänge | Menschen müssen passende Merkmale festlegen |
| One-Hot | Identität eines Worts | Keine gelernte semantische Nähe |
| TF-IDF | Gewichtete Wortvorkommen eines Dokuments | Verwandte Wörter bleiben getrennte Koordinaten |
| word2vec | Gelernte Verwendung eines Worts in Kontexten | Ein fester Vektor pro bekanntem Wort |
| Kontextuelle Embeddings | Wort bzw. Token im aktuellen Text | Nützlichkeit hängt vom Modell und Trainingsziel ab |

## WordNet und One-Hot

- WordNet: Menschlich gepflegte Datenbank mit Wortbedeutungen und Beziehungen
- Synset: Gruppe von Wörtern derselben Bedeutung, bspw. car und automobile
- Grenze: Neue Begriffe und Bedeutungen benötigen manuelle Pflege
- One-Hot: Festes Vokabular, pro Wort eine 1 an seiner Position und sonst 0
- Erkenntnis: Zahlen allein bedeuten noch keine semantische Ähnlichkeit

## Entwicklung der Textmodelle

| Meilenstein | Grundidee |
|---|---|
| LSA, 1990 | Latente Richtungen aus einer Term-Dokument-Matrix, kein neuronales Verfahren |
| word2vec, 2013 | Wortvektoren durch Kontextvorhersage lernen |
| Transformer, 2017 | Verarbeitung mit Self-Attention |
| BERT und GPT, ab 2018 | Kontextuelle Repräsentationen bzw. generative Sprachmodelle |
| Sentence-BERT, 2019 | Satzvektoren für Ähnlichkeitsvergleiche trainieren |

### Historischer Spam-Vergleich

| Vollständiges System | Precision | Recall |
|---|---|---|
| Manuelle Features plus MLP | 92,3 % | 53,3 % |
| Kontextueller Encoder plus MLP | 98,2 % | 85,4 % |

- Vergleich: Dieselben 6000 Testmails, Schwellenwerte vorher auf Validierungsdaten gewählt
- Achtung: Das ist kein gemessener Vergleich von TF-IDF gegen word2vec und kein isolierter Beweis für eine einzelne Modelländerung

## TF-IDF

- Definition: Ein Dokument wird durch die gewichteten Vorkommen seiner Wörter dargestellt
- Zweck: Wörter hervorheben, die im Dokument vorkommen und im Gesamtkorpus vergleichsweise selten sind
- Input: Trainingsdokumente zur Bestimmung von Vokabular und IDF, danach weitere Dokumente mit derselben Darstellung

| Begriff | Bedeutung |
|---|---|
| TF | Häufigkeit eines Begriffs im betrachteten Dokument |
| DF | Anzahl Trainingsdokumente, in denen der Begriff vorkommt |
| IDF | Gewichtung anhand der Seltenheit im Trainingskorpus |
| TF-IDF | TF mit IDF multiplizieren |

- Gezeigte Formel: `IDF = ln((1 + N) / (1 + DF)) + 1`
- N: Anzahl Trainingsdokumente
- Normalisierung: Im Folienbeispiel den gewichteten Dokumentvektor anschliessend auf Länge 1 bringen
- Bsp. claim prize prize enthält claim einmal und prize zweimal
- Neue Dokumente: Trainingsvokabular und IDF wiederverwenden, nicht anhand der Testdaten neu bestimmen
- Grenze: prize und reward bleiben verschiedene Koordinaten, obwohl sie ähnliche Bedeutungen haben können

## word2vec

- Definition: Verfahren, das Wortvektoren anhand beobachteter Wortumgebungen lernt
- Grundidee: Wörter mit ähnlichen Kontexten können ähnliche Verwendungen haben
- Trainingsziel im Beispiel: Aus einem zentralen Wort ein beobachtetes Kontextwort vorhersagen
- Ablauf: `Wort → Embedding → Kontextscores → Softmax → Loss`
- Lernen: Eingabe- und Ausgabegewichte werden angepasst, nicht nur ein separater Klassifikator
- Achtung: word2vec lernt Wortkontexte, keine Spam-Labels
- Verwendung: Wortvektoren bspw. zu einem E-Mail-Vektor mitteln und darauf einen eigenen Spam-Klassifikator trainieren
- Bsp. Mittelwert von `[0, 1]` und `[1, 2]` ergibt `[0,5, 1,5]`, die Dimension bleibt 2
- Erkenntnis: Ähnliche Kontexte sind ein Hinweis auf verwandte Verwendung, kein Beweis für Synonymie

## Skalarprodukt und Cosine Similarity

| Begriff | Erklärung |
|---|---|
| Skalarprodukt | Entsprechende Einträge multiplizieren und addieren |
| Vektorlänge | Grösse des Vektors |
| Cosine Similarity | Skalarprodukt geteilt durch beide Vektorlängen |
| Format | `cos(a, b) = (a · b) / (Länge(a) × Länge(b))` |

### Bsp.

`[1, 2] · [0, 1] = 2`

`Cosine = 2 / √5 ≈ 0,894`

- Skalarprodukt: Hängt von Richtung und Länge ab
- Cosine: Vergleicht Richtungen nach Längennormalisierung
- Wertebereich: Zwischen −1 und 1 für Vektoren mit Länge grösser als 0
- Normierte Vektoren: Bei Länge 1 sind Skalarprodukt und Cosine gleich
- Achtung: Cosine ist keine Klassenwahrscheinlichkeit und kein Beweis für Wahrheit oder Eignung
- word2vec: Verwendet beim Training rohe Skalarprodukte als Kontextscores
- Visualisierung: Ein 2D-Plot kann viel Information verlieren, im Folienbeispiel bleiben nur 32,1 % erhalten
- Erkenntnis: Ähnlichkeiten möglichst im ursprünglichen Vektorraum beurteilen

## Grenzen statischer Embeddings und Mittelwerte

- Statisch: Ein bekanntes Wort erhält unabhängig vom aktuellen Satz denselben Vektor
- Bsp. bank hat bei Geldinstitut und Flussufer denselben word2vec-Vektor
- Mean Pooling statischer Wortvektoren: Ignoriert die Reihenfolge der Wörter
- Bsp. your bank paid the supplier und the supplier paid your bank enthalten dieselben Wörter, aber vertauschte Zahlungsrollen
- Ergebnis: Identische statische Mittelwerte trotz anderer Bedeutung
- Erkenntnis: Ein nachgelagerter fester Klassifikator kann bereits verlorene Information nicht zurückholen
- Kontextuelle Embeddings: Verwenden den aktuellen Text, um Tokenrepräsentationen anzupassen

## Tokens und Tokenisierung

- Token: Verarbeitungseinheit eines Textmodells, bspw. Wort, Wortstück oder Satzzeichen
- Subwords: Wiederverwendbare Fragmente, aus denen auch lange oder unbekannte Wörter zusammengesetzt werden können
- Token-ID: Nummer eines Vokabulareintrags
- Position: Stelle des Tokens innerhalb des aktuellen Textes
- Achtung: Token-ID und Position sind unterschiedliche Informationen
- Erkenntnis: Tokenisierung ermöglicht Verarbeitung, sie ist noch kein Sprachverständnis

## Embedding-Dimension und Positionsinformation

- Embedding: Numerischer Vektor zur Darstellung eines Tokens oder Texts
- Embedding-Dimension d: Anzahl Zahlen pro Vektor
- Bei T Tokens entsteht eine Matrix mit `T × d` Werten
- Mehr Tokens: Mehr Zeilen
- Grössere Dimension: Mehr Werte pro Token
- Lookup: Token-ID zunächst einem gelernten Vektor zuordnen
- Positionsinformation: Macht die Reihenfolge für das Modell verfügbar
- Achtung: Mehr Dimensionen garantieren keine bessere Bedeutungserfassung

## Self-Attention

- Definition: Eine Tokenposition aktualisiert ihre Darstellung anhand der Informationen anderer Positionen derselben Eingabe
- Zweck: Kontextabhängige Beziehungen zwischen Tokens berücksichtigen
- Ablauf: Passung bewerten, Scores mit Softmax in Gewichte umwandeln und Informationsvektoren gewichtet mischen
- Selbstbezug: Auch die eigene Position kann berücksichtigt werden
- Attention Head: Lernt eigene Parameter zur Bewertung und Mischung von Informationen
- Achtung: Die Gewichte beziehen sich auf Tokenpositionen, nicht auf Jobkategorien

### Bsp.

| Gewicht | Informationsvektor |
|---|---|
| 0,665 | [1, 0] |
| 0,245 | [0, 1] |
| 0,090 | [1, 1] |

Gewichtete Summe: ungefähr `[0,755, 0,335]`

- Erkenntnis: Attention erzeugt eine neue Repräsentation durch Informationsmischung
- Achtung: Attention ist ein Baustein, noch kein vollständiges Modell

## Transformer

- Definition: Architektur, die Attention mit weiteren Bausteinen wie tokenweisen neuronalen Netzen kombiniert
- Zweck: Repräsentationen über mehrere Schichten kontextabhängig verfeinern
- Reihenfolge: Positionsinformationen machen die Textreihenfolge verfügbar
- Parallelisierung: Positionen können innerhalb einer Schicht beim Training parallel verarbeitet werden
- Grenze: Aufeinanderfolgende Schichten bleiben abhängig, autoregressive Generierung erfolgt weiterhin Token für Token
- Erkenntnis: Architektur beschreibt den Aufbau, ein trainiertes Modell benötigt zusätzlich gelernte Gewichte

## BERT

- Bedeutung: Bidirectional Encoder Representations from Transformers
- Aufbau: Encoder-only-Transformer
- Bidirektional: Verwendet Kontext vor und nach einer Tokenposition
- Folienmodell: Zwölf Schichten, zwölf Attention Heads pro Schicht und 768 Werte pro Token
- Zweck: Kontextuelle Textrepräsentationen erzeugen

### Masked Language Modeling

- Beim Vortraining werden Tokens verdeckt und aus dem Kontext vorhergesagt
- Trainingsziel ist die ursprüngliche Token-ID
- Einordnung: Selbstüberwacht, die Zielwerte stammen aus dem Text selbst
- Das verwendete mBERT wurde auf Wikipedia in 104 Sprachen vortrainiert
- Achtung: Dafür werden keine Jobkategorie-Labels benötigt
- Ergänzung: Das ursprüngliche BERT-Vortraining umfasst auch Satzpaarinformationen

## Mean Pooling und Textklassifikation mit Frozen BERT

- Input: Tokenisierter Text
- Encoder: Frozen BERT erzeugt einen kontextuellen Vektor pro Token
- Pooling: Inhaltsvektoren mitteln, Padding und spezielle Markierungen ausschliessen
- Output: Ein Vektor mit 768 Werten pro Inserat
- Klassifikator: Lernt anhand dieses Vektors und der Joblabels die Kategorien
- Frozen: Encodergewichte bleiben fest
- Fine-Tuning: Würde zusätzlich Encodergewichte verändern
- Erkenntnis: BERT liefert hier Merkmale, keine fertigen Jobkategorie-Wahrscheinlichkeiten
- Achtung: Pooling ändert die Form der Darstellung, nicht das ursprüngliche Trainingsziel

## Klassifikationsexperiment mit Stelleninseraten

- Datengrundlage: Aus 30'000 gelieferten Inseraten entsteht nach Auswahl und Prüfung eine Kohorte mit 5011 Inseraten
- Aufteilung: 3328 Training, 918 Validierung und 765 Test, zeitlich geordnet
- Trainingsklassen: 139 Jobkategorien
- Vergleich: Häufigkeitsbaseline, TF-IDF und Frozen mBERT
- Fairness: Dieselbe Kohorte und für TF-IDF und BERT derselbe gekürzte Text
- Eingabegrenze: 128 Tokens inkl. spezieller Tokens, höchstens 126 Inhaltstokens
- Kürzung: 757 von 765 Testinseraten werden gekürzt, wichtige Informationen können fehlen
- Unbekannte Testlabels: Werden mitbewertet, können aber von einem Klassifikator mit festen Trainingsklassen nicht als neue Kategorie ausgegeben werden

| Modell | Accuracy | Macro-F1 | Top-3-Trefferquote |
|---|---|---|---|
| Häufigkeitsbaseline | 4,6 % | 0,001 | 12,9 % |
| TF-IDF | 24,7 % | 0,195 | 37,9 % |
| Frozen mBERT | 31,6 % | 0,235 | 47,6 % |

- Accuracy: Ist die erste Kategorie richtig?
- Macro-F1: Gleichgewichteter Mittelwert der F1-Werte pro Kategorie
- Top-3-Trefferquote: Ist das richtige Label unter den ersten drei Vorschlägen?
- Interpretation: BERT verbessert die Ergebnisse, trotzdem fehlt das richtige Label bei 401 von 765 Testinseraten in den ersten drei Vorschlägen
- Erkenntnis: Ein besseres Modell kann absolut betrachtet weiterhin häufig unbrauchbare Vorschläge liefern
- Achtung: Das einzelne gezeigte Security-Inserat ist eine Trainingsillustration, kein eigener Testnachweis

## Original Transformer, BERT und GPT

| Modelltyp | Aufbau | Kontext und Aufgabe |
|---|---|---|
| Original Transformer | Encoder und Decoder mit Cross-Attention | Quelltext repräsentieren und Zieltext generieren, bspw. Übersetzung |
| BERT | Encoder-only | Kontext auf beiden Seiten nutzen und Text repräsentieren |
| GPT-artig | Causaler Decoder | Bisherigen Präfix lesen und nächste Tokens generieren |

- Erkenntnis: GPT ist ein separates Modell, kein fehlender Decoder, der an BERT angehängt wird

## Causale Textgenerierung

- Causal: Bei einer Position nur den bisherigen Präfix verwenden, keine zukünftigen Tokens
- Ablauf: Präfix lesen, Verteilung über das nächste Token berechnen, Token auswählen, anhängen und wiederholen
- Trainingsziel: Die nächsten Tokens stehen bereits im Trainingstext
- Output: Schrittweise erzeugte Textfortsetzung
- Achtung: Wahrscheinliche Fortsetzung bedeutet nicht automatisch sachlich richtige Aussage

## Unterschiedliche Trainingsziele

| Aufgabe | Was soll gelernt werden? |
|---|---|
| BERT-Vortraining | Verdeckte Tokens aus dem Kontext vorhersagen |
| GPT-artiges Vortraining | Nächstes Token aus dem Präfix vorhersagen |
| Jobklassifikation | Vorgegebene Jobkategorien vorhersagen |
| Ähnlichkeitslernen | Verwandte bzw. relevante Texte passend zueinander darstellen |
| Retrieval | Für eine Anfrage relevante Dokumente hoch einstufen |

- Erkenntnis: Ähnliche Architekturbausteine bedeuten nicht dieselbe Eignung für jede Aufgabe

## Sentence-BERT und Satzembeddings

- Definition: Satzrepräsentationen, die anhand von Textpaaren für Ähnlichkeitsvergleiche trainiert werden
- Zweck: Ganze Sätze oder Abschnitte als vergleichbare Vektoren darstellen
- Gezeigtes Training: Zwei Textzweige verwenden gemeinsame Encodergewichte, Encoder und Trainingsklassifikator werden angepasst
- Danach: Trainingsklassifikator kann entfallen, Textvektoren werden bspw. mit Cosine verglichen
- Vorteil: Dokumentvektoren können vorberechnet und für viele Anfragen wiederverwendet werden
- Vergleich: Gemeinsames BERT-Scoring müsste für jedes Anfrage-Dokument-Paar erneut berechnet werden
- Achtung: Plain BERT plus Mittelwert ist nicht automatisch ein für Ähnlichkeit trainierter Satzencoder
- Merksatz: Pooling verändert die Form, Ähnlichkeitstraining verändert die gelernte Geometrie

## E5-small und semantische Suche

- Definition: Vortrainierter Textencoder, der für Retrieval angepasst wurde
- Zweck: Passende Anfragen und Dokumente näher zusammenbringen als Ablenker
- Output: 384 Werte pro Text, im Beispiel auf Länge 1 normalisiert
- Cosine: Bei diesen normierten Vektoren gleich dem Skalarprodukt
- Kursdemo: Fertiges E5 wird frozen verwendet, vorangegangenes Retrieval-Training ist bereits erfolgt

| Begriff | Bedeutung |
|---|---|
| SentenceTransformers | Toolkit für Satz- und Textencoder |
| Distillation | Kleineres Modell lernt von einem grösseren Lehrermodell |
| Kontrastives Lernen | Passende Paare gegenüber unpassenden bevorzugen |
| Asymmetrische Suche | Anfrage und Dokument haben unterschiedliche Rollen, bspw. kurz gegenüber lang |

- Modellrezept: Anfrage mit query: und Dokument mit passage: kennzeichnen
- Eingabebudget: 512 Tokens, im Folienbeispiel 508 Inserat-Tokens plus Präfix und spezielle Tokens
- Achtung: Längenbegrenzungen und Kürzung prüfen, fehlender Text kann wichtige Anforderungen enthalten
- Entwicklungslinie: BERT, RoBERTa, XLM-R und MiniLM beschreiben Beziehungen zwischen Modellen, nicht überall unverändert kopierte Gewichte
- Erkenntnis: Auch ein Retrieval-Modell kann irrelevante Treffer hoch einstufen

## BM25: Keyword-Suche

- Definition: Nicht neuronales Rankingverfahren für genaue Worttreffer
- Zweck: Dokumente anhand ihrer Übereinstimmung mit Suchbegriffen gewichten
- Anwendung: Exakte Technologien und Begriffe wie C# oder ABAP können entscheidend sein

| Bestandteil | Wirkung |
|---|---|
| IDF | Seltene Begriffe erhalten mehr Gewicht |
| TF-Sättigung | Zusätzliche Nennungen helfen mit abnehmendem Zusatznutzen |
| Längennormalisierung | Dokumentlänge relativ zur mittleren Länge berücksichtigen |

- Grenze: Gleiche Wörter beweisen keine gleiche Tätigkeit
- Bsp. Daten abfragen ist nicht automatisch Datenpipelines entwickeln
- Erkenntnis: Keyword- und Embedding-Suche liefern unterschiedliche, potenziell ergänzende Signale

## Hybride Suche und Reciprocal Rank Fusion

- Hybride Suche: Ergebnisse verschiedener Suchverfahren kombinieren
- RRF: Rangpositionen zusammenführen, statt rohe Scores unterschiedlicher Skalen zu addieren
- Format im Beispiel: Beitrag je Liste `1 / (60 + Rang)`
- Fehlender Treffer: Beitrag 0
- Bsp. Rang 1 in beiden Listen ergibt `1/61 + 1/61 ≈ 0,03279`
- Vorteil: Unterschiedlich skalierte Scores müssen nicht direkt vergleichbar gemacht werden
- Grenze: Gemeinsame Fehler können verstärkt werden
- Erkenntnis: Übereinstimmung zweier Verfahren ist keine Relevanzgarantie

## Suchqualität: Precision@K, Recall@K und MRR

| Kennzahl | Berechnung | Aussage |
|---|---|---|
| Precision@K | Relevante Treffer unter den ersten K geteilt durch K | Wie nützlich sind die angezeigten Plätze? |
| Recall@K | Gefundene relevante Dokumente unter den ersten K geteilt durch alle relevanten Dokumente | Wie viel Relevantes wurde gefunden? |
| RR@K | `1 / Rang des ersten relevanten Treffers`, innerhalb K, sonst 0 | Wie früh erscheint der erste relevante Treffer? |
| MRR@K | Mittelwert von RR@K über mehrere Anfragen | Wie früh erscheint im Mittel der erste relevante Treffer? |

### Bsp.

Drei angezeigte Treffer, einer davon relevant auf Rang 2, insgesamt zwei relevante Dokumente im Katalog

- `Precision@3 = 1/3`
- `Recall@3 = 1/2`
- `RR@3 = 1/2`

- Erkenntnis: Gleiche MRR bei unterschiedlichem Recall ist möglich, MRR berücksichtigt nur den ersten relevanten Treffer
- Unterschied zur Klassifikation: Top-3-Klassifikation sucht ein richtiges Klassenlabel, Retrieval kann mehrere relevante Dokumente pro Anfrage haben

## Ergebnisse der Suchdemo interpretieren

| Verfahren | Precision@3 | Recall@3 | MRR@3 |
|---|---|---|---|
| BM25 | 0,417 | 0,625 | 0,667 |
| Plain BERT | 0,417 | 0,625 | 0,812 |
| E5 | 0,583 | 0,875 | 0,938 |
| Hybrid RRF | 0,500 | 0,750 | 0,938 |

- Grundlage: Acht Anfragen, 24 reale Inseratausschnitte und vorläufige Relevanzlabels
- Interpretation: E5 hat hier die höchste Precision und den höchsten Recall, MRR ist mit Hybrid gleich hoch
- Hybrid gegenüber E5: Recall sechsmal gleich, zweimal schlechter und nie besser
- Schlussfolgerung: E5 weiter validieren, noch keine allgemeine Produktionsentscheidung
- Bsp. data engineer: E5 findet einen passenden Pipeline-Treffer auf Rang 2, der unpassende erste Treffer bleibt bestehen
- Erkenntnis: Bessere Gesamtergebnisse bedeuten nicht, dass jeder einzelne Treffer relevant ist

## Relevanz, Eignung und fairer Vergleich

- Relevanz zuerst definieren, bspw. data engineer als Entwicklung oder Betrieb von Pipelines, ETL oder ELT
- Mehrdeutige Anfragen und unklare Relevanzlabels prüfen, Kennzahlen hängen von diesen Urteilen ab
- Für faire Vergleiche Katalog, Textbasis, Einstellungen und K konstant halten
- Weitere Prüfung: Frische repräsentative Anfragen, überprüfte Labels, Latenz, Suchdauer und tatsächlicher Aufgabenerfolg
- Jobeignung: Ort, Schichten, Lizenzen und andere Anforderungen im vollständigen Inserat gesondert prüfen
- Fehlende Information: Bedeutet unbekannt, nicht erfüllt
- Erkenntnis: Hohe Ähnlichkeit liefert einen Kandidaten, noch keine abschliessende Eignungsentscheidung

## Bandits: Aus Entscheidungen und Feedback lernen

- Definition: Ein Multi-Armed Bandit wählt wiederholt zwischen Optionen mit unbekanntem erwarteten Reward
- Arm: Eine auswählbare Option, bspw. ein angezeigtes Inserat
- Reward: Beobachtetes Ergebnis, bspw. Klick 1 oder kein Klick 0
- Policy: Regel, die entscheidet, welche Option gewählt wird
- Besonderheit: Feedback entsteht nur für die gewählte Option
- Erkenntnis: Die Reaktion auf das nicht angezeigte Inserat bleibt unbekannt
- Achtung: Ein Klick ist keine objektiv richtige Antwort und kein Beweis für Jobeignung

## Exploration und Exploitation

| Begriff | Bedeutung |
|---|---|
| Exploration | Optionen ausprobieren, um mehr über sie zu lernen |
| Exploitation | Die aktuell am besten eingeschätzte Option nutzen |

- Problem: Frühe Zufallsergebnisse können die falsche Option als Spitzenreiter erscheinen lassen
- Bsp. zwei Klicks bei zehn Anzeigen ergeben beobachtete 20 %, aber keine sicher bekannte tatsächliche Klickwahrscheinlichkeit
- Erkenntnis: Ohne weiteres Ausprobieren kann eine bessere Option dauerhaft unterschätzt bleiben

## Epsilon-greedy

- Definition: Meist den aktuellen Spitzenreiter nutzen, gelegentlich zufällig explorieren
- Ablauf: Jede Option zunächst ausprobieren, danach mit Wahrscheinlichkeit ε zufällig wählen und sonst den aktuellen Spitzenreiter verwenden
- Bei `ε = 0,1`: Im Mittel ungefähr 10 % Exploration und 90 % Exploitation
- Achtung: Das ist keine feste Quote, Exploration kann zufällig auch den Spitzenreiter auswählen
- Vorteil: Weitere Informationen über bislang schlechter eingeschätzte Optionen sammeln
- Grenze: Exploration kostet möglicherweise kurzfristigen Reward und garantiert keine schnelle Korrektur

## Bandit, Baseline und A/B-Test

| Vorgehen | Verhalten bzw. Ziel |
|---|---|
| Immer A | Feste Baseline ohne Lernen |
| Zufällige Auswahl | Probiert Optionen aus, passt sich aber nicht an |
| A/B-Test | Geplanten Unterschied zwischen Varianten schätzen |
| Bandit | Während der Auswahl lernen und Entscheidungen zur Reward-Suche anpassen |

- Erkenntnis: A/B-Test und Bandit verfolgen unterschiedliche Ziele, keines ist automatisch besser

## Klicksimulation richtig interpretieren

- Annahmen: A hat 8 % und B 12 % Klickwahrscheinlichkeit
- Achtung: Diese Wahrscheinlichkeiten sind erfunden, keine aus echten Klicklogs geschätzten Werte
- Policy: Kennt die Raten nicht und lernt nur aus beobachteten Rückmeldungen
- Vergleich: 200 unabhängige Durchläufe mit je 2000 Besuchern und vorher festgelegten Einstellungen

| Regel | Mittlere simulierte Klickzahl |
|---|---|
| Immer A | 159,6 |
| Zufällig | 199,6 |
| Epsilon-greedy mit ε = 0,1 | 228,0 |

- Interpretation: Epsilon-greedy ist in dieser Simulation durchschnittlich besser
- Einzelner schlechter Lauf: Epsilon-greedy erzielt 162 gegenüber 206 Klicks bei zufälliger Auswahl
- Erkenntnis: Ein besserer Mittelwert garantiert keinen besseren einzelnen Durchlauf
- Achtung: Mehr simulierte Klicks beweisen keine besseren realen Jobmatches, Bewerbungen oder Einstellungen

## Sequenzielles Reinforcement Learning

| Begriff | Bedeutung |
|---|---|
| Agent | Entscheidendes System |
| State | Relevante aktuelle Situation |
| Action | Gewählte Handlung |
| Reward | Rückmeldung zur Handlung |
| Policy | Regel zur Auswahl einer Handlung |

- Bandit: Einfacher RL-Fall mit wiederholten Entscheidungen und unmittelbarem Feedback
- Sequenzielles RL: Eine Aktion beeinflusst zusätzlich spätere Situationen und Entscheidungsmöglichkeiten
- Bsp. Ein Jobassistent fragt nach dem gewünschten Arbeitsort, die Antwort verbessert den Informationsstand für die nächste Empfehlung
- Abwägung: Sofort empfehlen ist schneller, Rückfragen können später passendere Ergebnisse ermöglichen
- Erkenntnis: Die gesamte Interaktion bewerten, nicht nur den ersten Klick
- Achtung: Das Folienbeispiel ist ein hypothetischer Ablauf, kein bereits trainierter RL-Agent

## RLHF: Reinforcement Learning from Human Feedback

- Definition: Menschliche Präferenzen werden verwendet, um ein Sprachmodell weiter zu verbessern
- Klassischer Ablauf: Aus Text und Beispielantworten lernen, Antworten vergleichen lassen, ein Reward-Modell trainieren und damit das Sprachmodell weitertrainieren

| Phase | Was geschieht? |
|---|---|
| Sprachvortraining | Textmuster lernen, bspw. nächste Tokens |
| Lernen aus Beispielantworten | Gewünschte Antworten als Trainingsziele verwenden |
| Präferenzen sammeln | Menschen vergleichen Antworten |
| Reward-Modell trainieren | Separates Modell lernt aus Vergleichen einen Bewertungsscore |
| Policy verbessern | Sprachmodell wird anhand dieser Scores weitertrainiert |

- Einordnung: Das Reward-Modell lernt überwacht, das anschliessende Policy-Training verwendet Rewards
- Achtung: Nicht jede Trainingsphase ist Reinforcement Learning und nicht jedes Sprachmodell verwendet genau diesen Ablauf
- Prompt-Beispiele: Verändern normalerweise den aktuellen Kontext, nicht die Modellgewichte
- Erkenntnis: Eine bevorzugte Antwort kann sachlich falsch sein, Präferenz ist keine Wahrheitsgarantie

## Reward und Geschäftsziel

- Problem: Ein guter Lernalgorithmus kann das falsche Ziel optimieren
- Bsp. Mehr Klicks bedeuten nicht automatisch passendere Stellen oder erfolgreichere Bewerbungen
- Bsp. Eine überzeugend formulierte Antwort kann bevorzugt werden und trotzdem falsch sein
- Vorgehen: Geschäftsziel und Reward unterscheiden, Baselines vergleichen, Fehler untersuchen und tatsächliche Ergebnisse überwachen
- Erkenntnis: Gute Kennzahlen oder hoher Reward ersetzen weder fachliche Prüfung noch menschliche Verantwortung

## Zentrale Merksätze für Verständnisfragen

| Aussage | Einordnung |
|---|---|
| Mehr Schichten ergeben automatisch ein besseres Modell | Falsch, Generalisierung und Validierungsleistung entscheiden |
| Backpropagation verändert die Gewichte | Sie berechnet Gradienten, der Optimizer verändert Parameter |
| Eine Feature Map ist eine Klassenwahrscheinlichkeit | Falsch, sie enthält Filterantworten |
| Pooling hat trainierbare Gewichte | Gewöhnliches Max Pooling nicht |
| Frozen bedeutet untrainiert | Falsch, vortrainierte Gewichte bleiben unverändert |
| Drei Graustufenkanäle erzeugen Farbinformation | Falsch, die Kanäle enthalten dieselbe Information |
| Ähnliche Embeddings beweisen gleiche Bedeutung | Falsch, Ähnlichkeit hängt vom Trainingsziel und Kontext ab |
| Mean Pooling macht Plain BERT automatisch zum Suchmodell | Falsch, Pooling verändert nicht das Trainingsziel |
| GPT nutzt bei der Vorhersage zukünftige Tokens | Falsch, causale Vorhersage verwendet nur den Präfix |
| Hybrid ist immer besser als ein einzelnes Suchverfahren | Falsch, Fusion kann Fehler verstärken |
| Gleiche MRR bedeutet gleichen Recall | Falsch, weitere relevante Treffer können unterschiedlich sein |
| Der frühe Spitzenreiter eines Bandits ist sicher die beste Option | Falsch, wenige Beobachtungen können täuschen |
| Mehr Reward bedeutet automatisch mehr Geschäftsnutzen | Falsch, das optimierte Signal kann vom eigentlichen Ziel abweichen |
| Bevorzugte LLM-Antwort bedeutet richtige Antwort | Falsch, menschliche Präferenz garantiert keine Wahrheit |
