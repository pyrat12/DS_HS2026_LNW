## Machine Learning
Folien 4 bis 5

- Definition: Ein Programm lernt aus Erfahrungen, wenn sich seine Leistung bei einer bestimmten Aufgabe dadurch verbessert
- Zweck: Regeln aus vorhandenen Daten lernen und auf neue Beobachtungen anwenden
- Erfahrung E: Daten bzw. Beispiele, aus denen gelernt wird
- Aufgabe T: Problem, das gelöst werden soll
- Leistung P: Kennzahl zur Beurteilung der Ergebnisse
- Bsp. Aus bisherigen Wohnungen den Zusammenhang zwischen Eigenschaften und Mietpreis lernen
- Erkenntnis: Entscheidend ist die Leistung bei neuen Daten, nicht nur bei bekannten Beispielen

### Grundbegriffe

| Begriff | Bedeutung | Bsp. |
|---|---|---|
| Beobachtung | Einzelner Datensatz bzw. eine Zeile | Eine Wohnung |
| Feature x | Eingangsmerkmal für das Modell | Fläche und Zimmerzahl |
| Target y | Vorherzusagende Zielvariable | Mietpreis |
| Prediction ŷ | Vom Modell geschätzter Zielwert | Vorhergesagter Mietpreis |
| Modell | Gelernte Zuordnung von Eingaben zu Ausgaben | Zusammenhang zwischen Fläche und Preis |
| Generalisierung | Übertragung auf neue Beobachtungen | Preis einer bisher unbekannten Wohnung schätzen |

- Anwendungen: Preisprognosen, Betrugserkennung, Wartung, Empfehlungen, Kundensegmentierung und Bilderkennung
- Achtung: Nicht jede Anwendung benötigt dieselbe Lernmethode oder Bewertungskennzahl
- Erkenntnis: Fragestellung und Zielvariable bestimmen die Auswahl des Verfahrens

## Drei Lernparadigmen
Folie 6

| Paradigma | Datengrundlage | Ziel | Bsp. |
|---|---|---|---|
| Supervised Learning | Features und bekannte Zielwerte | Zielwerte neuer Beobachtungen vorhersagen | Mietpreis oder Betrug vorhersagen |
| Unsupervised Learning | Features ohne vorgegebene Zielwerte | Gruppen und Strukturen finden | Kundensegmente erkennen |
| Reinforcement Learning | Aktionen und Rückmeldungen aus einer Umgebung | Strategie mit hoher langfristiger Belohnung lernen | Spielstrategie oder Robotersteuerung |

- Supervised: Lernt anhand bekannter Eingaben und Antworten
- Unsupervised: Findet Muster ohne vorgegebene richtige Antwort
- Reinforcement: Lernt durch Handlungen und deren Konsequenzen
- Erkenntnis: Die Art der verfügbaren Rückmeldung unterscheidet die Lernparadigmen

## Loss Functions
Folien 10 bis 13

- Definition: Eine Verlustfunktion bewertet die Abweichung einer Vorhersage vom tatsächlichen Ergebnis
- Zweck: Dem Lernverfahren vorgeben, welche Fehler wie stark bestraft werden
- Training: Modellparameter so bestimmen, dass das Trainingsziel möglichst klein wird
- Durchschnittlicher Verlust: Verluste der einzelnen Trainingsbeobachtungen zusammenfassen und durch ihre Anzahl teilen

| Verlustfunktion | Berechnung bzw. Grundidee | Eigenschaft |
|---|---|---|
| Squared Loss | `(y − ŷ)²` | Grosse Fehler werden besonders stark bestraft |
| Absolute Loss | `abs(y − ŷ)` | Fehler wachsen linear, weniger empfindlich gegenüber Extremwerten |
| 0-1 Loss | 0 bei richtiger und 1 bei falscher Klasse | Bewertet nur richtig oder falsch |
| Log-Loss | Tatsächlicher Klasse eine hohe Wahrscheinlichkeit zuweisen | Sichere falsche Vorhersagen werden stark bestraft |
| Hinge Loss | Falsche und zu knapp richtige Klassifikationen bestrafen | Verwendet einen Sicherheitsabstand zur Entscheidungsgrenze |

- Bsp. Squared Loss: Fehler 2 ergibt Verlust 4, Fehler 10 ergibt Verlust 100
- Bsp. Absolute Loss: Fehler 2 ergibt Verlust 2, Fehler 10 ergibt Verlust 10
- Trainingsfehler: Häufig optimistisch, weil das Modell an dieselben Daten angepasst wurde
- Generalisierungsfehler: Fehler bei neuen, bisher nicht verwendeten Beobachtungen
- Erkenntnis: Ein kleiner Trainingsverlust garantiert keine guten Vorhersagen bei neuen Daten

## Loss, Training Objective und Evaluation Metric
Folie 14

| Begriff | Bedeutung | Bsp. |
|---|---|---|
| Loss | Fehler einer einzelnen Vorhersage | Quadrierte Abweichung |
| Training Objective | Kriterium, das beim Lernen optimiert wird | Summe der Verluste mit möglicher Zusatzstrafe |
| Evaluation Metric | Kennzahl zur Bewertung des fertigen Modells | RMSE, R², Accuracy und AUC |

| Modell | Typisches Trainingsziel | Typische Bewertung |
|---|---|---|
| Lineare Regression | Summe quadrierter Residuen minimieren | RMSE und R² |
| Logistische Regression | Log-Loss minimieren | Accuracy, F1 und AUC |
| Entscheidungsbaum | Fehler bzw. Unreinheit durch Splits reduzieren | RMSE oder Klassifikationskennzahlen |
| SVM | Hinge Loss und Komplexitätsstrafe berücksichtigen | Accuracy und AUC |
| Gradient Boosting | Gewählte Verlustfunktion schrittweise reduzieren | Abhängig von der Aufgabe |

- Achtung: Nicht jedes Verfahren minimiert alle Modellfehler gleichzeitig global
- Bsp. Ein Baum wählt Splits schrittweise anhand eines lokalen Kriteriums
- Erkenntnis: Optimiertes Trainingsziel und berichtete Bewertungskennzahl müssen nicht identisch sein

## Train-Test Split
Folie 15

- Definition: Daten in getrennte Trainings- und Testdaten aufteilen
- Zweck: Die Leistung bei bisher unbekannten Beobachtungen abschätzen
- Trainingsdaten: Modell anpassen und innerhalb dieses Bereichs Einstellungen auswählen
- Testdaten: Abschliessende Bewertung des ausgewählten Modells
- Bsp. 80 % Training und 20 % Test
- random_state: Macht eine zufällige Aufteilung reproduzierbar
- stratify: Erhält bei Klassifikation die Klassenanteile ungefähr in beiden Teilmengen
- Achtung: Wiederholte Modellauswahl anhand des Testsets macht dessen Bewertung optimistisch
- Erkenntnis: Testdaten müssen bis zur abschliessenden Bewertung von der Modellauswahl getrennt bleiben

## k-fold Cross-Validation
Folie 15

- Definition: Trainingsdaten in k Teilmengen aufteilen und das Modell wiederholt trainieren und validieren
- Ablauf: Jeweils k − 1 Teilmengen zum Training und eine Teilmenge zur Validierung verwenden
- Wiederholung: Jede Teilmenge wird einmal zur Validierung verwendet
- Ergebnis: Validierungskennzahlen über die k Durchläufe zusammenfassen
- Bsp. Bei 5 Folds entstehen fünf Validierungsergebnisse
- Zweck: Modelle und Hyperparameter mit weniger Abhängigkeit von einer einzelnen Aufteilung vergleichen
- Achtung: Vorverarbeitung innerhalb jedes Durchlaufs nur an dessen Trainingsanteil anpassen
- Bsp. Skalierung mit einer Pipeline durchführen, damit keine Validierungsinformationen einfliessen
- Erkenntnis: Cross-Validation unterstützt die Modellauswahl, das Testset bleibt für die abschliessende Bewertung

## Underfitting und Overfitting
Folie 16

| Zustand | Trainingsfehler | Testfehler | Bedeutung |
|---|---|---|---|
| Underfitting | Hoch | Hoch | Modell erfasst wichtige Zusammenhänge nicht |
| Good Fit | Angemessen klein | Angemessen klein | Modell erfasst Muster und generalisiert |
| Overfitting | Sehr klein | Deutlich höher | Modell passt sich auch an zufällige Besonderheiten an |

- Underfitting: Bspw. zu einfaches Modell für einen komplexen Zusammenhang
- Overfitting: Bspw. sehr tiefer Baum, der fast jede Trainingsbeobachtung einzeln abbildet
- Gegenmassnahmen bei Overfitting: Komplexität begrenzen, regularisieren und Hyperparameter mit Cross-Validation wählen
- Achtung: Mehr Modellkomplexität verbessert nicht automatisch die Testleistung
- Erkenntnis: Gute Anpassung und gute Generalisierung gemeinsam beurteilen

## Bias-Variance Trade-off
Folie 17

- Bias: Systematische Abweichung durch zu vereinfachende Modellannahmen
- Variance: Empfindlichkeit des Modells gegenüber Änderungen der Trainingsdaten
- Irreduzibler Fehler: Zufällige Streuung, die mit den verfügbaren Informationen nicht entfernt werden kann
- Grundidee bei quadriertem Fehler: Erwarteter Vorhersagefehler setzt sich aus Bias², Varianz und irreduziblem Fehler zusammen

| Eigenschaft | Hoher Bias | Hohe Varianz |
|---|---|---|
| Typisches Modell | Zu einfach | Sehr flexibel |
| Problem | Wichtige Zusammenhänge fehlen | Zufällige Besonderheiten werden mitgelernt |
| Typischer Bezug | Underfitting | Overfitting |
| Bsp. | Jeden Mietpreis durch denselben Mittelwert vorhersagen | Sehr komplexe Kurve durch wenige Wohnungen legen |

- Erkenntnis: Ziel ist ein günstiger Gesamtfehler, nicht die isolierte Minimierung von Bias oder Varianz

## Lineare Regression
Folien 18 bis 19

- Definition: Numerische Zielvariable als gewichtete Kombination von Eingangsmerkmalen modellieren
- Zweck: Zusammenhänge beschreiben und numerische Werte vorhersagen
- Einfache Regression: `y = β₀ + β₁x + ε`
- Multiple Regression: `y = β₀ + β₁x₁ + ... + βₚxₚ + ε`

| Bestandteil | Bedeutung |
|---|---|
| β₀ | Intercept bzw. Achsenabschnitt |
| β₁ bis βₚ | Koeffizienten der Eingangsmerkmale |
| ε | Nicht durch das Modell erklärter Fehler |
| ŷ | Vorhergesagter Wert ohne unbekannten Fehlerterm |

- Einfacher Koeffizient: Geschätzte Änderung des durchschnittlichen y bei einer zusätzlichen Einheit x
- Multipler Koeffizient: Änderung bei einer zusätzlichen Einheit des Merkmals, wenn die übrigen berücksichtigten Merkmale konstant bleiben
- Achtung: Linear bedeutet linear in den Koeffizienten
- Bsp. Auch x² oder log(x) können als Merkmale in einer linearen Regression vorkommen
- Erkenntnis: Die Bezeichnung linear schliesst umgeformte Eingangsmerkmale nicht aus

## Ordinary Least Squares
Folie 20

- Definition: Koeffizienten durch Minimierung der Summe quadrierter Residuen bestimmen
- Residuum: `eᵢ = yᵢ − ŷᵢ`
- Trainingsziel: `SSE = Σ(yᵢ − ŷᵢ)²` minimieren
- SSE bzw. RSS: Bezeichnungen für die Summe quadrierter Residuen
- Steigung bei einem Prädiktor: `β̂₁ = Σ((xᵢ − x̄)(yᵢ − ȳ)) / Σ(xᵢ − x̄)²`
- Intercept: `β̂₀ = ȳ − β̂₁x̄`
- Eigenschaft: Die OLS-Gerade mit Intercept verläuft durch den Punkt `(x̄, ȳ)`
- Achtung: Quadrierte Fehler machen OLS empfindlich gegenüber einflussreichen Extremwerten
- Erkenntnis: OLS wählt die Gerade mit der kleinsten Summe quadrierter vertikaler Abweichungen

## Koeffizienten interpretieren
Folie 21

- Bsp. Mietpreis: `ŷ = 166.79 + 30.84 × Fläche`
- Steigung: Eine zusätzliche Fläche von 1 m² entspricht im Modell durchschnittlich CHF 30.84 mehr Monatsmiete
- Intercept: Geschätzter Mietpreis bei einer Fläche von 0 m²
- Achtung: Ein rechnerisch vorhandener Intercept muss fachlich nicht sinnvoll interpretierbar sein
- Vorhersage: Bei 95 m² ergibt sich eine Monatsmiete von ca. CHF 3'097
- Extrapolation: Vorhersage ausserhalb des beobachteten Wertebereichs
- Erkenntnis: Vorhersagen ausserhalb der vorhandenen Daten sind besonders vorsichtig zu beurteilen

## R² und RMSE
Folie 22

| Kennzahl | Berechnung | Interpretation |
|---|---|---|
| R² | `1 − SSE / SST` | Verbesserung gegenüber der Vorhersage mit dem Mittelwert der bewerteten Zielwerte |
| RMSE | `sqrt(Σ(yᵢ − ŷᵢ)² / n)` | Wurzel des mittleren quadrierten Vorhersagefehlers |

- SST: Summe der quadrierten Abweichungen der tatsächlichen Werte von ihrem Mittelwert
- R² = 1: Perfekte Vorhersage
- R² = 0: Gleiche Fehlerquadratsumme wie die Mittelwertvorhersage
- R² < 0: Schlechter als die Mittelwertvorhersage
- Bsp. SSE 20 und SST 100 ergeben R² von 0.80
- Achtung: Bei gewöhnlicher OLS mit Intercept liegt Trainings-R² zwischen 0 und 1, Test-R² kann negativ sein
- RMSE-Einheit: Entspricht der Einheit der Zielvariable
- Bsp. RMSE von 2.24 L/100 km beschreibt die Fehlergrösse in derselben Einheit wie der Treibstoffverbrauch
- Achtung: RMSE ist nicht die mittlere absolute Abweichung
- Erkenntnis: RMSE beschreibt die Fehlergrösse, R² bewertet relativ zu einer Mittelwertvorhersage

## Confidence Interval und Prediction Interval
Folie 23

| Intervall | Was wird beschrieben? | Bsp. |
|---|---|---|
| Confidence Interval | Unsicherheit des geschätzten durchschnittlichen Zielwerts | Durchschnittsmiete aller Wohnungen mit 95 m² |
| Prediction Interval | Unsicherheit einer einzelnen neuen Beobachtung | Mietpreis einer neuen Wohnung mit 95 m² |

- Prediction Interval: Berücksichtigt zusätzlich die Streuung einzelner Beobachtungen um den geschätzten Mittelwert
- Breite: Das übliche Vorhersageintervall ist bei gleichem Niveau und denselben Eingaben breiter als das Konfidenzintervall
- Bsp. Durchschnittsmiete relativ genau geschätzt, einzelne Wohnung wegen Lage und Ausstattung trotzdem deutlich teurer
- Achtung: Ein 95%-Konfidenzverfahren erzeugt bei wiederholter Anwendung Intervalle, die den tatsächlichen Parameter in 95 % der Fälle enthalten
- Erkenntnis: Durchschnitt und einzelne neue Beobachtung besitzen unterschiedliche Unsicherheit

## Annahmen und Diagnostik der linearen Regression
Folie 24

| Annahme | Bedeutung | Prüfung |
|---|---|---|
| Passende funktionale Form | Erwarteter Zielwert wird durch die gewählten Modellterme angemessen beschrieben | Residuen gegen vorhergesagte Werte |
| Unabhängigkeit | Beobachtungen bzw. Fehler sind nicht unerwartet voneinander abhängig | Erhebungsdesign und zeitliche Reihenfolge |
| Normalverteilung der Fehler | Für bestimmte exakte Tests und Intervalle bei kleinen Stichproben erforderlich | QQ-Plot der Residuen |
| Homoskedastizität | Fehlervarianz bleibt über den Wertebereich konstant | Residuen gegen vorhergesagte Werte |
| Keine perfekte Multikollinearität | Kein Prädiktor ist eine exakte lineare Kombination anderer Prädiktoren | Datenstruktur, Korrelationen und VIF |

- Heteroskedastizität: Streuung der Fehler verändert sich, bspw. trichterförmiges Residuenmuster
- Starke Multikollinearität: Kann Koeffizienten instabil und Standardfehler gross machen
- Achtung: Normalverteilung der Eingangsvariablen ist keine allgemeine OLS-Voraussetzung
- Achtung: OLS-Koeffizienten können auch ohne normalverteilte Fehler berechnet werden
- Erkenntnis: Modellschätzung, Vorhersage und statistische Inferenz stellen teilweise unterschiedliche Anforderungen

## Multiple Regression
Folie 25

- Definition: Mehrere Eingangsmerkmale gemeinsam zur Erklärung einer numerischen Zielvariable verwenden
- Zweck: Zusammenhänge unter Berücksichtigung weiterer Merkmale untersuchen
- Ceteris paribus: Andere im Modell enthaltene Merkmale konstant halten
- Bsp. Flächenkoeffizient beschreibt den Zusammenhang zwischen Fläche und Mietpreis bei gleicher Zimmerzahl
- p-Wert: Prüft unter den jeweiligen Annahmen die Vereinbarkeit der Daten mit einer Nullhypothese, häufig Koeffizient gleich 0
- Achtung: Ein kleiner p-Wert bedeutet nicht automatisch einen grossen praktischen Einfluss
- Achtung: Ein Koeffizient beschreibt nicht automatisch einen kausalen Effekt
- Erkenntnis: Koeffizienten immer zusammen mit Einheit, Modellmerkmalen und fachlichem Kontext interpretieren

## Maximum Likelihood
Folie 26

- Definition: Parameter so wählen, dass die beobachteten Daten unter dem Modell möglichst plausibel sind
- Likelihood: Bewertet die Plausibilität der beobachteten Daten bei gegebenen Parametern
- Bsp. Bei siebenmal Kopf und dreimal Zahl liegt die Maximum-Likelihood-Schätzung für die Kopf-Wahrscheinlichkeit bei 0.7
- Lineare Regression: Bei unabhängigen normalverteilten Fehlern mit gleicher Varianz liefert Maximum Likelihood dieselben Regressionskoeffizienten wie OLS
- Logistische Regression: Wird über Maximum Likelihood bzw. Minimierung der negativen Log-Likelihood angepasst
- Erkenntnis: Maximum Likelihood ist ein allgemeines Schätzprinzip für unterschiedliche Modelle

## Gradient Descent
Folie 27

- Definition: Parameter schrittweise entgegen der Steigung der Verlustfunktion verändern
- Zweck: Einen möglichst kleinen Wert des Trainingsziels finden
- Ablauf: Verlust berechnen, Gradient bestimmen und Parameter entgegen dem Gradienten aktualisieren
- Lernrate η: Bestimmt die Grösse eines Schritts
- Kleine Lernrate: Kann viele Schritte benötigen
- Grosse Lernrate: Kann das Minimum überspringen oder das Verfahren instabil machen
- Stochastic Gradient Descent: Verwendet einzelne Beobachtungen
- Mini-Batch Gradient Descent: Verwendet kleine Teilmengen
- Erkenntnis: Gradient Descent ist ein Optimierungsverfahren, kein eigenständiges Vorhersagemodell

## Regression Trees
Folien 28 bis 30

- Definition: Merkmalsraum durch aufeinanderfolgende Entscheidungsregeln in Bereiche aufteilen
- Zweck: Nichtlineare Zusammenhänge und Interaktionen für numerische Vorhersagen erfassen
- Split: Ein Merkmal und einen Schwellenwert auswählen
- Blatt: Endbereich des Baums
- Vorhersage bei quadriertem Fehler: Mittelwert der Trainingszielwerte im jeweiligen Blatt

### Auswahl eines Splits

- Schritt 1: Mögliche Merkmale und Schwellenwerte prüfen
- Schritt 2: Beobachtungen in zwei Kindknoten aufteilen
- Schritt 3: Mittelwerte der Zielvariable in beiden Kindknoten berechnen
- Schritt 4: Summe quadrierter Abweichungen innerhalb der Kindknoten bestimmen
- Schritt 5: Split mit der kleinsten gemeinsamen Fehlerquadratsumme auswählen
- Rekursiv: Vorgehen in den Kindknoten wiederholen
- Greedy: Jeweils aktuell besten Split auswählen, ohne alle vollständigen Bäume zu vergleichen

### Bsp. Split berechnen

| Beobachtung x | Zielwert y |
|---|---|
| 1 | 10 |
| 2 | 12 |
| 3 | 14 |
| 4 | 30 |
| 5 | 34 |

- Ausgangsknoten: Mittelwert 20 und SSE 496
- Split bei x ≤ 3.5: Linke Zielwerte 10, 12 und 14, rechte Zielwerte 30 und 34
- Linkes Blatt: Vorhersage 12 und SSE 8
- Rechtes Blatt: Vorhersage 32 und SSE 8
- Gesamtfehler nach Split: 16
- Erkenntnis: Ein guter Split erzeugt Teilgruppen mit möglichst ähnlichen Zielwerten

## Pruning
Folie 31

- Definition: Einen gewachsenen Baum nachträglich verkleinern
- Zweck: Unnötige Komplexität und Overfitting reduzieren
- Cost-Complexity-Pruning: Trainingsfehler mit einer Strafe für die Anzahl Blätter kombinieren
- α = 0: Keine zusätzliche Komplexitätsstrafe
- Grösseres α: Kleinere Bäume werden bevorzugt
- Hyperparameter: `ccp_alpha`
- Auswahl: Geeigneten Wert mit Cross-Validation bestimmen
- Achtung: Zu starkes Pruning kann Underfitting verursachen
- Erkenntnis: Ein kleinerer Baum kann bei neuen Daten besser sein als der grösste Trainingsbaum

## Hyperparameter von Bäumen
Folie 32

| Parameter | Bedeutung | Wirkung einer Erhöhung |
|---|---|---|
| max_depth | Maximale Tiefe | Mehr Komplexität möglich |
| min_samples_split | Mindestanzahl Beobachtungen für einen Split | Weniger Splits möglich |
| min_samples_leaf | Mindestanzahl Beobachtungen pro Blatt | Grössere und weniger spezielle Blätter |
| max_leaf_nodes | Maximale Anzahl Blätter | Mehr Komplexität möglich |
| max_features | Berücksichtigte Merkmale pro Split | Mehr Auswahl, bei Random Forest häufig stärker korrelierte Bäume |
| ccp_alpha | Stärke der Komplexitätsstrafe | Stärkeres Pruning |
| criterion | Kriterium zur Splitbewertung | Abhängig von Regression oder Klassifikation |

- Regression: Bspw. `squared_error`
- Klassifikation: Bspw. `gini` oder `entropy`
- Achtung: `min_samples_split` begrenzt den Elternknoten, `min_samples_leaf` die resultierenden Blätter
- Erkenntnis: Mehr Komplexität senkt häufig den Bias, kann aber die Varianz erhöhen

## Ensemble Learning
Folie 33

- Definition: Mehrere Modelle zu einer gemeinsamen Vorhersage kombinieren
- Zweck: Fehler einzelner Modelle teilweise ausgleichen
- Mittelung: Reduziert besonders den nicht gemeinsam auftretenden Anteil der Modellschwankungen
- Korrelation: Sehr ähnliche Modelle machen häufig ähnliche Fehler
- Erkenntnis: Viele möglichst unterschiedliche gute Modelle können gemeinsam stabiler vorhersagen

### Bagging und Boosting

| Verfahren | Grundidee | Typische Wirkung |
|---|---|---|
| Bagging | Modelle auf unterschiedlichen Stichproben trainieren und kombinieren | Vor allem Varianz reduzieren |
| Boosting | Modelle nacheinander auf noch nicht gut erfasste Muster ausrichten | Systematische Fehler schrittweise reduzieren |

- Achtung: Ensemble-Verfahren verbessern nicht automatisch jedes Modell oder jeden Datensatz
- Erkenntnis: Vielfalt der Modelle und passende Einstellungen bleiben entscheidend

## Bagging
Folie 34

- Definition: Bootstrap Aggregation
- Bootstrap: Stichproben mit Zurücklegen aus den Trainingsdaten ziehen
- Ablauf: Mehrere Bootstrap-Stichproben erzeugen, je ein Modell trainieren und Vorhersagen kombinieren
- Regression: Vorhersagen mitteln
- Klassifikation: Bspw. Klassen abstimmen lassen oder Klassenwahrscheinlichkeiten mitteln
- Vorteil: Einzelne instabile Bäume werden durch die Kombination stabiler

### Out-of-Bag-Beobachtungen

- Definition: Beobachtungen, die in einer bestimmten Bootstrap-Stichprobe nicht gezogen wurden
- Anteil: Bei grossen Stichproben ungefähr 37 % pro Bootstrap-Stichprobe
- Verwendung: Interne Fehlerabschätzung mit den jeweils nicht verwendeten Beobachtungen
- Achtung: Out-of-Bag-Bewertung ersetzt kein abschliessendes unabhängiges Testset
- Erkenntnis: Bootstrap erzeugt Unterschiede zwischen den trainierten Modellen

## Random Forest
Folie 35

- Definition: Kombination vieler Bäume mit Bootstrap-Stichproben und zufälliger Merkmalsauswahl
- Zweck: Leistungsfähige Bäume kombinieren und ihre gemeinsame Varianz reduzieren
- Zufällige Merkmale: Pro Split nur eine Auswahl der Merkmale berücksichtigen
- Wirkung: Bäume werden unterschiedlicher und weniger stark korreliert
- Regression: Baumvorhersagen mitteln
- Klassifikation: In scikit-learn Klassenwahrscheinlichkeiten mitteln und daraus die Klasse bestimmen

| Parameter | Bedeutung |
|---|---|
| n_estimators | Anzahl Bäume |
| max_features | Merkmalsauswahl pro Split |
| max_depth | Maximale Tiefe der einzelnen Bäume |
| min_samples_leaf | Mindestanzahl Beobachtungen je Blatt |

- Mehr Bäume: Stabilisieren die Vorhersage, erhöhen aber den Rechenaufwand
- Feature Importance: Beschreibt die Bedeutung von Merkmalen für das Modell
- Achtung: Unreinheitsbasierte Importance kann bestimmte Merkmale bevorzugen
- Permutation Importance: Prüft, wie stark sich die Bewertung verschlechtert, wenn ein Merkmal durchmischt wird
- Erkenntnis: Zufällige Merkmalsauswahl ergänzt Bagging um zusätzliche Vielfalt

## Gradient Boosting
Folie 36

- Definition: Ein Ensemble schrittweise um weitere kleine Modelle ergänzen
- Zweck: Verbleibende Fehler des bisherigen Modells reduzieren
- Grundidee: Neue Vorhersage = bisherige Vorhersage + Lernrate × Korrekturmodell
- Bei quadriertem Fehler: Der nächste Baum lernt die aktuellen Residuen
- Bei anderen Verlustfunktionen: Der nächste Baum lernt entsprechende negative Gradienten bzw. Pseudoresiduen
- Bsp. Tatsächlicher Wert 10 und Vorhersage 7 ergeben Residuum 3
- Bsp. Bei Korrektur 3 und Lernrate 0.1 steigt die Vorhersage von 7 auf 7.3

| Parameter | Bedeutung |
|---|---|
| learning_rate | Stärke des Beitrags jedes neuen Baums |
| n_estimators | Anzahl aufeinanderfolgender Bäume |
| max_depth | Komplexität der einzelnen Bäume |

- Kleine Lernrate: Benötigt häufig mehr Bäume
- Achtung: Zu viele oder zu komplexe Bäume können Overfitting verursachen
- Erkenntnis: Boosting lernt aufeinanderfolgende Korrekturen, Bagging kombiniert unabhängiger trainierte Modelle

## Regressionsmodelle vergleichen
Folie 37

| Modell | Stärke | Grenze |
|---|---|---|
| Lineare Regression | Gut interpretierbare Koeffizienten und geringer Rechenaufwand | Gewählte funktionale Form muss passen |
| Einzelner Baum | Verständliche Regeln und nichtlineare Beziehungen | Kann instabil sein und overfitten |
| Random Forest | Flexible Beziehungen und geringere Varianz als einzelne Bäume | Gesamtes Modell schwerer interpretierbar |
| Gradient Boosting | Leistungsfähige schrittweise Fehlerkorrektur | Einstellungen und Rechenaufwand beachten |

- Achtung: Aussagen zu Datenbedarf und Leistung sind typische Tendenzen, keine festen Regeln
- Erkenntnis: Modelle auf derselben Aufgabe mit geeigneter Cross-Validation vergleichen

## Logistic Regression
Folien 38 bis 39

- Definition: Klassifikationsmodell, das einen linearen Score in eine Wahrscheinlichkeit umwandelt
- Zweck: Wahrscheinlichkeit einer Klasse vorhersagen
- Score: `z = β₀ + β₁x₁ + ... + βₚxₚ`
- Sigmoid-Funktion: `p = 1 / (1 + exp(−z))`
- Wertebereich: Zwischen 0 und 1
- Training: Maximum Likelihood bzw. Minimierung des Log-Loss

### Bsp. Vorhersage

- Gegeben: β₀ = −5, β₁ = 1 und x = 6
- Score: `−5 + 6 = 1`
- Wahrscheinlichkeit: Ca. 0.73
- Entscheidung bei Schwelle 0.5: Klasse 1
- Positiver Koeffizient: Bei konstanten übrigen Merkmalen steigt die Wahrscheinlichkeit mit dem Merkmal
- Achtung: Ein Koeffizient ist keine konstante Änderung der Wahrscheinlichkeit in Prozentpunkten
- Erkenntnis: Trotz des Namens wird logistische Regression für Klassifikation verwendet

## Classification Trees
Folien 40 bis 43

- Definition: Beobachtungen durch aufeinanderfolgende Regeln in möglichst reine Klassenbereiche aufteilen
- Zweck: Klassen und Klassenwahrscheinlichkeiten vorhersagen

| Bestandteil | Bedeutung |
|---|---|
| Root Node | Ausgangsknoten mit allen Trainingsbeobachtungen |
| Internal Node | Knoten mit weiterer Entscheidung |
| Branch | Verbindung entsprechend einem Entscheidungsergebnis |
| Leaf | Endknoten ohne weiteren Split |
| Depth | Anzahl Kanten auf dem längsten Weg von der Wurzel zu einem Blatt |

- Bsp. Zuerst Geschlecht, danach Alter oder Reiseklasse prüfen
- Vorhersage im Blatt: Häufigste Klasse
- Wahrscheinlichkeit im Blatt: Anteil der jeweiligen Klasse unter den Trainingsbeobachtungen dieses Blatts
- Bsp. 145 Überlebende und 125 Nichtüberlebende ergeben eine Überlebenswahrscheinlichkeit von ca. 0.537
- Erkenntnis: Ein Klassifikationsbaum liefert nicht nur Klassen, sondern auch geschätzte Wahrscheinlichkeiten

## Gini Impurity
Folien 41 bis 42

- Definition: Mass für die Durchmischung der Klassen innerhalb eines Knotens
- Berechnung: `Gini = 1 − Σpₖ²`
- pₖ: Anteil der Klasse k im Knoten
- Reiner Knoten: Alle Beobachtungen gehören derselben Klasse an, Gini 0
- Zwei gleich grosse Klassen: Gini 0.5
- Achtung: Der Maximalwert 0.5 gilt für zwei Klassen

### Splitbewertung

- Vorgehen: Unreinheit des Elternknotens mit der gewichteten Unreinheit der Kindknoten vergleichen
- Gewichtung: Anteil der Beobachtungen im jeweiligen Kindknoten
- Ziel: Möglichst grosse Abnahme der Unreinheit
- Bsp. Elternknoten mit 40 % Überlebenden und 60 % Nichtüberlebenden besitzt Gini 0.48
- Bsp. Gewichtete Unreinheit nach einem Split von 0.36 ergibt Verbesserung 0.12
- Alternative: Entropie als anderes Mass der Unreinheit
- Erkenntnis: Gute Splits erzeugen möglichst reine Teilgruppen

## Wachstum und Begrenzung eines Klassifikationsbaums
Folie 43

- Schritt 1: Mögliche Merkmale und Schwellenwerte prüfen
- Schritt 2: Abnahme der Unreinheit berechnen
- Schritt 3: Besten zulässigen Split auswählen
- Schritt 4: In den Kindknoten wiederholen
- Stopp: Bspw. maximale Tiefe, zu wenige Beobachtungen oder reiner Knoten
- Pruning: Gewachsenen Baum bei Bedarf verkleinern
- Erkenntnis: Begrenzungsregeln verhindern zu spezielle Entscheidungen anhand weniger Trainingsfälle

## Random Forest und Boosting für Klassifikation
Folie 44

| Verfahren | Aufbau | Vorhersage |
|---|---|---|
| Random Forest Classifier | Viele Bäume mit Bootstrap und zufälligen Merkmalen | Klassenwahrscheinlichkeiten mitteln |
| Gradient Boosting Classifier | Kleine Bäume nacheinander ergänzen | Klassifikationsverlust schrittweise reduzieren |

- Bsp. Drei Bäume liefern Wahrscheinlichkeiten 0.8, 0.6 und 0.3
- Mittelwert: Ca. 0.567
- Entscheidung bei Schwelle 0.5: Positive Klasse
- Achtung: Gradient Boosting für Klassifikation passt Pseudoresiduen des gewählten Verlusts an
- Erkenntnis: Ensemble-Grundprinzipien gelten für Regression und Klassifikation, Ziel und Ausgabe unterscheiden sich

## k-nearest Neighbours
Folie 45

- Definition: Neue Beobachtung anhand der k nächsten Trainingsbeobachtungen einordnen
- Zweck: Ähnliche Fälle für die Vorhersage verwenden
- Ablauf: k wählen, Distanzen berechnen, nächste Nachbarn auswählen und deren Klassen auswerten
- Distanz: Bspw. euklidische Distanz
- Klassifikation: Häufigste Klasse der Nachbarn
- Wahrscheinlichkeit: Bei gleicher Gewichtung Anteil der jeweiligen Klasse unter den Nachbarn
- Bsp. Drei blaue Nachbarn unter fünf ergeben Wahrscheinlichkeit 0.6 für blau

| Einstellung | Typische Wirkung |
|---|---|
| Kleines k | Flexiblere Grenzen und höhere Empfindlichkeit gegenüber einzelnen Fällen |
| Grosses k | Glattere Grenzen, lokale Unterschiede können verloren gehen |

- Achtung: Merkmale mit unterschiedlichen Grössenordnungen vorher geeignet skalieren
- Achtung: Bei vielen Merkmalen kann die Aussagekraft von Distanzen abnehmen
- Ergänzung: Für Regression kann der Mittelwert der Zielwerte der Nachbarn verwendet werden
- Erkenntnis: k und Distanzberechnung bestimmen wesentlich, welche Beobachtungen als ähnlich gelten

## Naive Bayes
Folie 46

- Definition: Klassen anhand des Bayes-Theorems und einer vereinfachenden Unabhängigkeitsannahme vorhersagen
- Bayes-Theorem: `P(y|x) = P(x|y) × P(y) / P(x)`
- Prior P(y): Grundhäufigkeit einer Klasse vor Berücksichtigung der Merkmale
- Likelihood P(x|y): Plausibilität der beobachteten Merkmale innerhalb einer Klasse
- Posterior P(y|x): Aktualisierte Klassenwahrscheinlichkeit nach Berücksichtigung der Merkmale
- Naive Annahme: Merkmale sind innerhalb einer gegebenen Klasse voneinander unabhängig
- Klassenscore: `P(y) × P(x₁|y) × ... × P(xₚ|y)`
- Entscheidung: Klasse mit dem höchsten Score auswählen

### Bsp. Golf spielen

- Merkmale: Sonnig, kühl, hohe Luftfeuchtigkeit und windig
- Score nein: 0.0206
- Score ja: 0.0053
- Entscheidung: Nein
- Normierung: `P(nein|x) = 0.0206 / (0.0206 + 0.0053)`, ca. 0.795
- Achtung: Die Unabhängigkeitsannahme gilt bedingt auf die Klasse, nicht allgemein zwischen allen Merkmalen
- Erkenntnis: Naive Bayes kombiniert Grundhäufigkeiten mit der Verteilung der Merkmale innerhalb jeder Klasse

## Support Vector Machines
Folie 47

- Definition: Klassen durch eine Entscheidungsgrenze mit möglichst grossem Sicherheitsabstand trennen
- Margin: Abstand zwischen Entscheidungsgrenze und den begrenzenden Beobachtungen
- Support Vectors: Trainingsbeobachtungen, die die Lösung wesentlich bestimmen
- Soft Margin: Erlaubt Beobachtungen innerhalb des Sicherheitsbereichs und mögliche Fehlklassifikationen
- Kernel: Ermöglicht nichtlineare Entscheidungsgrenzen über eine geeignete Ähnlichkeitsfunktion

| Parameter | Wirkung |
|---|---|
| Kleines C | Stärkere Regularisierung und grössere Toleranz gegenüber Verletzungen der Margin |
| Grosses C | Verletzungen der Margin werden stärker bestraft |

- Achtung: Bei Soft Margin können Support Vectors auch innerhalb des Sicherheitsbereichs oder auf der falschen Seite liegen
- Achtung: Skalierung der Merkmale ist häufig wichtig
- Erkenntnis: SVM verbindet eine trennende Grenze mit einer Kontrolle ihrer Komplexität

## Confusion Matrix
Folie 48

- Definition: Gegenüberstellung tatsächlicher und vorhergesagter Klassen
- Zweck: Richtige Entscheidungen und verschiedene Fehlertypen unterscheiden

| Tatsächlich | Vorhergesagt negativ | Vorhergesagt positiv |
|---|---|---|
| Negativ | TN, True Negative | FP, False Positive |
| Positiv | FN, False Negative | TP, True Positive |

- TP: Positiver Fall richtig erkannt
- TN: Negativer Fall richtig erkannt
- FP: Negativer Fall fälschlich als positiv eingestuft
- FN: Positiver Fall übersehen
- Achtung: Zuerst festlegen, welche Klasse als positiv gilt
- Erkenntnis: Unterschiedliche Fehlerarten können unterschiedliche praktische Folgen besitzen

## Kennzahlen für Klassifikation
Folie 48

| Kennzahl | Berechnung | Frage |
|---|---|---|
| Accuracy | `(TP + TN) / n` | Welcher Anteil aller Vorhersagen ist richtig? |
| Precision | `TP / (TP + FP)` | Welcher Anteil der positiv vorhergesagten Fälle ist tatsächlich positiv? |
| Recall bzw. TPR | `TP / (TP + FN)` | Welcher Anteil aller tatsächlichen positiven Fälle wurde erkannt? |
| FPR | `FP / (FP + TN)` | Welcher Anteil aller tatsächlichen negativen Fälle wurde fälschlich positiv eingestuft? |
| F1 | `2 × Precision × Recall / (Precision + Recall)` | Wie gut werden Precision und Recall gemeinsam erreicht? |

### Bsp. aus den Folien

- Gegeben: TN 75, FP 22, FN 24 und TP 50
- Gesamtzahl: 171
- Accuracy: `125 / 171`, ca. 73.1 %
- Precision: `50 / 72`, ca. 69.4 %
- Recall: `50 / 74`, ca. 67.6 %
- FPR: `22 / 97`, ca. 22.7 %
- F1: `100 / 146`, ca. 68.5 %

### Ungleiche Klassenverteilung

- Problem: Die häufige Klasse kann eine hohe Accuracy ermöglichen, obwohl die seltene Klasse kaum erkannt wird
- Bsp. Bei 99 % normalen Transaktionen erreicht immer normal bereits 99 % Accuracy
- Achtung: Kennzahlen anhand der Fragestellung und Fehlerkosten auswählen
- Erkenntnis: Accuracy allein reicht bei unausgeglichenen Klassen häufig nicht aus

## Classification Threshold
Folie 49

- Definition: Schwellenwert zur Umwandlung einer Wahrscheinlichkeit in eine Klasse
- Regel: Positive Klasse vorhersagen, wenn `p ≥ τ`
- Häufiger Ausgangspunkt: τ = 0.5
- Höhere Schwelle: Weniger positive Vorhersagen, Recall und FPR können nicht steigen
- Tiefere Schwelle: Mehr positive Vorhersagen, Recall und FPR können nicht sinken
- Achtung: Precision steigt bei höherer Schwelle nicht zwingend
- Erkenntnis: Die Schwelle anhand der Kosten falscher positiver und falscher negativer Entscheidungen auswählen

### Fehlerkosten

- Bsp. Hohe Kosten eines übersehenen Betrugs sprechen eher für eine tiefere Schwelle
- Bsp. Hohe Kosten unnötiger Sperrungen sprechen eher für eine höhere Schwelle
- Gesamtkosten bei konstanten Fehlerkosten: `FP × Kosten_FP + FN × Kosten_FN`
- Achtung: Schwellenwahl auf Validierungsdaten durchführen und danach auf Testdaten bewerten
- Erkenntnis: Die beste Schwelle muss nicht 0.5 sein

## ROC Curve und AUC
Folie 50

- ROC Curve: Stellt Recall bzw. TPR gegen FPR für unterschiedliche Schwellenwerte dar
- X-Achse: False Positive Rate
- Y-Achse: True Positive Rate
- Punkt: Entspricht einem bestimmten Schwellenwert
- Idealer Bereich: Hohe TPR bei niedriger FPR, also nahe oben links
- Diagonale: Entspricht im Erwartungswert einer zufälligen Rangordnung

| AUC | Interpretation |
|---|---|
| 1 | Perfekte Rangordnung der Klassen |
| 0.5 | Zufällige Rangordnung im Erwartungswert |
| Unter 0.5 | Schlechtere Rangordnung als zufällig |

- AUC: Fläche unter der ROC-Kurve
- Interpretation: Wahrscheinlichkeit, dass ein zufälliger positiver Fall einen höheren Score erhält als ein zufälliger negativer Fall
- Gleichstand: Zählt bei dieser Interpretation zur Hälfte
- Bsp. AUC 0.95 bedeutet, dass positive Fälle in diesem Paarvergleich sehr häufig höher eingeordnet werden
- Achtung: AUC 0.95 bedeutet nicht 95 % richtige Klassifikationen
- Achtung: AUC bewertet die Rangordnung, nicht die Genauigkeit der Wahrscheinlichkeiten oder einen einzelnen Schwellenwert
- Erkenntnis: ROC und AUC vergleichen Trennfähigkeit über verschiedene Schwellenwerte

## Unsupervised Learning
Folien 51 bis 52

- Definition: Strukturen in Daten ohne vorgegebene Zielvariable lernen
- Zweck: Ähnliche Beobachtungen gruppieren oder viele Merkmale kompakter beschreiben
- Clustering: Gruppen ähnlicher Beobachtungen bilden
- Dimensionsreduktion: Viele Merkmale durch weniger neue Merkmale zusammenfassen
- Bsp. Kundensegmente, Auffälligkeiten und grafische Darstellung hochdimensionaler Daten
- Bewertung: Interne Kriterien, Stabilität und fachliche Interpretierbarkeit
- Erkenntnis: Ohne vorgegebene Zielwerte gibt es meist keine einfache Bewertung wie richtig oder falsch

## k-means Clustering
Folie 53

- Definition: Beobachtungen in k Gruppen mit jeweils einem Zentrum aufteilen
- Zweck: Gruppen mit möglichst kleinen quadrierten Abständen zum eigenen Zentrum bilden
- Schritt 1: Anzahl Cluster k festlegen
- Schritt 2: Startzentren wählen
- Schritt 3: Jede Beobachtung dem nächsten Zentrum zuordnen
- Schritt 4: Jedes Zentrum als Mittelwert seiner zugeordneten Beobachtungen neu berechnen
- Wiederholung: Zuordnung und Zentren aktualisieren, bis sich die Lösung kaum noch verändert
- Achtung: Unterschiedliche Startwerte können unterschiedliche Ergebnisse erzeugen
- Erkenntnis: k-means sucht kompakte Gruppen anhand der gewählten Merkmale und Distanzen

## Elbow Method
Folie 54

- Definition: Innerhalb der Cluster verbleibende Fehlerquadratsumme für verschiedene k vergleichen
- X-Achse: Anzahl Cluster
- Y-Achse: Summe quadrierter Abstände der Beobachtungen zu ihren Clusterzentren
- Grundidee: Eine Stelle suchen, ab der zusätzliche Cluster nur noch wenig Verbesserung bringen
- Elbow: Knick in der Kurve als mögliche Wahl von k
- Achtung: Das bestmögliche Kriterium wird mit mehr Clustern kleiner, bei k = n kann es 0 werden
- Achtung: Ein eindeutiger Knick muss nicht vorhanden sein
- Vorbereitung: Merkmale geeignet skalieren und mehrere Startlösungen verwenden
- Erkenntnis: Den Knick mit Stabilität und fachlicher Bedeutung der Cluster gemeinsam beurteilen

## Hierarchical Clustering
Folie 55

- Definition: Beobachtungen schrittweise zu einer Hierarchie von Gruppen zusammenführen
- Gezeigtes Vorgehen: Agglomeratives Clustering
- Schritt 1: Jede Beobachtung bildet einen eigenen Cluster
- Schritt 2: Die jeweils nächsten Cluster zusammenführen
- Wiederholung: Bis alle Beobachtungen zu einem Cluster verbunden sind
- Dendrogramm: Baumdarstellung der Zusammenführungen
- Höhe: Beschreibt die Entfernung bzw. das verwendete Verknüpfungskriterium bei der Zusammenführung
- Schnitt: Eine horizontale Schnittlinie legt die resultierenden Gruppen fest
- Achtung: Distanzmass, Skalierung und Verknüpfungsmethode beeinflussen das Ergebnis
- Erkenntnis: Das Dendrogramm zeigt, auf welchen Ebenen Gruppen miteinander verbunden werden

## Principal Component Analysis
Folie 56

- Definition: Numerische Merkmale durch neue, zueinander orthogonale Achsen zusammenfassen
- Zweck: Dimensionen reduzieren und möglichst viel der vorhandenen Varianz erhalten
- PC1: Richtung mit der grössten Varianz
- PC2: Richtung mit der grössten verbleibenden Varianz, orthogonal zu PC1
- Weitere Komponenten: Entsprechend mit abnehmendem Varianzanteil
- Explained Variance: Anteil der Gesamtvarianz, den eine Komponente beschreibt
- Bsp. PC1 erklärt 88 % der Varianz zweier Merkmale
- Achtung: Hohe Varianz bedeutet nicht automatisch hohe fachliche Bedeutung
- Achtung: Ohne geeignete Skalierung können Merkmale mit grosser Streuung dominieren
- Erkenntnis: PCA erzeugt neue kombinierte Merkmale und wählt nicht einfach vorhandene Spalten aus

## Zentrale Unterscheidungen für Verständnisfragen

| Frage | Kernantwort |
|---|---|
| Supervised oder Unsupervised? | Mit bekannten Zielwerten lernen oder ohne Zielwerte Strukturen finden |

| Training oder Test? | Modell anpassen oder abschliessend bei unbekannten Daten bewerten |
| Cross-Validation oder Testset? | Modelleinstellungen auswählen oder das ausgewählte Modell abschliessend prüfen |

| Bias oder Variance? | Systematische Vereinfachung oder Instabilität zwischen Trainingsstichproben |
| Residuum oder Vorhersage? | Tatsächlicher Wert minus Schätzung oder geschätzter Zielwert |
| R² oder RMSE? | Relative Leistung gegenüber Mittelwertvorhersage oder Fehlergrösse in der Zieleinheit |
| Konfidenzintervall oder Vorhersageintervall? | Unsicherheit des durchschnittlichen Zielwerts oder einer einzelnen neuen Beobachtung |
| OLS oder Maximum Likelihood? | Fehlerquadratsumme minimieren oder Plausibilität der beobachteten Daten maximieren |
| Modell oder Gradient Descent? | Vorhersageregel oder Verfahren zur Parameteroptimierung |
| Regressionsbaum oder Klassifikationsbaum? | Numerische Zielwerte oder Klassenbereiche vorhersagen |
| Pruning oder tiefen Baum behalten? | Komplexität reduzieren oder stärkere Anpassung an Trainingsdaten zulassen |
| Bagging oder Boosting? | Unterschiedliche Modelle kombinieren oder aufeinanderfolgende Korrekturen lernen |
| Bagging oder Random Forest? | Bootstrap-Modelle kombinieren oder zusätzlich zufällige Merkmale pro Split auswählen |
| Lineare oder logistische Regression? | Numerischen Zielwert oder Klassenwahrscheinlichkeit modellieren |
| Precision oder Recall? | Zuverlässigkeit positiver Vorhersagen oder Erkennung tatsächlicher positiver Fälle |
| FP oder FN? | Falscher Alarm oder übersehener positiver Fall |
| Accuracy oder AUC? | Richtige Klassen bei einer Entscheidung oder Rangordnung über Schwellenwerte beurteilen |
| Höhere oder tiefere Schwelle? | Weniger oder mehr positive Vorhersagen |
| k-NN oder k-means? | Zielwerte anhand bekannter Nachbarn vorhersagen oder Gruppen ohne Zielwerte bilden |
| Clustering oder PCA? | Beobachtungen gruppieren oder Merkmalsdimensionen reduzieren |
| Modellparameter oder Hyperparameter? | Aus Daten gelernte Werte oder vor dem Training gesetzte Einstellungen |
