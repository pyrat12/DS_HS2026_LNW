# Foundations

### Machine Learning (ML) – Aus Daten lernen

- **Definition:** Machine Learning bedeutet, dass ein Computer aus Beispielen Muster und Regeln lernt, statt für jeden Einzelfall eine ausdrücklich programmierte Regel zu erhalten
- **Zweck:** Gelernte Zusammenhänge für Vorhersagen auf neuen Beobachtungen nutzen oder bisher unbekannte Strukturen in Daten entdecken
- **Datenaufbau:** Jede Zeile beschreibt eine Beobachtung, bspw. ein Auto oder eine Wohnung, die Spalten enthalten deren Merkmale
- **Features `X`:** Eingabevariablen, mit denen das Modell arbeitet, bspw. Hubraum und Zylinderzahl
- **Zielvariable `y`:** Gesuchte Antwort beim überwachten Lernen, bspw. der tatsächliche Treibstoffverbrauch
- **Modell:** Gelernte Regel, die Eingaben mit einer Vorhersage verbindet → `ŷ = f̂(X)`, dabei ist `ŷ` der vorhergesagte und `y` der tatsächliche Wert
- **Interpretation:** Entscheidend ist, ob das Modell bei neuen, bisher ungesehenen Beobachtungen brauchbare Ergebnisse liefert, gute Ergebnisse auf bereits bekannten Trainingsdaten allein reichen nicht aus

### Supervised Learning – Überwachtes Lernen mit bekannten Antworten

- **Definition:** Das Modell lernt anhand von Beispielen, bei denen sowohl die Features `X` als auch die zugehörige Zielvariable `y` bekannt sind
- **Zweck und Anwendung:** Verwenden, wenn eine bestimmte Zahl oder Klasse für neue Beobachtungen vorhergesagt werden soll
- **Input:** Trainingsdaten mit Eingaben und bekannten Antworten, bei neuen Beobachtungen werden dem Modell nur die benötigten Eingaben übergeben
- **Bsp.:** Aus Hubraum, Zylinderzahl und bekanntem Verbrauch vorhandener Autos lernen, um den Verbrauch weiterer Autos vorherzusagen
- **Zwei Hauptaufgaben:** Regression für Zahlenwerte und Klassifikation für Kategorien

### Regression – Einen Zahlenwert vorhersagen

- **Definition:** Supervised Learning mit einer numerischen Zielvariable, deren Zahlenwert vorhergesagt werden soll
- **Wann anwenden:** Bei Fragen wie «Wie viel?», «Wie hoch?» oder «Wie lange?», bspw. Mietpreis in CHF oder Verbrauch in L/100 km
- **Methoden aus den Aufgaben:** Lineare Regression, Regressionsbaum, Random Forest Regression und Gradient Boosting Regression
- **Output:** Ein geschätzter Zahlenwert pro Beobachtung, bspw. ein vorhergesagter Mietpreis
- **Interpretation:** Vorhersagen mit tatsächlichen Werten vergleichen, RMSE beschreibt die Fehlergrösse in der Einheit der Zielvariable und R² die Leistung relativ zur Mittelwertvorhersage

### Klassifikation – Eine Klasse vorhersagen

- **Definition:** Supervised Learning mit einer kategorialen Zielvariable, jede Beobachtung soll einer Klasse zugeordnet werden
- **Wann anwenden:** Bei Fragen wie «Welche Klasse?» oder «Ja oder nein?», bspw. Service nötig oder nicht nötig
- **Binäre Klassifikation:** Zwei Klassen, bspw. `0 = kein Service` und `1 = Service nötig`
- **Mehrklassenklassifikation:** Mehr als zwei Klassen, bspw. verschiedene Tierarten
- **Methoden aus den Aufgaben:** Logistische Regression, Klassifikationsbaum, Random Forest Classification und Gradient Boosting Classification
- **Output:** Eine vorhergesagte Klasse und je nach Methode zusätzlich Wahrscheinlichkeiten für die Klassen
- **Interpretation:** Die vorhergesagten Klassen mit den tatsächlichen Klassen vergleichen, bspw. über Confusion Matrix, Accuracy, Precision und Recall, ROC/AUC bewertet die Rangfolge anhand von Wahrscheinlichkeiten oder Scores
- **Stolperstein:** Zahlen können Kategorien codieren, `0` und `1` als Serviceklassen machen die Aufgabe nicht zur Regression, auch die logistische Regression ist trotz ihres Namens ein Klassifikationsverfahren

### Unsupervised Learning – Unüberwachtes Lernen ohne Zielvariable

- **Definition:** Das Modell erhält nur Features `X`, keine vorgegebenen Antworten `y`, und sucht selbst nach Strukturen in den Daten
- **Zweck und Anwendung:** Verwenden, wenn ähnliche Beobachtungen gruppiert oder viele Merkmale kompakt zusammengefasst werden sollen
- **Clustering:** Ähnliche Beobachtungen zu Gruppen zusammenfassen, bspw. Kundensegmente anhand ihres Kaufverhaltens, Methoden sind k-Means und hierarchisches Clustering
- **Dimensionsreduktion:** Viele Merkmale durch wenige neue Merkmale zusammenfassen, bspw. mit PCA, um Daten übersichtlicher darzustellen
- **Output:** Bspw. Clusterzugehörigkeiten oder neue zusammenfassende Merkmale
- **Interpretation:** Prüfen, ob Gruppen oder Zusammenfassungen stabil, nachvollziehbar und für die Fragestellung sinnvoll sind, Cluster haben zunächst keine vorgegebene fachliche Bedeutung
- **Unterschied zur Klassifikation:** Klassifikation lernt bereits bekannte Klassen, Clustering sucht Gruppen ohne vorgegebene Klassenzuordnung

### Reinforcement Learning – Lernen durch Belohnung

- **Definition:** Ein Agent führt Aktionen in einer Umgebung aus und lernt anhand der erhaltenen Belohnungen eine Strategie
- **Zweck und Anwendung:** Entscheidungen über mehrere Schritte verbessern, bspw. bei Spielen oder der Steuerung eines Roboters
- **Input und Output:** Der Agent beobachtet die Umgebung, wählt Aktionen und erhält Rückmeldungen, daraus lernt er eine Handlungsstrategie
- **Interpretation:** Eine gute Strategie erzielt langfristig eine hohe kumulierte Belohnung, die unmittelbar grösste Belohnung muss dafür nicht die beste Entscheidung sein
- **Einordnung:** Gehört zu den drei Lernparadigmen, wird gemäss Folien im Kurs nicht weiter behandelt

# Model training & evaluation

## Loss Functions – Wie stark wird eine falsche Vorhersage bestraft?

- **Definition:** Eine Loss Function (Verlustfunktion) bewertet den Fehler einer einzelnen Vorhersage, ein kleiner Loss bedeutet eine gute Vorhersage gemäss der gewählten Funktion
- **Zweck und Anwendung:** Beim Modelltraining festlegen, welche Fehler wie stark zählen, das Modell soll sein Trainingsziel möglichst klein machen
- **Input:** Tatsächlicher Wert `y` und Vorhersage `ŷ`, bei bestimmten Klassifikationsverfahren stattdessen eine vorhergesagte Wahrscheinlichkeit oder ein Score
- **Fehler bei Regression:** `e = y − ŷ`, ein positiver Fehler bedeutet unterschätzt, ein negativer Fehler bedeutet überschätzt
- **Squared Loss – Quadratischer Verlust:** `L = (y − ŷ)²`, grosse Abweichungen werden besonders stark bestraft, bspw. Fehler 2 → Loss 4 und Fehler 10 → Loss 100
- **Absolute Loss – Absoluter Verlust:** `L = |y − ŷ|`, Fehler zählen entsprechend ihrer absoluten Grösse, bspw. Fehler −3 → Loss 3, weniger empfindlich gegenüber Extremfehlern als Squared Loss
- **0–1 Loss – Richtig oder falsch:** Richtige Klasse → Loss 0, falsche Klasse → Loss 1, berücksichtigt weder die Sicherheit der Vorhersage noch unterschiedliche Fehlerkosten
- **Log-Loss – Wahrscheinlichkeiten bewerten:** Bestraft besonders stark, wenn das Modell der tatsächlichen Klasse eine sehr kleine Wahrscheinlichkeit zuordnet, bspw. bei tatsächlich Klasse 1 ist `p = 0.9` besser als `p = 0.1`, wird bei logistischer Regression verwendet
- **Hinge Loss – Klassifikation mit Sicherheitsabstand:** Bestraft falsche Vorhersagen und auch richtige Vorhersagen mit zu geringem Abstand zur Entscheidungsgrenze, wird bei SVM verwendet
- **Interpretation:** Kleinere Loss-Werte sind besser, ein Modell mit kleinem Trainings-Loss kann trotzdem bei neuen Beobachtungen schlecht abschneiden
- **Stolperstein:** Die Zahlenwerte verschiedener Loss Functions sind nicht direkt vergleichbar, sie bewerten Fehler unterschiedlich und können unterschiedliche Einheiten haben

## Loss, Trainingsziel und Bewertungskennzahl – Drei unterschiedliche Rollen
- **Loss:** Bewertet eine einzelne Vorhersage, bspw. ein quadrierter Fehler für ein Auto
- **Trainingsziel (Objective):** Grösse, die der Lernalgorithmus optimiert, häufig Summe oder Mittelwert der Loss-Werte, ggf. ergänzt um eine Strafe für Modellkomplexität
- **Bewertungskennzahl (Evaluation Metric):** Fasst zusammen, wie gut das trainierte Modell auf ausgewählten Daten abschneidet, bspw. RMSE, R², Accuracy, F1 oder AUC
- **Wann anwenden:** Beim Lesen einer Aufgabenstellung oder eines Modell-Outputs unterscheiden, wie das Modell lernt und wie seine Leistung anschliessend beurteilt wird
- **Bsp. lineare Regression:** OLS minimiert beim Training die Summe der quadrierten Fehler, anschliessend beurteilen wir die Testvorhersagen bspw. mit RMSE und R²
- **Bsp. logistische Regression:** Das Training basiert auf Log-Loss, die Bewertung kann dagegen bspw. mit Recall oder AUC erfolgen
- **Bsp. Entscheidungsbaum:** Wählt schrittweise Splits, welche SSE bei Regression bzw. Gini oder Entropie bei Klassifikation möglichst stark reduzieren, die abschliessende Bewertung erfolgt mit passenden Regressions- oder Klassifikationskennzahlen
- **Zusammenhang mit MSE:** Squared Loss bewertet einen einzelnen quadrierten Fehler, MSE ist der Mittelwert dieser quadrierten Fehler über mehrere Beobachtungen
- **Interpretation:** Das optimierte Trainingsziel und die ausgegebene Bewertungskennzahl müssen nicht identisch sein, auch die Richtung unterscheidet sich, bspw. kleiner RMSE ist besser, grösseres R² ist besser
- **Generalisierung:** Gesucht ist eine gute Leistung auf neuen, ungesehenen Daten, diese schätzen wir mit zurückgehaltenen Testdaten oder Cross-Validation
- **Stolperstein:** Trainingswerte sind häufig zu optimistisch, weil das Modell dieselben Beobachtungen bereits zum Lernen verwendet hat

## MSE und RMSE – Fehler numerischer Vorhersagen beurteilen

- **Definition:** MSE und RMSE messen, wie stark numerische Vorhersagen von den tatsächlichen Werten abweichen, bspw. beim vorhergesagten Treibstoffverbrauch oder Mietpreis
- **Zweck und Anwendung:** Die Vorhersagequalität einer Regression beurteilen und Modelle auf denselben Beobachtungen miteinander vergleichen
- **Input und Voraussetzungen:** Tatsächliche Werte und Vorhersagen müssen numerisch, vollständig, gleich lang und in derselben Reihenfolge angeordnet sein, sodass jedes Wertepaar dieselbe Beobachtung beschreibt
- **Fehler:** Tatsächlicher Wert minus Vorhersage, also `observed - predicted` → positiv bedeutet unterschätzt, negativ bedeutet überschätzt
- **MSE (Mean Squared Error):** Jeden Fehler quadrieren und anschliessend den Mittelwert bilden → `np.mean(errors ** 2)` verhindert, dass sich positive und negative Fehler aufheben, und gewichtet grosse Abweichungen stärker
- **RMSE (Root Mean Squared Error):** Quadratwurzel des MSE → `np.sqrt(mse)` bringt den Fehler zurück in die Einheit der Zielvariable
- **Bsp. aus der Aufgabe:** Tatsächliche Verbrauchswerte `[6, 10]` und Vorhersagen `[5, 7]` ergeben Fehler `[1, 3]` → MSE = `(1² + 3²) / 2 = 5`, RMSE = `√5 ≈ 2.24`
- **Interpretation:** Der RMSE beträgt rund `2.24 L/100 km` und beschreibt die quadratisch gemittelte Fehlergrösse, nicht die Abweichung jeder einzelnen Vorhersage und auch nicht den mittleren absoluten Fehler
- **Wertebereich und Einheit:** Beide Kennzahlen liegen zwischen 0 und unbegrenzt, kleinere Werte sind besser, 0 bedeutet ausschliesslich exakte Vorhersagen → MSE hier in `(L/100 km)²`, RMSE in `L/100 km`
- **Stolpersteine:** MSE und RMSE zeigen keine Über- oder Unterschätzungsrichtung und reagieren empfindlich auf einzelne grosse Fehler, ob ein RMSE gut ist, hängt vom Anwendungsfall und einer Vergleichsbasis ab
- **Prüfung:** Für die Leistung auf unbekannten Daten tatsächliche Testwerte mit den zugehörigen Testvorhersagen vergleichen, ein kleiner Trainingsfehler allein belegt keine gute Vorhersagequalität

```python
#Verwenden, wenn tatsächliche Zahlenwerte und zugehörige Vorhersagen vorliegen und ihre Fehlergrösse beurteilt werden soll
import numpy as np                                                              #Einmaliger Import für numerische Berechnungen

observed = np.array([6, 10])                                                    #Tatsächliche Verbrauchswerte aus der Aufgabe in L/100 km
predicted = np.array([5, 7])                                                    #Vorhersagen derselben Autos in derselben Reihenfolge

errors = observed - predicted                                                   #Fehler je Auto berechnen, positiv = unterschätzt
mse = np.mean(errors ** 2)                                                      #Fehler quadrieren und mitteln, grosse Fehler zählen stärker
rmse = np.sqrt(mse)                                                             #Wurzel ziehen, damit die Einheit wieder L/100 km ist

print('Einzelne Fehler (L/100 km):', errors)                                    #Richtung und Grösse der einzelnen Abweichungen anzeigen
print(f'MSE: {mse:.2f} (L/100 km)²')                                            #Mittleren quadratischen Fehler anzeigen
print(f'RMSE: {rmse:.2f} L/100 km')                                             #Fehlergrösse in der ursprünglichen Verbrauchseinheit anzeigen

#ANPASSEN: observed durch tatsächliche Werte ersetzen, bei einer Testbewertung bspw. y_test
#ANPASSEN: predicted durch die zugehörigen Vorhersagen ersetzen, gleiche Reihenfolge und Länge sicherstellen
#ANPASSEN: L/100 km durch die Einheit der Zielvariable ersetzen, beim MSE die quadrierte Einheit verwenden
```

## Train/Test-Split – Daten zum Lernen und zur abschliessenden Bewertung aufteilen

- **Definition:** Teilt vorhandene Daten in Trainingsdaten zum Lernen und zurückgehaltene Testdaten zur abschliessenden Bewertung auf
- **Zweck und Anwendung:** Vor dem Modelltraining verwenden, um später zu prüfen, wie gut das Modell bei bisher ungesehenen Beobachtungen vorhersagt
- **Input und Voraussetzungen:** Features `X` und zugehörige Zielwerte `y`, jede Zeile in `X` muss zur gleichen Beobachtung in `y` gehören, fehlende Werte in den benötigten Spalten werden hier gemäss Aufgabe vorab entfernt
- **Features `X`:** Eingaben als zweidimensionaler DataFrame, hier `engine_size` und `cylinders`, doppelte eckige Klammern wählen mehrere Spalten aus
- **Zielvariable `y`:** Gesuchte Antwort als eindimensionale Series, hier `combined`, der kombinierte Stadt-/Autobahnverbrauch in L/100 km
- **`train_test_split(X, y, ...)`:** Teilt Features und Zielwerte gemeinsam auf, damit ihre Zuordnung erhalten bleibt, standardmässig werden die Zeilen zuvor zufällig gemischt
- **`test_size=0.2`:** Reserviert 20 % für den Test, die übrigen 80 % dienen dem Training, der Parameter bezeichnet den Testanteil
- **`random_state=42`:** Macht die zufällige Aufteilung bei unveränderten Daten reproduzierbar, 42 ist ein frei gewählter Startwert und keine Qualitätseinstellung
- **Output:** `X_train` enthält die Trainingsmerkmale und `y_train` deren Zielwerte, `X_test` enthält die Testmerkmale und `y_test` deren tatsächliche Zielwerte
- **Interpretation im Autodatensatz:** 701 vollständige Autos ergeben 560 Trainingsautos und 141 Testautos, die Aufteilung selbst trainiert und bewertet noch kein Modell
- **Weitere Verwendung:** Mit `fit(X_train, y_train)` lernen, mit `predict(X_test)` vorhersagen und die Vorhersagen anschliessend mit `y_test` vergleichen
- **Validierung:** Modelle und Einstellungen innerhalb der Trainingsdaten vergleichen, bspw. mit Cross-Validation, die Testdaten für die abschliessende Bewertung zurückhalten
- **Stolpersteine:** Die Zielvariable gehört nicht zusätzlich in die Features, aus Daten gelernte Vorverarbeitung wie Skalierung nur auf Trainingsdaten anpassen, bei Zeitreihen oder zusammengehörenden Beobachtungen ist eine zufällige Zeilenaufteilung ggf. ungeeignet

```python
#Verwenden, wenn ein numerischer Zielwert vorhergesagt werden soll und Trainings- sowie Testdaten vorbereitet werden müssen
import pandas as pd                                                             #Einmaliger Import zum Einlesen und Bearbeiten der Daten
from sklearn.model_selection import train_test_split                            #Funktion zur gemeinsamen Aufteilung von Features und Zielwerten

df_ap03spick_cars = pd.read_csv(
    'car_fuel_consumption_2025.csv'
)                                                                               #Autodaten einmal laden, Pfad relativ zum Arbeitsverzeichnis des Kernels

cars_features = ['engine_size', 'cylinders']                                    #Eingaben festlegen, Hubraum und Anzahl Zylinder
cars_target = 'combined'                                                        #Zielvariable festlegen, Verbrauch in L/100 km

df_ap03spick_cars_model = df_ap03spick_cars.dropna(
    subset=cars_features + [cars_target]
).copy()                                                                        #Nur Zeilen mit vollständigen Modellvariablen übernehmen, Ausgangsdaten erhalten

X_cars = df_ap03spick_cars_model[cars_features]                                 #Features als zweidimensionalen DataFrame auswählen
y_cars = df_ap03spick_cars_model[cars_target]                                   #Zielwerte als eindimensionale Series auswählen

X_train_cars, X_test_cars, y_train_cars, y_test_cars = train_test_split(
    X_cars, y_cars,                                                             #Features und zugehörige Zielwerte gemeinsam aufteilen
    test_size=0.2,                                                              #20 % Testdaten, die übrigen 80 % Trainingsdaten
    random_state=42                                                             #Zufallsaufteilung reproduzierbar machen
)

display(df_ap03spick_cars_model[cars_features + [cars_target]].head())          #Eingaben und Zielvariable nebeneinander kontrollieren
print('Vollständige Beobachtungen:', len(df_ap03spick_cars_model))
print('Trainingsdaten:', len(X_train_cars))                                     #Erwartet bei diesen Daten 560 Autos
print('Testdaten:', len(X_test_cars))                                           #Erwartet bei diesen Daten 141 Autos

#ANPASSEN: Dateipfad und DataFrame-Namen durch den Prüfungsdatensatz ersetzen, vorhandene Daten nicht erneut laden
#ANPASSEN: cars_features und cars_target durch die verlangten Eingabespalten und die Zielvariable ersetzen
#ANPASSEN: test_size und random_state gemäss Aufgabenstellung setzen, vorbereitete Splits direkt weiterverwenden


#Verwenden, um Daten für eine Regression in Trainingsdaten und Testdaten aufzuteilen
import pandas as pd
from sklearn.model_selection import train_test_split

#1. Daten laden
df_ap03spick_cars = pd.read_csv('car_fuel_consumption_2025.csv')                #CSV-Datei als DataFrame laden
                                                                                #ACHTUNG: Ist der DataFrame bereits gegeben, diese Ladezeile weglassen und den vorhandenen Namen verwenden

#2. Eingaben und Ziel festlegen
feature_cols = ['engine_size', 'cylinders']                                     #Mit diesen Spalten soll das Modell vorhersagen
target_col = 'combined'                                                         #Diese Spalte soll das Modell vorhersagen
#ANPASSEN: In feature_cols die verlangten Eingabespalten eintragen
#ANPASSEN: In target_col die vorherzusagende Spalte eintragen
#ACHTUNG: Die Zielspalte darf nicht zusätzlich in feature_cols stehen


#3. Zeilen mit fehlenden Angaben ausschliessen
df_model = df_ap03spick_cars.dropna(
    subset=feature_cols + [target_col]                                          #Nur die für das Modell benötigten Spalten prüfen
).copy()                                                                        #Separate Tabelle erstellen, ursprünglichen DataFrame erhalten
#ANPASSEN: Diese Bereinigung nur verwenden, wenn sie zur Aufgabe passt


#4. Eingaben X und Zielwerte y auswählen
X_cars = df_model[feature_cols]                                                 #X enthält die Eingabespalten, hier Hubraum und Zylinderzahl
y_cars = df_model[target_col]                                                   #y enthält die richtige Antwort, hier den tatsächlichen Verbrauch


#5. Daten gemeinsam aufteilen
X_train_cars, X_test_cars, y_train_cars, y_test_cars = train_test_split(
    X_cars, y_cars,                                                             #Eingaben und Antworten gemeinsam aufteilen, Zuordnung bleibt erhalten
    test_size=0.2,                                                              #20 % zum Testen zurückhalten, 80 % zum Lernen verwenden
    random_state=42                                                             #Bei jeder Ausführung dieselbe zufällige Aufteilung erhalten
)
#ANPASSEN: test_size gemäss Aufgabe setzen, bspw. 0.3 für 30 % Testdaten
#ANPASSEN: random_state gemäss Aufgabe setzen, falls vorgegeben
#ANPASSEN: Falls die Prüfung X_train, X_test, y_train und y_test verlangt, die vier Ausgaben entsprechend umbenennen


#6. Auswahl und Anzahl kontrollieren
display(df_model[feature_cols + [target_col]].head())                           #Eingaben und richtige Antwort nebeneinander anzeigen
print('Verwendete Beobachtungen:', len(df_model))
print('Trainingsdaten:', len(X_train_cars))                                     #Mit diesen Beobachtungen lernt das Modell später
print('Testdaten:', len(X_test_cars))                                           #Diese Beobachtungen werden erst zur abschliessenden Bewertung verwendet

#VORAUSSETZUNG: Eingabespalten sind hier bereits numerisch, Textkategorien müssten passend umgewandelt werden
#ACHTUNG: Sind X_train, X_test, y_train und y_test bereits vorbereitet, Schritte 1–5 überspringen und direkt mit diesen Variablen das Modell trainieren
```

## Cross-Validation (CV) – Ein Modell innerhalb der Trainingsdaten prüfen

- **Definition:** Cross-Validation teilt die Trainingsdaten mehrfach in einen Teil zum Lernen und einen Teil zum Validieren auf, so wird das Modell auf Beobachtungen geprüft, die es im jeweiligen Durchlauf nicht zum Lernen verwendet hat
- **Zweck und Anwendung:** Die Vorhersagequalität abschätzen sowie verschiedene Modelle oder Einstellungen vergleichen, ohne dafür die abschliessenden Testdaten zu verwenden
- **Input und Voraussetzungen:** Ein Modell sowie zusammengehörende Trainingsmerkmale und Trainingszielwerte, hier `X_train_cars` und `y_train_cars` aus dem vorherigen Eintrag
- **5-fold CV:** Die Trainingsdaten werden in fünf ungefähr gleich grosse Teile (Folds) aufgeteilt, pro Durchlauf lernt ein neues Modell auf vier Teilen und wird auf dem fünften geprüft, jeder Teil dient einmal zur Validierung
- **`cross_val_score()`:** Übernimmt das wiederholte Trainieren und Bewerten automatisch, ein vorheriger Aufruf von `fit()` ist nicht erforderlich
- **`cv=5`:** Führt fünf Durchläufe durch und liefert fünf Bewertungswerte, bei dieser Regression werden die Folds standardmässig ohne zusätzliches Mischen gebildet
- **`scoring='neg_mean_squared_error'`:** Liefert den negativen MSE, weil scikit-learn bei Scores grundsätzlich grössere Werte als besser behandelt
- **`np.sqrt(-scores)`:** Entfernt zuerst das Minuszeichen und zieht danach für jeden Fold die Wurzel → fünf RMSE-Werte in der Einheit der Zielvariable
- **`.mean()`:** Berechnet den Durchschnitt der fünf Fold-RMSE, genau wie in der Dozentenlösung
- **Output und Interpretation:** Die einzelnen RMSE zeigen, wie gut das Modell in den verschiedenen Validierungsteilen abschneidet, der mittlere CV-RMSE fasst diese Ergebnisse zusammen, kleinere Werte sind besser
- **Bsp. zur Interpretation:** Ein angenommener mittlerer CV-RMSE von `1.7 L/100 km` beschreibt die Fehlergrösse auf zurückgehaltenen Validierungsautos, nicht einen Fehler von genau 1.7 bei jedem Auto
- **Unterschiede zwischen Folds:** Stark unterschiedliche RMSE zeigen, dass das Ergebnis von den verwendeten Lern- und Validierungsbeobachtungen abhängt
- **Stolpersteine:** Keine Testdaten in diese CV übergeben, eine nötige Skalierung oder andere gelernte Vorverarbeitung muss innerhalb jedes Folds gelernt werden, dafür verwendet man eine Pipeline
- **Weiteres Vorgehen:** Nach der Modellauswahl das gewählte Modell auf allen Trainingsdaten trainieren und abschliessend auf den Testdaten bewerten, `cross_val_score()` liefert die Scores und kein fertig trainiertes Modell zur weiteren Verwendung zurück

```python
#Verwenden, um die Qualität eines Regressionsmodells mit Cross-Validation auf den Trainingsdaten zu prüfen
from sklearn.linear_model import LinearRegression                               #Lineares Regressionsmodell
from sklearn.model_selection import cross_val_score                             #Modell mehrfach trainieren und auf zurückgehaltenen Teilen prüfen
#VORAUSSETZUNG: NumPy ist als np importiert, X_train_cars und y_train_cars sind bereits vorbereitet


#1. Modell festlegen
cv_model_cars = LinearRegression()                                              #Hier soll eine lineare Regression geprüft werden
#ANPASSEN: LinearRegression() durch das verlangte Regressionsmodell inkl. Einstellungen ersetzen


#2. Cross-Validation durchführen
cv_scores_cars = cross_val_score(
    cv_model_cars,                                                              #Dieses Modell wird in jedem Durchlauf neu trainiert
    X_train_cars,                                                               #Nur Eingaben der Trainingsdaten verwenden
    y_train_cars,                                                               #Zugehörige richtige Antworten der Trainingsdaten verwenden
    cv=5,                                                                       #Fünf Durchläufe, jeweils vier Teile lernen und einen Teil prüfen
    scoring='neg_mean_squared_error'                                            #Pro Durchlauf den negativen MSE zurückgeben
)
#ANPASSEN: X_train_cars und y_train_cars durch die Trainingsvariablen aus der Prüfung ersetzen
#ANPASSEN: cv gemäss Aufgabe setzen, scoring und die Umrechnung unten müssen zusammenpassen


#3. Negative MSE in RMSE umrechnen
cv_rmse_cars = np.sqrt(-cv_scores_cars)                                         #Minuszeichen entfernen, danach Wurzel ziehen
mean_cv_rmse_cars = cv_rmse_cars.mean()                                         #Durchschnitt der fünf einzelnen RMSE berechnen


#4. Ergebnisse anzeigen
print('RMSE je Fold:', cv_rmse_cars.round(2))                                   #Zeigt die Fehlergrösse in jedem Validierungsteil
print(f'Mittlerer CV-RMSE: {mean_cv_rmse_cars:.2f} L/100 km')
#ANPASSEN: L/100 km durch die Einheit der Zielvariable ersetzen

#ACHTUNG: Bei einer anderen Bewertungskennzahl die RMSE-Umrechnung in Schritt 3 nicht ungeprüft übernehmen
#ACHTUNG: Dieser Block benötigt kein vorheriges fit(), die CV übernimmt das Training selbst
```

## Underfitting und Overfitting – Zu wenig oder zu viel aus Trainingsdaten lernen

- **Definition:** Underfitting bedeutet, dass ein Modell wichtige Muster nicht ausreichend erfasst, Overfitting bedeutet, dass es sich zu stark an Besonderheiten und zufällige Schwankungen der Trainingsdaten anpasst
- **Zweck und Anwendung:** Trainings- und Validierungsergebnisse vergleichen, um zu beurteilen, ob die Modellkomplexität zur Aufgabe passt, bspw. bei der Wahl der maximalen Baumtiefe
- **Input und Voraussetzungen:** Vergleichbare Kennzahlen auf Trainingsdaten und zurückgehaltenen Validierungsdaten, bspw. Trainings-RMSE und CV-RMSE, bei beiden gilt kleiner = besser
- **Underfitting – Modell zu einfach:** Bereits auf den Trainingsdaten entstehen grosse Fehler, auch der Validierungsfehler ist hoch, bspw. ein Baum mit nur einer Aufteilung, der wichtige Unterschiede zwischen Autos nicht erfasst
- **Passendes Modell:** Erfasst wesentliche Muster und liefert auch auf Validierungsdaten brauchbare Vorhersagen, Trainings- und Validierungsfehler sind im Anwendungskontext ausreichend klein
- **Overfitting – Zu starke Anpassung:** Sehr kleiner Trainingsfehler, aber deutlich grösserer Validierungsfehler, bspw. ein sehr tiefer Baum, der einzelne Trainingsbeobachtungen nahezu auswendig lernt
- **Interpretation beim Erhöhen der Komplexität:** Sinkt der Trainingsfehler weiter, während der CV-Fehler wieder steigt, spricht das für zunehmendes Overfitting
- **Interpretation bei sinkendem CV-Fehler:** Sinkt neben dem Trainingsfehler auch der CV-Fehler, verbessert die zusätzliche Komplexität hier die Validierungsleistung, ein komplexeres Modell ist nicht automatisch overfitted
- **Interpretation ähnlicher Fehler:** Ähnliche Trainings- und CV-Fehler allein bedeuten noch kein gutes Modell, beide können hoch sein und auf Underfitting hindeuten
- **Stolperstein:** Ein gewisser Abstand zwischen Trainings- und CV-Fehler ist üblich, es gibt keinen allgemein gültigen Grenzwert, ab dem Overfitting sicher vorliegt
- **Bei Underfitting:** Ggf. ein flexibleres Modell, mehr erlaubte Baumtiefe oder geeignetere Features verwenden
- **Bei Overfitting:** Ggf. die Komplexität begrenzen, einen Baum beschneiden (Pruning), Regularisierung einsetzen oder zusätzliche geeignete Trainingsdaten nutzen
- **Prüfung:** Modell und Einstellungen anhand der Validierung bzw. CV auswählen, der kleinste Trainingsfehler allein ist kein Auswahlkriterium, die abschliessenden Testdaten nicht wiederholt zur Optimierung verwenden

## Bias–Variance Trade-off – Modellkomplexität und Vorhersagefehler ausbalancieren

- **Definition:** Der Bias–Variance Trade-off beschreibt den Ausgleich zwischen systematischen Fehlern eines Modells und seiner Empfindlichkeit gegenüber unterschiedlichen Trainingsstichproben
- **Zweck und Anwendung:** Verstehen, weshalb ein einfacheres oder komplexeres Modell nicht automatisch besser ist, und die Modellkomplexität mithilfe von Validierung bzw. CV auswählen
- **Bias – Systematischer Fehler:** Entsteht durch zu einfache oder unpassende Modellannahmen, das Modell erfasst den tatsächlichen Zusammenhang nicht ausreichend → hoher Bias kann zu Underfitting führen
- **Bsp. für hohen Bias:** Für jede Wohnung denselben durchschnittlichen Mietpreis vorhersagen, obwohl sich die Preise nach Wohnfläche und Lage unterscheiden
- **Variance – Schwankung des gelernten Modells:** Beschreibt, wie stark sich die Vorhersagen verändern würden, wenn das Modell mit einer anderen Trainingsstichprobe trainiert würde → hohe Variance kann zu Overfitting führen
- **Bsp. für hohe Variance:** Ein sehr tiefer Baum passt sich stark an einzelne Trainingswohnungen an, mit anderen Trainingswohnungen können deutlich andere Regeln und Vorhersagen entstehen
- **Noise – Unvermeidbares Rauschen:** Zufällige Schwankungen oder Einflüsse, die mit den vorhandenen Features nicht vorhergesagt werden können, auch ein gutes Modell kann diesen Anteil nicht vollständig beseitigen
- **Zusammenhang bei quadratischem Fehler:** Der erwartete Vorhersagefehler lässt sich unter den entsprechenden Annahmen in `Bias² + Variance + Noise` zerlegen
- **Typische Tendenz:** Mehr Modellkomplexität kann den Bias senken, erhöht aber häufig die Variance, weniger Komplexität kann die Vorhersagen stabilisieren, aber wichtige Muster übersehen
- **Interpretation:** Gesucht ist eine möglichst kleine Summe der Fehleranteile, weder möglichst wenig Bias noch möglichst wenig Variance allein
- **Stolperstein:** Variance meint hier die Schwankung der Modellvorhersagen zwischen verschiedenen Trainingsstichproben, nicht die Streuung der Zielwerte innerhalb eines Datensatzes
- **Bias reduzieren:** Ggf. geeignetere Features oder ein flexibleres Modell verwenden
- **Variance reduzieren:** Ggf. mehr geeignete Trainingsdaten, ein einfacheres Modell, Regularisierung oder die Mittelung mehrerer Modelle verwenden, bspw. Bagging
- **Prüfung:** Bias und Variance werden durch einen einzelnen Trainings-/Testvergleich nicht direkt berechnet, Trainings- und CV-Ergebnisse liefern Hinweise auf Underfitting und Overfitting, Einstellungen anhand der CV-Leistung vergleichen

## Einfache lineare Regression mit OLS – Eine Gerade an Daten anpassen

- **Definition:** Die einfache lineare Regression beschreibt den Zusammenhang zwischen einer Eingabevariable `x` und einer numerischen Zielvariable `y` durch eine Gerade → `ŷ = intercept + slope × x`
- **Zweck und Anwendung:** Einen Zahlenwert anhand eines Merkmals schätzen, bspw. den monatlichen Mietpreis anhand der Wohnfläche, wenn eine Gerade den Zusammenhang sinnvoll beschreibt
- **Input und Voraussetzungen:** Zusammengehörende numerische x- und y-Werte, vollständig und in derselben Reihenfolge, x muss mindestens zwei unterschiedliche Werte enthalten
- **OLS (Ordinary Least Squares):** Wählt Steigung und Achsenabschnitt so, dass die Summe der quadrierten Residuen möglichst klein wird
- **Residuum:** Tatsächlicher Wert minus Modellvorhersage → `y − ŷ`, positiv bedeutet unterschätzt, negativ bedeutet überschätzt
- **SSE (Sum of Squared Errors):** Summe der quadrierten Residuen, im Gegensatz zum MSE wird nicht durch die Anzahl Beobachtungen geteilt
- **Steigung `slope`:** `Σ((x − x̄) × (y − ȳ)) / Σ((x − x̄)²)` → geschätzte Veränderung von y pro zusätzlicher Einheit x
- **Achsenabschnitt `intercept`:** `ȳ − slope × x̄` → vorhergesagter y-Wert bei `x = 0`, kann ausserhalb des sinnvollen Anwendungsbereichs liegen
- **Bsp. aus der Aufgabe:** Wohnflächen `[40, 60, 80]` m² und Mieten `[1200, 1600, 2000]` CHF ergeben `slope = 20` und `intercept = 400` → vorhergesagte Miete = `400 + 20 × Wohnfläche`
- **Interpretation der Steigung:** Ein zusätzlicher m² ist im Modell mit durchschnittlich 20 CHF mehr Monatsmiete verbunden, das belegt keinen kausalen Einfluss
- **Interpretation des Achsenabschnitts:** Bei 0 m² würde das Modell 400 CHF vorhersagen, eine Wohnung mit 0 m² liegt aber ausserhalb der beobachteten Daten und ist hier fachlich nicht sinnvoll
- **Vorhersage für 95 m²:** `400 + 20 × 95 = 2300 CHF`, dies ist Extrapolation, weil 95 m² ausserhalb der beobachteten 40–80 m² liegen
- **Interpolation:** Vorhersage innerhalb des beobachteten x-Bereichs, bspw. für 70 m²
- **Stolpersteine:** Extrapolation ist besonders unsicher, Ausreisser können die OLS-Gerade stark beeinflussen, eine gute Anpassung an drei Trainingspunkte belegt keine gute Vorhersagequalität für neue Wohnungen

```python
#Verwenden, wenn eine einfache OLS-Gerade aus gegebenen Zahlenpaaren berechnet und ein neuer Wert vorhergesagt werden soll
#VORAUSSETZUNG: NumPy ist bereits als np importiert


#1. Zusammengehörende Eingaben und Zielwerte eintragen
ols_demo_x = np.array([40, 60, 80])                                             #Wohnflächen in m²
ols_demo_y = np.array([1200, 1600, 2000])                                       #Zugehörige Monatsmieten in CHF, gleiche Reihenfolge wie die Wohnflächen
#ANPASSEN: Beide Zahlenlisten durch die gegebenen Werte ersetzen, jedes x muss zum y an derselben Position gehören


#2. Steigung und Achsenabschnitt berechnen
ols_demo_x_mean = ols_demo_x.mean()                                             #Durchschnittliche Wohnfläche
ols_demo_y_mean = ols_demo_y.mean()                                             #Durchschnittliche Monatsmiete

ols_demo_numerator = np.sum(
    (ols_demo_x - ols_demo_x_mean)
    * (ols_demo_y - ols_demo_y_mean)
)                                                                               #Abweichungen von den Mittelwerten paarweise multiplizieren und aufsummieren

ols_demo_denominator = np.sum(
    (ols_demo_x - ols_demo_x_mean) ** 2
)                                                                               #Abweichungen der Wohnflächen vom Mittelwert quadrieren und aufsummieren

ols_demo_slope = ols_demo_numerator / ols_demo_denominator                      #Steigung der Geraden berechnen
ols_demo_intercept = (
    ols_demo_y_mean - ols_demo_slope * ols_demo_x_mean
)                                                                               #Achsenabschnitt berechnen
#ACHTUNG: Sind alle x-Werte gleich, ist der Nenner 0 und diese Berechnung nicht möglich
#ANPASSEN: In diesem Schritt nichts ändern, wenn du nur die Zahlenlisten in Schritt 1 ersetzt
#ANPASSEN: Falls du ols_demo_x oder ols_demo_y umbenennst, diese Namen auch hier ersetzen
#ANPASSEN: Kommentare zu Wohnfläche und Monatsmiete an die Variablen deiner Aufgabe anpassen


#3. Zielwert für eine neue Eingabe vorhersagen
ols_demo_new_x = 95                                                             #Gesuchte Wohnfläche in m²
ols_demo_prediction = (
    ols_demo_intercept + ols_demo_slope * ols_demo_new_x
)                                                                               #Neue Wohnfläche in die gelernte Gerade einsetzen
#ANPASSEN: 95 durch den Eingabewert aus der Aufgabenstellung ersetzen


#4. Ergebnisse anzeigen
print(f'Steigung: {ols_demo_slope:.2f} CHF pro zusätzlichem m²')                #Änderung des vorhergesagten y-Werts pro zusätzlicher Einheit x
print(f'Achsenabschnitt: {ols_demo_intercept:.2f} CHF')                         #Vorhergesagter y-Wert bei x = 0
print(f'Vorhergesagte Monatsmiete für {ols_demo_new_x} m²: {ols_demo_prediction:.2f} CHF')
#ANPASSEN: CHF durch die Einheit von y und m² durch die Einheit von x ersetzen
#ANPASSEN: Monatsmiete durch die vorhergesagte Grösse ersetzen, bspw. Treibstoffverbrauch
#HINWEIS: :.2f zeigt zwei Nachkommastellen, bei Bedarf bspw. durch :.3f ersetzen

if ols_demo_x.min() <= ols_demo_new_x <= ols_demo_x.max():                      #Prüfen, ob das neue x zwischen dem kleinsten und grössten beobachteten x liegt
    print('Interpolation: Die neue Eingabe liegt innerhalb des beobachteten Bereichs')
else:
    print('Extrapolation: Die neue Eingabe liegt ausserhalb des beobachteten Bereichs')
#ANPASSEN: Die if-Abfrage unverändert übernehmen, wenn du nur die Zahlen in Schritt 1 und 3 ersetzt
#ANPASSEN: Falls du Variablen umbenennst, die entsprechenden Namen auch in den Ausgaben und der if-Abfrage ersetzen

#ANPASSEN: Wohnfläche, Monatsmiete und Einheiten in den Ausgaben durch die Grössen aus deiner Aufgabe ersetzen
```

## Einfache OLS mit statsmodels – Mietpreis anhand der Wohnfläche schätzen
- **Definition:** Passt eine Gerade an die Trainingsdaten an, welche den Mietpreis `price` anhand der Wohnfläche `area` schätzt → `vorhergesagter Preis = Achsenabschnitt + Steigung × Fläche`
- **Zweck und Anwendung:** Verwenden, wenn eine numerische Zielvariable durch genau eine Eingabevariable erklärt oder vorhergesagt werden soll
- **Input und Voraussetzungen:** Trainingstabelle mit numerischer Eingabe- und Zielspalte, hier `train_apartments` mit `area` und `price`, die benötigten Werte sind bereits vollständig
- **`smf.ols('price ~ area', data=train_apartments)`:** Legt das Modell fest, links von `~` steht die Zielvariable, rechts die Eingabevariable, ein Achsenabschnitt wird automatisch eingeschlossen
- **`.fit()`:** Schätzt die Koeffizienten anhand der Trainingsdaten, sodass die Summe der quadrierten Residuen möglichst klein wird
- **`.params`:** Zeigt die geschätzten Koeffizienten, `Intercept` ist der Achsenabschnitt und `area` die Steigung
- **`.rsquared`:** Zeigt das R² auf den Trainingsdaten, bei OLS mit Achsenabschnitt und nicht konstanter Zielvariable liegt es zwischen 0 und 1, grösser bedeutet bessere Anpassung an diese Trainingsdaten
- **`.predict(...)`:** Berechnet mit dem trainierten Modell Vorhersagen für eine Tabelle mit der benötigten Eingabespalte, ein erneutes Training ist dafür nicht nötig
- **Interpretation der Steigung:** Der Koeffizient von `area` beschreibt, um wie viele CHF sich die vorhergesagte Monatsmiete pro zusätzlichem m² verändert, bei einer positiven Steigung steigt sie
- **Interpretation des Achsenabschnitts:** Vorhergesagte Miete bei 0 m², liegt dieser Wert ausserhalb des sinnvollen Datenbereichs, hat er keine brauchbare fachliche Bedeutung
- **Interpretation von R²:** Ein angenommener Wert von `0.70` bedeutet, dass das Modell 70 % der Mietpreisvariabilität in den Trainingsdaten erklärt, nicht dass 70 % der einzelnen Preise richtig vorhergesagt werden
- **Interpretation der Grafik:** Jeder Punkt ist eine Trainingswohnung, die Linie zeigt die Modellvorhersagen, vertikale Abstände zwischen Punkten und Linie entsprechen den Residuen
- **Stolpersteine:** Ein Zusammenhang belegt keine Ursache, ein hohes Trainings-R² garantiert keine gute Leistung auf neuen Wohnungen, die abschliessende Testbewertung folgt separat

```python
#Verwenden, um einen numerischen Zielwert anhand einer Eingabespalte mit einer OLS-Geraden zu modellieren
import statsmodels.formula.api as smf                                           #OLS über eine Formel wie 'price ~ area' verwenden
import matplotlib.pyplot as plt                                                 #Punkte und Regressionsgerade zeichnen
#VORAUSSETZUNG: pd und np sind importiert, train_apartments ist bereits vorbereitet


#1. Modell festlegen und auf den Trainingsdaten trainieren
ols_simple_apartments = smf.ols(
    'price ~ area',                                                             #Links: vorherzusagende Spalte, rechts: Eingabespalte
    data=train_apartments                                                       #Nur die Trainingswohnungen zum Lernen verwenden
).fit()                                                                         #Steigung und Achsenabschnitt berechnen
#ANPASSEN: 'price ~ area' durch 'deine_zielspalte ~ deine_eingabespalte' ersetzen
#ANPASSEN: train_apartments durch die vorbereitete Trainingstabelle ersetzen
#ACHTUNG: Diese Formel enthält automatisch einen Achsenabschnitt, sm.add_constant() ist hier nicht nötig


#2. Koeffizienten und Trainings-R² anzeigen
display(ols_simple_apartments.params.rename('Koeffizient'))                     #Intercept = Achsenabschnitt, area = Änderung der Miete pro zusätzlichem m²
print(f'Trainings-R²: {ols_simple_apartments.rsquared:.3f}')                    #Anpassung an die Trainingsdaten, noch keine Testbewertung
#ANPASSEN: Falls du den Modellnamen änderst, ihn auch hier und in Schritt 3 ersetzen


#3. Eingabewerte für eine durchgehende Gerade erstellen
ols_plot_apartments = pd.DataFrame({
    'area': np.linspace(
        train_apartments['area'].min(),                                         #Beim kleinsten Flächenwert im Training beginnen
        train_apartments['area'].max(),                                         #Beim grössten Flächenwert im Training aufhören
        100                                                                     #100 gleichmässig verteilte Flächenwerte erstellen
    )
})
ols_plot_apartments['predicted_price'] = (
    ols_simple_apartments.predict(ols_plot_apartments)
)                                                                               #Für jeden Flächenwert die Miete mit dem trainierten Modell schätzen
#ANPASSEN: 'area' überall in Schritt 3 durch deine Eingabespalte ersetzen
#HINWEIS: predicted_price ist eine neue Ergebnisspalte nur in der separaten Zeichentabelle


#4. Trainingswohnungen und gelernte Gerade darstellen
fig, ax = plt.subplots(figsize=(8, 4))

ax.scatter(
    train_apartments['area'],                                                   #Wohnfläche auf der x-Achse
    train_apartments['price'],                                                  #Tatsächliche Monatsmiete auf der y-Achse
    alpha=0.4, label='Trainingswohnungen'                                       #Durchsichtige Punkte machen Überlagerungen besser sichtbar
)
ax.plot(
    ols_plot_apartments['area'],
    ols_plot_apartments['predicted_price'],
    color='red', label='OLS-Vorhersage'                                         #Gelernte Regressionsgerade zeichnen
)

ax.set_xlabel('Wohnfläche (m²)')
ax.set_ylabel('Monatsmiete (CHF)')
ax.set_title('Monatsmiete anhand der Wohnfläche')
ax.legend()
plt.tight_layout()
plt.show()

#ANPASSEN: In Schritt 4 DataFrame-Namen sowie 'area' und 'price' an deine Daten anpassen
#ANPASSEN: Achsenbeschriftungen, Einheiten, Titel und 'Trainingswohnungen' passend zur Aufgabe ersetzen
```

## Konfidenz- und Vorhersageintervalle – Unsicherheit einer OLS-Vorhersage einschätzen
- **Definition:** Ein Intervall ergänzt die einzelne Modellvorhersage um einen Bereich, der die Unsicherheit berücksichtigt
- **Zweck und Anwendung:** Verwenden, wenn nicht nur ein geschätzter Wert, sondern auch ein Unsicherheitsbereich verlangt wird, dabei zwischen dem durchschnittlichen Zielwert und einer einzelnen neuen Beobachtung unterscheiden
- **Input und Voraussetzungen:** Ein bereits trainiertes OLS-Modell und eine Tabelle mit den neuen Eingabewerten, die Spaltennamen müssen zur Modellformel passen, hier `area`
- **Konfidenzintervall (CI):** Beschreibt die Unsicherheit bei der Schätzung des durchschnittlichen Mietpreises für Wohnungen mit einer bestimmten Fläche
- **Vorhersageintervall (PI):** Beschreibt die Unsicherheit für den Mietpreis einer einzelnen neuen Wohnung mit dieser Fläche
- **Unterschied:** Das Vorhersageintervall ist breiter, weil es zusätzlich die Streuung einzelner Wohnungen um den geschätzten Durchschnitt berücksichtigt
- **`.get_prediction(...)`:** Berechnet Vorhersagen inkl. ihrer Unsicherheit für die übergebenen Eingaben
- **`.summary_frame(alpha=0.05)`:** Gibt eine Ergebnistabelle für ein Niveau von 95 % zurück → Niveau = `1 − alpha`, bspw. `alpha=0.01` für 99 %
- **`mean`:** Geschätzter durchschnittlicher Mietpreis, zugleich die Punktvorhersage für eine einzelne Wohnung
- **`mean_ci_lower` und `mean_ci_upper`:** Untere und obere Grenze des Konfidenzintervalls für den durchschnittlichen Mietpreis
- **`obs_ci_lower` und `obs_ci_upper`:** Untere und obere Grenze des Vorhersageintervalls für eine einzelne neue Wohnung
- **Interpretationsbeispiel mit erfundenen Zahlen:** Punktvorhersage 3000 CHF, CI 2900–3100 CHF und PI 2000–4000 CHF → der durchschnittliche Mietpreis ist genauer geschätzt als der Preis einer einzelnen Wohnung
- **Bedeutung von 95 %:** Bei wiederholter Datenerhebung würden unter passenden Modellannahmen ungefähr 95 % der so berechneten Konfidenzintervalle den wahren durchschnittlichen Mietpreis abdecken, die Vorhersageintervalle würden ungefähr 95 % der jeweiligen neuen Einzelwerte abdecken
- **Stolpersteine:** Für eine einzelne Wohnung das Vorhersageintervall verwenden, nicht das engere Konfidenzintervall, die Gültigkeit hängt von den Modellannahmen ab und ist bei Extrapolation besonders unsicher

```python
#Verwenden, wenn für neue Eingaben eine OLS-Vorhersage inkl. Konfidenz- und Vorhersageintervall verlangt wird
#VORAUSSETZUNG: pd ist importiert und ols_simple_apartments wurde bereits auf den Trainingsdaten trainiert

#1. Neue Eingabe als Tabelle erfassen
new_apartments_interval = pd.DataFrame({
    'area': [95]                                                                #Vorhersage für eine Wohnung mit 95 m²
})
#ANPASSEN: 'area' durch die Eingabespalte deiner Modellformel ersetzen, Schreibweise muss genau übereinstimmen
#ANPASSEN: 95 durch den gesuchten Eingabewert ersetzen, mehrere Werte bspw. als [70, 95, 120] eintragen
#ACHTUNG: Verwendet dein Modell mehrere Eingabespalten, müssen hier alle benötigten Spalten enthalten sein


#2. Vorhersage und Intervalle berechnen
intervals_apartments = (
    ols_simple_apartments
    .get_prediction(new_apartments_interval)                                    #Trainiertes Modell auf die neuen Eingaben anwenden
    .summary_frame(alpha=0.05)                                                  #95%-Intervalle berechnen, weil 1 − 0.05 = 0.95
)
#ANPASSEN: ols_simple_apartments durch den Namen deines bereits trainierten OLS-Modells ersetzen
#ANPASSEN: alpha gemäss Aufgabe setzen, 0.05 für 95 % oder 0.01 für 99 %
#HINWEIS: Kein erneutes fit() nötig, das bestehende Modell wird verwendet


#3. Eingaben und passende Ergebnisse nebeneinander anzeigen
interval_results_apartments = pd.concat(
    [
        new_apartments_interval.reset_index(drop=True),                         #Neue Eingaben mit fortlaufender Zeilennummer
        intervals_apartments[
            ['mean', 'mean_ci_lower', 'mean_ci_upper',
             'obs_ci_lower', 'obs_ci_upper']
        ].reset_index(drop=True)                                                #Zugehörige Vorhersagen und Intervallgrenzen
    ],
    axis=1                                                                      #Tabellen nebeneinander verbinden
)

display(interval_results_apartments.round(2))                                   #Mietpreise und Grenzen hier in CHF anzeigen

#HINWEIS: mean = Punktvorhersage, mean_ci_* = Intervall für den Durchschnitt, obs_ci_* = Intervall für einen Einzelwert
#ANPASSEN: .round(2) bei Bedarf ändern, bspw. .round(0) für ganze CHF
```

## Multiple lineare Regression – Einen Zahlenwert anhand mehrerer Merkmale schätzen
- **Definition:** Erweitert die lineare Regression um mehrere Eingabevariablen, hier `vorhergesagte Miete = Achsenabschnitt + Flächenkoeffizient × Fläche + Zimmerkoeffizient × Zimmerzahl`
- **Zweck und Anwendung:** Verwenden, wenn mehrere Merkmale gemeinsam zur Erklärung oder Vorhersage eines numerischen Zielwerts dienen sollen
- **Input und Voraussetzungen:** Trainingstabelle mit vollständigen numerischen Modellvariablen, hier `train_apartments` mit Zielvariable `price` und Eingaben `area` sowie `rooms`
- **`smf.ols('price ~ area + rooms', data=...).fit()`:** Trainiert ein Modell mit zwei Eingaben, das `+` nimmt beide Merkmale auf, der Achsenabschnitt ist automatisch enthalten
- **`.params`:** Zeigt den Achsenabschnitt und einen Koeffizienten je Eingabevariable
- **Interpretation von `area`:** Geschätzte Änderung der Monatsmiete in CHF pro zusätzlichem m² bei gleichbleibender Zimmerzahl
- **Interpretation von `rooms`:** Geschätzte Änderung der Monatsmiete in CHF pro zusätzlichem Zimmer bei gleichbleibender Wohnfläche
- **Ceteris paribus:** Alle anderen im Modell enthaltenen Eingabevariablen werden gedanklich konstant gehalten, die Koeffizienten beschreiben Zusammenhänge und belegen keine Ursache
- **Trainings-R²:** Beschreibt die Anpassung an die Trainingsdaten, bei OLS mit Achsenabschnitt kann es durch zusätzliche Eingaben auf denselben Trainingsbeobachtungen nicht sinken, ein höheres Trainings-R² allein belegt deshalb keine bessere Vorhersagequalität
- **Korrelation der Eingaben:** `.corr()` zeigt hier den linearen Zusammenhang zwischen Wohnfläche und Zimmerzahl, starke Korrelation kann die einzelnen Koeffizienten unsicher und empfindlich gegenüber Änderungen der Daten machen
- **Stolperstein:** Ein Flächenkoeffizient kann sich gegenüber dem einfachen Modell ändern, weil die Zimmerzahl nun zusätzlich berücksichtigt wird, die beiden Koeffizienten beantworten unterschiedliche Fragen
- **Vorhersage:** Für eine neue Wohnung müssen beide Eingaben vorliegen, bspw. Fläche und Zimmerzahl, die Spaltennamen müssen zur Modellformel passen
- **Interpretation des Outputs:** Koeffizienten immer mit Einheit und «bei gleichen anderen Modellmerkmalen» erklären, die abschliessende Vorhersagequalität beurteilen wir später anhand der Testdaten

```python
#Verwenden, wenn ein numerischer Zielwert anhand mehrerer Eingabespalten mit OLS vorhergesagt werden soll
#VORAUSSETZUNG: smf und pd sind importiert, train_apartments ist bereits vorbereitet

#1. Modell mit mehreren Eingaben trainieren
ols_multiple_apartments = smf.ols(
    'price ~ area + rooms',                                                     #Mietpreis aus Wohnfläche und Zimmerzahl vorhersagen
    data=train_apartments                                                       #Nur die Trainingswohnungen zum Lernen verwenden
).fit()                                                                         #Achsenabschnitt und beide Koeffizienten schätzen
#ANPASSEN: Links von ~ die Zielspalte eintragen, rechts die Eingabespalten mit + verbinden
#ANPASSEN: train_apartments durch deine Trainingstabelle ersetzen
#ACHTUNG: Zielspalte nicht zusätzlich rechts von ~ als Eingabe aufnehmen


#2. Koeffizienten und Trainings-R² anzeigen
display(ols_multiple_apartments.params.rename('Koeffizient'))                   #Intercept und Änderung des Preises pro Einheit der jeweiligen Eingabe
print(f'Trainings-R²: {ols_multiple_apartments.rsquared:.3f}')
#ANPASSEN: Falls du den Modellnamen änderst, ihn auch hier und in Schritt 4 ersetzen
#HINWEIS: Jeder Koeffizient gilt bei konstant gehaltenen anderen Eingaben des Modells


#3. Zusammenhang zwischen den Eingabespalten prüfen
display(train_apartments[['area', 'rooms']].corr().round(2))                    #Pearson-Korrelation zwischen Fläche und Zimmerzahl anzeigen
#ANPASSEN: DataFrame und Spaltenliste durch deine Trainingstabelle und numerischen Eingabespalten ersetzen
#ACHTUNG: Hier nur Eingabespalten vergleichen, die Zielspalte gehört nicht in diese Liste


#4. Optional: Eine neue Beobachtung vorhersagen
new_apartment_multiple = pd.DataFrame({
    'area': [95],                                                               #Beispielwohnung mit 95 m²
    'rooms': [3.5]                                                              #Beispielwohnung mit 3.5 Zimmern
})

multiple_prediction_apartments = (
    ols_multiple_apartments.predict(new_apartment_multiple)
)                                                                               #Beide Angaben gemeinsam in das trainierte Modell einsetzen

display(
    new_apartment_multiple.assign(
        predicted_price=multiple_prediction_apartments
    ).round(2)
)                                                                               #Neue Eingaben und vorhergesagte Monatsmiete in CHF nebeneinander anzeigen
#ANPASSEN: In Schritt 4 alle vom Modell benötigten Eingabespalten und die gesuchten Werte eintragen
#ANPASSEN: predicted_price bei Bedarf passend zur Zielvariable benennen
#HINWEIS: Schritt 4 weglassen, wenn keine Vorhersage für neue Beobachtungen verlangt ist
```

## Residuenanalyse – Annahmen der linearen Regression prüfen
- **Definition:** Untersucht die Abweichungen zwischen tatsächlichen Werten und Modellvorhersagen → Residuum = `y − ŷ`
- **Zweck und Anwendung:** Nach dem OLS-Training prüfen, ob Muster in den Fehlern auf eine unpassende Modellform oder verletzte Annahmen hinweisen, besonders relevant für Konfidenzintervalle und statistische Tests
- **Input und Voraussetzungen:** Ein trainiertes OLS-Modell, `.fittedvalues` liefert dessen Vorhersagen für die Trainingsdaten und `.resid` die zugehörigen Residuen
- **Residuenplot:** Zeigt vorhergesagte Werte auf der x-Achse und Residuen auf der y-Achse, jeder Punkt entspricht einer Trainingsbeobachtung
- **Positive Residuen:** Der tatsächliche Wert liegt über der Vorhersage → Modell unterschätzt
- **Negative Residuen:** Der tatsächliche Wert liegt unter der Vorhersage → Modell überschätzt
- **Günstiges Muster:** Ungeordnete Punktwolke um die Nulllinie mit ungefähr gleichbleibender vertikaler Streuung
- **Gekrümmtes Muster:** Hinweis, dass das Modell den durchschnittlichen Zusammenhang nicht ausreichend beschreibt, ggf. fehlen passende nichtlineare Terme
- **Trichterform:** Die Streuung der Residuen verändert sich mit den Vorhersagewerten → Hinweis auf ungleiche Fehlervarianz (Heteroskedastizität)
- **Einzelne weit entfernte Punkte:** Grosse Vorhersagefehler genauer untersuchen, nicht automatisch löschen
- **Q–Q-Plot:** Vergleicht die Quantile der Residuen mit denen einer Normalverteilung, Punkte ungefähr auf der Referenzgeraden sprechen für annähernd normalverteilte Fehler
- **Abweichungen von der Geraden:** Systematische Krümmung oder starke Abweichungen an den Enden können auf Schiefe oder ungewöhnlich ausgeprägte Verteilungsenden hinweisen
- **`sm.qqplot(..., line='s')`:** Zeichnet den Normal-Q–Q-Plot mit einer anhand von Mittelwert und Standardabweichung angepassten Referenzgeraden
- **Normalitätsannahme:** Betrifft die Fehler des Modells, nicht zwingend die ursprünglichen Eingaben oder Zielwerte, Normalität ist für bestimmte exakte Tests und Intervalle relevant, nicht für die reine Berechnung der OLS-Koeffizienten
- **Interpretationsgrenze:** Die Plots liefern Hinweise und beweisen die Annahmen nicht, insbesondere kann der Q–Q-Plot keine Unabhängigkeit der Beobachtungen nachweisen
- **Unabhängigkeit:** Anhand der Datenerhebung und ggf. zeitlicher oder räumlicher Zusammenhänge beurteilen
- **Stolperstein:** Residuenanalyse ist eine Modelldiagnose, sie ersetzt keine Bewertung der Vorhersagen auf zurückgehaltenen Daten

```python
#Verwenden, um nach dem OLS-Training Fehlermuster und die angenäherte Normalverteilung der Residuen zu untersuchen
import statsmodels.api as sm                                                    #Q–Q-Plot für die Residuen erstellen
#VORAUSSETZUNG: plt ist importiert, ols_simple_apartments ist bereits trainiert

#1. Zu untersuchendes Modell auswählen
ols_diagnostic_model = ols_simple_apartments                                    #Hier das einfache Modell mit Wohnfläche untersuchen
#ANPASSEN: Durch dein trainiertes OLS-Modell ersetzen, bspw. ols_multiple_apartments
#HINWEIS: Die folgenden Schritte verwenden automatisch das hier gewählte Modell, kein erneutes fit() nötig


#2. Vorhersagen und Residuen der Trainingsdaten holen
ols_fitted = ols_diagnostic_model.fittedvalues                                  #Vom Modell geschätzte Mietpreise der Trainingswohnungen
ols_residuals = ols_diagnostic_model.resid                                      #Tatsächliche Miete minus geschätzte Miete
#ANPASSEN: Hier nichts ändern, wenn du nur das Modell in Schritt 1 ersetzt


#3. Residuen gegen vorhergesagte Werte zeichnen
fig, ax = plt.subplots(figsize=(8, 4))
ax.scatter(ols_fitted, ols_residuals, alpha=0.4)                                #Pro Trainingswohnung Vorhersage und zugehörigen Fehler darstellen
ax.axhline(0, color='black', linestyle='--')                                    #Nulllinie, oberhalb unterschätzt und unterhalb überschätzt das Modell

ax.set_xlabel('Vorhergesagte Monatsmiete (CHF)')
ax.set_ylabel('Residuum: tatsächliche − vorhergesagte Miete (CHF)')
ax.set_title('Residuen gegen vorhergesagte Mietpreise')
plt.tight_layout()
plt.show()
#ANPASSEN: Mietpreis und CHF in Beschriftungen und Titel durch die Zielvariable und Einheit deiner Aufgabe ersetzen
#HINWEIS: Nach Krümmung, Trichterform und einzelnen grossen Fehlern suchen


#4. Residuen mit einer Normalverteilung vergleichen
fig, ax = plt.subplots(figsize=(6, 6))
sm.qqplot(ols_residuals, line='s', ax=ax)                                       #Residuen mit theoretischen Normalverteilungsquantilen vergleichen

ax.set_xlabel('Theoretische Quantile der Normalverteilung')
ax.set_ylabel('Quantile der Residuen (CHF)')
ax.set_title('Normal-Q–Q-Plot der Residuen')
plt.tight_layout()
plt.show()
#ANPASSEN: CHF durch die Einheit deiner Zielvariable ersetzen
#HINWEIS: Punkte ungefähr auf der Geraden sprechen für annähernde Normalität, Unabhängigkeit wird hier nicht geprüft
```

## Testbewertung mit RMSE und R² – Vorhersagen für unbekannte Beobachtungen beurteilen
- **Definition:** Vergleicht Modellvorhersagen mit den tatsächlichen Zielwerten zurückgehaltener Testdaten
- **Zweck und Anwendung:** Abschliessend beurteilen, wie gut ein bereits trainiertes Modell bei bisher ungesehenen Beobachtungen vorhersagt
- **Input und Voraussetzungen:** Trainiertes Modell, Testtabelle mit allen benötigten Eingabespalten und tatsächlichen Zielwerten, hier `test_apartments`, das Modell wird nicht erneut auf den Testdaten trainiert
- **`.predict(test_apartments)`:** Verwendet die gelernte Modellformel, um für jede Testwohnung einen Mietpreis vorherzusagen
- **RMSE:** Quadratwurzel des mittleren quadratischen Fehlers, kleiner ist besser, Einheit entspricht der Zielvariable, hier CHF
- **RMSE interpretieren:** Ein angenommener RMSE von 500 CHF beschreibt eine quadratisch gemittelte Fehlergrösse von 500 CHF, nicht jede Vorhersage weicht genau um 500 CHF ab, grosse Fehler zählen stärker
- **R²:** `1 − SSE / SST`, vergleicht die quadrierten Vorhersagefehler mit den quadrierten Abweichungen vom Mittelwert der tatsächlichen Testzielwerte
- **R² = 1:** Perfekte Vorhersagen auf diesen Testdaten
- **R² = 0:** Gleich grosse Summe quadrierter Fehler wie bei einer konstanten Vorhersage des Testmittelwerts
- **R² < 0:** Schlechter als diese Mittelwertvorhersage, negative Werte sind bei der Testbewertung möglich
- **R² interpretieren:** Grössere Werte sind besser, R² hat keine Einheit und ist keine prozentuale Trefferquote, die Berechnung setzt für eine sinnvolle Interpretation unterschiedliche tatsächliche Zielwerte voraus
- **Vergleich der Modelle:** Bei denselben Testbeobachtungen bedeutet kleinerer RMSE auch grösseres R², hier vergleichen wir das einfache und das multiple OLS-Modell
- **Einzelne Vorhersagen:** Tatsächliche und vorhergesagte Mieten nebeneinander zeigen, wo das Modell über- oder unterschätzt, die ersten Zeilen allein repräsentieren aber nicht die gesamte Modellleistung
- **Stolpersteine:** Testvorhersagen mit den zugehörigen Testzielwerten vergleichen, nicht mit Trainingszielwerten, kein `fit()` mit Testdaten durchführen
- **Weitere Modellwahl:** Die beiden Modelle sind vor diesem Vergleich festgelegt, für weitere Anpassungen CV auf den Trainingsdaten verwenden, wiederholte Optimierung anhand des Tests macht dessen Bewertung zu optimistisch

```python
#Verwenden, um bereits trainierte OLS-Modelle abschliessend auf unbekannten Testdaten mit RMSE und R² zu bewerten
from sklearn.metrics import mean_squared_error, r2_score                        #Kennzahlen für numerische Vorhersagen
#VORAUSSETZUNG: np und pd sind importiert, beide OLS-Modelle und test_apartments sind bereits vorhanden

#1. Testdaten und tatsächliche Zielwerte festlegen
evaluation_test = test_apartments                                               #Zurückgehaltene Wohnungen, die nicht zum Training verwendet wurden
evaluation_target = 'price'                                                     #Spalte mit den tatsächlichen Monatsmieten
evaluation_unit = 'CHF'                                                         #Einheit der Zielvariable für die Ausgabe

y_actual_test = evaluation_test[evaluation_target]                              #Richtige Antworten der Testbeobachtungen auswählen
#ANPASSEN: test_apartments durch deine vorbereitete Testtabelle ersetzen
#ANPASSEN: 'price' durch deine Zielspalte und 'CHF' durch deren Einheit ersetzen
#ACHTUNG: Bei getrennt vorbereiteten X_test und y_test stattdessen evaluation_test = X_test und y_actual_test = y_test setzen


#2. Bereits trainierte Modelle auswählen
evaluation_models = {
    'Einfache OLS': ols_simple_apartments,                                      #Bereits trainiertes Modell mit Wohnfläche
    'Multiple OLS': ols_multiple_apartments                                     #Bereits trainiertes Modell mit Wohnfläche und Zimmerzahl
}
#ANPASSEN: Rechts vom Doppelpunkt die Namen deiner bereits trainierten OLS-Modelle einsetzen
#ANPASSEN: Links die passenden Anzeigenamen eintragen, bei nur einem Modell die andere Zeile entfernen


#3. Für jedes Modell Testvorhersagen und Kennzahlen berechnen
evaluation_rows = []                                                            #Ergebnisse aller Modelle sammeln
evaluation_predictions = pd.DataFrame({
    'Tatsächlicher Wert': y_actual_test
})                                                                              #Tabelle für den Vergleich einzelner Vorhersagen beginnen

for model_name, fitted_model in evaluation_models.items():
    y_predicted_test = fitted_model.predict(evaluation_test)                    #Zielwerte der Testbeobachtungen vorhersagen, kein erneutes Training

    test_rmse = np.sqrt(
        mean_squared_error(y_actual_test, y_predicted_test)
    )                                                                           #Fehlergrösse in der Einheit der Zielvariable berechnen

    test_r2 = r2_score(y_actual_test, y_predicted_test)                         #Vorhersagequalität relativ zum Testmittelwert berechnen

    evaluation_rows.append({
        'Modell': model_name,
        'Test-RMSE': test_rmse,
        'Test-R²': test_r2
    })                                                                          #Eine Ergebniszeile für das aktuelle Modell speichern

    evaluation_predictions[model_name] = np.asarray(
        y_predicted_test
    )                                                                           #Vorhersagen in derselben Zeilenreihenfolge neben die tatsächlichen Werte setzen
#ANPASSEN: In Schritt 3 nichts ändern, wenn du nur Testdaten, Zielwerte und OLS-Modelle in Schritt 1–2 ersetzt
#VORAUSSETZUNG: Die Testdaten enthalten alle Eingabespalten, die deine OLS-Modellformeln benötigen


#4. Gesamtergebnis und einzelne Vorhersagen anzeigen
evaluation_results = pd.DataFrame(evaluation_rows).set_index('Modell')

print(f'Test-RMSE in {evaluation_unit}, kleiner ist besser')
print('Test-R² ohne Einheit, grösser ist besser')
display(evaluation_results.round(3))                                            #Beide Modelle anhand aller Testbeobachtungen vergleichen

print(f'Tatsächliche Werte und Vorhersagen in {evaluation_unit}:')
display(evaluation_predictions.head(10).round(2))                               #Die ersten zehn Testbeobachtungen vergleichen
#ANPASSEN: head(10) für mehr oder weniger Beispielzeilen ändern, bspw. head(5)
#HINWEIS: Die Kennzahlen verwenden alle Testbeobachtungen, head(10) begrenzt nur die angezeigte Kontrolle
```

## Maximum Likelihood (MLE) – Parameter wählen, welche die beobachteten Daten am plausibelsten machen
- **Definition:** Maximum Likelihood Estimation schätzt Modellparameter so, dass die bereits beobachteten Daten unter dem Modell möglichst wahrscheinlich sind
- **Zweck und Anwendung:** Parameter eines statistischen Modells aus Daten bestimmen, bspw. eine Erfolgswahrscheinlichkeit oder die Koeffizienten einer logistischen Regression
- **Grundidee:** Die Daten stehen fest, verschiedene mögliche Parameterwerte werden verglichen, gewählt wird der Wert mit der höchsten Likelihood
- **Beispiel aus den Folien:** Bei zehn unabhängigen Münzwürfen werden siebenmal Kopf und dreimal Zahl beobachtet, gesucht ist die unbekannte Wahrscheinlichkeit `p` für Kopf
- **Likelihood einer bestimmten Reihenfolge:** `L(p) = p⁷ × (1 − p)³`, dabei ist `p` die Kopfwahrscheinlichkeit und `1 − p` die Zahlwahrscheinlichkeit
- **Ergebnis:** Die Likelihood ist bei `p = 7 / 10 = 0.7` maximal, die geschätzte Kopfwahrscheinlichkeit beträgt somit 70 %
- **Reihenfolge nicht vorgegeben:** Für genau siebenmal Kopf in beliebiger Reihenfolge kommt ein konstanter Kombinationsfaktor hinzu, der maximierende Wert bleibt `p = 0.7`
- **Output und Interpretation:** Der geschätzte Parameter macht die beobachteten Daten innerhalb des gewählten Modells am plausibelsten, die Likelihood ist keine Wahrscheinlichkeit dafür, dass dieser Parameter wahr ist
- **Vergleich:** Bei denselben Daten und demselben Modell ist eine höhere Likelihood besser, ihre absolute Grösse ist kein allgemeines Mass für gute Vorhersagen
- **Bezug zu OLS:** Bei unabhängigen, normalverteilten Fehlern mit gleicher Varianz liefert MLE dieselbe Regressionsgerade wie OLS
- **Bezug zur logistischen Regression:** Das Training basiert auf Maximum Likelihood bzw. der Minimierung des negativen Log-Likelihoods (Log-Loss), die Berechnung übernimmt später `.fit()`
- **Stolperstein:** Maximum Likelihood beschreibt das Schätzprinzip, nicht den konkreten Rechenweg zur Suche nach dem besten Parameter, dafür kann ein numerisches Optimierungsverfahren verwendet werden

```python
#Verwenden, um die MLE-Grundidee anhand unabhängiger Versuche mit zwei möglichen Ergebnissen nachzuvollziehen

#1. Beobachtete Anzahlen eintragen
mle_successes = 7                                                               #Siebenmal Kopf beobachtet
mle_failures = 3                                                                #Dreimal Zahl beobachtet
#ANPASSEN: 7 und 3 durch die beobachteten Anzahlen ersetzen, bspw. Erfolge und Misserfolge
#VORAUSSETZUNG: Unabhängige Versuche mit derselben Erfolgswahrscheinlichkeit, insgesamt mindestens ein Versuch


#2. Vorgegebene mögliche Wahrscheinlichkeiten vergleichen
mle_candidates = [0.5, 0.7, 0.9]                                                #Diese möglichen Kopfwahrscheinlichkeiten prüfen
#ANPASSEN: Liste durch die in der Aufgabe vorgegebenen Wahrscheinlichkeiten ersetzen

for mle_p in mle_candidates:
    mle_likelihood = (
        mle_p ** mle_successes
        * (1 - mle_p) ** mle_failures
    )                                                                           #Wahrscheinlichkeit einer bestimmten Folge mit diesen Anzahlen berechnen

    print(f'p = {mle_p:.2f} → Likelihood = {mle_likelihood:.6f}')
#ANPASSEN: In Schritt 2 die Formel unverändert lassen, wenn nur Anzahlen und Kandidaten geändert werden
#ACHTUNG: Die Formel gilt für dieses Beispiel mit zwei möglichen Ergebnissen, nicht allgemein für jedes ML-Modell


#3. MLE-Schätzung für dieses Modell berechnen
mle_p_hat = mle_successes / (mle_successes + mle_failures)                      #Beobachteter Erfolgsanteil maximiert hier die Likelihood

print(f'MLE-Schätzung der Erfolgswahrscheinlichkeit: {mle_p_hat:.3f}')
print(f'Das entspricht {mle_p_hat:.1%}')
#ANPASSEN: In Schritt 3 nichts ändern, wenn du nur die Anzahlen in Schritt 1 ersetzt
#HINWEIS: Die direkte Schätzung kann auch zwischen den geprüften Kandidaten liegen
```

## Gradient Descent – Modellparameter schrittweise verbessern
- **Definition:** Gradient Descent (Gradientenabstieg) verändert Parameter schrittweise in die Richtung, in der das Trainingsziel kleiner wird
- **Zweck und Anwendung:** Ein Trainingsziel numerisch minimieren, bspw. wenn die passenden Modellparameter nicht direkt mit einer einfachen Formel berechnet werden
- **Input:** Eine zu minimierende Funktion, ihr Gradient, Startwerte und eine Lernrate, zusätzlich eine Abbruchregel oder vorgegebene Anzahl Schritte
- **Gradient:** Beschreibt die Richtung und Stärke des steilsten Anstiegs, zum Minimieren gehen wir in die entgegengesetzte Richtung
- **Update-Regel:** `neuer Parameter = alter Parameter − Lernrate × Gradient`
- **Lernrate `η`:** Bestimmt die Schrittgrösse, zu kleine Werte führen zu langsamer Annäherung, zu grosse Werte können das Minimum überspringen oder zu instabilen Schritten führen
- **Ablauf:** Gradient am aktuellen Parameterwert berechnen, Parameter aktualisieren und wiederholen, bis ein Abbruchkriterium erreicht ist
- **Bsp. aus dem Concept-Notebook:** `Loss(b) = (b − 3)²`, der Gradient lautet `2 × (b − 3)`, das Minimum liegt bei `b = 3`
- **Erster Schritt:** Start bei `b = 0` und Lernrate `0.2` → Gradient = `−6` → neuer Wert = `0 − 0.2 × (−6) = 1.2`
- **Interpretation des Outputs:** Nähert sich b dem Wert 3 und der Loss dem Wert 0, bewegt sich das Verfahren zum Minimum dieses Beispiels, eine feste Anzahl Schritte garantiert noch keine vollständige Konvergenz
- **Batch Gradient Descent:** Berechnet jeden Schritt anhand aller Trainingsbeobachtungen
- **Stochastic Gradient Descent:** Verwendet jeweils eine Trainingsbeobachtung, Mini-Batch Gradient Descent verwendet eine kleine Gruppe, die Schritte können dadurch stärker schwanken
- **Abgrenzung zu MLE:** MLE legt fest, welche Parameter gesucht werden, Gradient Descent ist ein möglicher Rechenweg zur Optimierung
- **Stolperstein:** Bei komplizierten Funktionen ist das globale Minimum nicht allgemein garantiert, dieses einfache Beispiel hat genau ein Minimum

```python
#Verwenden, um Gradientenabstieg für eine Funktion mit einem Parameter schrittweise nachzurechnen

#1. Einstellungen festlegen
gd_b = 0.0                                                                      #Startwert
gd_learning_rate = 0.2                                                          #Schrittgrösse
gd_steps = 8                                                                    #Anzahl Schritte
#ANPASSEN: Startwert, Lernrate und Schrittzahl gemäss Aufgabe ersetzen


#2. Ausgangszustand anzeigen
gd_loss = (gd_b - 3) ** 2                                                       #Loss beim Startwert berechnen
#ANPASSEN: Bei anderer Loss-Funktion diese Formel ersetzen, dieselbe Formel unten verwenden

print(f'Start: b = {gd_b:.4f}, Loss = {gd_loss:.4f}')


#3. Parameter schrittweise aktualisieren
for gd_step in range(1, gd_steps + 1):
    gd_gradient = 2 * (gd_b - 3)                                                #Ableitung der Loss-Funktion am aktuellen b
    #ANPASSEN: Bei anderer Loss-Funktion deren Ableitung einsetzen

    gd_b = gd_b - gd_learning_rate * gd_gradient                                #Parameter entgegen dem Gradienten verändern

    gd_loss = (gd_b - 3) ** 2                                                   #Loss nach dem Schritt berechnen
    #ANPASSEN: Dieselbe Loss-Formel wie in Schritt 2 einsetzen

    print(f'Schritt {gd_step}: b = {gd_b:.4f}, Loss = {gd_loss:.4f}')


#4. Ergebnis anzeigen
print(f'Parameter nach {gd_steps} Schritten: b = {gd_b:.4f}')
```

## Regressionsbaum – Blattvorhersagen und Qualität eines Splits berechnen
- **Definition:** Ein Regressionsbaum teilt Beobachtungen durch aufeinanderfolgende Ja-/Nein-Fragen auf, bspw. «Hubraum ≤ 2 Liter?», am Ende eines Pfads steht ein Blatt mit einer festen numerischen Vorhersage
- **Zweck und Anwendung:** Numerische Zielwerte anhand von Regeln vorhersagen, dabei können auch nichtlineare Zusammenhänge und Kombinationen mehrerer Merkmale erfasst werden
- **Blattvorhersage:** Bei quadratischem Fehler sagt jedes Blatt den Mittelwert der Trainingszielwerte in diesem Blatt vorher, dieser Mittelwert minimiert deren Summe quadrierter Fehler
- **SSE (Sum of Squared Errors):** Summe der quadrierten Abweichungen vom jeweiligen Blattmittelwert → `Σ(y − Blattmittelwert)²`
- **Qualität eines Splits:** SSE des linken und rechten Kindes addieren, bei denselben Beobachtungen ist der Split mit der kleineren Gesamt-SSE besser
- **Input für die Berechnung:** Zielwerte der Beobachtungen und ihre Zuordnung zum linken oder rechten Kind für jeden zu vergleichenden Split
- **Aufbau des Baums:** Verschiedene Eingabespalten und Grenzwerte prüfen, die beste Aufteilung wählen und in den Kindern wiederholen, dieses Vorgehen ist schrittweise (greedy) und rekursiv
- **Bsp. aus der Aufgabe:** Links `[10, 12, 14]` und rechts `[30, 34]` → Blattvorhersagen 12 und 32, SSE links = 8 und rechts = 8 → Gesamt-SSE = 16
- **Alternativer Split:** Links `[10, 12]` und rechts `[14, 30, 34]` → Blattvorhersagen 11 und 26 → Gesamt-SSE = 226
- **Interpretation:** Der erste Split ist besser, weil seine Gesamt-SSE mit 16 kleiner ist als 226, ohne Aufteilung beträgt die SSE 496
- **Wertebereich und Einheit:** SSE ist mindestens 0, kleinere Werte bedeuten geringere Trainingsfehler, die Einheit ist die quadrierte Einheit der Zielvariable
- **Stolpersteine:** Geteilt wird anhand einer Eingabespalte und ihres Grenzwerts, die Zielwerte dienen zur Bewertung der Aufteilung, eine gute Trainingsaufteilung belegt noch keine gute Leistung auf neuen Daten

```python
#Verwenden, wenn für vorgegebene Baumaufteilungen Blattvorhersagen und die Gesamt-SSE verglichen werden sollen
#VORAUSSETZUNG: NumPy ist bereits als np importiert

#1. Zielwerte für den ersten Split eintragen
tree_split_left = np.array([10, 12, 14])                                        #Zielwerte im linken Kind
tree_split_right = np.array([30, 34])                                           #Zielwerte im rechten Kind
#ANPASSEN: Beide Listen durch die Zielwerte der vorgegebenen ersten Aufteilung ersetzen


#2. Blattvorhersagen und SSE des ersten Splits berechnen
tree_left_mean = tree_split_left.mean()                                         #Vorhersage links = Durchschnitt der linken Zielwerte
tree_right_mean = tree_split_right.mean()                                       #Vorhersage rechts = Durchschnitt der rechten Zielwerte

tree_left_sse = np.sum((tree_split_left - tree_left_mean) ** 2)
tree_right_sse = np.sum((tree_split_right - tree_right_mean) ** 2)
tree_split_sse = tree_left_sse + tree_right_sse                                 #Fehler beider Kinder zusammenzählen
#HINWEIS: Diese Berechnungen bleiben gleich, wenn du die Zahlenlisten ersetzt


#3. Zielwerte für den alternativen Split eintragen
tree_alt_left = np.array([10, 12])
tree_alt_right = np.array([14, 30, 34])
#ANPASSEN: Dieselben Beobachtungen wie in Schritt 1 verwenden, aber gemäss alternativem Split zuordnen

tree_alt_sse = (
    np.sum((tree_alt_left - tree_alt_left.mean()) ** 2)
    + np.sum((tree_alt_right - tree_alt_right.mean()) ** 2)
)                                                                               #SSE beider alternativen Kinder zusammenzählen

#4. SSE ohne Aufteilung als Vergleich berechnen
tree_parent_values = np.concatenate([tree_split_left, tree_split_right])        #Alle Zielwerte zusammenführen
tree_parent_sse = np.sum(
    (tree_parent_values - tree_parent_values.mean()) ** 2
)                                                                               #Fehler bei nur einer gemeinsamen Mittelwertvorhersage

#5. Ergebnisse anzeigen
print(f'Ohne Split: SSE = {tree_parent_sse:.2f}')
print(f'Split 1: Blattvorhersagen = {tree_left_mean:.2f} und {tree_right_mean:.2f}')
print(f'Split 1: SSE links = {tree_left_sse:.2f}, SSE rechts = {tree_right_sse:.2f}')
print(f'Split 1: Gesamt-SSE = {tree_split_sse:.2f}')
print(f'Alternativer Split: Gesamt-SSE = {tree_alt_sse:.2f}')
print('Besser ist die Aufteilung mit der kleineren Gesamt-SSE')

#ACHTUNG: Jedes Kind muss mindestens einen Zielwert enthalten
```

## Regressionsbaum trainieren und lesen – Verbrauch anhand von Entscheidungsregeln vorhersagen
- **Definition:** Ein Regressionsbaum lernt Aufteilungsregeln aus Trainingsdaten und sagt in jedem Blatt einen festen Zahlenwert vorher, bei quadratischem Fehler den Mittelwert der Trainingszielwerte dieses Blatts
- **Zweck und Anwendung:** Verwenden, wenn ein numerischer Zielwert anhand eines oder mehrerer Merkmale mit verständlichen Entscheidungsregeln vorhergesagt werden soll
- **Input und Voraussetzungen:** Numerische Eingabespalten und passende Zielwerte, hier die bereits vorbereiteten `X_train_cars` und `y_train_cars`, neue Beobachtungen benötigen dieselben Eingabespalten in derselben Reihenfolge
- **`DecisionTreeRegressor(...)`:** Erstellt das Modell, standardmässig werden Splits anhand des quadratischen Fehlers gewählt
- **`max_depth=2`:** Erlaubt höchstens zwei Aufteilungen auf einem Pfad von der Wurzel zum Blatt, nicht insgesamt nur zwei Aufteilungen im ganzen Baum
- **`random_state=42`:** Macht Zufallsschritte beim Training reproduzierbar
- **`.fit(X_train_cars, y_train_cars)`:** Lernt die Regeln ausschliesslich anhand der Trainingsdaten
- **`.predict(...)`:** Führt neue Beobachtungen durch die gelernten Regeln und gibt die Blattvorhersagen zurück
- **Baum lesen:** Beim obersten Knoten beginnen, ist die Bedingung `Merkmal ≤ Grenzwert` erfüllt, nach links gehen, sonst nach rechts, bis ein Blatt erreicht ist
- **`samples`:** Anzahl Trainingsbeobachtungen im Knoten
- **`value`:** Mittelwert der Trainingszielwerte im Knoten, im Blatt die Vorhersage für alle dort ankommenden Beobachtungen
- **`squared_error`:** Mittlere quadratische Abweichung der Trainingszielwerte vom Knotenmittelwert, nicht die SSE und nicht der Testfehler
- **Interpretation im vorgegebenen Beispiel:** Das Auto mit 2.0 Litern Hubraum und vier Zylindern wird laut Musterlösung mit rund 8.95 L/100 km vorhergesagt, dieser Wert ist der mittlere Verbrauch der Trainingsautos im erreichten Blatt
- **Verwendete Merkmale:** Nur Merkmale, die tatsächlich in den Aufteilungsregeln erscheinen, beeinflussen die Vorhersage dieses Baums, nicht jedes übergebene Merkmal muss verwendet werden
- **Stolpersteine:** Blattwerte sind Schätzungen, keine sicheren Einzelwerte, ein tieferer Baum kann Trainingsdetails lernen, die passende Komplexität anhand von CV wählen
- **Prüfung:** Zum Bewerten unbekannter Beobachtungen `X_test` vorhersagen und mit `y_test` vergleichen, das blosse Zeichnen des Baums bewertet seine Testleistung noch nicht

```python
#Verwenden, um einen numerischen Zielwert mit einem Regressionsbaum vorherzusagen und seine Regeln anzuzeigen
from sklearn.tree import DecisionTreeRegressor, plot_tree
#VORAUSSETZUNG: pd und plt sind importiert, die Trainingsdaten der Autos sind bereits vorbereitet


#1. Regressionsbaum erstellen
reg_tree_cars = DecisionTreeRegressor(
    max_depth=2,                                                                #Höchstens zwei Aufteilungen je Pfad erlauben
    random_state=42                                                             #Training reproduzierbar machen
)
#ANPASSEN: max_depth und random_state gemäss Aufgabe setzen


#2. Baum trainieren
reg_tree_cars.fit(X_train_cars, y_train_cars)                                   #Aus Trainingsmerkmalen und tatsächlichem Verbrauch lernen
#ANPASSEN: X_train_cars und y_train_cars durch deine vorbereiteten Trainingsvariablen ersetzen


#3. Gelernten Baum zeichnen
fig, ax = plt.subplots(figsize=(11, 5))
plot_tree(
    reg_tree_cars,
    feature_names=list(X_train_cars.columns),                                   #Merkmalsnamen in derselben Reihenfolge wie beim Training anzeigen
    filled=True, rounded=True, precision=2, ax=ax                               #Knoten einfärben und Zahlen auf zwei Nachkommastellen anzeigen
)
ax.set_title('Regressionsbaum für den Treibstoffverbrauch')
plt.tight_layout()
plt.show()
#ANPASSEN: X_train_cars durch dieselbe Trainingstabelle wie in Schritt 2 ersetzen
#ANPASSEN: Titel an die Zielvariable anpassen, bei grösseren Bäumen figsize erhöhen


#4. Optional: Neue Beobachtung vorhersagen
new_car_tree = pd.DataFrame({
    'engine_size': [2.0],                                                       #Hubraum des neuen Autos
    'cylinders': [4]                                                            #Zylinderzahl des neuen Autos
})
new_car_tree = new_car_tree[X_train_cars.columns]                               #Spalten in dieselbe Reihenfolge wie beim Training bringen

new_car_tree_prediction = reg_tree_cars.predict(new_car_tree)

display(
    new_car_tree.assign(
        predicted_combined=new_car_tree_prediction
    )
)                                                                               #Neue Eingaben und vorhergesagten Verbrauch in L/100 km nebeneinander anzeigen

#ANPASSEN: In Schritt 4 alle Eingabespalten und Werte für die neue Beobachtung einsetzen
#ANPASSEN: X_train_cars wie in Schritt 2 ersetzen, predicted_combined passend zur Zielvariable benennen
#HINWEIS: Schritt 4 weglassen, wenn keine neue Einzelbeobachtung vorhergesagt werden soll
#HINWEIS: Für Testvorhersagen stattdessen reg_tree_cars.predict(X_test_cars) verwenden und mit y_test_cars bewerten
```

### Pruning mit ccp_alpha – Einen Regressionsbaum vereinfachen
- **Definition:** Pruning beschneidet einen Baum, indem Zweige entfernt werden, deren Verbesserung der Trainingsanpassung im Verhältnis zur zusätzlichen Komplexität zu klein ist
- **Zweck und Anwendung:** Einen übersichtlicheren Baum erhalten und eine zu starke Anpassung an Trainingsdetails begrenzen
- **Input und Voraussetzungen:** Bereits vorbereitete Trainingsmerkmale und Trainingszielwerte, hier `X_train_cars` und `y_train_cars`
- **Grundidee:** Trainingsfehler und Anzahl Blätter gemeinsam berücksichtigen → `Baumfehler + alpha × Anzahl Blätter`, scikit-learn verwendet dabei einen gewichteten, auf die Beobachtungszahl normierten Fehler
- **`ccp_alpha=0.0`:** Keine zusätzliche Pruning-Strafe, der Baum wächst bis zu den übrigen Stoppregeln
- **Grösseres `ccp_alpha`:** Stärkere Bestrafung zusätzlicher Blätter, dadurch kann der Baum kleiner werden, nicht jede Erhöhung verändert sofort den Baum
- **Unterschied zu `max_depth`:** `max_depth` begrenzt die Tiefe bereits beim Wachsen, `ccp_alpha` steuert das anschliessende Beschneiden
- **`.get_n_leaves()`:** Liefert die Anzahl Blätter, weniger Blätter bedeuten weniger unterschiedliche Vorhersageregeln
- **Output:** Für jede Einstellung werden Blattanzahl und Trainings-RMSE angezeigt, der RMSE verwendet hier bewusst Trainingsvorhersagen und beschreibt keine Testleistung
- **Interpretation gemäss Musterlösung:** Bei `alpha=0.05` sinkt die Blattanzahl von 37 auf 8, während der Trainings-RMSE nur von ungefähr 1.43 auf 1.52 L/100 km steigt → deutlich weniger Regeln bei etwas schlechterer Trainingsanpassung
- **Interpretationsgrenze:** Ein höherer Trainingsfehler beweist keine schlechtere Leistung auf neuen Daten, ob Pruning die Vorhersagen verbessert, muss mit Validierung bzw. CV geprüft werden
- **Stolpersteine:** Ein zu grosses alpha kann zu Underfitting führen, passende Werte hängen auch von der Grössenordnung und Einheit der Zielvariable ab, deshalb `0.05` nicht ungeprüft auf andere Aufgaben übertragen

```python
#Verwenden, um zu untersuchen, wie Pruning die Blattanzahl und den Trainingsfehler eines Regressionsbaums verändert
from sklearn.metrics import mean_squared_error                                  #RMSE aus tatsächlichen Werten und Vorhersagen berechnen
#VORAUSSETZUNG: np, pd und DecisionTreeRegressor sind importiert, Trainingsdaten sind vorbereitet


#1. Zu vergleichende Pruning-Einstellungen festlegen
pruning_alphas_cars = [0.0, 0.05]                                               #Ohne Pruning und mit Pruning vergleichen
#ANPASSEN: Liste durch die verlangten ccp_alpha-Werte ersetzen


#2. Für jede Einstellung einen neuen Baum trainieren
pruning_rows_cars = []                                                          #Ergebnisse der Einstellungen sammeln

for pruning_alpha in pruning_alphas_cars:
    pruning_tree_cars = DecisionTreeRegressor(
        ccp_alpha=pruning_alpha,                                                #Aktuellen Wert aus der Liste verwenden
        random_state=42                                                         #Training reproduzierbar machen
    )
    #ANPASSEN: random_state gemäss Aufgabe setzen, weitere Parameter nur ergänzen, wenn verlangt

    pruning_tree_cars.fit(X_train_cars, y_train_cars)                           #Baum ausschliesslich mit Trainingsdaten lernen lassen
    pruning_train_prediction = pruning_tree_cars.predict(
        X_train_cars
    )                                                                           #Dieselben Trainingsbeobachtungen vorhersagen
    #ANPASSEN: X_train_cars an beiden Stellen und y_train_cars durch deine Trainingsvariablen ersetzen

    pruning_train_rmse = np.sqrt(
        mean_squared_error(y_train_cars, pruning_train_prediction)
    )                                                                           #Trainingsfehler in der Einheit der Zielvariable berechnen
    #ANPASSEN: y_train_cars wie beim fit() durch deine Trainingszielwerte ersetzen

    pruning_rows_cars.append({
        'ccp_alpha': pruning_alpha,
        'Blätter': pruning_tree_cars.get_n_leaves(),                            #Anzahl verbleibender Blattregeln
        'Trainings-RMSE': pruning_train_rmse
    })


#3. Ergebnisse vergleichen
print('Trainings-RMSE in L/100 km, kleiner ist besser')
display(pd.DataFrame(pruning_rows_cars).round(3))
#ANPASSEN: L/100 km durch die Einheit deiner Zielvariable ersetzen

#ACHTUNG: Hier wird nur die Trainingsanpassung verglichen, alpha anhand von CV auswählen
#HINWEIS: Die Schleife verändert reg_tree_cars aus dem vorherigen Eintrag nicht
```

## Bagging – Vorhersagen vieler Bäume mitteln
- **Definition:** Bagging (Bootstrap Aggregation) trainiert mehrere Modelle auf zufällig gezogenen Varianten der Trainingsdaten und kombiniert ihre Vorhersagen
- **Zweck und Anwendung:** Schwankungsanfällige Modelle wie tiefe Entscheidungsbäume stabilisieren, Bagging reduziert vor allem die Variance
- **Input und Voraussetzungen:** Vorbereitete numerische Trainingsmerkmale und zugehörige Zielwerte, hier `X_train_cars` und `y_train_cars`
- **Bootstrap-Stichprobe:** Trainingszeilen werden zufällig mit Zurücklegen gezogen, dadurch können Beobachtungen mehrfach vorkommen und andere fehlen
- **`BaggingRegressor(...)`:** Verwendet standardmässig Regressionsbäume als Einzelmodelle
- **`n_estimators=50`:** Trainiert 50 Bäume, jeder erhält standardmässig eine Bootstrap-Stichprobe mit so vielen Ziehungen, wie die Trainingstabelle Zeilen hat
- **`bootstrap=True`:** Aktiviert das Ziehen mit Zurücklegen, dies ist die Standardeinstellung
- **Vorhersage bei Regression:** Mittelwert der Einzelvorhersagen, bspw. Baumvorhersagen 8, 10 und 9 → Bagging-Vorhersage 9
- **Interpretation des Outputs:** Die Vorhersage ist ein geschätzter Zahlenwert in der Einheit der Zielvariable, hier Verbrauch in L/100 km, sie kombiniert die Einschätzungen aller Bäume
- **Warum Mittelung hilft:** Einzelne Bäume reagieren unterschiedlich auf Trainingsdaten, durch Mittelung gleichen sich Teile dieser Schwankungen aus, stark ähnliche Bäume begrenzen den Nutzen
- **Stolpersteine:** Mehr Bäume bedeuten mehr Rechenaufwand und keine garantierte Verbesserung, Bagging beseitigt weder ungeeignete Features noch Data Leakage
- **Bewertung:** Eine plausible Einzelvorhersage belegt keine gute Modellqualität, Bagging später anhand der CV-Ergebnisse mit anderen Modellen vergleichen

```python
#Verwenden, um numerische Vorhersagen durch Mittelung vieler auf Bootstrap-Stichproben trainierter Bäume zu stabilisieren
from sklearn.ensemble import BaggingRegressor
#VORAUSSETZUNG: pd ist importiert, X_train_cars und y_train_cars sind vorbereitet

#1. Bagging-Modell erstellen
bagging_cars = BaggingRegressor(
    n_estimators=50,                                                            #50 Regressionsbäume trainieren
    bootstrap=True,                                                             #Trainingszeilen mit Zurücklegen ziehen
    random_state=42                                                             #Zufallsschritte reproduzierbar machen
)
#ANPASSEN: n_estimators und random_state gemäss Aufgabe setzen
#HINWEIS: bootstrap=True entspricht dem Bootstrap-Prinzip von Bagging


#2. Modell trainieren
bagging_cars.fit(X_train_cars, y_train_cars)                                    #Jeden Baum auf seiner eigenen Bootstrap-Stichprobe trainieren
#ANPASSEN: X_train_cars und y_train_cars durch deine vorbereiteten Trainingsvariablen ersetzen


#3. Optional: Eine neue Beobachtung erfassen
new_car_bagging = pd.DataFrame({
    'engine_size': [2.0],                                                       #Hubraum des neuen Autos
    'cylinders': [4]                                                            #Zylinderzahl des neuen Autos
})
new_car_bagging = new_car_bagging[X_train_cars.columns]                         #Dieselbe Spaltenreihenfolge wie beim Training verwenden
#ANPASSEN: Spaltennamen und Werte durch alle benötigten Eingaben der neuen Beobachtung ersetzen
#ANPASSEN: X_train_cars wie in Schritt 2 durch deine Trainingstabelle ersetzen


#4. Vorhersage berechnen und kontrollieren
bagging_new_prediction = bagging_cars.predict(new_car_bagging)                  #Vorhersagen der 50 Bäume automatisch mitteln

display(
    new_car_bagging.assign(
        predicted_combined=bagging_new_prediction
    )
)                                                                               #Eingaben und vorhergesagten Verbrauch in L/100 km nebeneinander anzeigen
#ANPASSEN: predicted_combined bei Bedarf passend zur Zielvariable benennen
#HINWEIS: Schritte 3–4 weglassen, wenn keine neue Einzelbeobachtung vorhergesagt werden soll
#HINWEIS: Für Testvorhersagen bagging_cars.predict(X_test_cars) verwenden, mit y_test_cars bewerten
```

## Random Forest Regression – Bagging mit zufälliger Merkmalsauswahl
- **Definition:** Ein Random Forest kombiniert viele Regressionsbäume, die auf Bootstrap-Stichproben trainiert werden und bei jeder Aufteilung nur eine zufällige Auswahl der Eingabemerkmale berücksichtigen
- **Zweck und Anwendung:** Numerische Zielwerte vorhersagen und die Schwankungen einzelner Bäume durch Mittelung reduzieren, geeignet auch für nichtlineare Zusammenhänge
- **Input und Voraussetzungen:** Numerische Trainingsmerkmale und passende Zielwerte, hier `X_train_cars` und `y_train_cars`
- **Neu gegenüber Bagging:** Zusätzlich zur zufälligen Zeilenauswahl wird die Merkmalsauswahl pro Split eingeschränkt, dadurch können die Bäume unterschiedlicher werden
- **`n_estimators=100`:** Trainiert 100 Bäume, die endgültige Vorhersage ist der Mittelwert ihrer Vorhersagen
- **`max_features=1`:** Berücksichtigt pro Split genau ein zufällig ausgewähltes Merkmal, hier entweder Hubraum oder Zylinderzahl
- **Wichtiger Unterschied:** Die ganze Zahl `1` bedeutet ein Merkmal, die Dezimalzahl `1.0` bedeutet 100 % aller Merkmale
- **`max_depth`:** Begrenzt bei Bedarf die Tiefe jedes einzelnen Baums, ohne Angabe gilt keine feste Tiefengrenze
- **`oob_score=True`:** Aktiviert die Out-of-Bag-Bewertung, dafür muss Bootstrap-Sampling eingeschaltet sein, beim Random Forest ist es standardmässig aktiv
- **Out-of-Bag (OOB):** Eine Beobachtung fehlt im Durchschnitt bei grossen Stichproben in ungefähr 37 % der Bootstrap-Stichproben, für ihre OOB-Vorhersage werden nur Bäume verwendet, die diese Beobachtung nicht zum Lernen erhalten haben
- **`.oob_score_`:** Liefert beim RandomForestRegressor standardmässig das R² der OOB-Vorhersagen, nicht den RMSE und nicht das R² auf dem separaten Testdatensatz
- **Interpretation:** Ein OOB-R² von bspw. 0.69 beschreibt eine Fehlerreduktion von ungefähr 69 % gegenüber der Vorhersage des Trainingszielmittelwerts, basierend auf den OOB-Vorhersagen
- **Wertebereich:** Grössere R²-Werte sind besser, 1 bedeutet perfekte Vorhersagen, 0 entspricht der Mittelwertvorhersage und negative Werte sind möglich
- **Stolpersteine:** OOB ist eine interne Bewertung anhand der Trainingsdaten und ersetzt keinen abschliessenden Test, bei wenigen Bäumen können einzelne Beobachtungen zu wenige OOB-Vorhersagen erhalten

```python
#Verwenden, um einen numerischen Zielwert mit einem Random Forest vorherzusagen und optional intern über OOB zu bewerten
from sklearn.ensemble import RandomForestRegressor
#VORAUSSETZUNG: pd ist importiert, X_train_cars und y_train_cars sind vorbereitet

#1. Random Forest erstellen
forest_cars = RandomForestRegressor(
    n_estimators=100,                                                           #100 Bäume erstellen
    max_features=1,                                                             #Pro Split genau ein zufällig gewähltes Merkmal prüfen
    oob_score=True,                                                             #Zusätzlich OOB-R² berechnen
    random_state=42                                                             #Zufallsschritte reproduzierbar machen
)
#ANPASSEN: n_estimators, max_features und random_state gemäss Aufgabe setzen
#ANPASSEN: Falls eine maximale Baumtiefe verlangt wird, bspw. max_depth=5 ergänzen
#ANPASSEN: Wird keine OOB-Bewertung benötigt, oob_score=True und Schritt 3 weglassen
#ACHTUNG: max_features=1 bedeutet ein Merkmal, max_features=1.0 bedeutet alle Merkmale


#2. Modell trainieren
forest_cars.fit(X_train_cars, y_train_cars)                                     #Bäume auf Bootstrap-Stichproben der Trainingsdaten trainieren
#ANPASSEN: X_train_cars und y_train_cars durch deine vorbereiteten Trainingsvariablen ersetzen


#3. OOB-Bewertung anzeigen
print(f'OOB-R²: {forest_cars.oob_score_:.3f}')                                  #Bewertung mit Bäumen, die die jeweilige Beobachtung nicht gelernt haben
#ACHTUNG: Nur verwenden, wenn oob_score=True und bootstrap=True gesetzt sind, bootstrap=True gilt hier standardmässig


#4. Optional: Neue Beobachtung vorhersagen
new_car_forest = pd.DataFrame({
    'engine_size': [2.0],
    'cylinders': [4]
})
new_car_forest = new_car_forest[X_train_cars.columns]                           #Spalten wie beim Training anordnen
#ANPASSEN: Spalten und Werte durch alle benötigten Eingaben der neuen Beobachtung ersetzen
#ANPASSEN: X_train_cars wie in Schritt 2 ersetzen


forest_new_prediction = forest_cars.predict(new_car_forest)                     #Vorhersagen aller Bäume automatisch mitteln

display(
    new_car_forest.assign(
        predicted_combined=forest_new_prediction
    )
)                                                                               #Eingaben und vorhergesagten Verbrauch in L/100 km nebeneinander anzeigen
#ANPASSEN: predicted_combined bei Bedarf passend zur Zielvariable benennen
#HINWEIS: Schritt 4 weglassen, wenn keine neue Einzelbeobachtung verlangt ist
#HINWEIS: Für Testvorhersagen forest_cars.predict(X_test_cars) verwenden, mit y_test_cars bewerten
```

## Gradient Boosting Regression – Vorhersagefehler schrittweise korrigieren
- **Definition:** Gradient Boosting baut Bäume nacheinander auf, jeder neue Baum soll die bisherige Gesamtvorhersage verbessern
- **Zweck und Anwendung:** Numerische Zielwerte vorhersagen und Zusammenhänge schrittweise erfassen, die das bisherige Modell noch nicht ausreichend beschreibt
- **Input und Voraussetzungen:** Numerische Trainingsmerkmale und passende Zielwerte, hier `X_train_cars` und `y_train_cars`
- **Unterschied zu Bagging und Random Forest:** Dort werden separat trainierte Bäume gemittelt, beim Boosting bauen die Bäume aufeinander auf und ihre Beiträge werden addiert
- **Bei quadratischem Loss:** Das Modell beginnt mit dem Mittelwert der Trainingszielwerte, jeder weitere Baum lernt die aktuellen Residuen → `tatsächlicher Wert − bisherige Vorhersage`
- **Aktualisierung:** `neue Vorhersage = bisherige Vorhersage + learning_rate × Vorhersage des neuen Baums`
- **Beispiel aus den Folien:** Tatsächlicher Wert 10, bisherige Vorhersage 7, neuer Baum sagt Korrektur 3 vorher, Lernrate 0.1 → neue Vorhersage = `7 + 0.1 × 3 = 7.3`
- **`n_estimators=100`:** Führt 100 Boosting-Schritte mit jeweils einem neuen Baum durch
- **`max_depth=2`:** Begrenzt jeden einzelnen Baum auf höchstens zwei Aufteilungen pro Pfad, nicht das gesamte Ensemble
- **`learning_rate=0.1`:** Übernimmt jeweils 10 % des Beitrags eines neuen Baums, kleinere Lernraten benötigen häufig mehr Bäume
- **`loss='squared_error'`:** Verwendet quadratischen Fehler, das ist die Standardeinstellung, bei anderen Loss Functions werden passende negative Gradienten statt einfacher Residuen gelernt
- **Output und Interpretation:** `.predict()` liefert die kombinierte Vorhersage des gesamten Ensembles in der Einheit der Zielvariable, hier L/100 km
- **Bewertung:** Lernrate, Baumtiefe und Anzahl Bäume gemeinsam anhand von CV beurteilen, mehr Bäume verbessern die Validierungsleistung nicht automatisch
- **Stolpersteine:** Boosting kann ebenfalls overfitten, eine kleinere Lernrate allein garantiert kein besseres Modell, eine plausible Einzelvorhersage ersetzt keine Bewertung auf zurückgehaltenen Beobachtungen

```python
#Verwenden, um einen numerischen Zielwert durch schrittweise Fehlerkorrektur mit vielen kleinen Bäumen vorherzusagen
from sklearn.ensemble import GradientBoostingRegressor
#VORAUSSETZUNG: pd ist importiert, X_train_cars und y_train_cars sind vorbereitet


#1. Boosting-Modell erstellen
boosting_cars = GradientBoostingRegressor(
    n_estimators=100,                                                           #100 Bäume nacheinander hinzufügen
    max_depth=2,                                                                #Jeden einzelnen Baum auf Tiefe 2 begrenzen
    learning_rate=0.1,                                                          #Jeweils 10 % des neuen Baumbeitrags übernehmen
    random_state=42                                                             #Zufallsschritte reproduzierbar machen
)
#ANPASSEN: n_estimators, max_depth, learning_rate und random_state gemäss Aufgabe ersetzen
#HINWEIS: Standardmässig wird quadratischer Fehler verwendet, die Bäume lernen dabei die aktuellen Residuen


#2. Modell trainieren
boosting_cars.fit(X_train_cars, y_train_cars)                                   #Bäume und ihre Korrekturen aus den Trainingsdaten lernen
#ANPASSEN: X_train_cars und y_train_cars durch deine vorbereiteten Trainingsvariablen ersetzen


#3. Optional: Neue Beobachtung erfassen
new_car_boosting = pd.DataFrame({
    'engine_size': [2.0],                                                       #Hubraum des neuen Autos
    'cylinders': [4]                                                            #Zylinderzahl des neuen Autos
})
new_car_boosting = new_car_boosting[X_train_cars.columns]                       #Dieselbe Spaltenreihenfolge wie beim Training verwenden
#ANPASSEN: Spalten und Werte durch alle benötigten Eingaben der neuen Beobachtung ersetzen
#ANPASSEN: X_train_cars wie in Schritt 2 ersetzen


#4. Vorhersage berechnen und anzeigen
boosting_new_prediction = boosting_cars.predict(new_car_boosting)               #Gesamtvorhersage aus Startwert und Baumkorrekturen berechnen

display(
    new_car_boosting.assign(
        predicted_combined=boosting_new_prediction
    )
)                                                                               #Eingaben und vorhergesagten Verbrauch in L/100 km nebeneinander anzeigen
#ANPASSEN: predicted_combined bei Bedarf passend zur Zielvariable benennen
#HINWEIS: Schritte 3–4 weglassen, wenn keine neue Einzelbeobachtung verlangt ist
#HINWEIS: Für Testvorhersagen boosting_cars.predict(X_test_cars) verwenden, mit y_test_cars bewerten
```

## Regressionsmodelle vergleichen – Mit CV auswählen und abschliessend testen
- **Zweck und Anwendung:** Mehrere Modelle auf denselben Trainingsdaten vergleichen und das Modell mit dem kleinsten mittleren CV-RMSE auswählen
- **Input und Voraussetzungen:** Vorbereitete Trainings- und Testdaten sowie die zu vergleichenden Modelle, die Eingaben müssen für alle Modelle geeignet sein
- **Verglichene Modelle:** Lineare Regression, beschnittener Regressionsbaum, Bagging, Random Forest und Gradient Boosting
- **Lineare Regression:** Beschreibt den Zielwert als gewichtete Summe der Eingaben, Koeffizienten lassen sich direkt interpretieren
- **Einzelner Baum:** Lernt verständliche Regeln und nichtlineare Zusammenhänge, tiefe Bäume können stark auf Trainingsdetails reagieren
- **Bagging und Random Forest:** Mitteln viele Baumvorhersagen und reduzieren dadurch vor allem deren Schwankungen, Random Forest ergänzt zufällige Merkmalsauswahl pro Split
- **Boosting:** Fügt kleine Bäume nacheinander als Korrekturen hinzu, Komplexität hängt u.a. von Baumtiefe, Lernrate und Anzahl Bäume ab
- **Einordnung:** Das sind typische Eigenschaften, keine Methode ist auf jedem Datensatz die beste
- **CV-Vergleich:** Für jedes Modell dieselben Trainingsdaten, Folds und dieselbe Bewertungskennzahl verwenden, hier fünf Folds und mittlerer Fold-RMSE
- **Auswahl:** Kleinster mittlerer CV-RMSE gewinnt, Unterschiede zwischen den Folds und geringe Unterschiede zwischen Modellen mitberücksichtigen
- **Abschliessendes Training:** Das ausgewählte Modell nochmals auf allen Trainingsdaten trainieren, weil `cross_val_score()` kein fertig trainiertes Modell zurückgibt
- **Abschliessender Test:** Testmerkmale vorhersagen und diese Vorhersagen mit den tatsächlichen Testzielwerten vergleichen
- **Interpretation gemäss Musterlösung:** Random Forest erreicht ungefähr 1.55 L/100 km CV-RMSE, Bagging 1.56 und Boosting 1.57, der Wald gewinnt rechnerisch, die Unterschiede zwischen den Ensembles sind aber klein
- **Interpretation der Testbewertung:** Der anschliessende Test-RMSE schätzt die Fehlergrösse des ausgewählten Modells auf ungesehenen Autos, er muss nicht genau dem CV-RMSE entsprechen
- **Stolperstein:** Nicht anhand des Testfehlers immer wieder Modelle oder Einstellungen ändern, die Auswahl erfolgt vor der abschliessenden Testbewertung

```python
#Verwenden, um Regressionsmodelle mit CV zu vergleichen und das ausgewählte Modell abschliessend zu testen
#VORAUSSETZUNG: np, pd, die verwendeten Modellklassen, cross_val_score und mean_squared_error sind bereits importiert
#VORAUSSETZUNG: Trainings- und Testdaten sind bereits vorbereitet


#1. Vorbereitete Daten zuordnen
comparison_X_train = X_train_cars
comparison_y_train = y_train_cars
comparison_X_test = X_test_cars
comparison_y_test = y_test_cars
comparison_unit = 'L/100 km'
#ANPASSEN: Rechts die vier Trainings- und Testvariablen deiner Aufgabe einsetzen und die Einheit ersetzen


#2. Modelle und vorgegebene Einstellungen festlegen
comparison_models = {
    'Linear': LinearRegression(),
    'Pruned tree': DecisionTreeRegressor(ccp_alpha=0.05, random_state=42),
    'Bagging': BaggingRegressor(n_estimators=50, random_state=42),
    'Forest': RandomForestRegressor(
        n_estimators=100, max_features=1, random_state=42
    ),
    'Boosting': GradientBoostingRegressor(
        n_estimators=100, max_depth=2, learning_rate=0.1, random_state=42
    )
}
#ANPASSEN: Nur die verlangten Modelle aufnehmen und ihre Parameter gemäss Aufgabe setzen
#HINWEIS: Links stehen frei wählbare Anzeigenamen, rechts die Modellaufrufe


#3. Jedes Modell mit derselben Cross-Validation prüfen
comparison_folds = 5
#ANPASSEN: Anzahl Folds gemäss Aufgabe setzen

comparison_rows = []

for model_name, candidate_model in comparison_models.items():
    candidate_scores = cross_val_score(
        candidate_model,
        comparison_X_train, comparison_y_train,             #Nur Trainingsdaten für die Auswahl verwenden
        cv=comparison_folds,
        scoring='neg_mean_squared_error'                    #Negativen MSE je Fold berechnen
    )
    candidate_rmse = np.sqrt(-candidate_scores)             #Negativen MSE in RMSE umrechnen

    comparison_rows.append({
        'Modell': model_name,
        'CV-RMSE': candidate_rmse.mean()                    #Durchschnitt der einzelnen Fold-RMSE
    })
#HINWEIS: Bei diesem RMSE-Vergleich bleibt Schritt 3 nach Anpassung der Daten und Modelle unverändert
#ACHTUNG: Bei anderer Bewertungskennzahl müssen scoring, Umrechnung und Auswahlrichtung zusammenpassen


#4. Modelle nach CV-RMSE sortieren und Gewinner auswählen
comparison_results = (
    pd.DataFrame(comparison_rows)
    .set_index('Modell')
    .sort_values('CV-RMSE')                                 #Kleinster Fehler steht oben
)

print(f'CV-RMSE in {comparison_unit}, kleiner ist besser')
display(comparison_results.round(3))                        #Nur die Anzeige runden, nicht die Werte für die Auswahl

best_model_name = comparison_results.index[0]               #Name des Modells mit kleinstem CV-RMSE
best_regression_model = comparison_models[best_model_name]  #Zugehöriges Modell auswählen


#5. Gewinner auf allen Trainingsdaten trainieren und testen
best_regression_model.fit(comparison_X_train, comparison_y_train)
best_test_prediction = best_regression_model.predict(comparison_X_test)

best_test_rmse = np.sqrt(
    mean_squared_error(comparison_y_test, best_test_prediction)
)

print('Ausgewähltes Modell:', best_model_name)
print(f'Abschliessender Test-RMSE: {best_test_rmse:.3f} {comparison_unit}')
#HINWEIS: Schritte 4–5 bleiben beim RMSE-Vergleich unverändert, die Daten wurden in Schritt 1 zugeordnet
#ACHTUNG: Testdaten erst nach der Auswahl verwenden, aus diesem Testergebnis keine neue Modellsuche starten
```

## Permutation Importance – Welche Eingaben nutzt das Modell für seine Vorhersagen?
- **Definition:** Mischt die Werte einer Eingabespalte zufällig zwischen den Beobachtungen und misst, wie stark die Modellleistung dadurch abnimmt, das Modell wird dabei nicht neu trainiert
- **Zweck und Anwendung:** Nach dem Training untersuchen, auf welche Merkmale sich das Modell für seine Vorhersagen stützt
- **Input und Voraussetzungen:** Ein trainiertes Modell, eine passende Eingabetabelle und die zugehörigen tatsächlichen Zielwerte, hier der bereits trainierte Random Forest `forest_cars`
- **Ablauf:** Modellleistung auf unveränderten Daten berechnen, eine Spalte mischen, erneut bewerten und die Differenz bestimmen, für jede Spalte wiederholen
- **`scoring='r2'`:** Bewertet hier mit R² → Importance = ursprüngliches R² minus R² nach dem Mischen
- **`n_repeats=5`:** Mischt jede Spalte fünfmal, um die Abhängigkeit von einer einzelnen zufälligen Mischung zu verringern
- **`.importances_mean`:** Mittlere Abnahme des R² pro Merkmal
- **`.importances_std`:** Streuung der Abnahmen zwischen den Wiederholungen, kein Konfidenzintervall
- **Interpretation positiver Werte:** Je stärker das R² nach dem Mischen sinkt, desto stärker nutzt das Modell dieses Merkmal auf den ausgewerteten Daten
- **Interpretation nahe 0:** Das Mischen verändert die Leistung kaum, das Merkmal trägt für dieses Modell und diese Daten wenig zusätzliche Vorhersageinformation bei
- **Interpretation negativer Werte:** Nach dem Mischen ist die Leistung besser, möglich durch Zufall oder ein Merkmal, das die Vorhersagen beeinträchtigt
- **Bsp.::** Importance `0.20` bedeutet eine Abnahme des R² um 0.20, nicht 20 % Merkmalsanteil, die Werte müssen sich nicht zu 1 summieren
- **Datenauswahl gemäss Aufgabe:** Die Übung verwendet Trainingsdaten, diese Importance kann zu optimistisch sein, auf zurückgehaltenen Daten lässt sich die Bedeutung für ungesehene Beobachtungen untersuchen
- **Korrelierte Merkmale:** Können sich gegenseitig ersetzen, dadurch kann ein relevantes Merkmal beim einzelnen Mischen weniger wichtig erscheinen
- **Stolpersteine:** Importance beschreibt die Nutzung durch das Modell, keine Ursache-Wirkungs-Beziehung, sie zeigt auch nicht, ob grössere Merkmalswerte die Vorhersage erhöhen oder senken

```python
#Verwenden, um bei einem trainierten Regressionsmodell die Bedeutung einzelner Eingaben durch zufälliges Mischen zu untersuchen
from sklearn.inspection import permutation_importance
#VORAUSSETZUNG: pd ist importiert und forest_cars ist bereits trainiert


#1. Modell und auszuwertende Daten festlegen
importance_model = forest_cars                                                  #Bereits trainierten Random Forest verwenden
importance_X = X_train_cars                                                     #Gemäss Übung die Trainingsmerkmale auswerten
importance_y = y_train_cars                                                     #Zugehörige tatsächliche Verbrauchswerte
#ANPASSEN: forest_cars durch dein bereits trainiertes Regressionsmodell ersetzen
#ANPASSEN: X_train_cars und y_train_cars durch das zusammengehörende Datenpaar aus der Aufgabe ersetzen
#HINWEIS: Für eine Auswertung auf Testdaten hier stattdessen X_test_cars und y_test_cars einsetzen
#ACHTUNG: Bei Testdaten die Ergebnisse nicht für weitere Modelloptimierung verwenden


#2. Jede Eingabespalte mehrfach mischen und die Leistungsabnahme messen
permutation_result = permutation_importance(
    importance_model,
    importance_X, importance_y,
    scoring='r2',                                                               #Abnahme des R² messen
    n_repeats=5,                                                                #Jedes Merkmal fünfmal mischen
    random_state=42                                                             #Zufällige Mischungen reproduzierbar machen
)
#ANPASSEN: n_repeats und random_state gemäss Aufgabe setzen
#ANPASSEN: Falls eine andere Kennzahl verlangt wird, scoring und die Ergebnisbeschriftungen in Schritt 3 gemeinsam ändern
#HINWEIS: Hier kein fit() aufrufen, das bestehende Modell bleibt unverändert


#3. Ergebnisse den Eingabespalten zuordnen und sortieren
permutation_table = pd.DataFrame({
    'Merkmal': importance_X.columns,                                            #Spaltennamen in der Reihenfolge der ausgewerteten Eingaben
    'Mittlere Abnahme R²': permutation_result.importances_mean,
    'Standardabweichung': permutation_result.importances_std
}).sort_values('Mittlere Abnahme R²', ascending=False)                          #Grösste Leistungsabnahme zuerst anzeigen

display(permutation_table.round(3))
#HINWEIS: Bei unverändertem scoring='r2' bleibt Schritt 3 gleich
#ANPASSEN: Bei anderer Kennzahl 'Mittlere Abnahme R²' an beiden Stellen passend umbenennen
```
