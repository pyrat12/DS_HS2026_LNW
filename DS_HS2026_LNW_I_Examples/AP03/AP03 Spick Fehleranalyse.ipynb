# AP03 – Teil A Konzeptfragen

## Case 1

### A1 – Regression oder Klassifikation und passende Kennzahl
**Aufgabe:** Ein Gebrauchtwagenhändler möchte vorhersagen:
- Zu welchem Preis lässt sich ein Auto verkaufen?
- Benötigt das Auto vor dem Verkauf eine Reparatur?

**a) Welche Lernaufgabe liegt jeweils vor?**
- **Preis → Regression:** Die Zielvariable ist ein numerischer Betrag in CHF
- **Reparaturbedarf → Klassifikation:** Die Zielvariable enthält die Klassen «Reparatur nötig» und «keine Reparatur nötig»

**b) Nenne je eine passende Kennzahl für die Bewertung auf Testdaten**
- **Preis → RMSE:** Beschreibt die Fehlergrösse der Preisvorhersagen in CHF, kleiner ist besser
- **Reparaturbedarf → Recall:** Beschreibt den Anteil tatsächlich reparaturbedürftiger Autos, die erkannt werden, grösser ist besser

**Mögliche Varianten:** Mietpreis vs. Kündigung ja/nein, Lieferdauer vs. verspätet ja/nein, Verbrauch vs. Defekt ja/nein

### A2 – OLS-Koeffizienten interpretieren und eine Vorhersage berechnen

**Aufgabe:** Ein multiples lineares Regressionsmodell schätzt den Verkaufspreis eines Autos:

`Preis = 20000 − 1200 × Alter − 0.05 × Kilometerstand + 4000 × Elektroauto`

Dabei gilt `Elektroauto = 1` für Elektroautos und `0` für andere Autos

**a) Interpretiere den Koeffizienten von Elektroauto**
- Ein Elektroauto wird im Modell durchschnittlich um 4000 CHF teurer geschätzt als ein anderes Auto, bei gleichem Alter und Kilometerstand

**b) Berechne den geschätzten Preis eines vierjährigen Elektroautos mit 60000 km**
- `20000 − 1200 × 4 − 0.05 × 60000 + 4000 × 1 = 16200 CHF`
- Das Modell schätzt den Verkaufspreis auf 16200 CHF

**Mögliche Varianten der Koeffizientenfrage**
- **Alter:** Pro zusätzlichem Jahr sinkt der geschätzte Preis um 1200 CHF, bei gleichem Kilometerstand und Elektroauto-Status
- **Kilometerstand:** Pro zusätzlichem Kilometer sinkt der geschätzte Preis um 0.05 CHF, bei gleichem Alter und Elektroauto-Status
- **Wichtig:** Immer Einheit und konstant gehaltene andere Merkmale nennen, der Zusammenhang belegt keine Ursache

### A3 – Recall, Accuracy und Fehlerfolgen beurteilen

**Aufgabe:** Ein Modell erkennt den Reparaturbedarf von 200 Autos, Klasse 1 bedeutet «Reparatur nötig»

| Tatsächliche Klasse | Vorhergesagt 0 | Vorhergesagt 1 |
|---|---:|---:|
| 0: Keine Reparatur nötig | 150 | 10 |
| 1: Reparatur nötig | 20 | 20 |

Ein Kollege sagt: «85 % Accuracy, das Modell ist sehr gut»

**a) Berechne und interpretiere den Recall für Klasse 1**

- **Gesucht:** Welcher Anteil der tatsächlich reparaturbedürftigen Autos wird erkannt?
- **Richtige Zeile wählen:** «Tatsächlich 1: Reparatur nötig», darin stehen alle Autos, die wirklich eine Reparatur brauchen
- **TP = 20:** Reparatur nötig und als reparaturbedürftig vorhergesagt → richtig erkannt
- **FN = 20:** Reparatur nötig, aber als «keine Reparatur nötig» vorhergesagt → übersehen
- **Insgesamt reparaturbedürftig:** `TP + FN = 20 + 20 = 40 Autos`
- **Recall:** `richtig erkannt / tatsächlich reparaturbedürftig = 20 / 40 = 0.50 = 50 %`
- **Interpretation:** Von 40 tatsächlich reparaturbedürftigen Autos erkennt das Modell 20, die anderen 20 werden übersehen

**b) Weshalb genügt die Accuracy nicht als Qualitätsurteil?**
- 160 von 200 Autos benötigen keine Reparatur, immer «keine Reparatur» vorherzusagen würde bereits 80 % Accuracy ergeben
- Die Accuracy von 85 % verdeckt, dass die Hälfte der Reparaturfälle übersehen wird

**c) Welcher Fehler ist kritischer? Begründe mit dem Anwendungsfall**
- **FN:** Ein reparaturbedürftiges Auto wird als unproblematisch eingestuft und könnte mit einem sicherheitsrelevanten Defekt verkauft werden
- **FP:** Ein funktionierendes Auto wird unnötig kontrolliert, dadurch entstehen zusätzliche Kosten
- Bei sicherheitsrelevanten Defekten sind FN besonders kritisch, deshalb ist ein hoher Recall wichtig

**Mögliche Variante: Precision statt Recall**
- **Gesucht:** Welcher Anteil der als reparaturbedürftig vorhergesagten Autos braucht tatsächlich eine Reparatur?
- **Richtige Spalte wählen:** «Vorhergesagt 1», darin stehen alle Autos, für die das Modell eine Reparatur empfiehlt
- **TP = 20:** Reparatur vorhergesagt und tatsächlich nötig → richtige Empfehlung
- **FP = 10:** Reparatur vorhergesagt, aber tatsächlich nicht nötig → Fehlalarm
- **Insgesamt als reparaturbedürftig vorhergesagt:** `TP + FP = 20 + 10 = 30 Autos`
- **Precision:** `richtige Empfehlungen / alle Reparaturempfehlungen = 20 / 30 ≈ 0.667 = 66.7 %`
- **Interpretation:** Von 30 Autos mit Reparaturempfehlung brauchen 20 tatsächlich eine Reparatur, bei den übrigen 10 ist die Empfehlung ein Fehlalarm

## Ergänzende Denk- & Rechenaufgaben

### 1. Preisunterschied aus einer Merkmalsänderung berechnen

**Aufgabe:** Im multiplen Preismodell beträgt der Koeffizient von `mileage_km` −0.03, zwei Velos unterscheiden sich nur im Kilometerstand, Velo B ist 2500 km mehr gefahren

**Frage:** Wie unterscheiden sich die vorhergesagten Preise?

- **Bedeutung des Koeffizienten:** Pro zusätzlichem Kilometer sinkt der geschätzte Preis um 0.03 CHF, bei gleichen anderen Modellmerkmalen
- **Preisunterschied:** `−0.03 × 2500 = −75 CHF`
- **Antwort:** Velo B wird um 75 CHF günstiger geschätzt als Velo A, die absoluten Verkaufspreise lassen sich daraus nicht bestimmen

### 2. Mehrere Merkmalsänderungen gemeinsam berücksichtigen

**Aufgabe:** Ein Preismodell enthält folgende Koeffizienten:
- Alter: −120 CHF pro Jahr
- Kilometerstand: −0.02 CHF pro km
- Ausstattung: +200 CHF pro Stufe

Velo B ist gegenüber Velo A zwei Jahre älter, 3000 km mehr gefahren und eine Ausstattungsstufe besser, alle anderen Merkmale sind gleich

**Frage:** Welches Velo wird teurer geschätzt und um wie viel?

- **Alter:** Zwei zusätzliche Jahre verändern den geschätzten Preis um `−120 × 2 = −240 CHF`
- **Kilometerstand:** 3000 zusätzliche km verändern ihn um `−0.02 × 3000 = −60 CHF`
- **Ausstattung:** Eine zusätzliche Stufe verändert ihn um `+200 CHF`
- **Gesamter Unterschied B gegenüber A:** `−240 − 60 + 200 = −100 CHF`
- **Antwort:** Velo B wird trotz besserer Ausstattung um 100 CHF günstiger geschätzt
- **Warum kein Achsenabschnitt nötig ist:** Beide Vorhersagen enthalten denselben Achsenabschnitt, bei der Differenz fällt er weg

### 3. Binäre Eingabe umcodieren

**Aufgabe:** Das Modell lautet:

`Preis = 1000 − 100 × Alter + 1600 × is_ebike`

Bisher gilt `is_ebike = 1` für E-Bikes und `0` für normale Velos, neu wird stattdessen `is_normal = 1` für normale Velos und `0` für E-Bikes verwendet

**Frage:** Wie lautet dieselbe Vorhersageregel mit der neuen Codierung?

- **Zusammenhang:** `is_ebike = 1 − is_normal`
- **Einsetzen:** `Preis = 1000 − 100 × Alter + 1600 × (1 − is_normal)`
- **Vereinfachen:** `Preis = 2600 − 100 × Alter − 1600 × is_normal`
- **Interpretation:** Normale Velos werden bei gleichem Alter um 1600 CHF günstiger geschätzt als E-Bikes
- **Wichtig:** Referenzgruppe, Achsenabschnitt und Vorzeichen des Gruppenkoeffizienten ändern sich, die vorhergesagten Preise bleiben gleich

### 4. Confusion Matrix aus einem Text rekonstruieren

**Aufgabe:** Von 100 Velos benötigen tatsächlich 20 einen Service, das Modell erkennt davon 15 und empfiehlt insgesamt für 25 Velos einen Service, Klasse 1 bedeutet «Service nötig»

**a) Bestimme TP, FN, FP und TN**

- **TP = 15 (True Positive – richtig positiv):** Diese Servicefälle werden korrekt erkannt
- **FN = 5 (False Negative – falsch negativ):** Von 20 tatsächlichen Servicefällen werden nur 15 erkannt, also bleiben `20 − 15 = 5` unentdeckt
- **FP = 10 (False Positive – falsch positiv):** Von 25 Serviceempfehlungen sind nur 15 richtig, also sind `25 − 15 = 10` Fehlalarme
- **TN = 70 (True Negative – richtig negativ):** Tatsächlich benötigen 80 Velos keinen Service, davon erhalten zehn einen Fehlalarm, also bleiben `80 − 10 = 70` korrekte negative Vorhersagen

| Tatsächliche Klasse | Vorhergesagt 0 | Vorhergesagt 1 |
|---|---:|---:|
| 0: Kein Service nötig | 70 | 10 |
| 1: Service nötig | 5 | 15 |

**b) Berechne und interpretiere Recall und Precision**

- **Recall:** Von 20 tatsächlichen Servicefällen werden 15 erkannt → `15 / 20 = 75 %`
- **Precision:** Von 25 Serviceempfehlungen sind 15 berechtigt → `15 / 25 = 60 %`

### 5. Zwei Modelle mit gleicher Accuracy beurteilen

**Aufgabe:** Zwei Modelle werden auf denselben 200 Velos geprüft, davon benötigen 40 tatsächlich einen Service

| Modell | TN | FP | FN | TP |
|---|---:|---:|---:|---:|
| A | 150 | 10 | 20 | 20 |
| B | 135 | 25 | 5 | 35 |

**a) Welches Modell hat die höhere Accuracy?**

- **Modell A:** `Accuracy = (150 + 20) / 200 = 85 %`
- **Modell B:** `Accuracy = (135 + 35) / 200 = 85 %`
- **Antwort:** Beide haben dieselbe Accuracy, jeweils 170 richtige und 30 falsche Vorhersagen

**b) Welches Modell eignet sich eher, wenn übersehene Servicefälle besonders kritisch sind?**

- **Recall A:** Von 40 Servicefällen werden 20 erkannt → `20 / 40 = 50 %`
- **Recall B:** Von 40 Servicefällen werden 35 erkannt → `35 / 40 = 87.5 %`
- **Antwort:** Modell B übersieht deutlich weniger Servicefälle, verursacht dafür mehr unnötige Kontrollen
- **Schlussfolgerung:** Gleiche Accuracy bedeutet nicht dieselbe praktische Eignung, die Fehlerarten müssen berücksichtigt werden

### 6. Fehlerzahlen aus dem Recall ableiten

**Aufgabe:** Unter den Testvelos benötigen tatsächlich 50 einen Service, der Recall für diese Klasse beträgt 80 %

**a) Wie viele Servicefälle wurden erkannt und wie viele übersehen?**

- **Recall bedeutet:** 80 % der tatsächlich servicebedürftigen Velos wurden erkannt
- **TP:** `0.80 × 50 = 40` erkannte Servicefälle
- **FN:** `50 − 40 = 10` übersehene Servicefälle

**b) Kann daraus auch die Precision berechnet werden?**

- Nein, die Anzahl Fehlalarme FP ist unbekannt
- Für Precision benötigen wir alle Serviceempfehlungen, also die richtigen Empfehlungen TP und die Fehlalarme FP

### 7. Gleicher Recall, unterschiedliche Precision

**Aufgabe:** Zwei Modelle erkennen jeweils 18 von 20 tatsächlichen Servicefällen, Modell A empfiehlt insgesamt 30 Velos einen Service und Modell B insgesamt 20

**a) Vergleiche Recall und Precision**

- **Recall beider Modelle:** Jeweils 18 von 20 tatsächlichen Servicefällen erkannt → `18 / 20 = 90 %`
- **Precision A:** 18 richtige unter 30 Empfehlungen → `18 / 30 = 60 %`
- **Precision B:** 18 richtige unter 20 Empfehlungen → `18 / 20 = 90 %`

**b) Welches Modell verursacht weniger unnötige Kontrollen?**

- **Fehlalarme A:** `30 − 18 = 12`
- **Fehlalarme B:** `20 − 18 = 2`
- **Antwort:** Modell B erkennt gleich viele Servicefälle, erzeugt aber zehn Fehlalarme weniger
- **Schlussfolgerung:** Gleicher Recall bedeutet nicht gleich zuverlässige positive Vorhersagen

### 8. Modelle anhand konkreter Fehlerkosten auswählen

**Aufgabe:** Für die Modelle aus Aufgabe 5 kostet ein übersehener Servicefall 200 CHF und eine unnötige Kontrolle 20 CHF

**Frage:** Welches Modell verursacht geringere Fehlerkosten?

- **Zuordnung:** Übersehener Servicefall = FN, unnötige Kontrolle = FP
- **Formel:** `Gesamtkosten = FN × 200 CHF + FP × 20 CHF`
- **Modell A:** `20 × 200 + 10 × 20 = 4200 CHF`
- **Modell B:** `5 × 200 + 25 × 20 = 1500 CHF`
- **Antwort:** Modell B verursacht unter diesen Annahmen 2700 CHF weniger Fehlerkosten
- **Interpretation:** Die zusätzlichen Fehlalarme sind hier weniger teuer als die vermiedenen übersehenen Servicefälle, obwohl beide Modelle dieselbe Accuracy erreichen

### 9. Bedeutung der positiven Klasse ändern

**Aufgabe:** Bisher bedeutet Klasse 1 «Service nötig», neu bedeutet Klasse 1 «kein Service nötig»

**Frage:** Bedeutet ein hoher Recall weiterhin, dass viele Servicefälle erkannt werden?

- **Antwort:** Nein, Recall bezieht sich auf die jeweils als positiv definierte Klasse
- **Neue Bedeutung:** Anteil der tatsächlich nicht servicebedürftigen Velos, die korrekt als «kein Service nötig» erkannt werden
- **TP neu:** Kein Service nötig und korrekt als «kein Service nötig» vorhergesagt
- **FN neu:** Kein Service nötig, aber fälschlich als servicebedürftig vorhergesagt
- **Vorgehen:** Vor jeder Berechnung zuerst die Bedeutung der positiven Klasse prüfen, danach die Fallzahlen zuordnen

## Weitere Fragen

### 1. Lernaufgabe und passende Kennzahl bestimmen

- **Mögliche Frage:** Ein Unternehmen möchte den Verkaufspreis eines Produkts und dessen Reparaturbedarf vorhersagen, welche Lernaufgaben liegen vor und welche Kennzahlen passen?
- **Antwort:** Preis → Regression, weil ein Zahlenwert vorhergesagt wird, Bewertung bspw. mit RMSE oder R² auf Testdaten
- **Antwort:** Reparaturbedarf ja/nein → Klassifikation, weil eine Klasse vorhergesagt wird, Bewertung bspw. mit Recall, Precision oder AUC
- **Wichtig:** Die Einordnung mit der konkreten Zielvariable begründen, eine als 0/1 gespeicherte Klasse bleibt eine Klassifikationsaufgabe

### 2. Features und Zielvariable unterscheiden

- **Mögliche Frage:** Welche Angaben dienen als Eingaben und welche als Zielvariable?
- **Antwort:** Features X sind die zur Vorhersage verwendeten Merkmale, y ist die gesuchte Antwort, bspw. Alter und Kilometerstand als Eingaben und Verkaufspreis als Ziel
- **Wichtig:** Die Zielvariable darf nicht zusätzlich als Eingabe verwendet werden, die Features müssen zum vorgesehenen Vorhersagezeitpunkt verfügbar sein

### 3. Lineare Regressionskoeffizienten interpretieren

- **Mögliche Frage:** Der Koeffizient von `age_years` beträgt −100, was bedeutet das?
- **Antwort:** Pro zusätzlichem Altersjahr sinkt der geschätzte Verkaufspreis im Modell durchschnittlich um 100 CHF, bei gleichbleibenden anderen enthaltenen Merkmalen
- **Mögliche Frage:** Der Koeffizient von `is_ebike` beträgt 1700, was bedeutet das?
- **Antwort:** Ein E-Bike wird bei gleichem Alter, Kilometerstand und gleicher Ausstattung im Durchschnitt um 1700 CHF teurer geschätzt als ein konventionelles Velo
- **Wichtig:** Einheit und Vergleich nennen, bei multipler Regression die anderen Modellmerkmale konstant halten, keine Kausalität behaupten

### 4. Eine Vorhersage von Hand berechnen

- **Mögliche Frage:** Berechnen Sie den Preis anhand einer vorgegebenen Regressionsgleichung und der Merkmale eines neuen Produkts
- **Vorgehen:** Achsenabschnitt plus jeden Koeffizienten mal zugehörigem Eingabewert zusammenzählen
- **Beispiel mit erfundenen Werten:** `Preis = 1000 − 100 × Alter + 1500 × E-Bike`, ein zweijähriges E-Bike ergibt `1000 − 100 × 2 + 1500 × 1 = 2300 CHF`
- **Wichtig:** Kategorien korrekt codieren, Einheiten und Vorzeichen beachten

### 5. Achsenabschnitt und OLS erklären

- **Mögliche Frage:** Was bedeutet der Achsenabschnitt und wie bestimmt OLS die Gerade?
- **Antwort:** Der Achsenabschnitt ist der vorhergesagte Zielwert, wenn alle numerischen Eingaben 0 sind, diese Situation muss fachlich nicht sinnvoll sein
- **Antwort:** OLS wählt die Koeffizienten so, dass die Summe der quadrierten Residuen möglichst klein wird
- **Wichtig:** Residuum = tatsächlicher Wert minus Vorhersage, positiv bedeutet unterschätzt und negativ überschätzt

### 6. R² und RMSE interpretieren

- **Mögliche Frage:** Das Trainings-R² beträgt 0.80, was bedeutet das?
- **Antwort:** Das OLS-Modell mit Achsenabschnitt erklärt 80 % der Zielwertvariabilität in den Trainingsdaten, das bedeutet weder 80 % richtige Vorhersagen noch automatisch gute Testleistung
- **Mögliche Frage:** Der Test-RMSE beträgt 400 CHF, was bedeutet das?
- **Antwort:** Die quadratisch gemittelte Fehlergrösse auf den Testdaten beträgt 400 CHF, einzelne Vorhersagen können stärker oder schwächer abweichen
- **Mögliche Frage:** Kann das Test-R² negativ sein?
- **Antwort:** Ja, dann ist die Summe der quadrierten Vorhersagefehler grösser als bei einer konstanten Vorhersage des Testmittelwerts
- **Wichtig:** RMSE kleiner = besser, R² grösser = besser, nur auf vergleichbaren Daten sinnvoll vergleichen

### 7. Interpolation und Extrapolation unterscheiden

- **Mögliche Frage:** Das Modell wurde mit Wohnflächen von 40 bis 80 m² trainiert, wie ist eine Vorhersage für 95 m² einzuordnen?
- **Antwort:** Extrapolation, weil die neue Eingabe ausserhalb des beobachteten Bereichs liegt, der gelernte Zusammenhang muss dort nicht weiter gelten
- **Wichtig:** Eine Vorhersage innerhalb dieses Bereichs wäre Interpolation, auch diese ist nicht automatisch fehlerfrei

### 8. Konfidenz- und Vorhersageintervall unterscheiden

- **Mögliche Frage:** Welches Intervall verwenden Sie für die durchschnittliche Miete aller 95-m²-Wohnungen und welches für eine einzelne neue Wohnung?
- **Antwort:** Konfidenzintervall für den durchschnittlichen Mietpreis, Vorhersageintervall für die einzelne neue Wohnung
- **Mögliche Frage:** Warum ist das Vorhersageintervall breiter?
- **Antwort:** Es berücksichtigt zusätzlich die Streuung einzelner Wohnungen um den geschätzten Durchschnitt
- **Wichtig:** Die Punktvorhersage kann identisch sein, obwohl die Unsicherheitsbereiche unterschiedlich breit sind

### 9. Regressionsannahmen anhand von Grafiken beurteilen

- **Mögliche Frage:** Was bedeutet eine Trichterform im Residuenplot?
- **Antwort:** Hinweis auf eine veränderliche Fehlervarianz, die Streuung der Fehler ist nicht überall gleich
- **Mögliche Frage:** Was deutet ein gekrümmtes Residuenmuster an?
- **Antwort:** Das Modell beschreibt den durchschnittlichen Zusammenhang möglicherweise unzureichend
- **Mögliche Frage:** Kann ein Q–Q-Plot die Unabhängigkeit nachweisen?
- **Antwort:** Nein, er untersucht die Verteilungsform der Residuen, Unabhängigkeit muss anhand der Datenerhebung und möglicher Abhängigkeiten beurteilt werden

### 10. Train/Test-Split und Cross-Validation erklären

- **Mögliche Frage:** Warum reicht es nicht, das Modell auf seinen Trainingsdaten zu bewerten?
- **Antwort:** Das Modell hat diese Beobachtungen bereits zum Lernen verwendet, seine Leistung darauf ist häufig zu optimistisch
- **Mögliche Frage:** Wozu dient Cross-Validation und wozu der Testdatensatz?
- **Antwort:** CV dient dem Vergleich von Modellen und Einstellungen innerhalb der Trainingsdaten, Testdaten dienen der abschliessenden Bewertung nach der Auswahl
- **Mögliche Frage:** Wie funktioniert 5-fold CV?
- **Antwort:** Trainingsdaten in fünf Teile aufteilen, jeweils auf vier Teilen lernen und auf dem fünften validieren, jeder Teil wird einmal zur Validierung verwendet
- **Wichtig:** Bei Klassifikation erhält eine geschichtete Aufteilung die Klassenanteile ungefähr

### 11. Underfitting und Overfitting erkennen

- **Mögliche Frage:** Der Trainingsfehler sinkt mit zunehmender Baumtiefe, der CV-Fehler steigt wieder, was bedeutet das?
- **Antwort:** Hinweis auf Overfitting, der komplexere Baum passt sich stärker an Trainingsdetails an und sagt ungesehene Validierungsfälle schlechter vorher
- **Mögliche Frage:** Trainings- und CV-Fehler sind beide hoch, was könnte vorliegen?
- **Antwort:** Underfitting, das Modell ist möglicherweise zu einfach oder die Features liefern zu wenig passende Information
- **Wichtig:** Ähnliche Trainings- und Validierungsfehler allein beweisen kein gutes Modell

### 12. Bias und Variance unterscheiden

- **Mögliche Frage:** Warum kann ein sehr einfacher Baum schlechter sein als ein mitteltiefer Baum?
- **Antwort:** Er kann wichtige Muster übersehen und dadurch einen hohen Bias haben
- **Mögliche Frage:** Warum können sehr tiefe Bäume instabil sein?
- **Antwort:** Ihre Regeln können sich bei anderen Trainingsstichproben stark verändern, das bezeichnet hohe Variance
- **Wichtig:** Gesucht ist ein guter Ausgleich, zusätzlich bleibt nicht vorhersagbares Rauschen

### 13. Confusion Matrix lesen und Kennzahlen berechnen

- **Mögliche Frage:** Ein Service-Modell hat TN = 175, FP = 6, FN = 28 und TP = 31, berechnen und interpretieren Sie den Recall
- **Antwort:** `Recall = 31 / (31 + 28) ≈ 52.5 %`, das Modell erkennt nur etwas mehr als die Hälfte der tatsächlich servicebedürftigen Velos
- **Mögliche Frage:** Was unterscheidet Precision und Recall?
- **Antwort:** Precision fragt nach dem Anteil echter positiver Fälle unter den positiven Vorhersagen, Recall nach dem Anteil erkannter Fälle unter allen tatsächlichen positiven Fällen
- **Formeln:** `Precision = TP / (TP + FP)`, `Recall = TP / (TP + FN)`
- **Wichtig:** Bei der üblichen Anordnung sind Zeilen tatsächlich und Spalten vorhergesagt → `[[TN, FP], [FN, TP]]`

### 14. Hohe Accuracy kritisch beurteilen

- **Mögliche Frage:** Das Service-Modell erreicht ungefähr 86 % Accuracy, weshalb genügt das nicht als Qualitätsurteil?
- **Antwort:** Bei vielen Velos ohne Servicebedarf kann bereits die ständige Vorhersage «kein Service» eine hohe Accuracy erreichen, trotzdem bleiben alle Servicefälle unerkannt
- **Vorgehen:** Klassenanteile und einfache Baseline prüfen, zusätzlich Recall, Precision und die Fehlerzahlen betrachten
- **Wichtig:** Welche Kennzahl besonders wichtig ist, hängt von der konkreten Entscheidung ab

### 15. Fehlerfolgen und Schwellenwerte erklären

- **Mögliche Frage:** Welcher Fehler ist beim Erkennen eines sicherheitsrelevanten Servicebedarfs kritischer?
- **Antwort:** Häufig FN, ein tatsächlich servicebedürftiges Velo wird übersehen, FP verursacht dagegen eine unnötige Kontrolle, die Begründung hängt vom Anwendungskontext ab
- **Mögliche Frage:** Was passiert beim Senken des Schwellenwerts?
- **Antwort:** Mehr positive Vorhersagen, Recall kann nicht sinken, gleichzeitig können mehr Fehlalarme entstehen
- **Mögliche Frage:** Wie berücksichtigen Sie unterschiedliche Fehlerkosten?
- **Antwort:** `Gesamtkosten = Kosten pro FN × FN + Kosten pro FP × FP`, Schwellenwert anhand von Validierungsdaten und fachlichen Kosten auswählen

### 16. Logistische Regression interpretieren

- **Mögliche Frage:** Was unterscheidet Score, Wahrscheinlichkeit und Klasse?
- **Antwort:** Der Score ist die gewichtete Summe der Eingaben, die Sigmoid-Funktion wandelt ihn in eine Wahrscheinlichkeit um, ein Schwellenwert macht daraus eine Klasse
- **Mögliche Frage:** Was bedeutet ein positiver Koeffizient?
- **Antwort:** Grössere Werte dieser Eingabe erhöhen die geschätzte Wahrscheinlichkeit für Klasse 1, bei konstanten anderen Eingaben
- **Wichtig:** Der Koeffizient ist keine konstante Änderung in Prozentpunkten, eine Wahrscheinlichkeit ist keine sichere Einzelentscheidung

### 17. ROC und AUC interpretieren

- **Mögliche Frage:** Ein Modell erreicht AUC 0.90, bedeutet das 90 % richtige Klassifikationen?
- **Antwort:** Nein, AUC bewertet die Rangtrennung, bei einem zufällig gewählten positiven und negativen Fall erhält der positive Fall mit entsprechender Wahrscheinlichkeit den höheren Score, Gleichstände zählen zur Hälfte
- **Mögliche Frage:** Was bedeuten die Achsen der ROC-Kurve?
- **Antwort:** X = Falsch-Positiv-Rate, Y = Recall bzw. Richtig-Positiv-Rate
- **Mögliche Frage:** Verändert ein anderer fester Schwellenwert die AUC?
- **Antwort:** Bei unveränderten Wahrscheinlichkeiten nicht, er verändert aber die vorhergesagten Klassen und damit die Confusion Matrix

### 18. Einen Entscheidungsbaum erklären

- **Mögliche Frage:** Wie liest man einen Baum und wie entsteht die Blattvorhersage?
- **Antwort:** Beim Wurzelknoten beginnen und den Bedingungen folgen, bei einer Bedingung «≤» geht erfüllt nach links und sonst nach rechts
- **Regression:** Ein Blatt sagt bei quadratischem Fehler den Mittelwert seiner Trainingszielwerte vorher
- **Klassifikation:** Die Klassenanteile im Blatt ergeben die geschätzten Wahrscheinlichkeiten, die häufigste Klasse wird vorhergesagt
- **Mögliche Frage:** Warum ist der E-Bike-Status als erster Split plausibel?
- **Antwort:** Er kann Gruppen mit deutlich unterschiedlichen Verkaufspreisen trennen und dadurch die quadrierten Trainingsfehler stark reduzieren

### 19. Gini und Pruning erklären

- **Mögliche Frage:** Ist ein Knoten mit 40 % Klasse 1 und 60 % Klasse 0 rein?
- **Antwort:** Nein, beide Klassen sind enthalten, bei zwei Klassen ist Gini nur bei ausschliesslich einer Klasse gleich 0
- **Mögliche Frage:** Welcher Split ist besser?
- **Antwort:** Der mit der grössten Abnahme der grössengewichteten Gini-Unreinheit, bei Regression der mit der kleinsten Gesamt-SSE der Kinder
- **Mögliche Frage:** Was bewirkt ein grösseres ccp_alpha?
- **Antwort:** Zusätzliche Blätter werden stärker bestraft, dadurch kann ein kleinerer Baum entstehen, die passende Einstellung mit CV auswählen

### 20. Bagging, Random Forest und Boosting unterscheiden

- **Bagging:** Trainiert Bäume auf Bootstrap-Stichproben und kombiniert ihre Vorhersagen
- **Random Forest:** Ergänzt die zufällige Merkmalsauswahl bei jedem Split, damit die Bäume weniger ähnlich werden
- **Boosting:** Baut kleine Bäume nacheinander als Verbesserungen des bisherigen Modells auf
- **Mögliche Frage:** Warum kann ein Wald stabiler sein als ein einzelner Baum?
- **Antwort:** Durch Mittelung gleichen sich Teile der Schwankungen einzelner Bäume aus
- **Wichtig:** Bei Regression mittelt der Wald Zahlenvorhersagen, bei Klassifikation mittelt scikit-learn Klassenwahrscheinlichkeiten

### 21. OOB und Feature Importance interpretieren

- **Mögliche Frage:** Was bedeutet eine Out-of-Bag-Bewertung?
- **Antwort:** Eine Trainingsbeobachtung wird nur durch Bäume vorhergesagt, die sie nicht in ihrer Bootstrap-Stichprobe hatten, das liefert eine interne Bewertung
- **Mögliche Frage:** Bedeutet Feature Importance 0.74, dass ein Merkmal 74 % höhere Preise verursacht?
- **Antwort:** Nein, die impurity-basierte Importance beschreibt seinen relativen Beitrag zur Verringerung der Unreinheit über die Baumaufteilungen
- **Mögliche Frage:** Was zeigt Permutation Importance?
- **Antwort:** Wie stark die gewählte Modellleistung nach dem zufälligen Mischen eines Merkmals abnimmt
- **Wichtig:** Importance beschreibt weder Kausalität noch die Richtung eines Effekts

### 22. Maximum Likelihood und Gradient Descent erklären

- **Mögliche Frage:** Was ist die Grundidee von Maximum Likelihood?
- **Antwort:** Parameter wählen, unter denen die bereits beobachteten Daten innerhalb des Modells am plausibelsten sind
- **Mögliche Frage:** Was macht Gradient Descent?
- **Antwort:** Parameter schrittweise entgegen dem Gradienten verändern, um das Trainingsziel zu verkleinern
- **Mögliche Frage:** Welche Rolle spielt die Lernrate?
- **Antwort:** Sie bestimmt die Schrittgrösse, zu klein kann langsam sein und zu gross kann das Minimum überspringen oder instabile Schritte erzeugen

### 23. SVM erklären

- **Mögliche Frage:** Was sucht eine SVM und was sind Support Vectors?
- **Antwort:** Eine Entscheidungsgrenze mit möglichst grosser Margin, die Stützvektoren sind die für diese Grenze entscheidenden Trainingsbeobachtungen
- **Mögliche Frage:** Was verändert C?
- **Antwort:** Ein grosses C bestraft Margin-Verletzungen stärker, ein kleines C erlaubt sie eher und regularisiert stärker
- **Mögliche Frage:** Wozu dient ein Kernel?
- **Antwort:** Er ermöglicht auch nichtlineare Entscheidungsgrenzen durch eine passende Ähnlichkeitsberechnung

# AP03 Spick Teil B – Fehleranalyse

### Prüfungs-Code auf falsche Ergebnisse kontrollieren

- **Auftrag zuerst lesen:** Welche Zielvariable, Eingaben, Modellklasse, Einstellungen und Kennzahlen sind ausdrücklich verlangt? Ein lauffähiger Code kann trotzdem die falsche Aufgabe lösen
- **Regression oder Klassifikation:** Numerischen Zielwert mit einem Regressionsmodell vorhersagen, Klassen mit einem Klassifikationsmodell, die Zahlenkodierung 0/1 macht Klassen nicht zu einer Regression
- **Zielvariable versehentlich in X:** Eingabespalten kontrollieren, die gesuchte Antwort darf nicht selbst als Eingabe enthalten sein, ein R² von 1 kann dadurch entstehen
- **Information zum Vorhersagezeitpunkt verfügbar:** Eingaben müssen bei der späteren Anwendung bereits bekannt sein, spätere Ergebnisse nicht als frühere Vorhersagegrundlage verwenden
- **Train/Test-Anteil vertauscht:** `test_size=0.2` bedeutet 20 % Testdaten und 80 % Trainingsdaten, nicht umgekehrt
- **Falsche Daten beim Training:** `.fit(X_train, y_train)` verwenden, Testdaten bleiben beim Lernen und bei der Modellauswahl zurückgehalten
- **Training statt Test bewertet:** Für die Leistung auf unbekannten Beobachtungen mit `X_test` vorhersagen und mit `y_test` vergleichen, Trainings-Accuracy von 1 beweist keine fehlerfreie Testleistung
- **Zuordnung verloren:** Eingaben, tatsächliche Zielwerte und Vorhersagen müssen dieselben Beobachtungen in derselben Reihenfolge beschreiben, bei neuen Eingaben auch die Spaltenreihenfolge prüfen
- **OLS-Konstante vergessen:** `sm.OLS()` benötigt für einen Achsenabschnitt eine zusätzliche Konstantenspalte, die Formelvariante `smf.ols('y ~ x', ...)` enthält ihn standardmässig
- **Parameter falsch interpretiert:** `max_depth` begrenzt die Tiefe je Baum, `n_estimators` zählt die Bäume bzw. Boosting-Schritte, `max_features=1` bedeutet ein Merkmal und `1.0` alle Merkmale
- **CV-Score falsch umgerechnet:** Bei `neg_mean_squared_error` zuerst das Minuszeichen entfernen und dann die Wurzel ziehen, AUC dagegen direkt mitteln
- **Beste Richtung verwechselt:** RMSE und Fehlerkosten minimieren, R² und AUC maximieren, Gini-Abnahme maximieren, gewichtete Gini-Unreinheit nach dem Split minimieren
- **Klassen statt Wahrscheinlichkeiten für AUC:** ROC/AUC aus Wahrscheinlichkeiten der positiven Klasse oder geeigneten Scores berechnen, die Spaltenreihenfolge von `predict_proba()` steht in `.classes_`
- **Confusion Matrix falsch gelesen:** Bei `labels=[0, 1]` gilt `[[TN, FP], [FN, TP]]`, Zeilen sind tatsächliche und Spalten vorhergesagte Klassen
- **Accuracy ohne Vergleichsbasis:** Klassenanteile und Recall prüfen, immer die häufigste Klasse vorherzusagen kann bereits eine hohe Accuracy liefern
- **Schwellenwert auf Testdaten optimiert:** Schwellenwert und Modelleinstellungen anhand von Validierungsdaten bzw. CV auswählen, erst danach abschliessend testen
- **Gelernte Vorverarbeitung auf allen Daten:** Skalierung oder Imputation nur auf Trainingsdaten lernen, innerhalb einer CV pro Fold neu lernen, dafür eine Pipeline verwenden
- **Plausibilitätscheck der Aufteilung:** Bei 1200 Beobachtungen und verlangten 80 % Training / 20 % Test müssen 960 Trainings- und 240 Testbeobachtungen entstehen, `test_size` bezeichnet immer den Testanteil
