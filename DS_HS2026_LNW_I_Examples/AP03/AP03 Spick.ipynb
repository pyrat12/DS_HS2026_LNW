# Foundations

### Machine Learning (ML) – Aus Daten lernen

- **Definition:** Machine Learning bedeutet, dass ein Computer aus Beispielen Muster und Regeln lernt, statt für jeden Einzelfall eine ausdrücklich programmierte Regel zu erhalten
- **Zweck:** Gelernte Zusammenhänge für Vorhersagen auf neuen Beobachtungen nutzen oder bisher unbekannte Strukturen in Daten entdecken
- **Datenaufbau:** Jede Zeile beschreibt eine Beobachtung, bspw. ein Auto oder eine Wohnung, die Spalten enthalten deren Merkmale
- **Features `X`:** Eingabevariablen, mit denen das Modell arbeitet, bspw. Hubraum und Zylinderzahl
- **Zielvariable `y`:** Gesuchte Antwort beim überwachten Lernen, bspw. der tatsächliche Treibstoffverbrauch
- **Modell:** Gelernte Regel, die Eingaben mit einer Vorhersage verbindet → `ŷ = f̂(X)`, dabei ist `ŷ` der vorhergesagte und `y` der tatsächliche Wert
- **Interpretation:** Entscheidend ist, ob das Modell bei neuen, bisher ungesehenen Beobachtungen brauchbare Ergebnisse liefert, gute Ergebnisse auf bereits bekannten Trainingsdaten allein reichen nicht aus

### Supervised Learning – Überwachtes Lernen mit bekannten Antworten

- **Definition:** Das Modell lernt anhand von Beispielen, bei denen sowohl die Features `X` als auch die zugehörige Zielvariable `y` bekannt sind
- **Zweck und Anwendung:** Verwenden, wenn eine bestimmte Zahl oder Klasse für neue Beobachtungen vorhergesagt werden soll
- **Input:** Trainingsdaten mit Eingaben und bekannten Antworten, bei neuen Beobachtungen werden dem Modell nur die benötigten Eingaben übergeben
- **Bsp.:** Aus Hubraum, Zylinderzahl und bekanntem Verbrauch vorhandener Autos lernen, um den Verbrauch weiterer Autos vorherzusagen
- **Zwei Hauptaufgaben:** Regression für Zahlenwerte und Klassifikation für Kategorien

### Regression – Einen Zahlenwert vorhersagen

- **Definition:** Supervised Learning mit einer numerischen Zielvariable, deren Zahlenwert vorhergesagt werden soll
- **Wann anwenden:** Bei Fragen wie «Wie viel?», «Wie hoch?» oder «Wie lange?», bspw. Mietpreis in CHF oder Verbrauch in L/100 km
- **Methoden aus den Aufgaben:** Lineare Regression, Regressionsbaum, Random Forest Regression und Gradient Boosting Regression
- **Output:** Ein geschätzter Zahlenwert pro Beobachtung, bspw. ein vorhergesagter Mietpreis
- **Interpretation:** Vorhersagen mit tatsächlichen Werten vergleichen, RMSE beschreibt die Fehlergrösse in der Einheit der Zielvariable und R² die Leistung relativ zur Mittelwertvorhersage

### Klassifikation – Eine Klasse vorhersagen

- **Definition:** Supervised Learning mit einer kategorialen Zielvariable, jede Beobachtung soll einer Klasse zugeordnet werden
- **Wann anwenden:** Bei Fragen wie «Welche Klasse?» oder «Ja oder nein?», bspw. Service nötig oder nicht nötig
- **Binäre Klassifikation:** Zwei Klassen, bspw. `0 = kein Service` und `1 = Service nötig`
- **Mehrklassenklassifikation:** Mehr als zwei Klassen, bspw. verschiedene Tierarten
- **Methoden aus den Aufgaben:** Logistische Regression, Klassifikationsbaum, Random Forest Classification und Gradient Boosting Classification
- **Output:** Eine vorhergesagte Klasse und je nach Methode zusätzlich Wahrscheinlichkeiten für die Klassen
- **Interpretation:** Die vorhergesagten Klassen mit den tatsächlichen Klassen vergleichen, bspw. über Confusion Matrix, Accuracy, Precision und Recall, ROC/AUC bewertet die Rangfolge anhand von Wahrscheinlichkeiten oder Scores
- **Stolperstein:** Zahlen können Kategorien codieren, `0` und `1` als Serviceklassen machen die Aufgabe nicht zur Regression, auch die logistische Regression ist trotz ihres Namens ein Klassifikationsverfahren

### Unsupervised Learning – Unüberwachtes Lernen ohne Zielvariable

- **Definition:** Das Modell erhält nur Features `X`, keine vorgegebenen Antworten `y`, und sucht selbst nach Strukturen in den Daten
- **Zweck und Anwendung:** Verwenden, wenn ähnliche Beobachtungen gruppiert oder viele Merkmale kompakt zusammengefasst werden sollen
- **Clustering:** Ähnliche Beobachtungen zu Gruppen zusammenfassen, bspw. Kundensegmente anhand ihres Kaufverhaltens, Methoden sind k-Means und hierarchisches Clustering
- **Dimensionsreduktion:** Viele Merkmale durch wenige neue Merkmale zusammenfassen, bspw. mit PCA, um Daten übersichtlicher darzustellen
- **Output:** Bspw. Clusterzugehörigkeiten oder neue zusammenfassende Merkmale
- **Interpretation:** Prüfen, ob Gruppen oder Zusammenfassungen stabil, nachvollziehbar und für die Fragestellung sinnvoll sind, Cluster haben zunächst keine vorgegebene fachliche Bedeutung
- **Unterschied zur Klassifikation:** Klassifikation lernt bereits bekannte Klassen, Clustering sucht Gruppen ohne vorgegebene Klassenzuordnung

### Reinforcement Learning – Lernen durch Belohnung

- **Definition:** Ein Agent führt Aktionen in einer Umgebung aus und lernt anhand der erhaltenen Belohnungen eine Strategie
- **Zweck und Anwendung:** Entscheidungen über mehrere Schritte verbessern, bspw. bei Spielen oder der Steuerung eines Roboters
- **Input und Output:** Der Agent beobachtet die Umgebung, wählt Aktionen und erhält Rückmeldungen, daraus lernt er eine Handlungsstrategie
- **Interpretation:** Eine gute Strategie erzielt langfristig eine hohe kumulierte Belohnung, die unmittelbar grösste Belohnung muss dafür nicht die beste Entscheidung sein
- **Einordnung:** Gehört zu den drei Lernparadigmen, wird gemäss Folien im Kurs nicht weiter behandelt

### Prüfungsfrage einordnen – Welche Lernaufgabe liegt vor?

- **Bekannte Zielvariable und gesuchter Zahlenwert** → Supervised Learning, Regression
- **Bekannte Zielvariable und gesuchte Klasse** → Supervised Learning, Klassifikation
- **Keine Zielvariable, gesucht sind Gruppen oder Strukturen** → Unsupervised Learning
- **Aktionen, Umgebung und Belohnungen** → Reinforcement Learning
- **Begründung:** Immer auf die konkrete Aufgabe beziehen, bspw. «Regression, weil der Verkaufspreis in CHF vorhergesagt wird» oder «Klassifikation, weil Servicebedarf als Ja-/Nein-Klasse vorhergesagt wird»

# Model training & evaluation

## Loss Functions – Wie stark wird eine falsche Vorhersage bestraft?

- **Definition:** Eine Loss Function (Verlustfunktion) bewertet den Fehler einer einzelnen Vorhersage, ein kleiner Loss bedeutet eine gute Vorhersage gemäss der gewählten Funktion
- **Zweck und Anwendung:** Beim Modelltraining festlegen, welche Fehler wie stark zählen, das Modell soll sein Trainingsziel möglichst klein machen
- **Input:** Tatsächlicher Wert `y` und Vorhersage `ŷ`, bei bestimmten Klassifikationsverfahren stattdessen eine vorhergesagte Wahrscheinlichkeit oder ein Score
- **Fehler bei Regression:** `e = y − ŷ`, ein positiver Fehler bedeutet unterschätzt, ein negativer Fehler bedeutet überschätzt
- **Squared Loss – Quadratischer Verlust:** `L = (y − ŷ)²`, grosse Abweichungen werden besonders stark bestraft, bspw. Fehler 2 → Loss 4 und Fehler 10 → Loss 100
- **Absolute Loss – Absoluter Verlust:** `L = |y − ŷ|`, Fehler zählen entsprechend ihrer absoluten Grösse, bspw. Fehler −3 → Loss 3, weniger empfindlich gegenüber Extremfehlern als Squared Loss
- **0–1 Loss – Richtig oder falsch:** Richtige Klasse → Loss 0, falsche Klasse → Loss 1, berücksichtigt weder die Sicherheit der Vorhersage noch unterschiedliche Fehlerkosten
- **Log-Loss – Wahrscheinlichkeiten bewerten:** Bestraft besonders stark, wenn das Modell der tatsächlichen Klasse eine sehr kleine Wahrscheinlichkeit zuordnet, bspw. bei tatsächlich Klasse 1 ist `p = 0.9` besser als `p = 0.1`, wird bei logistischer Regression verwendet
- **Hinge Loss – Klassifikation mit Sicherheitsabstand:** Bestraft falsche Vorhersagen und auch richtige Vorhersagen mit zu geringem Abstand zur Entscheidungsgrenze, wird bei SVM verwendet
- **Interpretation:** Kleinere Loss-Werte sind besser, ein Modell mit kleinem Trainings-Loss kann trotzdem bei neuen Beobachtungen schlecht abschneiden
- **Stolperstein:** Die Zahlenwerte verschiedener Loss Functions sind nicht direkt vergleichbar, sie bewerten Fehler unterschiedlich und können unterschiedliche Einheiten haben

## Loss, Trainingsziel und Bewertungskennzahl – Drei unterschiedliche Rollen
- **Loss:** Bewertet eine einzelne Vorhersage, bspw. ein quadrierter Fehler für ein Auto
- **Trainingsziel (Objective):** Grösse, die der Lernalgorithmus optimiert, häufig Summe oder Mittelwert der Loss-Werte, ggf. ergänzt um eine Strafe für Modellkomplexität
- **Bewertungskennzahl (Evaluation Metric):** Fasst zusammen, wie gut das trainierte Modell auf ausgewählten Daten abschneidet, bspw. RMSE, R², Accuracy, F1 oder AUC
- **Wann anwenden:** Beim Lesen einer Aufgabenstellung oder eines Modell-Outputs unterscheiden, wie das Modell lernt und wie seine Leistung anschliessend beurteilt wird
- **Bsp. lineare Regression:** OLS minimiert beim Training die Summe der quadrierten Fehler, anschliessend beurteilen wir die Testvorhersagen bspw. mit RMSE und R²
- **Bsp. logistische Regression:** Das Training basiert auf Log-Loss, die Bewertung kann dagegen bspw. mit Recall oder AUC erfolgen
- **Bsp. Entscheidungsbaum:** Wählt schrittweise Splits, welche SSE bei Regression bzw. Gini oder Entropie bei Klassifikation möglichst stark reduzieren, die abschliessende Bewertung erfolgt mit passenden Regressions- oder Klassifikationskennzahlen
- **Zusammenhang mit MSE:** Squared Loss bewertet einen einzelnen quadrierten Fehler, MSE ist der Mittelwert dieser quadrierten Fehler über mehrere Beobachtungen
- **Interpretation:** Das optimierte Trainingsziel und die ausgegebene Bewertungskennzahl müssen nicht identisch sein, auch die Richtung unterscheidet sich, bspw. kleiner RMSE ist besser, grösseres R² ist besser
- **Generalisierung:** Gesucht ist eine gute Leistung auf neuen, ungesehenen Daten, diese schätzen wir mit zurückgehaltenen Testdaten oder Cross-Validation
- **Stolperstein:** Trainingswerte sind häufig zu optimistisch, weil das Modell dieselben Beobachtungen bereits zum Lernen verwendet hat

## MSE und RMSE – Fehler numerischer Vorhersagen beurteilen

- **Definition:** MSE und RMSE messen, wie stark numerische Vorhersagen von den tatsächlichen Werten abweichen, bspw. beim vorhergesagten Treibstoffverbrauch oder Mietpreis
- **Zweck und Anwendung:** Die Vorhersagequalität einer Regression beurteilen und Modelle auf denselben Beobachtungen miteinander vergleichen
- **Input und Voraussetzungen:** Tatsächliche Werte und Vorhersagen müssen numerisch, vollständig, gleich lang und in derselben Reihenfolge angeordnet sein, sodass jedes Wertepaar dieselbe Beobachtung beschreibt
- **Fehler:** Tatsächlicher Wert minus Vorhersage, also `observed - predicted` → positiv bedeutet unterschätzt, negativ bedeutet überschätzt
- **MSE (Mean Squared Error):** Jeden Fehler quadrieren und anschliessend den Mittelwert bilden → `np.mean(errors ** 2)` verhindert, dass sich positive und negative Fehler aufheben, und gewichtet grosse Abweichungen stärker
- **RMSE (Root Mean Squared Error):** Quadratwurzel des MSE → `np.sqrt(mse)` bringt den Fehler zurück in die Einheit der Zielvariable
- **Bsp. aus der Aufgabe:** Tatsächliche Verbrauchswerte `[6, 10]` und Vorhersagen `[5, 7]` ergeben Fehler `[1, 3]` → MSE = `(1² + 3²) / 2 = 5`, RMSE = `√5 ≈ 2.24`
- **Interpretation:** Der RMSE beträgt rund `2.24 L/100 km` und beschreibt die quadratisch gemittelte Fehlergrösse, nicht die Abweichung jeder einzelnen Vorhersage und auch nicht den mittleren absoluten Fehler
- **Wertebereich und Einheit:** Beide Kennzahlen liegen zwischen 0 und unbegrenzt, kleinere Werte sind besser, 0 bedeutet ausschliesslich exakte Vorhersagen → MSE hier in `(L/100 km)²`, RMSE in `L/100 km`
- **Stolpersteine:** MSE und RMSE zeigen keine Über- oder Unterschätzungsrichtung und reagieren empfindlich auf einzelne grosse Fehler, ob ein RMSE gut ist, hängt vom Anwendungsfall und einer Vergleichsbasis ab
- **Prüfung:** Für die Leistung auf unbekannten Daten tatsächliche Testwerte mit den zugehörigen Testvorhersagen vergleichen, ein kleiner Trainingsfehler allein belegt keine gute Vorhersagequalität

```python
#Verwenden, wenn tatsächliche Zahlenwerte und zugehörige Vorhersagen vorliegen und ihre Fehlergrösse beurteilt werden soll
import numpy as np                                                              #Einmaliger Import für numerische Berechnungen

observed = np.array([6, 10])                                                    #Tatsächliche Verbrauchswerte aus der Aufgabe in L/100 km
predicted = np.array([5, 7])                                                    #Vorhersagen derselben Autos in derselben Reihenfolge

errors = observed - predicted                                                   #Fehler je Auto berechnen, positiv = unterschätzt
mse = np.mean(errors ** 2)                                                      #Fehler quadrieren und mitteln, grosse Fehler zählen stärker
rmse = np.sqrt(mse)                                                             #Wurzel ziehen, damit die Einheit wieder L/100 km ist

print('Einzelne Fehler (L/100 km):', errors)                                    #Richtung und Grösse der einzelnen Abweichungen anzeigen
print(f'MSE: {mse:.2f} (L/100 km)²')                                            #Mittleren quadratischen Fehler anzeigen
print(f'RMSE: {rmse:.2f} L/100 km')                                             #Fehlergrösse in der ursprünglichen Verbrauchseinheit anzeigen

#ANPASSEN: observed durch tatsächliche Werte ersetzen, bei einer Testbewertung bspw. y_test
#ANPASSEN: predicted durch die zugehörigen Vorhersagen ersetzen, gleiche Reihenfolge und Länge sicherstellen
#ANPASSEN: L/100 km durch die Einheit der Zielvariable ersetzen, beim MSE die quadrierte Einheit verwenden
```

## Train/Test-Split – Daten zum Lernen und zur abschliessenden Bewertung aufteilen

- **Definition:** Teilt vorhandene Daten in Trainingsdaten zum Lernen und zurückgehaltene Testdaten zur abschliessenden Bewertung auf
- **Zweck und Anwendung:** Vor dem Modelltraining verwenden, um später zu prüfen, wie gut das Modell bei bisher ungesehenen Beobachtungen vorhersagt
- **Input und Voraussetzungen:** Features `X` und zugehörige Zielwerte `y`, jede Zeile in `X` muss zur gleichen Beobachtung in `y` gehören, fehlende Werte in den benötigten Spalten werden hier gemäss Aufgabe vorab entfernt
- **Features `X`:** Eingaben als zweidimensionaler DataFrame, hier `engine_size` und `cylinders`, doppelte eckige Klammern wählen mehrere Spalten aus
- **Zielvariable `y`:** Gesuchte Antwort als eindimensionale Series, hier `combined`, der kombinierte Stadt-/Autobahnverbrauch in L/100 km
- **`train_test_split(X, y, ...)`:** Teilt Features und Zielwerte gemeinsam auf, damit ihre Zuordnung erhalten bleibt, standardmässig werden die Zeilen zuvor zufällig gemischt
- **`test_size=0.2`:** Reserviert 20 % für den Test, die übrigen 80 % dienen dem Training, der Parameter bezeichnet den Testanteil
- **`random_state=42`:** Macht die zufällige Aufteilung bei unveränderten Daten reproduzierbar, 42 ist ein frei gewählter Startwert und keine Qualitätseinstellung
- **Output:** `X_train` enthält die Trainingsmerkmale und `y_train` deren Zielwerte, `X_test` enthält die Testmerkmale und `y_test` deren tatsächliche Zielwerte
- **Interpretation im Autodatensatz:** 701 vollständige Autos ergeben 560 Trainingsautos und 141 Testautos, die Aufteilung selbst trainiert und bewertet noch kein Modell
- **Weitere Verwendung:** Mit `fit(X_train, y_train)` lernen, mit `predict(X_test)` vorhersagen und die Vorhersagen anschliessend mit `y_test` vergleichen
- **Validierung:** Modelle und Einstellungen innerhalb der Trainingsdaten vergleichen, bspw. mit Cross-Validation, die Testdaten für die abschliessende Bewertung zurückhalten
- **Stolpersteine:** Die Zielvariable gehört nicht zusätzlich in die Features, aus Daten gelernte Vorverarbeitung wie Skalierung nur auf Trainingsdaten anpassen, bei Zeitreihen oder zusammengehörenden Beobachtungen ist eine zufällige Zeilenaufteilung ggf. ungeeignet

```python
#Verwenden, wenn ein numerischer Zielwert vorhergesagt werden soll und Trainings- sowie Testdaten vorbereitet werden müssen
import pandas as pd                                                             #Einmaliger Import zum Einlesen und Bearbeiten der Daten
from sklearn.model_selection import train_test_split                            #Funktion zur gemeinsamen Aufteilung von Features und Zielwerten

df_ap03spick_cars = pd.read_csv(
    'car_fuel_consumption_2025.csv'
)                                                                               #Autodaten einmal laden, Pfad relativ zum Arbeitsverzeichnis des Kernels

cars_features = ['engine_size', 'cylinders']                                    #Eingaben festlegen, Hubraum und Anzahl Zylinder
cars_target = 'combined'                                                        #Zielvariable festlegen, Verbrauch in L/100 km

df_ap03spick_cars_model = df_ap03spick_cars.dropna(
    subset=cars_features + [cars_target]
).copy()                                                                        #Nur Zeilen mit vollständigen Modellvariablen übernehmen, Ausgangsdaten erhalten

X_cars = df_ap03spick_cars_model[cars_features]                                 #Features als zweidimensionalen DataFrame auswählen
y_cars = df_ap03spick_cars_model[cars_target]                                   #Zielwerte als eindimensionale Series auswählen

X_train_cars, X_test_cars, y_train_cars, y_test_cars = train_test_split(
    X_cars, y_cars,                                                             #Features und zugehörige Zielwerte gemeinsam aufteilen
    test_size=0.2,                                                              #20 % Testdaten, die übrigen 80 % Trainingsdaten
    random_state=42                                                             #Zufallsaufteilung reproduzierbar machen
)

display(df_ap03spick_cars_model[cars_features + [cars_target]].head())          #Eingaben und Zielvariable nebeneinander kontrollieren
print('Vollständige Beobachtungen:', len(df_ap03spick_cars_model))
print('Trainingsdaten:', len(X_train_cars))                                     #Erwartet bei diesen Daten 560 Autos
print('Testdaten:', len(X_test_cars))                                           #Erwartet bei diesen Daten 141 Autos

#ANPASSEN: Dateipfad und DataFrame-Namen durch den Prüfungsdatensatz ersetzen, vorhandene Daten nicht erneut laden
#ANPASSEN: cars_features und cars_target durch die verlangten Eingabespalten und die Zielvariable ersetzen
#ANPASSEN: test_size und random_state gemäss Aufgabenstellung setzen, vorbereitete Splits direkt weiterverwenden


#Verwenden, um Daten für eine Regression in Trainingsdaten und Testdaten aufzuteilen
import pandas as pd
from sklearn.model_selection import train_test_split

#1. Daten laden
df_ap03spick_cars = pd.read_csv('car_fuel_consumption_2025.csv')                #CSV-Datei als DataFrame laden
                                                                                #ACHTUNG: Ist der DataFrame bereits gegeben, diese Ladezeile weglassen und den vorhandenen Namen verwenden

#2. Eingaben und Ziel festlegen
feature_cols = ['engine_size', 'cylinders']                                     #Mit diesen Spalten soll das Modell vorhersagen
target_col = 'combined'                                                         #Diese Spalte soll das Modell vorhersagen
#ANPASSEN: In feature_cols die verlangten Eingabespalten eintragen
#ANPASSEN: In target_col die vorherzusagende Spalte eintragen
#ACHTUNG: Die Zielspalte darf nicht zusätzlich in feature_cols stehen


#3. Zeilen mit fehlenden Angaben ausschliessen
df_model = df_ap03spick_cars.dropna(
    subset=feature_cols + [target_col]                                          #Nur die für das Modell benötigten Spalten prüfen
).copy()                                                                        #Separate Tabelle erstellen, ursprünglichen DataFrame erhalten
#ANPASSEN: Diese Bereinigung nur verwenden, wenn sie zur Aufgabe passt


#4. Eingaben X und Zielwerte y auswählen
X_cars = df_model[feature_cols]                                                 #X enthält die Eingabespalten, hier Hubraum und Zylinderzahl
y_cars = df_model[target_col]                                                   #y enthält die richtige Antwort, hier den tatsächlichen Verbrauch


#5. Daten gemeinsam aufteilen
X_train_cars, X_test_cars, y_train_cars, y_test_cars = train_test_split(
    X_cars, y_cars,                                                             #Eingaben und Antworten gemeinsam aufteilen, Zuordnung bleibt erhalten
    test_size=0.2,                                                              #20 % zum Testen zurückhalten, 80 % zum Lernen verwenden
    random_state=42                                                             #Bei jeder Ausführung dieselbe zufällige Aufteilung erhalten
)
#ANPASSEN: test_size gemäss Aufgabe setzen, bspw. 0.3 für 30 % Testdaten
#ANPASSEN: random_state gemäss Aufgabe setzen, falls vorgegeben
#ANPASSEN: Falls die Prüfung X_train, X_test, y_train und y_test verlangt, die vier Ausgaben entsprechend umbenennen


#6. Auswahl und Anzahl kontrollieren
display(df_model[feature_cols + [target_col]].head())                           #Eingaben und richtige Antwort nebeneinander anzeigen
print('Verwendete Beobachtungen:', len(df_model))
print('Trainingsdaten:', len(X_train_cars))                                     #Mit diesen Beobachtungen lernt das Modell später
print('Testdaten:', len(X_test_cars))                                           #Diese Beobachtungen werden erst zur abschliessenden Bewertung verwendet

#VORAUSSETZUNG: Eingabespalten sind hier bereits numerisch, Textkategorien müssten passend umgewandelt werden
#ACHTUNG: Sind X_train, X_test, y_train und y_test bereits vorbereitet, Schritte 1–5 überspringen und direkt mit diesen Variablen das Modell trainieren
```

## Cross-Validation (CV) – Ein Modell innerhalb der Trainingsdaten prüfen

- **Definition:** Cross-Validation teilt die Trainingsdaten mehrfach in einen Teil zum Lernen und einen Teil zum Validieren auf, so wird das Modell auf Beobachtungen geprüft, die es im jeweiligen Durchlauf nicht zum Lernen verwendet hat
- **Zweck und Anwendung:** Die Vorhersagequalität abschätzen sowie verschiedene Modelle oder Einstellungen vergleichen, ohne dafür die abschliessenden Testdaten zu verwenden
- **Input und Voraussetzungen:** Ein Modell sowie zusammengehörende Trainingsmerkmale und Trainingszielwerte, hier `X_train_cars` und `y_train_cars` aus dem vorherigen Eintrag
- **5-fold CV:** Die Trainingsdaten werden in fünf ungefähr gleich grosse Teile (Folds) aufgeteilt, pro Durchlauf lernt ein neues Modell auf vier Teilen und wird auf dem fünften geprüft, jeder Teil dient einmal zur Validierung
- **`cross_val_score()`:** Übernimmt das wiederholte Trainieren und Bewerten automatisch, ein vorheriger Aufruf von `fit()` ist nicht erforderlich
- **`cv=5`:** Führt fünf Durchläufe durch und liefert fünf Bewertungswerte, bei dieser Regression werden die Folds standardmässig ohne zusätzliches Mischen gebildet
- **`scoring='neg_mean_squared_error'`:** Liefert den negativen MSE, weil scikit-learn bei Scores grundsätzlich grössere Werte als besser behandelt
- **`np.sqrt(-scores)`:** Entfernt zuerst das Minuszeichen und zieht danach für jeden Fold die Wurzel → fünf RMSE-Werte in der Einheit der Zielvariable
- **`.mean()`:** Berechnet den Durchschnitt der fünf Fold-RMSE, genau wie in der Dozentenlösung
- **Output und Interpretation:** Die einzelnen RMSE zeigen, wie gut das Modell in den verschiedenen Validierungsteilen abschneidet, der mittlere CV-RMSE fasst diese Ergebnisse zusammen, kleinere Werte sind besser
- **Bsp. zur Interpretation:** Ein angenommener mittlerer CV-RMSE von `1.7 L/100 km` beschreibt die Fehlergrösse auf zurückgehaltenen Validierungsautos, nicht einen Fehler von genau 1.7 bei jedem Auto
- **Unterschiede zwischen Folds:** Stark unterschiedliche RMSE zeigen, dass das Ergebnis von den verwendeten Lern- und Validierungsbeobachtungen abhängt
- **Stolpersteine:** Keine Testdaten in diese CV übergeben, eine nötige Skalierung oder andere gelernte Vorverarbeitung muss innerhalb jedes Folds gelernt werden, dafür verwendet man eine Pipeline
- **Weiteres Vorgehen:** Nach der Modellauswahl das gewählte Modell auf allen Trainingsdaten trainieren und abschliessend auf den Testdaten bewerten, `cross_val_score()` liefert die Scores und kein fertig trainiertes Modell zur weiteren Verwendung zurück

```python
#Verwenden, um die Qualität eines Regressionsmodells mit Cross-Validation auf den Trainingsdaten zu prüfen
from sklearn.linear_model import LinearRegression                               #Lineares Regressionsmodell
from sklearn.model_selection import cross_val_score                             #Modell mehrfach trainieren und auf zurückgehaltenen Teilen prüfen
#VORAUSSETZUNG: NumPy ist als np importiert, X_train_cars und y_train_cars sind bereits vorbereitet


#1. Modell festlegen
cv_model_cars = LinearRegression()                                              #Hier soll eine lineare Regression geprüft werden
#ANPASSEN: LinearRegression() durch das verlangte Regressionsmodell inkl. Einstellungen ersetzen


#2. Cross-Validation durchführen
cv_scores_cars = cross_val_score(
    cv_model_cars,                                                              #Dieses Modell wird in jedem Durchlauf neu trainiert
    X_train_cars,                                                               #Nur Eingaben der Trainingsdaten verwenden
    y_train_cars,                                                               #Zugehörige richtige Antworten der Trainingsdaten verwenden
    cv=5,                                                                       #Fünf Durchläufe, jeweils vier Teile lernen und einen Teil prüfen
    scoring='neg_mean_squared_error'                                            #Pro Durchlauf den negativen MSE zurückgeben
)
#ANPASSEN: X_train_cars und y_train_cars durch die Trainingsvariablen aus der Prüfung ersetzen
#ANPASSEN: cv gemäss Aufgabe setzen, scoring und die Umrechnung unten müssen zusammenpassen


#3. Negative MSE in RMSE umrechnen
cv_rmse_cars = np.sqrt(-cv_scores_cars)                                         #Minuszeichen entfernen, danach Wurzel ziehen
mean_cv_rmse_cars = cv_rmse_cars.mean()                                         #Durchschnitt der fünf einzelnen RMSE berechnen


#4. Ergebnisse anzeigen
print('RMSE je Fold:', cv_rmse_cars.round(2))                                   #Zeigt die Fehlergrösse in jedem Validierungsteil
print(f'Mittlerer CV-RMSE: {mean_cv_rmse_cars:.2f} L/100 km')
#ANPASSEN: L/100 km durch die Einheit der Zielvariable ersetzen

#ACHTUNG: Bei einer anderen Bewertungskennzahl die RMSE-Umrechnung in Schritt 3 nicht ungeprüft übernehmen
#ACHTUNG: Dieser Block benötigt kein vorheriges fit(), die CV übernimmt das Training selbst
```

## Underfitting und Overfitting – Zu wenig oder zu viel aus Trainingsdaten lernen

- **Definition:** Underfitting bedeutet, dass ein Modell wichtige Muster nicht ausreichend erfasst, Overfitting bedeutet, dass es sich zu stark an Besonderheiten und zufällige Schwankungen der Trainingsdaten anpasst
- **Zweck und Anwendung:** Trainings- und Validierungsergebnisse vergleichen, um zu beurteilen, ob die Modellkomplexität zur Aufgabe passt, bspw. bei der Wahl der maximalen Baumtiefe
- **Input und Voraussetzungen:** Vergleichbare Kennzahlen auf Trainingsdaten und zurückgehaltenen Validierungsdaten, bspw. Trainings-RMSE und CV-RMSE, bei beiden gilt kleiner = besser
- **Underfitting – Modell zu einfach:** Bereits auf den Trainingsdaten entstehen grosse Fehler, auch der Validierungsfehler ist hoch, bspw. ein Baum mit nur einer Aufteilung, der wichtige Unterschiede zwischen Autos nicht erfasst
- **Passendes Modell:** Erfasst wesentliche Muster und liefert auch auf Validierungsdaten brauchbare Vorhersagen, Trainings- und Validierungsfehler sind im Anwendungskontext ausreichend klein
- **Overfitting – Zu starke Anpassung:** Sehr kleiner Trainingsfehler, aber deutlich grösserer Validierungsfehler, bspw. ein sehr tiefer Baum, der einzelne Trainingsbeobachtungen nahezu auswendig lernt
- **Interpretation beim Erhöhen der Komplexität:** Sinkt der Trainingsfehler weiter, während der CV-Fehler wieder steigt, spricht das für zunehmendes Overfitting
- **Interpretation bei sinkendem CV-Fehler:** Sinkt neben dem Trainingsfehler auch der CV-Fehler, verbessert die zusätzliche Komplexität hier die Validierungsleistung, ein komplexeres Modell ist nicht automatisch overfitted
- **Interpretation ähnlicher Fehler:** Ähnliche Trainings- und CV-Fehler allein bedeuten noch kein gutes Modell, beide können hoch sein und auf Underfitting hindeuten
- **Stolperstein:** Ein gewisser Abstand zwischen Trainings- und CV-Fehler ist üblich, es gibt keinen allgemein gültigen Grenzwert, ab dem Overfitting sicher vorliegt
- **Bei Underfitting:** Ggf. ein flexibleres Modell, mehr erlaubte Baumtiefe oder geeignetere Features verwenden
- **Bei Overfitting:** Ggf. die Komplexität begrenzen, einen Baum beschneiden (Pruning), Regularisierung einsetzen oder zusätzliche geeignete Trainingsdaten nutzen
- **Prüfung:** Modell und Einstellungen anhand der Validierung bzw. CV auswählen, der kleinste Trainingsfehler allein ist kein Auswahlkriterium, die abschliessenden Testdaten nicht wiederholt zur Optimierung verwenden

## Bias–Variance Trade-off – Modellkomplexität und Vorhersagefehler ausbalancieren

- **Definition:** Der Bias–Variance Trade-off beschreibt den Ausgleich zwischen systematischen Fehlern eines Modells und seiner Empfindlichkeit gegenüber unterschiedlichen Trainingsstichproben
- **Zweck und Anwendung:** Verstehen, weshalb ein einfacheres oder komplexeres Modell nicht automatisch besser ist, und die Modellkomplexität mithilfe von Validierung bzw. CV auswählen
- **Bias – Systematischer Fehler:** Entsteht durch zu einfache oder unpassende Modellannahmen, das Modell erfasst den tatsächlichen Zusammenhang nicht ausreichend → hoher Bias kann zu Underfitting führen
- **Bsp. für hohen Bias:** Für jede Wohnung denselben durchschnittlichen Mietpreis vorhersagen, obwohl sich die Preise nach Wohnfläche und Lage unterscheiden
- **Variance – Schwankung des gelernten Modells:** Beschreibt, wie stark sich die Vorhersagen verändern würden, wenn das Modell mit einer anderen Trainingsstichprobe trainiert würde → hohe Variance kann zu Overfitting führen
- **Bsp. für hohe Variance:** Ein sehr tiefer Baum passt sich stark an einzelne Trainingswohnungen an, mit anderen Trainingswohnungen können deutlich andere Regeln und Vorhersagen entstehen
- **Noise – Unvermeidbares Rauschen:** Zufällige Schwankungen oder Einflüsse, die mit den vorhandenen Features nicht vorhergesagt werden können, auch ein gutes Modell kann diesen Anteil nicht vollständig beseitigen
- **Zusammenhang bei quadratischem Fehler:** Der erwartete Vorhersagefehler lässt sich unter den entsprechenden Annahmen in `Bias² + Variance + Noise` zerlegen
- **Typische Tendenz:** Mehr Modellkomplexität kann den Bias senken, erhöht aber häufig die Variance, weniger Komplexität kann die Vorhersagen stabilisieren, aber wichtige Muster übersehen
- **Interpretation:** Gesucht ist eine möglichst kleine Summe der Fehleranteile, weder möglichst wenig Bias noch möglichst wenig Variance allein
- **Stolperstein:** Variance meint hier die Schwankung der Modellvorhersagen zwischen verschiedenen Trainingsstichproben, nicht die Streuung der Zielwerte innerhalb eines Datensatzes
- **Bias reduzieren:** Ggf. geeignetere Features oder ein flexibleres Modell verwenden
- **Variance reduzieren:** Ggf. mehr geeignete Trainingsdaten, ein einfacheres Modell, Regularisierung oder die Mittelung mehrerer Modelle verwenden, bspw. Bagging
- **Prüfung:** Bias und Variance werden durch einen einzelnen Trainings-/Testvergleich nicht direkt berechnet, Trainings- und CV-Ergebnisse liefern Hinweise auf Underfitting und Overfitting, Einstellungen anhand der CV-Leistung vergleichen
