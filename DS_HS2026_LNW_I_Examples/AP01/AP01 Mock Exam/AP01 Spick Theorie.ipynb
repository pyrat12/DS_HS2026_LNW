## Data Science: Definition und Zweck
Folien 10 bis 13

- Definition: Data Science verbindet Programmierung, Mathematik und Statistik sowie Fachwissen, um aus Daten Erkenntnisse und Handlungsmöglichkeiten abzuleiten
- Zweck: Daten verstehen, Muster erkennen, Vorhersagen erstellen und Entscheidungen unterstützen
- Ablauf: Rohdaten → Erkenntnisse → Entscheidungen mit praktischem Nutzen
- Bedeutung: Mehr Daten, günstige Rechenleistung und verfügbare Werkzeuge ermöglichen umfangreichere datenbasierte Analysen
- Datenquellen: Bspw. Sensoren, Webseiten, Transaktionen und kontinuierliche Datenströme
- Erkenntnis: Daten sammeln oder ein Modell ausführen ist allein noch keine vollständige Data-Science-Arbeit

## Die drei Kompetenzen der Data Science
Folien 11 bis 13

| Kompetenz | Bedeutung | Bsp. |
|---|---|---|
| Hacking Skills | Daten technisch beschaffen, bearbeiten und verarbeiten | Python, SQL, Dateien, APIs und algorithmisches Denken |
| Math & Statistics Knowledge | Geeignete Methoden auswählen, anwenden und Ergebnisse korrekt interpretieren | Regression, statistische Kennzahlen und Modellbewertung |
| Substantive Expertise | Fachgebiet und konkrete Problemstellung verstehen | Geschäftsprozesse, Kundenverhalten und fachliche Anforderungen |

- Hacking Skills bedeutet hier praktische technische Fähigkeiten, nicht unerlaubtes Eindringen in Systeme
- Fachwissen hilft, sinnvolle Fragen und prüfbare Hypothesen zu formulieren
- Statistisches Wissen hilft, Ergebnisse fachlich richtig einzuordnen
- Erkenntnis: Erst die Verbindung aller drei Kompetenzen ergibt im dargestellten Modell Data Science

### Überschneidungen und Danger Zone

| Kombination | Einordnung | Grenze |
|---|---|---|
| Technik und Statistik | Machine Learning | Fachliche Fragestellung und Einordnung können fehlen |
| Statistik und Fachwissen | Traditionelle Forschung | Technische Verarbeitung grosser Datenmengen kann schwierig sein |
| Technik und Fachwissen | Danger Zone | Methoden werden ausgeführt, aber möglicherweise falsch interpretiert |
| Alle drei Kompetenzen | Data Science | Technische, methodische und fachliche Perspektive werden verbunden |

- Bsp. Danger Zone: Jemand führt eine Regression aus und präsentiert die Koeffizienten, versteht deren Bedeutung aber nicht
- Erkenntnis: Technisch lauffähige Analysen können trotzdem fachlich falsch sein

## Data Analytics Process Model
Folien 14 bis 16

- Definition: Prozessmodell, das ein Analyseprojekt von der Geschäftsfrage bis zur Nutzung der Ergebnisse strukturiert
- Zweck: Sicherstellen, dass Datenbeschaffung, Aufbereitung, Analyse und Interpretation zur ursprünglichen Fragestellung passen

| Schritt | Aufgabe | Bsp. |
|---|---|---|
| Identify business problem | Geschäftsproblem und Ziel klären | Welche Faktoren hängen mit hohen Ticketumsätzen zusammen? |
| Identify data sources | Geeignete Datenquellen finden | Tickets, Vorstellungen und Kinos |
| Select the data | Benötigte Datensätze und Merkmale auswählen | Verkaufsdatum, Preis und Kino |
| Clean the data | Fehler, fehlende Werte und Duplikate behandeln | Uneinheitliche Kategorien vereinheitlichen |
| Transform the data | Daten passend umformen und kombinieren | Tabellen verbinden und Jahreswerte ableiten |
| Analyze the data, create models | Kennzahlen, Zusammenhänge oder Modelle bestimmen | Umsatz pro Kino berechnen |
| Interpret, evaluate, deploy | Ergebnisse erklären, prüfen und nutzbar machen | Bericht oder Dashboard bereitstellen |

- Iterativ: Neue Erkenntnisse können zu früheren Schritten zurückführen
- Datenvorverarbeitung: Auswahl, Bereinigung und Transformation beanspruchen häufig einen grossen Teil des Aufwands
- Garbage in, garbage out: Schlechte oder ungeeignete Daten können auch mit guten Methoden schlechte Ergebnisse liefern
- CRISP-DM: Verwandtes verbreitetes Prozessmodell für Data-Mining-Projekte
- Erkenntnis: Eine Analyse beginnt mit einer sinnvollen Frage, nicht mit einem beliebigen Modell

|---|---|
| Datenquellen verwalten | Relationale Daten, SQL, NoSQL und Echtzeitdaten |

| Daten analysieren | Statistik und Machine Learning |
| Ergebnisse bereitstellen | Berichte, Dashboards, Webanwendungen und Entscheidungsunterstützung |

## Data Management: Daten verwalten
Folien 18 bis 19

- Definition: Datenmanagement umfasst Regeln, Verantwortlichkeiten und Tätigkeiten zum Umgang mit Daten
- Zweck: Daten nutzbar, geschützt, nachvollziehbar und angemessen verfügbar halten
- Aufgaben: Daten erfassen, organisieren, speichern, integrieren, schützen, bereitstellen, archivieren und bei Bedarf löschen
- Erkenntnis: Datenmanagement ist mehr als Daten speichern, auch Organisation, Qualität und Verantwortlichkeiten gehören dazu

## Data Governance und DAMA-DMBOK
Folie 19

- Data Governance: Legt fest, wer für Daten verantwortlich ist, welche Regeln gelten und wie deren Einhaltung sichergestellt wird
- Zweck: Einheitliche Entscheidungen und abgestimmte Datenmanagement-Aktivitäten ermöglichen
- Bsp. Regeln: Wer darf Kundendaten ändern, wer definiert gültige Mitgliedschaftstypen und wer behebt Qualitätsprobleme?
- DAMA-DMBOK: Strukturiertes Rahmenwerk für die verschiedenen Aufgabenbereiche des Datenmanagements
- Erkenntnis: Governance steht im Zentrum, weil die übrigen Datenbereiche miteinander abgestimmt werden müssen

| Bereich im Framework | Aufgabe |
|---|---|
| Data Governance | Regeln, Verantwortung und Steuerung |
| Data Architecture | Gesamtstruktur der Datenlandschaft |
| Data Modeling and Design | Datenstrukturen und Beziehungen entwerfen |
| Data Storage and Operations | Daten speichern und Systeme betreiben |
| Data Security | Daten schützen |
| Data Integration and Interoperability | Daten und Systeme zusammenführen |
| Documents and Content | Dokumente und Inhalte verwalten |
| Reference and Master Data | Referenzdaten und Stammdaten verwalten |
| Data Warehousing and Business Intelligence | Integrierte Daten für Analysen bereitstellen |
| Metadata | Informationen über Daten verwalten |
| Data Quality | Datenqualität sichern und verbessern |

### Governance und Management unterscheiden

- Governance: Legt Verantwortung, Regeln und Entscheidungsbefugnisse fest
- Management: Setzt die Regeln im täglichen Umgang mit Daten um
- Bsp. Governance definiert gültige Kundenkategorien, Datenmanagement kontrolliert und bereinigt die gespeicherten Werte

- Zweck: Eine verständliche und abgestimmte Datenlandschaft schaffen

## Data Modeling: Daten und Beziehungen beschreiben
Folien 21 bis 24

- Definition: Datenanforderungen untersuchen und als strukturiertes Modell darstellen
- Zweck: Fachliche Anforderungen verständlich machen und eine Grundlage für die Datenbankgestaltung schaffen
- Iterativ: Modelle werden mit zunehmendem Verständnis verfeinert
- Ebenen: Konzeptionelles, logisches und physisches Datenmodell

| Ebene | Leitfrage | Inhalt | Bsp. |
|---|---|---|---|
| Konzeptionell | Welche fachlichen Objekte hängen zusammen? | Grobe Entitäten und Beziehungen, technologieunabhängig | Kunde bestellt Produkt |
| Logisch | Welche Informationen und Beziehungen werden genau benötigt? | Entitäten, Attribute und detaillierte Beziehungen | Kunde mit Kunden-ID, Name und Bestellungen |
| Physisch | Wie wird das Modell in der Datenbank umgesetzt? | Tabellen, Spalten, Datentypen, Schlüssel, Constraints und Indizes | customers mit customerid als Primärschlüssel |

- Konzeptionell: Höchster Abstraktionsgrad, keine konkrete Datenbankimplementierung
- Logisch: Detaillierte Struktur der benötigten Informationen
- Physisch: Konkrete Umsetzung im gewählten Datenbanksystem
- Erkenntnis: Dieselbe fachliche Anforderung wird auf drei unterschiedlichen Detailebenen beschrieben

## Entitäten, Attribute und Beziehungen
Folie 25

| Begriff | Bedeutung | Bsp. |
|---|---|---|
| Entität | Fachlich relevantes Objekt oder Konzept | Kunde, Fahrrad oder Verkauf |
| Attribut | Eigenschaft einer Entität | Nachname, Marke oder Verkaufspreis |
| Beziehung | Verbindung zwischen Entitäten | Kunde tätigt Verkauf |
| Datensatz | Konkrete gespeicherte Ausprägung | Ein bestimmter Kunde mit seiner Kunden-ID |

- In relationalen Datenbanken werden Entitätstypen häufig durch Tabellen dargestellt
- Attribute entsprechen dabei typischerweise Spalten
- Einzelne Datensätze stehen in Zeilen
- Erkenntnis: Eine Tabelle beschreibt eine Art von Objekt, eine Zeile ein konkretes Exemplar

## Schlüssel und Integritätsregeln
Folien 24 bis 28

| Begriff | Zweck | Wichtige Eigenschaft |
|---|---|---|
| Primary Key, Primärschlüssel | Datensatz eindeutig identifizieren | Eindeutig und nicht NULL |
| Foreign Key, Fremdschlüssel | Auf einen Datensatz einer anderen oder derselben Tabelle verweisen | Verbindet Tabellen und sichert zulässige Referenzen |
| UNIQUE | Doppelte Werte bzw. Kombinationen verhindern | Eindeutigkeitsregel |
| NOT NULL | Fehlenden Wert verhindern | Angabe ist erforderlich |
| Constraint | Regel für zulässige Daten | Sichert Datenintegrität |

- Bsp. sales.customerid verweist auf customers.customerid
- Ein Fremdschlüssel kann mehrfach vorkommen, bspw. dieselbe Kunden-ID in mehreren Verkäufen
- NULL ist bei einem Fremdschlüssel nur erlaubt, wenn keine entsprechende NOT-NULL-Regel besteht
- Referenzielle Integrität: Ein nicht fehlender Fremdschlüssel muss auf einen vorhandenen zulässigen Schlüsselwert verweisen
- Erkenntnis: Schlüssel ermöglichen Beziehungen, Constraints verhindern bestimmte ungültige Datenzustände

## Normalisierung
Folie 25

- Definition: Daten so auf Tabellen aufteilen und organisieren, dass unnötige Redundanz reduziert wird
- Zweck: Widersprüche und Probleme beim Einfügen, Ändern oder Löschen vermeiden
- Bsp. Verkäufername und Kontaktdaten in salespersons speichern, Verkäufe verweisen über salespersonid darauf
- Vorteil: Eine Änderung der Telefonnummer muss nicht in jedem einzelnen Verkauf durchgeführt werden
- Erkenntnis: Wiederholte Fremdschlüssel sind sinnvoll, unnötig wiederholte Stammdaten können dagegen Inkonsistenzen verursachen

## ER-Diagramme
Folien 26 bis 28

- Definition: Grafische Darstellung von Entitäten, Attributen und Beziehungen
- Zweck: Datenstrukturen verständlich darstellen und vor der Umsetzung prüfen
- Leseregel: Objekte erkennen, ihre Attribute und Schlüssel bestimmen, anschliessend Beziehungen und Kardinalitäten lesen
- Notationen: Bspw. Chen, IDEF1X, Bachman, Martin bzw. Crow's Foot, min-max und UML
- Erkenntnis: Die Darstellungszeichen können sich unterscheiden, die beschriebenen fachlichen Zusammenhänge bleiben vergleichbar

## Kardinalitäten
Folien 27 bis 28

- Definition: Kardinalität beschreibt, wie viele Datensätze auf den jeweiligen Seiten einer Beziehung zugeordnet sein können

| Kardinalität | Bedeutung | Bsp. |
|---|---|---|
| 1:1 | Ein Datensatz ist höchstens einem auf der anderen Seite zugeordnet | Person und zugehöriger einzelner Datensatz in einer Zusatztabelle |
| 1:n | Ein Datensatz kann mehreren auf der anderen Seite zugeordnet sein | Kunde und Bestellungen |
| n:1 | Mehrere Datensätze gehören zu einem auf der anderen Seite | Bestellungen und Kunde |
| m:n | Mehrere Datensätze können mit mehreren anderen verbunden sein | Bestellungen und Produkte |

- 1:n und n:1 beschreiben dieselbe Beziehung aus entgegengesetzten Richtungen
- m:n wird in relationalen Datenbanken typischerweise durch eine Zwischentabelle umgesetzt
- Bsp. Bestellposition verbindet Bestellung und Produkt und kann zusätzlich Menge oder Preis enthalten
- Achtung: Die genannten Kardinalitäten beschreiben nicht allein, ob eine Beziehung verpflichtend oder optional ist
- Erkenntnis: Kardinalitäten beeinflussen auch die Zeilenanzahl bei JOINs

## Star Schema und Snowflake Schema
Folie 29

- Verwendung: Datenmodellierung im Data Warehouse
- Faktentabelle: Enthält messbare Ereignisse bzw. Kennzahlen und Verweise auf Dimensionen
- Dimensionstabellen: Beschreiben den Auswertungskontext, bspw. Produkt, Zeit oder Region

| Merkmal | Star Schema | Snowflake Schema |
|---|---|---|
| Aufbau | Zentrale Faktentabelle direkt mit Dimensionstabellen verbunden | Dimensionen in weitere zusammenhängende Tabellen aufgeteilt |
| Normalisierung | Dimensionen häufig weniger normalisiert | Dimensionen stärker normalisiert |
| Redundanz | Kann höher sein | Wird reduziert |
| Abfragen | Häufig einfacher und mit weniger JOINs | Häufig komplexer und mit mehr JOINs |
| Bsp. | Produktdimension enthält Kategorie direkt | Produkt verweist auf separate Kategorietabelle |

- Erkenntnis: Star vereinfacht häufig die Analyse, Snowflake reduziert Wiederholungen in den Dimensionen
- Achtung: Ein Schema ist nicht grundsätzlich immer schneller oder besser, Anforderungen und Umsetzung entscheiden

## Data Integration und ETL
Folie 37

- Zweck: Informationen systemübergreifend für Analysen und Entscheidungen verfügbar machen

## SQL-Datentypen
Folie 38

- Definition: Ein Datentyp bestimmt, welche Art von Wert eine Spalte speichern kann
- Zweck: Zulässige Werte, mögliche Operationen und Speicherung festlegen
- Erkenntnis: Dieselbe Zeichenfolge kann als Text eine andere Bedeutung und andere Operationen haben als als Zahl oder Datum

| Datentyp bzw. Gruppe | Verwendung | Bsp. |
|---|---|---|
| INTEGER | Ganze Zahlen | Anzahl Tickets |
| NUMERIC | Exakte Dezimalzahlen | Preis oder Umsatz |
| REAL bzw. DOUBLE PRECISION | Gleitkommazahlen | Messwerte |

## JOINs: Tabellen verknüpfen
Folie 40

- Definition: Zeilen mehrerer Tabellen anhand einer Verknüpfungsbedingung kombinieren
- Zweck: Zusammengehörige Informationen aus unterschiedlichen Tabellen gemeinsam auswerten
- Bsp. Verkäufe über salespersonid mit den zugehörigen Verkäufern verbinden

| JOIN | Welche Zeilen bleiben erhalten? |
|---|---|
| INNER JOIN | Nur passende Kombinationen beider Tabellen |
| LEFT JOIN | Alle Zeilen der linken Tabelle, passende Informationen von rechts |
| RIGHT JOIN | Alle Zeilen der rechten Tabelle, passende Informationen von links |
| FULL OUTER JOIN | Passende Kombinationen und nicht zugeordnete Zeilen beider Seiten |
| CROSS JOIN | Jede Zeile der einen Tabelle mit jeder Zeile der anderen |

- Fehlende Gegenseite bei einem OUTER JOIN: Deren Spalten werden mit NULL ergänzt
- CROSS JOIN: Bewusst alle Kombinationen bilden, dafür keine ON-Bedingung erforderlich
- 1:n-Beziehung: Ein linker Datensatz kann nach dem JOIN mehrfach vorkommen
- Achtung: Mehrere Beziehungen mit mehreren Treffern können Zeilen und damit Summen vervielfachen
- Erkenntnis: Den JOIN danach auswählen, welche Objekte auch ohne passenden Treffer erhalten bleiben sollen

### Wichtige Begriffe im JOIN

- ON: Legt fest, welche Zeilen zusammenpassen
- Tabellenalias: Optionaler Kurzname für eine Tabelle, vollständige Tabellennamen können verwendet werden
- Spaltenalias: Name einer Ergebnisspalte, bspw. total_sales_amount
- Mehrdeutige Spalte: Wenn beide Tabellen denselben Spaltennamen besitzen, den Tabellennamen voranstellen
- Tabelle1.*: Alle Spalten aus Tabelle1 auswählen
- NULL: Fehlender bzw. unbekannter Wert, weder 0 noch automatisch ein leerer Text

## SQL Agents und natürliche Sprache
Folien 41 bis 42

- Definition: KI-gestützte Systeme, die Fragen in natürlicher Sprache zur Abfrage einer Datenbank verwenden
- Bsp. LangChain SQL Agent mit einem Sprachmodell
- Zweck: Datenzugriff erleichtern und die notwendige manuelle SQL-Formulierung reduzieren
- Grundidee: Fachliche Frage formulieren, passende Datenbankabfrage erzeugen bzw. ausführen und Ergebnis zurückgeben
- Grenze: Falsche Tabellen, Beziehungen oder Interpretationen können zu falschen Antworten führen

## Data Quality Management
Folie 43

- Definition: Prozesse, Regeln und Werkzeuge zur Sicherung und Verbesserung der Datenqualität
- Zweck: Verlässliche Analysen, Entscheidungen und operative Abläufe ermöglichen
- Vorgehen: Daten untersuchen, Qualitätsregeln definieren, Probleme überwachen, beheben und Ergebnisse berichten
- Erkenntnis: Datenqualität ist eine dauerhafte Aufgabe, keine einmalige Bereinigung

### Die vier genannten Qualitätseigenschaften

| Eigenschaft | Bedeutung | Bsp. |
|---|---|---|
| Accurate | Werte sind sachlich korrekt | Preis korrekt erfasst, keine fehlerhafte Kategoriebezeichnung |
| Consistent | Werte und Darstellungen sind einheitlich und widerspruchsfrei | Standard, standard und STANDARD vereinheitlichen |
| Complete | Benötigte Angaben sind vorhanden | Preis und Mitgliedschaftstyp fehlen nicht |
| Reliable | Daten und ihre Herkunft sind vertrauenswürdig und verlässlich | Nachvollziehbare Quelle und kontrollierte Erfassung |

### Bsp. membership_type

Vorhandene Werte: Standard, standard, STANDARD, Standart, Premium, premium, Student, student, Senior

- Consistency-Problem: Dieselben Kategorien besitzen unterschiedliche Gross-/Kleinschreibung
- Accuracy-Problem: Standart ist eine fehlerhafte Schreibweise von Standard
- Bereinigung: Schreibweisen vereinheitlichen und Tippfehler anhand gültiger Kategorien korrigieren
- Achtung: Senior ist nur ungültig, wenn diese Kategorie gemäss Datendefinition nicht erlaubt ist
- Erkenntnis: Ein Problem kann mehrere Qualitätseigenschaften betreffen, Gross-/Kleinschreibung vereinheitlichen korrigiert keine Tippfehler

### Bestandteile des DQM

| Bestandteil | Aufgabe |
|---|---|
| Data Profiling | Datenstruktur, Werteverteilungen und Auffälligkeiten untersuchen |
| Data Rules | Regeln für zulässige und erwartete Daten festlegen |
| Data Monitoring | Qualität fortlaufend überwachen |
| Data Remediation | Erkannte Probleme korrigieren |
| Data Quality Reporting | Qualitätszustand und Probleme sichtbar machen |
| Data Discovery | Vorhandene Daten und ihre Eigenschaften auffinden und verstehen |

- Achtung: Ein technisch bereinigter Datensatz kann weiterhin sachlich falsche Werte enthalten
- Bsp. Ein Preis ist numerisch korrekt gespeichert, aber um eine Null zu hoch erfasst

## Master Data Management
Folie 44

- Definition: Zentrale und abgestimmte Verwaltung wichtiger Stammdaten eines Unternehmens
- Zweck: Eine konsistente, korrekte und verbindliche Referenz über Systeme und Abteilungen hinweg schaffen
- Typische Stammdaten: Kunden, Produkte, Mitarbeitende und Lieferanten
- Bsp. Derselbe Kunde soll im Shop, CRM und Reporting über eine eindeutige Identität zugeordnet werden können
- Erkenntnis: MDM verhindert, dass verschiedene Unternehmensbereiche mit widersprüchlichen Stammdaten arbeiten

### Stammdaten und Transaktionsdaten unterscheiden

| Datenart | Bedeutung | Bsp. |
|---|---|---|
| Stammdaten | Beschreiben zentrale Objekte | Kunde, Produkt und Verkäufer |
| Transaktionsdaten | Beschreiben einzelne Vorgänge oder Ereignisse | Kauf, Ticketverkauf und Service |

- Achtung: Stammdaten können sich ändern, sie sind nicht grundsätzlich unveränderlich

## Metadata Management
Folie 45

- Definition: Verwaltung von Informationen, die andere Daten beschreiben
- Zweck: Daten verständlich, auffindbar und korrekt nutzbar machen
- Bsp. Spaltenbeschreibung, Datentyp, Einheit, Herkunft, Verantwortlichkeit und Aktualisierungszeitpunkt
- Erkenntnis: Ein Wert ist ohne seine Bedeutung und seinen Kontext häufig nicht richtig interpretierbar

| Begriff | Bedeutung | Bsp. |
|---|---|---|
| Business Metadata | Fachliche Bedeutung | Umsatz ohne Mehrwertsteuer |
| Technical Metadata | Technische Struktur | Spaltenname, Datentyp und Tabellenbeziehung |
| Metadata Catalog | Verzeichnis vorhandener Daten und ihrer Beschreibungen | Datenkatalog mit Tabellen und Verantwortlichen |
| Data Lineage | Herkunft und Verarbeitungsschritte | Quelldatei → Bereinigung → Tabelle → Bericht |
| Metadata Standards | Einheitliche Regeln zur Beschreibung | Verbindliche Definition von Einheiten und Begriffen |

### Master Data und Metadata unterscheiden

- Master Data: Beschreibt fachliche Objekte, bspw. den Kunden mit ID 123
- Metadata: Beschreibt die Daten darüber, bspw. Bedeutung und Datentyp von customerid

## Data Privacy und Compliance
Folie 46

- Data Privacy: Verantwortlicher Umgang mit personenbezogenen Daten
- Compliance: Einhaltung geltender rechtlicher und organisatorischer Anforderungen
- Zweck: Personen schützen und Daten nur angemessen, nachvollziehbar und zulässig verwenden
- Bsp. Personenbezogene Daten nicht allein deshalb für jede beliebige Analyse nutzen, weil sie technisch verfügbar sind

| Aspekt | Grundidee |
|---|---|
| Consent | Einwilligung berücksichtigen, soweit sie für die Verarbeitung erforderlich ist |
| Transparency | Verarbeitung verständlich machen |
| Data Minimization | Nur erforderliche Daten verwenden |
| Access | Angemessenen Zugang und Betroffenenrechte berücksichtigen |
| Accuracy | Personenbezogene Angaben korrekt halten |
| Retention | Aufbewahrung und Löschung regeln |
| Accountability | Verantwortlichkeiten und Einhaltung nachweisen |
| Breach Notification | Vorgesehene Meldepflichten bei Datenschutzverletzungen beachten |

- GDPR: In den Folien als Beispiel eines rechtlichen Datenschutzrahmens genannt
- Erkenntnis: Technischer Zugriff allein bedeutet nicht, dass eine Datennutzung erlaubt ist

## Data Protection und Security
Folie 47

- Definition: Organisatorische und technische Massnahmen gegen unbefugten Zugriff, Veränderung, Verlust oder Diebstahl von Daten
- Zweck: Vertraulichkeit, Integrität und Verfügbarkeit sicherstellen

| Schutzziel | Bedeutung | Bsp. |
|---|---|---|
| Confidentiality, Vertraulichkeit | Nur Berechtigte dürfen Daten sehen | Zugriff auf Kundendaten beschränken |
| Integrity, Integrität | Daten bleiben korrekt und vor unzulässiger Veränderung geschützt | Unbefugte Preisänderungen verhindern |
| Availability, Verfügbarkeit | Daten sind bei Bedarf zugänglich | Wiederherstellung nach einem Ausfall ermöglichen |

| Massnahme | Zweck |
|---|---|
| Verschlüsselung | Inhalte für Unberechtigte unlesbar machen |
| Zugriffskontrolle | Berechtigungen festlegen und durchsetzen |
| Data Masking | Sensible Werte bei der Nutzung verdecken |
| Backup und Recovery | Daten sichern und wiederherstellen |
| Security Audits | Schutzmassnahmen überprüfen |
| Compliance Management | Sicherheitsanforderungen systematisch berücksichtigen |

### Privacy und Security unterscheiden

- Privacy: Welche personenbezogenen Daten werden warum und unter welchen Bedingungen verwendet?
- Security: Wie werden Daten und Systeme vor unbefugtem Zugriff, Manipulation oder Ausfall geschützt?
- Erkenntnis: Gut gesicherte Daten können trotzdem unzulässig verwendet werden, rechtmässige Verarbeitung benötigt trotzdem angemessene Sicherheit

## Data Lifecycle Management
Folie 48

- Definition: Daten während ihres gesamten Lebenszyklus nach festgelegten Regeln verwalten
- Zweck: Effizienten, sicheren und angemessenen Umgang von der Entstehung bis zur Löschung sicherstellen
- Ablauf: `Erstellung → Speicherung → Nutzung → Archivierung → Löschung`

| Phase | Wichtige Frage |
|---|---|
| Erstellung | Wie werden Daten erhoben und geprüft? |
| Speicherung | Wo und wie werden sie gespeichert und geschützt? |
| Nutzung | Wer verwendet sie für welchen Zweck? |
| Archivierung | Welche Daten müssen länger erhalten bleiben? |
| Löschung | Wann und wie werden Daten entfernt? |

- Erkenntnis: Daten müssen nicht für immer im aktiven System bleiben, Aufbewahrung und Löschung gehören zur Planung

## Data Archiving
Folie 49

- Definition: Nicht mehr aktiv benötigte Daten in einen separaten Speicher zur längerfristigen Aufbewahrung verschieben
- Zweck: Historische Informationen erhalten und aktive Systeme entlasten
- Vorteile: Weniger Last auf primären Speichern, mögliche Kostensenkung und Unterstützung von Aufbewahrungsanforderungen
- Bsp. Alte Verkaufsdaten bleiben für historische Auswertungen zugänglich, ohne ständig im aktiven Verkaufssystem benötigt zu werden

### Archiv und Backup unterscheiden

| Begriff | Hauptzweck |
|---|---|
| Archiv | Langfristige Aufbewahrung nicht mehr aktiv genutzter Daten |
| Backup | Wiederherstellung nach Verlust, Beschädigung oder Ausfall |

- Erkenntnis: Archivierung und Datensicherung ergänzen sich, haben aber unterschiedliche Hauptziele

## Data Analysis
Folie 50

- Definition: Daten untersuchen, bereinigen, transformieren und modellieren, um Informationen und Schlussfolgerungen zu gewinnen
- Zweck: Entscheidungen unterstützen und konkrete Fragen beantworten

| Bestandteil | Aufgabe |
|---|---|
| Data Preparation | Daten für die Analyse vorbereiten |
| Exploratory Data Analysis | Verteilungen, Muster und Auffälligkeiten untersuchen |
| Statistical Hypothesis Testing | Statistische Hypothesen prüfen |
| Predictive Modeling | Modelle für Vorhersagen entwickeln |
| Presentation of Results | Ergebnisse verständlich darstellen |

- Erkenntnis: Berechnung, Interpretation und verständliche Kommunikation gehören zusammen
- Achtung: Ein Zusammenhang allein beweist keine Ursache

## OLAP: Online Analytical Processing
Folie 51

- Definition: Mehrdimensionale Analyse grosser Datenbestände, häufig aus einem Data Warehouse
- Zweck: Daten interaktiv aus verschiedenen Perspektiven betrachten
- Bsp. Verkaufsumsatz nach Zeit, Region und Produktkategorie auswerten

| Begriff | Bedeutung | Bsp. |
|---|---|---|
| Dimension | Perspektive der Auswertung | Zeit, Region und Produkt |
| Hierarchie | Geordnete Stufen innerhalb einer Dimension | Jahr → Quartal → Monat |
| Level | Einzelne Hierarchiestufe | Monat |
| Measure | Auszuwertende Kennzahl | Umsatz oder Ticketanzahl |
| Fact | Gespeichertes messbares Ereignis bzw. Faktendatensatz | Einzelner Verkauf |
| Aggregate | Zusammengefasster Wert | Monatsumsatz |
| Cell | Wert für eine konkrete Kombination von Dimensionen | Umsatz im März in Zürich für Kategorie A |

### OLAP-Operationen

| Operation | Bedeutung | Bsp. |
|---|---|---|
| Slice | Einen bestimmten Dimensionswert auswählen | Nur das Jahr 2026 betrachten |
| Dice | Teilbereich anhand mehrerer Dimensionen auswählen | Zwei Regionen und drei Produktkategorien |
| Drill Down | Zu detaillierteren Ebenen wechseln | Jahresumsatz → Monatsumsatz |
| Drill Up bzw. Roll Up | Zu stärker zusammengefassten Ebenen wechseln | Monatsumsatz → Jahresumsatz |

- Erkenntnis: Dieselben Fakten können je nach Fragestellung unterschiedlich gefiltert, gruppiert und verdichtet werden

## Zentrale Unterscheidungen für Verständnisfragen

| Frage | Kernantwort |
|---|---|
| Warum reichen Programmierkenntnisse für Data Science nicht? | Methoden müssen statistisch korrekt und fachlich sinnvoll eingesetzt werden |
| Warum ist die Danger Zone problematisch? | Analysen können überzeugend aussehen, obwohl ihre Ergebnisse falsch verstanden werden |
| Warum ist der Analyseprozess iterativ? | Neue Ergebnisse können andere Daten oder eine geänderte Fragestellung erfordern |
| Governance oder Management? | Governance definiert Regeln und Verantwortung, Management setzt sie um |
| Architektur oder Datenmodell? | Architektur beschreibt die Datenlandschaft, ein Modell konkretisiert Datenstrukturen |
| Konzeptionell, logisch oder physisch? | Fachliche Übersicht, detaillierte Struktur oder konkrete technische Umsetzung |
| Primärschlüssel oder Fremdschlüssel? | Eigene Zeile eindeutig identifizieren oder auf einen zulässigen Datensatz verweisen |
| Warum Normalisierung? | Unnötige Wiederholungen und daraus entstehende Widersprüche reduzieren |
| Star oder Snowflake? | Direkt angebundene Dimensionen oder weiter normalisierte Dimensionstabellen |
| INNER JOIN oder LEFT JOIN? | Nur passende Kombinationen oder zusätzlich alle linken Datensätze behalten |
| Accurate oder consistent? | Sachlich richtige Werte oder einheitliche und widerspruchsfreie Darstellung |
| Complete oder reliable? | Benötigte Angaben vorhanden oder Daten und Herkunft verlässlich |
| Stammdaten oder Metadaten? | Fachliche Objekte oder Beschreibungen der Daten |
| Privacy oder Security? | Zulässige personenbezogene Nutzung oder Schutz der Daten und Systeme |
| Archiv oder Backup? | Langfristige Aufbewahrung oder Wiederherstellung nach einem Problem |
| Drill Down oder Drill Up? | Detaillierter auswerten oder stärker zusammenfassen |
