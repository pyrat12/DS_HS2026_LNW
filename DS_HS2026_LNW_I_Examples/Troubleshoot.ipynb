# GitHub

## GitHub-Fork aktualisieren

- Commit: Gespeicherter Änderungsstand, kann mehrere Dateien betreffen
- 1 commit behind: Deinem Branch fehlt ein Commit aus dem Original-Repository
- Eigenen Fork öffnen → `Sync fork` → `Update branch`
- Bestehenden Codespace danach mit `git pull --ff-only` aktualisieren, eigene Änderungen vorher speichern, committen und pushen

## Codespace öffnen

Auf GitHub anmelden → eigenen Fork öffnen → `Code` → `Codespaces` → bestehenden Codespace öffnen oder neuen erstellen

Zu viele Codespaces aktiv? Unter https://github.com/codespaces nicht benötigte Codespaces stoppen

Notebook öffnen, Python-Kernel auswählen und benötigte Setup-Zellen von oben nach unten ausführen. `Alle ausführen` nur verwenden, wenn keine alternativen oder unerwünschten Änderungsschritte enthalten sind

## GitHub-Zwischenstand speichern

```bash
git add .
git commit -m "Spicks aktualisiert"
git push
```

`add` merkt Änderungen vor, `commit` speichert sie lokal und `push` lädt die Commits auf GitHub hoch

## Docker-Container

Compose-Befehle im Ordner der `compose.yaml` bzw. `docker-compose.yml` ausführen

| Befehl | Bedeutung |
|---|---|
| `docker ps` | Laufende Container prüfen |
| `docker ps -a` | Auch gestoppte Container anzeigen |
| `docker compose up -d` | Container erstellen/starten |
| `docker compose restart` | Bestehende Container neu starten |
| `docker logs CONTAINERNAME` | Fehlermeldungen eines Containers ansehen |
| `clear` | Terminalanzeige leeren |

Nur wenn ein Container gezielt entfernt und neu erstellt werden soll:

```bash
docker rm -f CONTAINERNAME
docker compose up -d
```

Bsp.: `docker rm -f postgres_container pgadmin_container`

Achtung: Daten ausschliesslich im Container gehen beim Entfernen verloren, vorher prüfen, ob sie gesichert oder in einem Volume gespeichert sind
