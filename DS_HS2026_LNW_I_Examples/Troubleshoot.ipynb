### VS Code neu laden, wenn die Codefarben fehlen

`Cmd + Shift + P` drücken → `Developer: Reload Window` auswählen.

### Änderungen auf GitHub zwischenspeichern

Notebook zuerst mit `Cmd + S` speichern. Danach im Terminal ausführen:

```bash
git add .
git commit -m "Spicks aktualisiert"
git push
```
