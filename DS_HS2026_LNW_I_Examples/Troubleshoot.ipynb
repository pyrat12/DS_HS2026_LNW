### Welche Methode benötigt welchen Datentyp?

| Methode | Benötigte Daten |
|---|---|
| Mean, Median, Varianz, Standardabweichung, Quantile, Skewness und Kurtosis | Numerische Werte |
| Kovarianz und Pearson-Korrelation | Zwei oder mehr numerische Merkmale; IDs nicht als Messwerte behandeln |
| Spearman-Korrelation | Numerische oder sinnvoll geordnete Merkmale; ungeordnete Kategorien nicht willkürlich nummerieren |
| Mode und `value_counts()` | Kategorien oder numerische Werte |
| Kreuztabelle / `pd.crosstab()` | Kategorial auswertbare Merkmale, bspw. Ladeart und Steckertyp |
| Histogramm und QQ-Plot | Numerische Variable |
| Scatter Plot und Hexbin Plot | Zwei numerische Variablen |
| Boxplot und Violin Plot | Numerische Variable; optional eine Kategorie zur Gruppenbildung |
| Balkendiagramm | Kategorien mit zugehörigen Häufigkeiten oder Kennzahlen |
| One-Hot Encoding | Kategoriale Merkmale |
| Scaling, Standardisierung und mathematische Transformationen | Numerische Werte; zusätzliche Voraussetzungen der jeweiligen Methode beachten |

**Datentypen prüfen:** `df.dtypes` oder `df.info()` verwenden.

**Zahlentexte:** `"245 PS"` oder `"CHF 999.–"` zuerst bereinigen und numerisch umwandeln. Die Spalte umzubenennen reicht nicht.

**Zusammenhänge:** Bei Korrelation und Kovarianz müssen die Werte derselben Zeile zur selben Beobachtung gehören; Tabellen aus verschiedenen Quellen zuerst korrekt zuordnen.
