## Rules

## describe() richtig lesen

| Kennzahl | Bedeutung | Rule |
|---|---|---|
| count | Anzahl nicht fehlender Werte | Mit Gesamtzahl der Zeilen vergleichen |
| mean | Mittelwert | Empfindlich gegenüber sehr hohen oder tiefen Werten |
| std | Standardabweichung | Beschreibt die Streuung in derselben Einheit wie die Variable |
| min | Kleinster Wert | Auf fachliche Plausibilität prüfen |
| 25% | Q1, unteres Quartil | Rund 25 % der Werte liegen darunter oder darauf |
| 50% | Median | Rund die Hälfte liegt darunter oder darauf |
| 75% | Q3, oberes Quartil | Rund 75 % der Werte liegen darunter oder darauf |
| max | Grösster Wert | Auf mögliche Extremwerte oder Fehler prüfen |

- Fehlende Werte = Gesamtzahl der Zeilen − count
- Bsp. 142 Zeilen und count 134 bedeuten acht fehlende Werte
- count 134,00 bedeutet 134 Beobachtungen, die Nachkommastellen sind nur Ausgabeformatierung
- Zwischen Q1 und Q3 liegen ungefähr die mittleren 50 % der Werte
- Bei vielen gleichen Werten können tatsächliche Anteile von den genannten Prozentwerten abweichen

## Verteilungsform erkennen

| Beobachtung | Interpretation |
|---|---|
| Positive Schiefe, skew > 0 | Rechtsschief, langer Ausläufer nach rechts |
| Negative Schiefe, skew < 0 | Linksschief, langer Ausläufer nach links |
| Schiefe nahe 0 | Geringe gemessene Asymmetrie, Symmetrie oder Normalverteilung nicht bewiesen |
| Mean > Median | Hinweis auf Rechtsschiefe bzw. Einfluss hoher Werte |
| Mean < Median | Hinweis auf Linksschiefe bzw. Einfluss tiefer Werte |
| Mean ungefähr Median | Mit Symmetrie vereinbar, beweist sie aber nicht |
| Maximum weit oberhalb von Q3 | Hohe Extremwerte genauer untersuchen |
| Minimum weit unterhalb von Q1 | Tiefe Extremwerte genauer untersuchen |

- Für die Begründung mehrere Hinweise kombinieren, bspw. positive Schiefe, Mean über Median und sehr hohes Maximum
- Mean und Standardabweichung nicht zur Bestimmung der Schieferichtung vergleichen
- Rechtsschief bedeutet langer Ausläufer nach rechts, nicht dass die meisten Werte rechts liegen
- Linksschief bedeutet langer Ausläufer nach links, nicht dass die meisten Werte links liegen

### Musterantwort aus dem Mock Exam

- Die Preise sind deutlich rechtsschief, die Schiefe ist mit 9,19 positiv
- Der Mittelwert von 992,89 CHF liegt über dem Median von 899 CHF
- Das Maximum von 12'990 CHF liegt weit über Q3 von 1199 CHF und deutet auf einen möglichen Ausreisser hin
- Zur Überprüfung eignen sich Histogramm und Boxplot

## Passendes Lagemass wählen

| Situation | Geeignetes Mass | Begründung |
|---|---|---|
| Typischer Wert bei schiefer Verteilung oder Extremwerten | Median | Wird von einzelnen Extremwerten wenig beeinflusst |
| Durchschnittlicher Betrag über alle Beobachtungen | Mean | Berücksichtigt die Höhe jedes vorhandenen Werts |
| Häufigste Kategorie bzw. häufigster Wert | Mode | Zeigt die häufigste Ausprägung |
| Ungeordnete Kategorien wie Marke | Mode oder Häufigkeiten | Mittelwert und Median sind fachlich nicht sinnvoll |

- Median nicht automatisch immer verwenden, das Mass muss zur Frage passen
- Ein hoher Mittelwert kann durch wenige hohe Werte entstehen
- Ein hoher Median bedeutet, dass auch die Mitte der Verteilung hoch liegt
- Bei Gruppenvergleichen zusätzlich Anzahl und fehlende Werte je Gruppe betrachten

## Streuung interpretieren

| Beobachtung | Rule |
|---|---|
| Hohe Standardabweichung | Werte unterscheiden sich stark, Grössenordnung und Einheit mitberücksichtigen |
| Kleine Standardabweichung | Werte liegen vergleichsweise nahe am Mittelwert |
| Standardabweichung 0 | Alle vorhandenen Werte sind gleich |
| Varianz | Quadrierte Einheit, bspw. CHF² |
| Standardabweichung | Ursprüngliche Einheit, bspw. CHF |
| Spannweite | max − min, stark von Extremwerten beeinflusst |
| IQR | Q3 − Q1, Streuung der mittleren 50 % |

- Eine Standardabweichung über dem Mittelwert ist möglich und kein automatischer Fehler
- Mean ± std ist kein garantierter Wertebereich
- Grosse Standardabweichung beweist keine Schiefe
- Unterschiedliche Einheiten nicht direkt vergleichen, bspw. CHF und Quadratmeter
- pandas berechnet Varianz und Standardabweichung standardmässig mit ddof=1, also für die Stichprobe

## Ausreisser erkennen und behandeln

- IQR = Q3 − Q1
- Untere Grenze = Q1 − 1,5 × IQR
- Obere Grenze = Q3 + 1,5 × IQR
- Werte ausserhalb markieren potenzielle Ausreisser, keine automatisch falschen Werte

### Bsp. Mock Exam

- Q1 = 479 und Q3 = 1199 ergeben IQR = 720
- Obere Grenze = 1199 + 1,5 × 720 = 2279
- 12'990 liegt darüber und muss untersucht werden
- Dass dasselbe Gerät in anderer Farbe 1299 CHF kostet, liefert zusätzliche Evidenz für einen Erfassungsfehler

### Vorgehen

- Erst describe(), Boxplot und sortierte Extremwerte anschauen
- Dann Einheiten, ähnliche Datensätze und fachliche Plausibilität prüfen
- Bestätigte Fehler korrigieren oder begründet entfernen
- Echte Extremwerte nicht allein wegen ihrer Grösse löschen
- Median verwenden ersetzt nicht die Prüfung eines vermuteten Datenfehlers

## Kurtosis interpretieren

| pandas-Ergebnis | Bedeutung |
|---|---|
| Kurtosis > 0 | Schwerere Verteilungsenden als bei der Normalverteilung |
| Kurtosis < 0 | Leichtere Verteilungsenden als bei der Normalverteilung |
| Kurtosis ungefähr 0 | Ähnliche Kurtosis wie die Normalverteilung |

- pandas verwendet Exzess-Kurtosis, die Normalverteilung hat den Referenzwert 0 statt 3
- Hohe Kurtosis weist auf ausgeprägte Verteilungsenden hin, nicht auf eine bestimmte Anzahl Ausreisser
- Kurtosis und Schiefe allein beweisen keine Normalverteilung

## Fehlende Werte und Verzerrung

| Mechanismus | Erkennungsregel |
|---|---|
| MCAR | Fehlen hängt weder vom fehlenden Wert noch von anderen Variablen ab |
| MAR | Fehlen hängt von beobachteten Variablen ab, nach deren Berücksichtigung nicht zusätzlich vom fehlenden Wert selbst |
| MNAR | Fehlen hängt auch unter Berücksichtigung beobachteter Variablen vom fehlenden Wert selbst ab |

- Fehlen gezielt hohe Preise, wird der Mittelwert der verbleibenden Preise nach unten verzerrt
- Fehlen gezielt tiefe Preise, wird der Mittelwert der verbleibenden Preise nach oben verzerrt
- MCAR-Entfernung reduziert die Datenmenge, verursacht aber grundsätzlich keine systematische Auswahlverzerrung
- MAR und MNAR lassen sich aus beobachteten Daten allein im Allgemeinen nicht sicher unterscheiden
- Fehlend ist nicht 0, einen unbekannten Preis nicht ungeprüft durch 0 ersetzen
- Mean, Median und andere Kennzahlen ignorieren fehlende Werte normalerweise, deshalb count kontrollieren
- Imputation kann Verteilungen verändern, Mittelwert-Imputation verringert typischerweise die Streuung

## Korrelation und Kovarianz

| Ergebnis | Interpretation |
|---|---|
| Pearson r > 0 | Höheres X geht tendenziell mit höherem Y einher |
| Pearson r < 0 | Höheres X geht tendenziell mit tieferem Y einher |
| Pearson r nahe 0 | Wenig linearer Zusammenhang, andere Muster können bestehen |
| Betrag von r nahe 1 | Starker linearer Zusammenhang |
| Spearman nahe +1 bzw. −1 | Starker monotoner Zusammenhang, nicht zwingend linear |
| Positive Kovarianz | Tendenziell gemeinsame Veränderung in dieselbe Richtung |
| Negative Kovarianz | Tendenziell Veränderung in entgegengesetzte Richtungen |

- Pearson verwendet ursprüngliche Zahlenwerte, Spearman Rangplätze und ist weniger empfindlich gegenüber Extremwerten
- Stärke anhand des Betrags vergleichen, −0,9 ist stärker als +0,4
- In einer Korrelationsmatrix die Hauptdiagonale mit 1 nicht als stärksten Zusammenhang auswählen
- Kovarianzbeträge hängen von Einheiten und Skalierung ab, nicht wie Korrelationswerte vergleichen
- Korrelation 0 beweist keine Unabhängigkeit
- Korrelation beweist keine Ursache
- Immer Scatterplot ergänzen, Kennzahlen können nichtlineare Muster oder Ausreissereinflüsse verbergen
- pandas verwendet je Variablenpaar die vorhandenen vollständigen Wertepaare, die verwendete Anzahl kann unterschiedlich sein

### Interpretation formulieren

- Bsp. Spearman 0,76 bedeutet einen deutlichen positiven monotonen Zusammenhang
- Teurere Geräte werden tendenziell besser bewertet
- Das beweist nicht, dass ein höherer Preis bessere Bewertungen verursacht
- Korrelationswerte sind keine Prozentsteigerungen und keine Trefferquoten

## Gruppen und Kreuztabellen

- Pro Marke, Status oder Kategorie verlangt meist groupby()
- Anzahl Zeilen und Anzahl vorhandener Werte unterscheiden, size() zählt Zeilen und count() nicht fehlende Werte
- Ein Gruppenmittelwert kann von einem einzigen Extremwert stark beeinflusst werden
- Kleine Gruppen vorsichtig interpretieren, eine Marke mit sechs Geräten ist nicht gleich breit vertreten wie eine mit 50 Geräten
- Den Gesamtmittelwert nicht als einfachen Mittelwert unterschiedlich grosser Gruppen berechnen, Gruppengrössen berücksichtigen

### crosstab()

- Zeilen und Spalten zeigen Kategorien, jede Zelle zählt deren Kombination
- Standardmässig absolute Häufigkeiten, keine Prozentwerte
- Fehlende Werte werden standardmässig nicht mitgezählt
- Grössere absolute Anzahl beweist keinen höheren Anteil, unterschiedlich grosse Gruppen zusätzlich relativ vergleichen

| Prozentdarstellung | Bezug |
|---|---|
| normalize='index' | Innerhalb der jeweiligen Zeile |
| normalize='columns' | Innerhalb der jeweiligen Spalte |
| normalize='all' | Bezogen auf alle berücksichtigten Beobachtungen |

## cut und qcut

| Methode | Rule |
|---|---|
| cut | Klassen anhand von Zahlenintervallen bilden, bei automatisch gewählten Intervallen gleich breite Bereiche |
| qcut | Grenzen über Quantile bestimmen, ungefähr gleich viele Beobachtungen je Klasse |

- Gleich breite Intervalle bedeuten nicht gleich viele Beobachtungen
- Ungefähr gleich viele Beobachtungen bedeuten nicht gleich breite Intervalle
- Fehlende Zahlen erhalten keine Klasse und fehlen standardmässig in anschliessenden Häufigkeitsauswertungen
- Bei vielen gleichen Werten kann qcut gleiche Grenzen und damit Probleme erzeugen
- Anzahl Labels muss zur Anzahl gebildeter Klassen passen

## Passende Grafik wählen

| Fragestellung | Grafik |
|---|---|
| Verteilungsform, Schiefe und mehrere Gipfel | Histogramm |
| Median, Quartile und potenzielle Ausreisser | Boxplot |
| Verteilung zwischen Gruppen vergleichen | Gruppierte Boxplots |
| Annäherung an eine theoretische Verteilung | QQ-Plot |
| Häufigkeiten von Kategorien | Balkendiagramm |
| Zusammenhang zweier numerischer Variablen | Scatterplot |
| Viele überlagerte Punkte | Hexbin |
| Mehrere numerische Variablen paarweise vergleichen | Scatterplot-Matrix |
| Korrelationsmatrix übersichtlich darstellen | Heatmap |
| Entwicklung über geordnete Zeitpunkte | Liniendiagramm |

### Grafiken interpretieren

- Histogramm: Langer rechter Ausläufer bedeutet Rechtsschiefe, mehrere Gipfel können auf unterschiedliche Gruppen hinweisen
- Histogramm: Der höchste Balken zeigt das häufigste Intervall, nicht den höchsten Einzelwert
- Histogramm: Bin-Anzahl beeinflusst das Bild, bei unterschiedlichen Klassenbreiten Flächen bzw. Dichten beachten
- Boxplot: Linie in der Box ist der Median, Box reicht von Q1 bis Q3
- Boxplot: Punkte ausserhalb der Whisker sind potenzielle Ausreisser, Whisker sind nicht zwingend Minimum und Maximum
- QQ-Plot: Punkte ungefähr auf der Referenzgeraden sprechen für eine passende Vergleichsverteilung, systematische Abweichungen dagegen
- Scatterplot: Richtung, Form, Gruppen und Extremwerte gemeinsam betrachten
- Heatmap: Vorzeichen und Betrag beachten, starke Korrelation beweist keine Ursache

## Schneller Ablauf bei Interpretationsaufgaben

1. Variable und Einheit nennen, bspw. Gerätepreise in CHF
2. count mit Gesamtzahl vergleichen und fehlende Werte erkennen
3. Mitte mit Mean und Median beschreiben
4. Streuung mit std, Quartilen und IQR einordnen
5. Form mit skew und Mean-Median-Vergleich beurteilen
6. Minimum und Maximum auf Plausibilität prüfen
7. Geeignete Grafik zur Überprüfung nennen
8. Schlussfolgerung auf die konkrete Aufgabe beziehen und relevante Einschränkungen nennen

## Begriffe
### Statistische Grundbegriffe

- **Beobachtung:** Ein einzelner untersuchter Fall, bspw. ein Smartphone oder eine Ladestation
- **Variable:** Ein Merkmal einer Beobachtung, bspw. Preis, Marke oder Ladeleistung
- **Numerische Variable:** Zahlen mit sinnvoller rechnerischer Bedeutung, bspw. Preis oder Leistung
- **Kategoriale Variable:** Gruppenzugehörigkeit, bspw. Marke oder Steckertyp
- **Wichtig:** Eine numerische ID identifiziert einen Datensatz, ihr Mittelwert ist normalerweise nicht sinnvoll

### Stichprobe und Grundgesamtheit

- **Definition:** Die Grundgesamtheit umfasst alle interessierenden Beobachtungen, die Stichprobe ist der tatsächlich untersuchte Teil
- **Zweck:** Ergebnisse einer Stichprobe helfen, Aussagen über eine grössere Gruppe zu treffen
- **Bsp.:** 500 untersuchte Ladestationen sind eine Stichprobe aller Ladestationen der Schweiz
- **Wichtig:** Eine grosse Stichprobe ist nicht automatisch repräsentativ, die Auswahl der Beobachtungen ist entscheidend

### Univariate und multivariate Analyse

- **Univariat:** Eine Variable untersuchen, bspw. die Preisverteilung
- **Bivariat:** Zwei Variablen gemeinsam untersuchen, bspw. Preis und Speichergrösse
- **Multivariat:** Mehrere Variablen gemeinsam untersuchen, bspw. Preis, Speichergrösse und Marke
- **Verwendung:** Zuerst einzelne Variablen verstehen, danach Zusammenhänge und Gruppenunterschiede untersuchen

### Absolute und relative Häufigkeit

- **Definition:** Die absolute Häufigkeit zählt Beobachtungen, die relative Häufigkeit beschreibt ihren Anteil
- **Zweck:** Erkennen, welche Werte oder Kategorien häufig oder selten vorkommen
- **Bsp.:** 20 von 100 Geräten sind von Apple, absolute Häufigkeit 20, relative Häufigkeit 20 %
- **Interpretation:** Eine relative Häufigkeit von 0,25 entspricht 25 %
- **Wichtig:** Bei Prozentwerten immer prüfen, worauf sich der Anteil bezieht, bspw. alle Geräte oder nur Geräte einer Marke

### Anzahl gültiger Werte

- **Definition:** Anzahl der tatsächlich vorhandenen Werte einer Variable, fehlende Werte werden nicht mitgezählt
- **Zweck:** Beurteilen, auf wie vielen Beobachtungen eine Kennzahl beruht
- **Bsp.:** 100 Geräte, davon 8 ohne Preis, die Preisstatistik beruht auf 92 gültigen Preisen
- **Interpretation:** Unterschiedliche Anzahlen pro Spalte weisen auf unterschiedlich viele fehlende Werte hin

### Mittelwert – Mean

- **Definition:** Summe aller gültigen Werte geteilt durch ihre Anzahl
- **Zweck:** Den durchschnittlichen Wert beschreiben
- **Bsp.:** Die Werte 10, 20 und 30 haben den Mittelwert 20
- **Interpretation:** Ein durchschnittlicher Preis von 800 CHF bedeutet nicht, dass die meisten Geräte 800 CHF kosten
- **Wichtig:** Sehr hohe oder tiefe Werte können den Mittelwert stark beeinflussen, deshalb mit Median und Histogramm vergleichen

### Median – Zentralwert

- **Definition:** Der mittlere Wert der sortierten Daten, bei einer geraden Anzahl der Durchschnitt der beiden mittleren Werte
- **Zweck:** Einen zentralen Wert beschreiben, der weniger empfindlich gegenüber Ausreissern ist
- **Bsp.:** Bei 10, 20 und 300 ist der Median 20, der Mittelwert hingegen 110
- **Interpretation:** Ein Median von 800 CHF bedeutet, dass mindestens die Hälfte der gültigen Preise höchstens 800 CHF beträgt und mindestens die Hälfte mindestens 800 CHF
- **Verwendung:** Besonders hilfreich bei schiefen Verteilungen und extremen Werten

### Modus – Modalwert

- **Definition:** Der am häufigsten vorkommende Wert
- **Zweck:** Den häufigsten Zahlenwert oder die häufigste Kategorie bestimmen
- **Bsp.:** Bei 1, 2, 2, 8, 10, 12 und 15 ist der Modus 2, der Median hingegen 8
- **Interpretation:** Ein Modus von 22 kW bedeutet, dass 22 kW häufiger vorkommen als andere einzelne Ladeleistungen
- **Wichtig:** Es kann mehrere gleich häufige Modalwerte geben, auch Kategorien wie Marken besitzen einen Modus

### Quantile, Quartile und Perzentile

- **Definition:** Grenzwerte, die sortierte Daten nach bestimmten Anteilen unterteilen
- **Zweck:** Erkennen, in welchen Wertebereichen bestimmte Anteile der Beobachtungen liegen
- **Quartile:** Q1 entspricht dem 25%-Quantil, Q2 dem Median und Q3 dem 75%-Quantil
- **Perzentile:** Quantile als Prozentangabe, bspw. 95. Perzentil = 95%-Quantil
- **Bsp.:** Ein 95%-Quantil von 150 kW bedeutet, dass ungefähr 95 % der Ladeleistungen höchstens 150 kW betragen
- **Interpretation:** Hohe Quantile beschreiben den oberen Bereich der Verteilung, tiefe Quantile den unteren Bereich
- **Wichtig:** Quantile werden pro Variable separat berechnet, durch gleiche Werte und Zwischenwertberechnungen sind Prozentangaben teilweise nur ungefähr erfüllt

### Minimum, Maximum und Spannweite

- **Definition:** Minimum und Maximum sind der kleinste und grösste Wert, die Spannweite ist Maximum minus Minimum
- **Zweck:** Den gesamten beobachteten Wertebereich erkennen
- **Bsp.:** Minimum 100 CHF und Maximum 900 CHF ergeben eine Spannweite von 800 CHF
- **Interpretation:** Eine grosse Spannweite zeigt grosse Unterschiede zwischen den Extremwerten
- **Wichtig:** Die Spannweite berücksichtigt nur zwei Werte und reagiert stark auf Ausreisser

### Interquartilsabstand – IQR

- **Definition:** Q3 minus Q1, beschreibt die Breite des Wertebereichs der mittleren 50 % der Daten
- **Zweck:** Die Streuung beurteilen, ohne sich stark von Extremwerten beeinflussen zu lassen
- **Bsp.:** Q1 = 500 CHF und Q3 = 900 CHF ergeben einen IQR von 400 CHF
- **Interpretation:** Die mittleren 50 % der Preise liegen ungefähr zwischen 500 und 900 CHF
- **Vergleich:** Bei derselben Variable bedeutet ein grösserer IQR eine stärkere Streuung im mittleren Datenbereich

### Varianz

- **Definition:** Beschreibt die Streuung anhand der quadrierten Abweichungen vom Mittelwert
- **Zweck:** Unterschiede zwischen den Werten rechnerisch erfassen
- **Bsp.:** Bei Ladeleistungen in kW wird die Varianz in kW² angegeben
- **Interpretation:** Eine grössere Varianz bedeutet bei gleicher Einheit und vergleichbarer Skala eine stärkere Streuung
- **Wichtig:** Empfindlich gegenüber Ausreissern, die Standardabweichung ist meist leichter verständlich
- **Zusatz:** Für die Stichprobenvarianz wird durch n − 1 geteilt, wobei n die Anzahl gültiger Werte ist

### Standardabweichung

- **Definition:** Quadratwurzel der Varianz, beschreibt, wie stark die Werte um ihren Mittelwert streuen
- **Zweck:** Die Streuung in derselben Einheit wie die ursprünglichen Werte ausdrücken
- **Bsp.:** Zwei Gruppen haben beide einen Mittelwert von 50 kW, bei einer Standardabweichung von 5 kW liegen die Werte enger zusammen als bei 40 kW
- **Interpretation:** Es gibt keinen allgemein gültigen Grenzwert für eine hohe Standardabweichung, Einheit, Mittelwert und Datenbereich mitberücksichtigen
- **Wichtig:** Eine Standardabweichung von 0 bedeutet identische gültige Werte, Ausreisser können sie stark erhöhen
- **Verwendung:** Gemeinsam mit Mittelwert und Histogramm beurteilen, für jede numerische Spalte entsteht eine eigene Standardabweichung

### Schiefe – Skewness

- **Definition:** Beschreibt, ob eine Verteilung einen längeren Ausläufer nach rechts oder links besitzt
- **Zweck:** Die Asymmetrie einer Verteilung beurteilen
- **Bsp.:** Viele günstige Geräte und wenige sehr teure Geräte ergeben häufig eine rechtsschiefe Preisverteilung
- **Interpretation:** Positiv bedeutet rechtsschief, negativ linksschief, nahe 0 bedeutet geringe gemessene Schiefe
- **Zusammenhang:** Bei Rechtsschiefe liegt der Mittelwert häufig über dem Median, bei Linksschiefe häufig darunter
- **Wichtig:** Diese Beziehungen sind Hinweise, keine festen Regeln, zusätzlich das Histogramm betrachten

### Kurtosis – Exzess-Kurtosis

- **Definition:** Beschreibt die Ausprägung der Verteilungsenden und den Einfluss extremer Werte
- **Zweck:** Beurteilen, wie stark Extremwerte im Vergleich zu einer Normalverteilung ausgeprägt sind
- **Bsp.:** Eine Exzess-Kurtosis von +4 deutet auf ausgeprägte Extremwerte hin
- **Interpretation:** Positive Werte sprechen für schwerere Verteilungsenden, negative für leichtere Verteilungsenden als bei einer Normalverteilung
- **Wichtig:** pandas verwendet die Exzess-Kurtosis mit dem Referenzwert 0 für die Normalverteilung, ein Ergebnis nahe 0 beweist keine Normalverteilung

### Normalverteilung

- **Definition:** Eine symmetrische, glockenförmige Verteilung mit vielen Werten in der Mitte und wenigen an den Rändern
- **Zweck:** Dient als Referenzverteilung und als Voraussetzung bestimmter statistischer Verfahren
- **Eigenschaften:** Mittelwert, Median und Modus liegen bei der theoretischen Normalverteilung an derselben Stelle
- **Interpretation:** Nur bei ungefähr normalverteilten Daten liegen etwa 68 % der Werte innerhalb von einer Standardabweichung um den Mittelwert und etwa 95 % innerhalb von zwei
- **Wichtig:** Ähnliche Mittelwerte und Mediane oder eine Schiefe nahe 0 reichen als Nachweis nicht aus

### Ausreisser und Messfehler

- **Definition:** Ein Ausreisser weicht stark von den übrigen Werten ab, ein Messfehler ist ein falsch erfasster Wert
- **Zweck:** Auffällige Beobachtungen erkennen und ihre Bedeutung prüfen
- **Bsp.:** Ein Smartphone für 12’990 CHF kann ein Fehler oder ein tatsächlich sehr teures Spezialgerät sein
- **Interpretation:** Ein auffälliger Wert ist zunächst ein Prüfhinweis, kein automatischer Löschgrund
- **Verwendung:** Rohdaten, Einheiten und fachliche Plausibilität prüfen, zusätzlich Boxplot oder Histogramm betrachten

### Ausreisser erkennen mit der IQR-Regel

- **Definition:** Werte unter Q1 − 1,5 × IQR oder über Q3 + 1,5 × IQR werden als potenzielle Ausreisser markiert
- **Zweck:** Auffällige Werte anhand nachvollziehbarer statistischer Grenzen erkennen
- **Bsp.:** Q1 = 500, Q3 = 900 und IQR = 400 ergeben die Grenzen −100 und 1’500
- **Interpretation:** Ein Preis über 1’500 wäre hier statistisch auffällig, aber nicht automatisch falsch
- **Wichtig:** Die Regel berücksichtigt keine fachlichen Besonderheiten, stark schiefe Verteilungen können viele markierte Werte besitzen

### Fehlende Werte und Imputation

- **Definition:** Fehlende Werte enthalten keine verfügbare Information, Imputation ersetzt sie durch geschätzte Werte
- **Zweck:** Fehlende Angaben untersuchen und bei geeigneter Begründung Beobachtungen für Analysen erhalten
- **Bsp.:** Ein fehlender Preis wird durch den Median der vorhandenen Preise ersetzt
- **Interpretation:** Nach der Imputation sind mehr Werte auswertbar, die ergänzten Werte sind jedoch Schätzungen
- **Wichtig:** Einfaches Ersetzen kann Streuung und Zusammenhänge verändern, Löschen kann ebenfalls verzerren
- **Zusatz:** Wenn alle Werte fehlen, lässt sich aus dieser Spalte kein Mittelwert oder Median als Ersatz berechnen

### MCAR, MAR und MNAR

- **MCAR:** Werte fehlen vollständig zufällig, bspw. durch zufälligen technischen Datenverlust
- **MAR:** Das Fehlen hängt von beobachteten Merkmalen ab, nach deren Berücksichtigung nicht zusätzlich vom fehlenden Wert selbst
- **Bsp. MAR:** Ältere Personen beantworten eine Frage seltener, das Alter ist im Datensatz vorhanden
- **MNAR:** Das Fehlen hängt auch nach Berücksichtigung beobachteter Merkmale vom fehlenden Wert selbst ab, bspw. besonders hohe Einkommen werden häufiger verschwiegen
- **Zweck:** Eine passende Behandlung fehlender Werte wählen
- **Wichtig:** Die Mechanismen lassen sich nicht allein aus der Anzahl fehlender Werte sicher bestimmen

### Kovarianz

- **Definition:** Beschreibt, ob zwei numerische Variablen tendenziell gemeinsam oder gegenläufig variieren
- **Zweck:** Die Richtung einer gemeinsamen Veränderung untersuchen
- **Bsp.:** Eine positive Kovarianz zwischen Ladeleistung und Stromstärke bedeutet, dass hohe Ladeleistungen tendenziell mit hohen Stromstärken auftreten
- **Interpretation:** Positiv bedeutet gleichgerichtet, negativ gegenläufig, eine Kovarianz von 0 bedeutet keinen linearen Zusammenhang
- **Wichtig:** Die Grösse hängt von Einheiten und Skalierung ab, ein Wert wie 15’380 ist deshalb nicht automatisch stark
- **Kovarianzmatrix:** Zeigt alle Variablenpaare, die Hauptdiagonale enthält die jeweiligen Varianzen

### Pearson-Korrelation

- **Definition:** Misst Stärke und Richtung eines linearen Zusammenhangs zwischen zwei numerischen Variablen
- **Zweck:** Zusammenhänge auf einer einheitlichen Skala von −1 bis +1 vergleichen
- **Bsp.:** r = +0,90 bedeutet einen starken positiven linearen Zusammenhang, r = −0,90 einen starken negativen
- **Interpretation:** Das Vorzeichen zeigt die Richtung, der Betrag die Stärke, Werte nahe 0 zeigen einen schwachen oder fehlenden linearen Zusammenhang
- **Wichtig:** Ausreisser können das Ergebnis stark beeinflussen, nichtlineare Beziehungen können trotz r nahe 0 bestehen
- **Voraussetzung:** Zusammengehörige Werte müssen dieselbe Beobachtung betreffen, bei konstanten Variablen ist die Korrelation nicht definiert

### Spearman-Rangkorrelation

- **Definition:** Misst Stärke und Richtung eines monotonen Zusammenhangs anhand der Rangplätze
- **Zweck:** Beziehungen untersuchen, bei denen eine Variable mit der anderen tendenziell zunimmt oder abnimmt, ohne dass eine Gerade entsteht
- **Bsp.:** Mit zunehmender Speichergrösse steigen die Preise meist an, aber nicht gleichmässig
- **Interpretation:** ρ = +0,85 bedeutet einen starken positiven monotonen Zusammenhang, ρ nahe 0 zeigt keinen oder einen schwachen monotonen Zusammenhang
- **Unterschied zu Pearson:** Spearman verwendet Rangplätze und ist meist weniger empfindlich gegenüber extremen Zahlenwerten
- **Wichtig:** Ungeordnete Kategorien wie Marken dürfen nicht willkürlich nummeriert und als Rangfolge behandelt werden

### Korrelation und Kausalität

- **Definition:** Korrelation beschreibt einen Zusammenhang, Kausalität eine Ursache-Wirkungs-Beziehung
- **Zweck:** Statistische Ergebnisse nicht überinterpretieren
- **Bsp.:** Eisverkäufe und Sonnenbrände nehmen gemeinsam zu, warmes Wetter beeinflusst beide
- **Interpretation:** Eine hohe Korrelation beweist weder eine direkte Ursache noch die Richtung einer möglichen Wirkung

### Kontingenztabelle – Kreuztabelle

- **Definition:** Zeigt, wie häufig Kombinationen kategorialer Merkmale vorkommen
- **Zweck:** Kategorien vergleichen und mögliche Zusammenhänge erkennen
- **Bsp.:** Eine Zelle mit DC und CCS = 120 bedeutet, dass 120 Beobachtungen beide Merkmale besitzen
- **Interpretation:** Zeilen und Spalten zeigen Kategorien, die Zellen ihre gemeinsamen Häufigkeiten, Gesamtsummen die berücksichtigten Beobachtungen
- **Wichtig:** Bei unterschiedlich grossen Gruppen sind Anteile oft aussagekräftiger als absolute Anzahlen

### Histogramm

- **Definition:** Fasst numerische Werte in Intervalle zusammen und zeigt deren Häufigkeiten
- **Zweck:** Verteilungsform, Streuung, Schiefe und auffällige Werte erkennen
- **Bsp.:** Viele Ladeleistungen im unteren Bereich und wenige bis 400 kW ergeben einen langen rechten Ausläufer
- **Interpretation:** Die X-Achse zeigt Wertebereiche, die Y-Achse standardmässig die Anzahl Beobachtungen je Intervall
- **Wichtig:** Die Anzahl Intervalle beeinflusst die Darstellung, mehrere Gipfel können auf unterschiedliche Gruppen hinweisen

### Boxplot

- **Definition:** Zeigt Median, Quartile, Streuung und potenzielle Ausreisser einer numerischen Variable
- **Zweck:** Verteilungen kompakt darstellen und zwischen Kategorien vergleichen
- **Interpretation:** Die Linie in der Box ist der Median, die Box reicht von Q1 bis Q3 und umfasst ungefähr die mittleren 50 % der Werte
- **Whiskers:** Reichen standardmässig bis zu den äussersten beobachteten Werten innerhalb der 1,5-IQR-Grenzen, weiter entfernte Werte erscheinen als Punkte
- **Bsp.:** Eine Gruppe mit höherem Median und grösserer Box besitzt höhere zentrale Werte und mehr Streuung im mittleren Bereich
- **Wichtig:** Eine grössere Box bedeutet mehr Streuung, nicht mehr Beobachtungen

### Violinplot

- **Definition:** Zeigt die Verteilungsform einer numerischen Variable als spiegelbildliche Dichtekurve
- **Zweck:** Neben Lage und Streuung auch Häufungen und mehrere Gipfel erkennen
- **Bsp.:** Breite Stellen bei 11 und 22 kW zeigen zwei Bereiche mit vielen Beobachtungen
- **Interpretation:** Breite Stellen bedeuten eine hohe geschätzte Dichte, schmale Stellen eine geringe
- **Wichtig:** Die Form hängt von der Glättung ab, die Breite entspricht nicht automatisch der absoluten Gruppengrösse

### Balkendiagramm

- **Definition:** Vergleicht Kategorien anhand ihrer Häufigkeiten oder berechneter Kennzahlen
- **Zweck:** Grössenunterschiede und Rangfolgen übersichtlich darstellen
- **Bsp.:** Städte nach Anzahl Ladestationen oder Marken nach Medianpreis vergleichen
- **Interpretation:** Die Balkenlänge zeigt die auf der Zahlenachse angegebene Grösse
- **Wichtig:** Prüfen, ob Anzahlen, Prozentwerte oder Kennzahlen dargestellt werden, eine Top-10-Auswahl blendet alle übrigen Kategorien aus

### Liniendiagramm

- **Definition:** Verbindet Werte entlang einer sinnvoll geordneten X-Achse
- **Zweck:** Entwicklungen und Verläufe zeigen, bspw. über die Zeit oder nach Zimmerzahl
- **Bsp.:** Der Medianpreis steigt von 2-Zimmer- zu 4-Zimmer-Wohnungen
- **Interpretation:** Steigende Abschnitte zeigen zunehmende Y-Werte, fallende Abschnitte abnehmende Y-Werte
- **Wichtig:** Die X-Werte müssen sinnvoll geordnet sein, Verbindungslinien beweisen keine gemessenen Zwischenwerte

### QQ-Plot

- **Definition:** Vergleicht die Quantile der beobachteten Daten mit denen einer theoretischen Verteilung, häufig der Normalverteilung
- **Zweck:** Prüfen, ob die Verteilungsform ungefähr zur gewählten Referenz passt
- **Bsp.:** Liegen besonders hohe Ladeleistungen deutlich über der Referenzlinie, passt der obere Datenbereich schlecht zur Normalverteilung
- **Interpretation:** Punkte ungefähr auf einer Geraden sprechen für eine passende Verteilungsform, systematische Krümmungen dagegen
- **Wichtig:** Die Referenzlinie muss weder durch 0 verlaufen noch eine Steigung von 1 besitzen, der Plot ist kein Beweis für Normalverteilung

### Scatterplot – Streudiagramm

- **Definition:** Stellt jede Beobachtung als Punkt mit zwei numerischen Werten dar
- **Zweck:** Zusammenhänge, Gruppen, Streuung und Ausreisser erkennen
- **Interpretation:** Eine steigende Punktwolke spricht für einen positiven Zusammenhang, eine fallende für einen negativen, Krümmungen für eine nichtlineare Beziehung
- **Bsp.:** Punkte weit abseits der übrigen Punktwolke können auffällige Beobachtungen sein
- **Gruppierung:** Farben können Kategorien unterscheiden, ein Gesamttrend kann innerhalb einzelner Gruppen anders aussehen

### Scatterplot-Matrix

- **Definition:** Zeigt paarweise Streudiagramme mehrerer numerischer Variablen
- **Zweck:** Mehrere mögliche Zusammenhänge gleichzeitig überblicken
- **Interpretation:** Ausserhalb der Diagonale zeigt der Spaltenname die X-Variable und der Zeilenname die Y-Variable
- **Diagonale:** Zeigt die Verteilung jeder einzelnen Variable, bspw. als Histogramm
- **Wichtig:** Gespiegelte Felder zeigen dasselbe Variablenpaar mit vertauschten Achsen, viele Variablen machen die Matrix schnell unübersichtlich

### Hexbin-Plot

- **Definition:** Fasst Beobachtungen zweier numerischer Variablen in sechseckigen Bereichen zusammen
- **Zweck:** Bei vielen überlappenden Punkten erkennen, wo sich Beobachtungen häufen
- **Bsp.:** Ein dunkles Sechseck bei 400 V und 32 A bedeutet bei einer entsprechend gewählten Farbskala viele Beobachtungen in diesem Bereich
- **Interpretation:** Die Farblegende zeigt standardmässig die Anzahl Beobachtungen je Sechseck
- **Wichtig:** Eine intensive Farbe bedeutet hohe Häufigkeit, nicht automatisch hohe X- oder Y-Werte

### Korrelationsmatrix und Heatmap

- **Definition:** Eine Korrelationsmatrix enthält die Korrelationen aller ausgewählten Variablenpaare, die Heatmap stellt sie zusätzlich farbig dar
- **Zweck:** Starke und schwache Zusammenhänge schnell erkennen
- **Bsp.:** Eine Zelle mit −0,85 zeigt einen starken negativen Zusammenhang zwischen der Zeilen- und Spaltenvariable
- **Interpretation:** Für den stärksten Zusammenhang den höchsten Betrag ausserhalb der Hauptdiagonale suchen
- **Wichtig:** Die Matrix ist symmetrisch, Selbstkorrelationen auf der Diagonale sind 1, sofern definiert
- **Farben:** Die Bedeutung hängt von der Farbskala ab, deshalb die Legende und die Zahlen lesen
