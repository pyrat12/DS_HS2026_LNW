# AP02 – Statistik

## Rules für Statistik und Interpretation

### describe() richtig lesen

| Kennzahl | Bedeutung | Anschauliches Bsp |
|---|---|---|
| count | Anzahl vorhandener Werte in dieser Spalte, fehlende Werte werden nicht mitgezählt | 142 Smartphones, aber count = 134 bei Preis → acht Smartphones haben keinen Preis |
| mean | Durchschnitt, Summe der vorhandenen Werte geteilt durch deren Anzahl | Wenige sehr teure Smartphones können den Durchschnitt stark erhöhen |
| std | Standardabweichung, beschreibt die Streuung der Werte um den Mittelwert | Grosse Preisunterschiede führen zu einer hohen Standardabweichung in CHF |
| min | Kleinster vorhandener Wert | Ein Smartphone für 0 CHF genauer prüfen, möglicherweise Fehler oder Platzhalter |
| 25% | Q1, unteres Quartil | Q1 = 479 CHF → ungefähr ein Viertel der Geräte kostet höchstens 479 CHF |
| 50% | Median, mittlerer Wert der sortierten Daten | Median = 899 CHF → ungefähr die Hälfte kostet höchstens 899 CHF |
| 75% | Q3, oberes Quartil | Q3 = 1199 CHF → ungefähr drei Viertel kosten höchstens 1199 CHF |
| max | Grösster vorhandener Wert | Maximum = 12'990 CHF → prüfen, ob dieser Preis fachlich plausibel ist |

- Fehlende Werte berechnen mit Gesamtzahl der Zeilen minus count der untersuchten Spalte
- count = 134,00 bedeutet 134 vorhandene Werte, die Nachkommastellen sind nur Ausgabeformatierung
- Zwischen Q1 und Q3 liegen ungefähr die mittleren 50 % der Werte
- Bei vielen identischen Werten können die tatsächlichen Anteile an den Quartilsgrenzen abweichen
- Kennzahlen beschreiben nur die vorhandenen Werte, fehlende Preise können das Gesamtbild verzerren

### Verteilungsform erkennen

| Beobachtung | Interpretation |
|---|---|
| Positive Schiefe, skew > 0 | Rechtsschief → bspw. viele günstige Smartphones und wenige sehr teure, die hohen Preise bilden den langen Ausläufer nach rechts und ziehen den Durchschnitt nach oben |
| Negative Schiefe, skew < 0 | Linksschief → bspw. viele hohe Prüfungsnoten und wenige sehr tiefe, die tiefen Noten bilden den langen Ausläufer nach links und ziehen den Durchschnitt nach unten |
| Schiefe nahe 0 | Geringe gemessene Asymmetrie, Symmetrie oder Normalverteilung nicht bewiesen |
| Mean > Median | Hinweis auf Rechtsschiefe bzw. Einfluss hoher Werte |
| Mean < Median | Hinweis auf Linksschiefe bzw. Einfluss tiefer Werte |
| Mean ungefähr Median | Mit Symmetrie vereinbar, beweist sie aber nicht |
| Maximum weit oberhalb von Q3 | Hohe Extremwerte genauer untersuchen |
| Minimum weit unterhalb von Q1 | Tiefe Extremwerte genauer untersuchen |

- Merksatz: Die Richtung der Schiefe zeigt, auf welcher Seite der lange Ausläufer mit den extremen Werten liegt
- Für die Begründung mehrere Hinweise kombinieren, bspw. positive Schiefe, Mean über Median und sehr hohes Maximum
- Mean und Standardabweichung nicht zur Bestimmung der Schieferichtung vergleichen
- Rechtsschief bedeutet nicht, dass die meisten Werte rechts liegen, im Smartphone-Bsp. liegen viele Preise eher links und wenige hohe Preise weit rechts
- Linksschief bedeutet nicht, dass die meisten Werte links liegen, im Noten-Bsp. liegen viele Noten eher rechts und wenige tiefe Noten weit links

#### Musterantwort aus dem Mock Exam

- Die Preise sind deutlich rechtsschief, die Schiefe ist mit 9,19 positiv
- Der Mittelwert von 992,89 CHF liegt über dem Median von 899 CHF, einzelne hohe Preise ziehen den Durchschnitt nach oben
- Die mittleren 50 % der Preise liegen ungefähr zwischen 479 und 1199 CHF
- Das Maximum von 12'990 CHF liegt weit über diesem Bereich und sollte auf einen möglichen Fehler untersucht werden
- Histogramm und Boxplot eignen sich, um Verteilungsform und auffällige Preise genauer zu prüfen

### Passendes Lagemass wählen

Ein Lagemass beschreibt, wo die Mitte oder ein typischer Wert der Daten liegt

| Fragestellung | Geeignetes Mass | Erklärung und Bsp |
|---|---|---|
| Was ist ein typischer Preis bei schiefer Verteilung oder extremen Preisen | Median | Einzelne Luxusgeräte beeinflussen ihn wenig, deshalb beschreibt er die Mitte häufig besser |
| Wie hoch ist der durchschnittliche Preis aller Geräte | Mean | Jeder Preis fliesst mit seiner tatsächlichen Höhe ein, auch sehr hohe Preise |
| Welche Marke oder welcher Wert kommt am häufigsten vor | Mode | Bspw. Samsung ist die häufigste Marke |
| Welche Marke ist durchschnittlich vertreten | Kein numerischer Mittelwert | Marken sind Kategorien ohne Zahlenabstand, stattdessen Häufigkeiten oder Anteile vergleichen |

- Bsp. Preise 100, 200 und 3000 CHF → Mean = 1100 CHF, Median = 200 CHF
- 1100 CHF ist der korrekte Durchschnitt, beschreibt aber keines der drei Geräte besonders gut
- Wird ausdrücklich nach dem Durchschnitt gefragt, Mean verwenden, auch wenn die Verteilung schief ist
- Wird nach einem typischen Wert bei Ausreissern gefragt, Median wählen und seine geringere Empfindlichkeit begründen
- Bei Gruppenvergleichen zusätzlich die Anzahl Geräte und fehlenden Preise betrachten

### Streuung interpretieren

Streuung beschreibt, wie stark sich die Werte unterscheiden, zwei Gruppen können denselben Mittelwert und trotzdem sehr unterschiedliche Preise haben

| Kennzahl oder Beobachtung | Bedeutung und Bsp |
|---|---|
| Kleine Standardabweichung | Werte liegen vergleichsweise nahe am Mittelwert, bspw. viele Geräte um 500 CHF |
| Hohe Standardabweichung | Werte unterscheiden sich stark, bspw. günstige Geräte und sehr teure Luxusgeräte gemeinsam |
| Standardabweichung = 0 | Alle vorhandenen Werte sind gleich, sofern die Kennzahl berechnet werden kann |
| Varianz | Streuungsmass in quadrierter Einheit, bei Preisen also CHF² |
| Standardabweichung | Wurzel der Varianz, wieder in ursprünglicher Einheit, bei Preisen also CHF |
| Spannweite = max − min | Abstand zwischen kleinstem und grösstem Wert, reagiert stark auf einzelne Extremwerte |
| IQR = Q3 − Q1 | Breite des Bereichs mit den mittleren 50 % der Werte, weniger empfindlich gegenüber Extremwerten |

- Bsp. Q1 = 479 CHF und Q3 = 1199 CHF → IQR = 720 CHF
- Eine hohe std allein zeigt keine Rechts- oder Linksschiefe, auch eine symmetrische Verteilung kann stark streuen
- std darf grösser als mean sein, das ist kein automatischer Rechenfehler
- Mean ± std ist kein garantierter Wertebereich, einzelne Werte können deutlich ausserhalb liegen
- Ob eine std gross ist, hängt von der Grössenordnung ab, 100 CHF Streuung ist bei 200 CHF anders einzuordnen als bei 20'000 CHF
- Streuungen unterschiedlicher Einheiten nicht direkt vergleichen, bspw. Preis in CHF und Fläche in m²
- pandas berechnet std und Varianz standardmässig mit ddof=1, bei der Varianz wird durch n − 1 geteilt

### Ausreisser erkennen und behandeln

Ein Ausreisser ist ein auffälliger Wert im Vergleich zu den übrigen Daten, er kann ein echter Extremfall oder ein Datenfehler sein

| Berechnung | Bedeutung |
|---|---|
| IQR = Q3 − Q1 | Streuung der mittleren 50 % |
| Untere Grenze = Q1 − 1,5 × IQR | Kleinere Werte gelten nach dieser Regel als potenzielle Ausreisser |
| Obere Grenze = Q3 + 1,5 × IQR | Grössere Werte gelten nach dieser Regel als potenzielle Ausreisser |

#### Bsp. Mock Exam

- Q1 = 479 und Q3 = 1199 → IQR = 720 CHF
- Obere Grenze = 1199 + 1,5 × 720 = 2279 CHF
- Der Preis von 12'990 CHF liegt darüber und wird zur Prüfung markiert
- Dass dasselbe Gerät in anderer Farbe 1299 CHF kostet, ist ein zusätzlicher Hinweis auf einen Erfassungsfehler
- Die IQR-Regel allein beweist diesen Fehler noch nicht

#### Vorgehen bei einer Aufgabe

1. Auffällige Werte mit describe(), Boxplot und sortierter Ausgabe finden
2. Betroffene Datensätze vollständig ansehen, inkl. Produkt, Einheit und weiteren Merkmalen
3. Mit ähnlichen Datensätzen oder der Quelle vergleichen
4. Bestätigte Fehler korrigieren oder begründet entfernen
5. Echte Extremwerte grundsätzlich behalten, sofern die Aufgabenstellung nichts anderes verlangt

- Ein Luxusgerät kann tatsächlich sehr teuer sein, ein hoher Preis allein rechtfertigt keine Löschung
- Ein robuster Median verringert den Einfluss des Werts, behebt aber keinen Datenfehler
- Bei Unsicherheit Ergebnisse mit und ohne auffälligen Wert vergleichen und den Unterschied erklären

### Kurtosis interpretieren

Kurtosis beschreibt, wie stark extreme Abweichungen vom Mittelwert ins Gewicht fallen, sie zeigt keine Richtung wie die Schiefe

| pandas-Ergebnis | Einfache Interpretation |
|---|---|
| Kurtosis > 0 | Hinweis auf stärker ausgeprägte Extremwerte bzw. schwerere Verteilungsenden als bei der Normalverteilung |
| Kurtosis < 0 | Hinweis auf weniger ausgeprägte Extremwerte bzw. leichtere Verteilungsenden |
| Kurtosis ungefähr 0 | Ähnliche Kurtosis wie die Normalverteilung, deshalb aber nicht automatisch normalverteilt |

- pandas verwendet Exzess-Kurtosis, die Normalverteilung hat den Referenzwert 0 statt 3
- Bsp. Viele normale Smartphone-Preise und einzelne extrem hohe Preise können zu hoher Kurtosis führen
- Hohe Kurtosis bedeutet nicht, dass eine bestimmte Anzahl Ausreisser vorliegt
- Kurtosis und Schiefe zusammen mit Histogramm oder QQ-Plot beurteilen, einzelne Kennzahlen beweisen keine Normalverteilung

### Fehlende Werte und Verzerrung

Entscheidend ist, warum Werte fehlen, davon hängt ab, ob die vorhandenen Daten ein verzerrtes Bild zeigen

| Mechanismus | Einfache Erklärung | Anschauliches Bsp |
|---|---|---|
| MCAR | Das Fehlen ist unabhängig vom fehlenden Wert und von anderen Merkmalen | Ein technischer Zufallsfehler löscht Preise unabhängig von Marke und Preishöhe |
| MAR | Das Fehlen hängt mit bekannten Merkmalen zusammen, nach deren Berücksichtigung nicht zusätzlich mit dem unbekannten Wert | Bei einer Marke fehlen Preise häufiger, innerhalb der jeweiligen Marke hängt das Fehlen nicht zusätzlich von der Preishöhe ab |
| MNAR | Das Fehlen hängt auch nach Berücksichtigung bekannter Merkmale vom fehlenden Wert selbst ab | Besonders teure Geräte zeigen gezielt «Preis auf Anfrage», gerade wegen ihres hohen Preises fehlt die Zahl |

#### Was bedeutet das für die Auswertung

- Fehlen vor allem hohe Preise, wird der Durchschnitt der vorhandenen Preise tendenziell zu tief
- Fehlen vor allem tiefe Preise, wird der Durchschnitt der vorhandenen Preise tendenziell zu hoch
- Bei MCAR führt Löschen hauptsächlich zu weniger Daten und ungenaueren Ergebnissen, grundsätzlich aber nicht zu einer systematischen Auswahl bestimmter Preise
- Unterschiedliche Fehlendenanteile je Marke sind ein Hinweis auf einen Zusammenhang, beweisen allein aber keinen Mechanismus
- MAR und MNAR lassen sich aus den beobachteten Daten allein im Allgemeinen nicht sicher unterscheiden

#### Wichtige Regeln

- Fehlend bedeutet unbekannt, nicht 0 CHF
- Mean und Median ignorieren fehlende Werte standardmässig, deshalb immer count prüfen
- Imputation ergänzt Schätzwerte und stellt die tatsächlichen unbekannten Werte nicht wieder her
- Mittelwert-Imputation häuft Werte beim Mittelwert an und verringert typischerweise die Streuung
- Ein durch fehlende hohe Preise verzerrter Durchschnitt wird durch einfache Mittelwert-Imputation nicht automatisch korrigiert

### Korrelation und Kovarianz verstehen

Korrelation beschreibt, ob zwei Variablen gemeinsam steigen oder fallen und wie deutlich dieser Zusammenhang ist

| Ergebnis | Bedeutung | Anschauliches Bsp |
|---|---|---|
| Positive Korrelation | Höheres X geht tendenziell mit höherem Y einher | Geräte mit mehr Speicher kosten tendenziell mehr |
| Negative Korrelation | Höheres X geht tendenziell mit tieferem Y einher | Ältere Geräte haben tendenziell einen tieferen Wiederverkaufspreis |
| Pearson r nahe 0 | Wenig linearer Zusammenhang, ein anderes Muster kann trotzdem bestehen | Bei einem U-förmigen Zusammenhang kann r nahe 0 liegen |
| Betrag von Pearson r nahe 1 | Starker Zusammenhang entlang einer steigenden oder fallenden Geraden | Mit zunehmender Fläche steigt die Miete sehr gleichmässig |
| Spearman nahe +1 oder −1 | Starker Zusammenhang in durchgehend derselben Richtung, auch entlang einer Kurve | Mehr Speicher geht meist mit höherem Preis einher, aber nicht mit einem konstanten Preisaufschlag |

#### Pearson oder Spearman

| Methode | Wann passend | Worauf achten |
|---|---|---|
| Pearson | Linearen Zusammenhang numerischer Werte untersuchen | Verwendet tatsächliche Werte und reagiert empfindlich auf Extremwerte |
| Spearman | Rangfolge oder monotonen Zusammenhang untersuchen | Verwendet Rangplätze, ist weniger empfindlich gegenüber der Grösse einzelner Extremwerte |
| Kovarianz | Gemeinsame Veränderungsrichtung beschreiben | Vorzeichen ist interpretierbar, der Betrag hängt aber von Einheiten und Skalierung ab |

- Linear bedeutet ungefähr entlang einer Geraden
- Monoton bedeutet, dass Y mit steigendem X überwiegend in dieselbe Richtung geht, die Stärke der Veränderung darf variieren
- Bsp. Preis steigt mit Speicher zuerst stark und später nur noch wenig → monoton, aber nicht unbedingt linear
- Stärke anhand des Betrags vergleichen, −0,9 ist stärker als +0,4, das Minus zeigt nur die Richtung
- Korrelationswerte liegen zwischen −1 und +1, Kovarianzen haben diese feste Begrenzung nicht
- In einer Korrelationsmatrix die Hauptdiagonale mit 1 ignorieren, jede Variable korreliert dort mit sich selbst
- Korrelation 0 bedeutet nicht automatisch Unabhängigkeit, bspw. kann ein gekrümmter Zusammenhang bestehen
- Immer den Scatterplot ansehen, einzelne Extremwerte oder getrennte Gruppen können den Korrelationswert stark beeinflussen
- pandas verwendet bei der üblichen Korrelationsberechnung pro Variablenpaar die gemeinsam vorhandenen Werte, deshalb kann die Anzahl je Paar unterschiedlich sein

### Korrelation in einer Antwort formulieren

Eine gute Antwort nennt Richtung, Stärke, Bedeutung im Datensatz und eine Einschränkung

- Bsp. Spearman = 0,76 → deutlicher positiver monotoner Zusammenhang
- Passende Antwort → Teurere Smartphones werden tendenziell besser bewertet, die Rangfolgen von Preis und Bewertung stimmen relativ deutlich überein
- Einschränkung → Daraus folgt nicht, dass ein höherer Preis bessere Bewertungen verursacht, bspw. könnte die Ausstattung beide beeinflussen
- 0,76 bedeutet weder 76 % bessere Bewertungen noch 76 % richtige Vorhersagen
- Bei Pearson zusätzlich prüfen, ob die Punkte tatsächlich ungefähr einem geraden Verlauf folgen
- Begriffe wie schwach oder stark hängen auch vom Anwendungsgebiet ab, keine starren Grenzen als Naturgesetz behandeln

### Gruppen vergleichen

Formulierungen wie «pro Marke», «je Kategorie» oder «für jeden Status» verlangen meist eine getrennte Auswertung mit groupby()

| Frage | Passende Auswertung | Bsp |
|---|---|---|
| Wie viele Geräte gehören zu jeder Marke | size() | Alle Geräte je Marke zählen, auch solche ohne Preis |
| Wie viele Geräte haben je Marke einen bekannten Preis | count() auf der Preisspalte | Nur vorhandene Preise je Marke zählen |
| Welche Marke hat den höchsten durchschnittlichen Preis | mean() je Marke | Durchschnitt aller bekannten Preise je Marke |
| Welche Marke hat den höchsten typischen Preis bei Ausreissern | median() je Marke | Medianpreise vergleichen |
| Wie unterscheiden sich Preisverteilungen je Marke | Gruppierte Boxplots | Median, Streuung und potenzielle Ausreisser gemeinsam vergleichen |

- Kleine Gruppen vorsichtig interpretieren, ein Durchschnitt aus zwei Geräten ist weniger breit abgestützt als einer aus 100 Geräten
- Ein einzelnes sehr teures Gerät kann den Mittelwert einer kleinen Marke stark erhöhen
- Gleiche Gruppengrösse bedeutet nicht gleiche Anzahl bekannter Preise, fehlende Werte zusätzlich prüfen
- Den Gesamtmittelwert nicht einfach aus den Gruppenmittelwerten berechnen, wenn die Gruppen unterschiedlich gross sind
- Bsp. Marke A hat einen bekannten Preis von 100 CHF, Marke B neun bekannte Preise von je 1000 CHF → Gesamtmittelwert = 910 CHF, nicht 550 CHF
- Beim Zusammenführen von Preis-Mittelwerten mit der Anzahl bekannter Preise gewichten, nicht mit Zeilen ohne Preis
- Gruppenmediane lassen sich durch einfaches Mitteln nicht zu einem Gesamtmedian zusammenführen

### Kreuztabellen mit crosstab() lesen

Eine Kreuztabelle zeigt, wie häufig Kombinationen zweier Kategorien vorkommen

- Bsp. Zeilen = Marke, Spalten = günstig, mittel und teuer
- Die Zelle Apple und teuer zählt, wie viele Apple-Geräte zur teuren Klasse gehören
- Standardmässig stehen absolute Anzahlen in den Zellen
- Kategorienkombinationen mit fehlenden Ausgangswerten werden standardmässig nicht mitgezählt

| Darstellung | Bedeutung | Bsp |
|---|---|---|
| Ohne normalize | Absolute Anzahl je Kombination | 20 Apple-Geräte sind teuer |
| normalize='index' | Anteile innerhalb jeder Zeile, Zeile summiert sich zu 1 | Welcher Anteil der Apple-Geräte ist teuer |
| normalize='columns' | Anteile innerhalb jeder Spalte, Spalte summiert sich zu 1 | Welcher Anteil aller teuren Geräte stammt von Apple |
| normalize='all' | Anteile bezogen auf alle berücksichtigten Geräte | Welcher Anteil aller Geräte ist gleichzeitig Apple und teuer |

- Für Prozentwerte die normalisierten Anteile mit 100 multiplizieren
- 20 teure Apple-Geräte und zehn teure Sony-Geräte bedeuten nicht automatisch einen höheren teuren Anteil bei Apple
- Bsp. 20 von 100 Apple-Geräten = 20 %, zehn von 20 Sony-Geräten = 50 %
- Für einen fairen Vergleich unterschiedlich grosser Marken häufig Zeilenprozente verwenden, sofern Marken in den Zeilen stehen

### cut und qcut unterscheiden

Beide Methoden teilen numerische Werte in Klassen ein, sie verwenden aber unterschiedliche Regeln für die Grenzen

| Methode | Wann verwenden | Anschauliches Bsp |
|---|---|---|
| cut mit Anzahl Klassen | Zahlenbereich in ungefähr gleich breite Intervalle aufteilen | Preise zwischen 0 und 1500 CHF in drei etwa gleich breite Preisbereiche einteilen |
| cut mit vorgegebenen Grenzen | Fachlich festgelegte Preisbereiche verwenden | Günstig bis 500 CHF, mittel bis 1000 CHF, teuer darüber |
| qcut | Ungefähr gleich viele Beobachtungen pro Klasse erhalten | 120 Geräte nach Preis in drei Gruppen mit ungefähr je 40 Geräten einteilen |

- cut → Breite der Wertebereiche ist entscheidend, die Anzahl Geräte pro Bereich kann stark variieren
- qcut → Anzahl Geräte pro Gruppe ist entscheidend, die Preisbereiche können unterschiedlich breit sein
- Bei rechtsschiefen Preisen kann die teuerste qcut-Klasse einen sehr grossen Preisbereich abdecken
- Fehlende Preise erhalten keine Klasse und fehlen normalerweise in anschliessenden Häufigkeitstabellen
- Viele identische Preise können bei qcut zu gleichen Klassengrenzen führen, eventuell sind weniger unterschiedliche Klassen möglich
- Labels müssen zur tatsächlich erzeugten Anzahl Klassen passen
- Bei vorgegebenen cut-Grenzen prüfen, ob alle relevanten Werte abgedeckt sind und welche Intervallgrenze eingeschlossen wird

### Passende Grafik wählen

| Grafik | Wann verwenden | Anschauliches Bsp |
|---|---|---|
| Histogramm | Verteilung einer numerischen Variable untersuchen, häufige Wertebereiche, Schiefe und mehrere Gipfel erkennen | Wie verteilen sich Smartphone-Preise, gibt es viele günstige und wenige sehr teure Geräte → x-Achse Preisintervalle, y-Achse Anzahl Geräte |
| Boxplot | Median, mittlere 50 % der Werte, Streuung und potenzielle Ausreisser kompakt darstellen | Wie hoch ist der typische Smartphone-Preis und gibt es auffällig teure Geräte → Box zeigt Q1 bis Q3, Linie zeigt Median, Punkte ausserhalb der Whisker sind potenzielle Ausreisser |
| Gruppierte Boxplots | Eine numerische Variable zwischen Kategorien vergleichen | Sind Apple-Geräte typischerweise teurer als Samsung-Geräte und welche Marke hat die grössere Preisstreuung → ein Preis-Boxplot pro Marke |
| QQ-Plot | Prüfen, ob Daten ungefähr einer theoretischen Verteilung folgen, häufig der Normalverteilung | Sind Modellresiduen ungefähr normalverteilt → Punkte nahe der Referenzgeraden sprechen dafür, systematische Abweichungen dagegen |
| Balkendiagramm | Anzahlen oder Prozentanteile verschiedener Kategorien vergleichen | Welche Marke kommt am häufigsten vor → x-Achse Marken, y-Achse Anzahl Smartphones |
| Scatterplot | Zusammenhang zwischen zwei numerischen Variablen untersuchen, inkl. Richtung, Form und auffälligen Einzelwerten | Haben teurere Smartphones bessere Bewertungen → x-Achse Preis, y-Achse Bewertung, jeder Punkt ist ein Smartphone |
| Hexbin | Zusammenhang zwischen zwei numerischen Variablen bei sehr vielen überlagerten Punkten untersuchen | Wo häufen sich bei 100 000 Wohnungen bestimmte Kombinationen aus Wohnfläche und Miete → x-Achse Fläche, y-Achse Miete, Farbe zeigt Anzahl Wohnungen je Sechseck |
| Scatterplot-Matrix | Mehrere numerische Variablen gleichzeitig auf paarweise Zusammenhänge prüfen | Welche Zusammenhänge bestehen zwischen Preis, Speicher, Bildschirmgrösse und Akkukapazität → je Variablenpaar ein Scatterplot |
| Heatmap | Eine Korrelationsmatrix anhand von Farben schnell überblicken | Welche Merkmale hängen am stärksten mit dem Preis zusammen → Zeilen und Spalten zeigen Variablen, Farbe und Zahlen zeigen die Korrelationswerte |
| Liniendiagramm | Entwicklung über die Zeit und zeitliche Trends darstellen | Wie verändert sich der durchschnittliche Smartphone-Preis pro Monat → x-Achse Monate in zeitlicher Reihenfolge, y-Achse Durchschnittspreis |

### Schnelle Entscheidung an der Prüfung

- Eine numerische Spalte, bspw. Preis → Histogramm für die Verteilungsform, Boxplot für Median, Streuung und potenzielle Ausreisser
- Eine Kategorie und eine numerische Spalte, bspw. Marke und Preis → gruppierte Boxplots zum Vergleich der Preisverteilungen
- Häufigkeiten einer Kategorie, bspw. Anzahl je Marke → Balkendiagramm
- Zwei numerische Spalten, bspw. Preis und Bewertung → Scatterplot, bei starker Überlagerung Hexbin
- Viele numerische Spalten → Scatterplot-Matrix für Form und Einzelpunkte der Zusammenhänge, Korrelations-Heatmap für einen kompakten Überblick der Korrelationswerte
- Datum und numerischer Wert, bspw. Monat und Durchschnittspreis → Liniendiagramm
- Frage nach Normalverteilung → QQ-Plot, Histogramm ergänzend zur Beurteilung der Verteilungsform

### Wichtige Unterschiede

- Histogramm zeigt Zahlenintervalle, Balkendiagramm zeigt Kategorien
- Boxplot zeigt keine Gipfel, dafür ist ein Histogramm geeigneter
- Scatterplot zeigt auch gekrümmte Zusammenhänge, eine Pearson-Korrelationsmatrix fasst nur lineare Zusammenhänge zusammen
- Punkte ausserhalb der Boxplot-Whisker sind potenzielle Ausreisser, nicht automatisch Datenfehler

### Grafiken richtig interpretieren

| Grafik | Was konkret prüfen |
|---|---|
| Histogramm | Wo liegen viele Werte, auf welcher Seite ist der lange Ausläufer, gibt es einen oder mehrere Gipfel |
| Boxplot | Wo liegt der Median, wie breit ist der mittlere Wertebereich, gibt es auffällige Punkte ausserhalb der Whisker |
| Gruppierte Boxplots | Welche Gruppe hat den höheren Median, welche streut stärker, welche zeigt potenzielle Ausreisser |
| QQ-Plot | Folgen die Punkte ungefähr der Referenzgeraden oder weichen sie systematisch davon ab |
| Scatterplot | Steigen oder fallen die Punkte, ist das Muster gerade oder gekrümmt, gibt es Gruppen oder auffällige Einzelpunkte |
| Korrelations-Heatmap | Welche Variablenpaare haben grosse positive oder negative Werte, die Hauptdiagonale nicht berücksichtigen |

- Histogramm → Der höchste Balken zeigt das häufigste Preisintervall, nicht das teuerste Gerät
- Histogramm → Wenige breite Intervalle können Details verbergen, viele schmale Intervalle können zufällige Schwankungen betonen
- Histogramm → Bei unterschiedlich breiten Intervallen Häufigkeitsdichte und Balkenfläche beachten, Höhen allein können täuschen
- Mehrere Gipfel → Können auf gemischte Gruppen hinweisen, bspw. günstige Einsteigergeräte und teure Premiumgeräte
- Boxplot → Die Whisker enden beim üblichen 1,5-IQR-Verfahren am äussersten noch innerhalb der Grenzen liegenden Wert, nicht zwingend bei Minimum und Maximum
- Boxplot → Eine längere Box bedeutet einen grösseren IQR, also stärkere Streuung der mittleren 50 %
- QQ-Plot → Einzelne Abweichungen an den Enden können auf ungewöhnlich tiefe oder hohe Werte hinweisen
- Scatterplot → Ein einzelner extremer Punkt kann den Pearson-Wert stark verändern, deshalb Kennzahl und Grafik gemeinsam beurteilen

### Schneller Ablauf bei Interpretationsaufgaben

1. Variable und Einheit nennen → Bspw. Smartphone-Preise in CHF
2. Datenbasis prüfen → Wie viele Werte sind vorhanden und wie viele fehlen
3. Mitte beschreiben → Mean und Median nennen und ihren Unterschied erklären
4. Streuung einordnen → std, Quartile oder IQR verwenden und auf die Einheit beziehen
5. Verteilungsform beurteilen → Schiefe, Mean-Median-Vergleich und Grafik kombinieren
6. Auffällige Werte prüfen → Minimum und Maximum fachlich einordnen, Fehler nicht allein aus der Grösse ableiten
7. Passende Grafik nennen → Bspw. Histogramm für die Form und Boxplot für potenzielle Ausreisser
8. Konkrete Schlussfolgerung formulieren → Bspw. Median beschreibt den typischen Preis besser, weil einzelne hohe Preise den Durchschnitt anheben

- Nur die zur Frage passenden Punkte verwenden, keine unnötige Aufzählung aller Kennzahlen
- Aussagen mit konkreten Zahlen begründen, statt nur «stark gestreut» oder «rechtsschief» zu schreiben
- Beobachtung und Erklärung unterscheiden, ein auffälliger Preis ist sichtbar, ein Erfassungsfehler muss zusätzlich begründet werden

## Begriffe
### Statistische Grundbegriffe

- **Beobachtung:** Ein einzelner untersuchter Fall, bspw. ein Smartphone oder eine Ladestation
- **Variable:** Ein Merkmal einer Beobachtung, bspw. Preis, Marke oder Ladeleistung
- **Numerische Variable:** Zahlen mit sinnvoller rechnerischer Bedeutung, bspw. Preis oder Leistung
- **Kategoriale Variable:** Gruppenzugehörigkeit, bspw. Marke oder Steckertyp
- **Wichtig:** Eine numerische ID identifiziert einen Datensatz, ihr Mittelwert ist normalerweise nicht sinnvoll

### Stichprobe und Grundgesamtheit

- **Definition:** Die Grundgesamtheit umfasst alle interessierenden Beobachtungen, die Stichprobe ist der tatsächlich untersuchte Teil
- **Zweck:** Ergebnisse einer Stichprobe helfen, Aussagen über eine grössere Gruppe zu treffen
- **Bsp.:** 500 untersuchte Ladestationen sind eine Stichprobe aller Ladestationen der Schweiz
- **Wichtig:** Eine grosse Stichprobe ist nicht automatisch repräsentativ, die Auswahl der Beobachtungen ist entscheidend

### Univariate und multivariate Analyse

- **Univariat:** Eine Variable untersuchen, bspw. die Preisverteilung
- **Bivariat:** Zwei Variablen gemeinsam untersuchen, bspw. Preis und Speichergrösse
- **Multivariat:** Mehrere Variablen gemeinsam untersuchen, bspw. Preis, Speichergrösse und Marke
- **Verwendung:** Zuerst einzelne Variablen verstehen, danach Zusammenhänge und Gruppenunterschiede untersuchen

### Absolute und relative Häufigkeit

- **Definition:** Die absolute Häufigkeit zählt Beobachtungen, die relative Häufigkeit beschreibt ihren Anteil
- **Zweck:** Erkennen, welche Werte oder Kategorien häufig oder selten vorkommen
- **Bsp.:** 20 von 100 Geräten sind von Apple, absolute Häufigkeit 20, relative Häufigkeit 20 %
- **Interpretation:** Eine relative Häufigkeit von 0,25 entspricht 25 %
- **Wichtig:** Bei Prozentwerten immer prüfen, worauf sich der Anteil bezieht, bspw. alle Geräte oder nur Geräte einer Marke

### Anzahl gültiger Werte

- **Definition:** Anzahl der tatsächlich vorhandenen Werte einer Variable, fehlende Werte werden nicht mitgezählt
- **Zweck:** Beurteilen, auf wie vielen Beobachtungen eine Kennzahl beruht
- **Bsp.:** 100 Geräte, davon 8 ohne Preis, die Preisstatistik beruht auf 92 gültigen Preisen
- **Interpretation:** Unterschiedliche Anzahlen pro Spalte weisen auf unterschiedlich viele fehlende Werte hin

### Mittelwert – Mean

- **Definition:** Summe aller gültigen Werte geteilt durch ihre Anzahl
- **Zweck:** Den durchschnittlichen Wert beschreiben
- **Bsp.:** Die Werte 10, 20 und 30 haben den Mittelwert 20
- **Interpretation:** Ein durchschnittlicher Preis von 800 CHF bedeutet nicht, dass die meisten Geräte 800 CHF kosten
- **Wichtig:** Sehr hohe oder tiefe Werte können den Mittelwert stark beeinflussen, deshalb mit Median und Histogramm vergleichen

### Median – Zentralwert

- **Definition:** Der mittlere Wert der sortierten Daten, bei einer geraden Anzahl der Durchschnitt der beiden mittleren Werte
- **Zweck:** Einen zentralen Wert beschreiben, der weniger empfindlich gegenüber Ausreissern ist
- **Bsp.:** Bei 10, 20 und 300 ist der Median 20, der Mittelwert hingegen 110
- **Interpretation:** Ein Median von 800 CHF bedeutet, dass mindestens die Hälfte der gültigen Preise höchstens 800 CHF beträgt und mindestens die Hälfte mindestens 800 CHF
- **Verwendung:** Besonders hilfreich bei schiefen Verteilungen und extremen Werten

### Modus – Modalwert

- **Definition:** Der am häufigsten vorkommende Wert
- **Zweck:** Den häufigsten Zahlenwert oder die häufigste Kategorie bestimmen
- **Bsp.:** Bei 1, 2, 2, 8, 10, 12 und 15 ist der Modus 2, der Median hingegen 8
- **Interpretation:** Ein Modus von 22 kW bedeutet, dass 22 kW häufiger vorkommen als andere einzelne Ladeleistungen
- **Wichtig:** Es kann mehrere gleich häufige Modalwerte geben, auch Kategorien wie Marken besitzen einen Modus

### Quantile, Quartile und Perzentile

- **Definition:** Grenzwerte, die sortierte Daten nach bestimmten Anteilen unterteilen
- **Zweck:** Erkennen, in welchen Wertebereichen bestimmte Anteile der Beobachtungen liegen
- **Quartile:** Q1 entspricht dem 25%-Quantil, Q2 dem Median und Q3 dem 75%-Quantil
- **Perzentile:** Quantile als Prozentangabe, bspw. 95. Perzentil = 95%-Quantil
- **Bsp.:** Ein 95%-Quantil von 150 kW bedeutet, dass ungefähr 95 % der Ladeleistungen höchstens 150 kW betragen
- **Interpretation:** Hohe Quantile beschreiben den oberen Bereich der Verteilung, tiefe Quantile den unteren Bereich
- **Wichtig:** Quantile werden pro Variable separat berechnet, durch gleiche Werte und Zwischenwertberechnungen sind Prozentangaben teilweise nur ungefähr erfüllt

### Minimum, Maximum und Spannweite

- **Definition:** Minimum und Maximum sind der kleinste und grösste Wert, die Spannweite ist Maximum minus Minimum
- **Zweck:** Den gesamten beobachteten Wertebereich erkennen
- **Bsp.:** Minimum 100 CHF und Maximum 900 CHF ergeben eine Spannweite von 800 CHF
- **Interpretation:** Eine grosse Spannweite zeigt grosse Unterschiede zwischen den Extremwerten
- **Wichtig:** Die Spannweite berücksichtigt nur zwei Werte und reagiert stark auf Ausreisser

### Interquartilsabstand – IQR

- **Definition:** Q3 minus Q1, beschreibt die Breite des Wertebereichs der mittleren 50 % der Daten
- **Zweck:** Die Streuung beurteilen, ohne sich stark von Extremwerten beeinflussen zu lassen
- **Bsp.:** Q1 = 500 CHF und Q3 = 900 CHF ergeben einen IQR von 400 CHF
- **Interpretation:** Die mittleren 50 % der Preise liegen ungefähr zwischen 500 und 900 CHF
- **Vergleich:** Bei derselben Variable bedeutet ein grösserer IQR eine stärkere Streuung im mittleren Datenbereich

### Varianz

- **Definition:** Beschreibt die Streuung anhand der quadrierten Abweichungen vom Mittelwert
- **Zweck:** Unterschiede zwischen den Werten rechnerisch erfassen
- **Bsp.:** Bei Ladeleistungen in kW wird die Varianz in kW² angegeben
- **Interpretation:** Eine grössere Varianz bedeutet bei gleicher Einheit und vergleichbarer Skala eine stärkere Streuung
- **Wichtig:** Empfindlich gegenüber Ausreissern, die Standardabweichung ist meist leichter verständlich
- **Zusatz:** Für die Stichprobenvarianz wird durch n − 1 geteilt, wobei n die Anzahl gültiger Werte ist

### Standardabweichung

- **Definition:** Quadratwurzel der Varianz, beschreibt, wie stark die Werte um ihren Mittelwert streuen
- **Zweck:** Die Streuung in derselben Einheit wie die ursprünglichen Werte ausdrücken
- **Bsp.:** Zwei Gruppen haben beide einen Mittelwert von 50 kW, bei einer Standardabweichung von 5 kW liegen die Werte enger zusammen als bei 40 kW
- **Interpretation:** Es gibt keinen allgemein gültigen Grenzwert für eine hohe Standardabweichung, Einheit, Mittelwert und Datenbereich mitberücksichtigen
- **Wichtig:** Eine Standardabweichung von 0 bedeutet identische gültige Werte, Ausreisser können sie stark erhöhen
- **Verwendung:** Gemeinsam mit Mittelwert und Histogramm beurteilen, für jede numerische Spalte entsteht eine eigene Standardabweichung

### Schiefe – Skewness

- **Definition:** Beschreibt, ob eine Verteilung einen längeren Ausläufer nach rechts oder links besitzt
- **Zweck:** Die Asymmetrie einer Verteilung beurteilen
- **Bsp.:** Viele günstige Geräte und wenige sehr teure Geräte ergeben häufig eine rechtsschiefe Preisverteilung
- **Interpretation:** Positiv bedeutet rechtsschief, negativ linksschief, nahe 0 bedeutet geringe gemessene Schiefe
- **Zusammenhang:** Bei Rechtsschiefe liegt der Mittelwert häufig über dem Median, bei Linksschiefe häufig darunter
- **Wichtig:** Diese Beziehungen sind Hinweise, keine festen Regeln, zusätzlich das Histogramm betrachten

### Kurtosis – Exzess-Kurtosis

- **Definition:** Beschreibt die Ausprägung der Verteilungsenden und den Einfluss extremer Werte
- **Zweck:** Beurteilen, wie stark Extremwerte im Vergleich zu einer Normalverteilung ausgeprägt sind
- **Bsp.:** Eine Exzess-Kurtosis von +4 deutet auf ausgeprägte Extremwerte hin
- **Interpretation:** Positive Werte sprechen für schwerere Verteilungsenden, negative für leichtere Verteilungsenden als bei einer Normalverteilung
- **Wichtig:** pandas verwendet die Exzess-Kurtosis mit dem Referenzwert 0 für die Normalverteilung, ein Ergebnis nahe 0 beweist keine Normalverteilung

### Normalverteilung

- **Definition:** Eine symmetrische, glockenförmige Verteilung mit vielen Werten in der Mitte und wenigen an den Rändern
- **Zweck:** Dient als Referenzverteilung und als Voraussetzung bestimmter statistischer Verfahren
- **Eigenschaften:** Mittelwert, Median und Modus liegen bei der theoretischen Normalverteilung an derselben Stelle
- **Interpretation:** Nur bei ungefähr normalverteilten Daten liegen etwa 68 % der Werte innerhalb von einer Standardabweichung um den Mittelwert und etwa 95 % innerhalb von zwei
- **Wichtig:** Ähnliche Mittelwerte und Mediane oder eine Schiefe nahe 0 reichen als Nachweis nicht aus

### Ausreisser und Messfehler

- **Definition:** Ein Ausreisser weicht stark von den übrigen Werten ab, ein Messfehler ist ein falsch erfasster Wert
- **Zweck:** Auffällige Beobachtungen erkennen und ihre Bedeutung prüfen
- **Bsp.:** Ein Smartphone für 12’990 CHF kann ein Fehler oder ein tatsächlich sehr teures Spezialgerät sein
- **Interpretation:** Ein auffälliger Wert ist zunächst ein Prüfhinweis, kein automatischer Löschgrund
- **Verwendung:** Rohdaten, Einheiten und fachliche Plausibilität prüfen, zusätzlich Boxplot oder Histogramm betrachten

### Ausreisser erkennen mit der IQR-Regel

- **Definition:** Werte unter Q1 − 1,5 × IQR oder über Q3 + 1,5 × IQR werden als potenzielle Ausreisser markiert
- **Zweck:** Auffällige Werte anhand nachvollziehbarer statistischer Grenzen erkennen
- **Bsp.:** Q1 = 500, Q3 = 900 und IQR = 400 ergeben die Grenzen −100 und 1’500
- **Interpretation:** Ein Preis über 1’500 wäre hier statistisch auffällig, aber nicht automatisch falsch
- **Wichtig:** Die Regel berücksichtigt keine fachlichen Besonderheiten, stark schiefe Verteilungen können viele markierte Werte besitzen

### Fehlende Werte und Imputation

- **Definition:** Fehlende Werte enthalten keine verfügbare Information, Imputation ersetzt sie durch geschätzte Werte
- **Zweck:** Fehlende Angaben untersuchen und bei geeigneter Begründung Beobachtungen für Analysen erhalten
- **Bsp.:** Ein fehlender Preis wird durch den Median der vorhandenen Preise ersetzt
- **Interpretation:** Nach der Imputation sind mehr Werte auswertbar, die ergänzten Werte sind jedoch Schätzungen
- **Wichtig:** Einfaches Ersetzen kann Streuung und Zusammenhänge verändern, Löschen kann ebenfalls verzerren
- **Zusatz:** Wenn alle Werte fehlen, lässt sich aus dieser Spalte kein Mittelwert oder Median als Ersatz berechnen

### MCAR, MAR und MNAR

- **MCAR:** Werte fehlen vollständig zufällig, bspw. durch zufälligen technischen Datenverlust
- **MAR:** Das Fehlen hängt von beobachteten Merkmalen ab, nach deren Berücksichtigung nicht zusätzlich vom fehlenden Wert selbst
- **Bsp. MAR:** Ältere Personen beantworten eine Frage seltener, das Alter ist im Datensatz vorhanden
- **MNAR:** Das Fehlen hängt auch nach Berücksichtigung beobachteter Merkmale vom fehlenden Wert selbst ab, bspw. besonders hohe Einkommen werden häufiger verschwiegen
- **Zweck:** Eine passende Behandlung fehlender Werte wählen
- **Wichtig:** Die Mechanismen lassen sich nicht allein aus der Anzahl fehlender Werte sicher bestimmen

### Kovarianz

- **Definition:** Beschreibt, ob zwei numerische Variablen tendenziell gemeinsam oder gegenläufig variieren
- **Zweck:** Die Richtung einer gemeinsamen Veränderung untersuchen
- **Bsp.:** Eine positive Kovarianz zwischen Ladeleistung und Stromstärke bedeutet, dass hohe Ladeleistungen tendenziell mit hohen Stromstärken auftreten
- **Interpretation:** Positiv bedeutet gleichgerichtet, negativ gegenläufig, eine Kovarianz von 0 bedeutet keinen linearen Zusammenhang
- **Wichtig:** Die Grösse hängt von Einheiten und Skalierung ab, ein Wert wie 15’380 ist deshalb nicht automatisch stark
- **Kovarianzmatrix:** Zeigt alle Variablenpaare, die Hauptdiagonale enthält die jeweiligen Varianzen

### Pearson-Korrelation

- **Definition:** Misst Stärke und Richtung eines linearen Zusammenhangs zwischen zwei numerischen Variablen
- **Zweck:** Zusammenhänge auf einer einheitlichen Skala von −1 bis +1 vergleichen
- **Bsp.:** r = +0,90 bedeutet einen starken positiven linearen Zusammenhang, r = −0,90 einen starken negativen
- **Interpretation:** Das Vorzeichen zeigt die Richtung, der Betrag die Stärke, Werte nahe 0 zeigen einen schwachen oder fehlenden linearen Zusammenhang
- **Wichtig:** Ausreisser können das Ergebnis stark beeinflussen, nichtlineare Beziehungen können trotz r nahe 0 bestehen
- **Voraussetzung:** Zusammengehörige Werte müssen dieselbe Beobachtung betreffen, bei konstanten Variablen ist die Korrelation nicht definiert

### Spearman-Rangkorrelation

- **Definition:** Misst Stärke und Richtung eines monotonen Zusammenhangs anhand der Rangplätze
- **Zweck:** Beziehungen untersuchen, bei denen eine Variable mit der anderen tendenziell zunimmt oder abnimmt, ohne dass eine Gerade entsteht
- **Bsp.:** Mit zunehmender Speichergrösse steigen die Preise meist an, aber nicht gleichmässig
- **Interpretation:** ρ = +0,85 bedeutet einen starken positiven monotonen Zusammenhang, ρ nahe 0 zeigt keinen oder einen schwachen monotonen Zusammenhang
- **Unterschied zu Pearson:** Spearman verwendet Rangplätze und ist meist weniger empfindlich gegenüber extremen Zahlenwerten
- **Wichtig:** Ungeordnete Kategorien wie Marken dürfen nicht willkürlich nummeriert und als Rangfolge behandelt werden

### Korrelation und Kausalität

- **Definition:** Korrelation beschreibt einen Zusammenhang, Kausalität eine Ursache-Wirkungs-Beziehung
- **Zweck:** Statistische Ergebnisse nicht überinterpretieren
- **Bsp.:** Eisverkäufe und Sonnenbrände nehmen gemeinsam zu, warmes Wetter beeinflusst beide
- **Interpretation:** Eine hohe Korrelation beweist weder eine direkte Ursache noch die Richtung einer möglichen Wirkung

### Kontingenztabelle – Kreuztabelle

- **Definition:** Zeigt, wie häufig Kombinationen kategorialer Merkmale vorkommen
- **Zweck:** Kategorien vergleichen und mögliche Zusammenhänge erkennen
- **Bsp.:** Eine Zelle mit DC und CCS = 120 bedeutet, dass 120 Beobachtungen beide Merkmale besitzen
- **Interpretation:** Zeilen und Spalten zeigen Kategorien, die Zellen ihre gemeinsamen Häufigkeiten, Gesamtsummen die berücksichtigten Beobachtungen
- **Wichtig:** Bei unterschiedlich grossen Gruppen sind Anteile oft aussagekräftiger als absolute Anzahlen

### Histogramm

- **Definition:** Fasst numerische Werte in Intervalle zusammen und zeigt deren Häufigkeiten
- **Zweck:** Verteilungsform, Streuung, Schiefe und auffällige Werte erkennen
- **Bsp.:** Viele Ladeleistungen im unteren Bereich und wenige bis 400 kW ergeben einen langen rechten Ausläufer
- **Interpretation:** Die X-Achse zeigt Wertebereiche, die Y-Achse standardmässig die Anzahl Beobachtungen je Intervall
- **Wichtig:** Die Anzahl Intervalle beeinflusst die Darstellung, mehrere Gipfel können auf unterschiedliche Gruppen hinweisen

### Boxplot

- **Definition:** Zeigt Median, Quartile, Streuung und potenzielle Ausreisser einer numerischen Variable
- **Zweck:** Verteilungen kompakt darstellen und zwischen Kategorien vergleichen
- **Interpretation:** Die Linie in der Box ist der Median, die Box reicht von Q1 bis Q3 und umfasst ungefähr die mittleren 50 % der Werte
- **Whiskers:** Reichen standardmässig bis zu den äussersten beobachteten Werten innerhalb der 1,5-IQR-Grenzen, weiter entfernte Werte erscheinen als Punkte
- **Bsp.:** Eine Gruppe mit höherem Median und grösserer Box besitzt höhere zentrale Werte und mehr Streuung im mittleren Bereich
- **Wichtig:** Eine grössere Box bedeutet mehr Streuung, nicht mehr Beobachtungen

### Violinplot

- **Definition:** Zeigt die Verteilungsform einer numerischen Variable als spiegelbildliche Dichtekurve
- **Zweck:** Neben Lage und Streuung auch Häufungen und mehrere Gipfel erkennen
- **Bsp.:** Breite Stellen bei 11 und 22 kW zeigen zwei Bereiche mit vielen Beobachtungen
- **Interpretation:** Breite Stellen bedeuten eine hohe geschätzte Dichte, schmale Stellen eine geringe
- **Wichtig:** Die Form hängt von der Glättung ab, die Breite entspricht nicht automatisch der absoluten Gruppengrösse

### Balkendiagramm

- **Definition:** Vergleicht Kategorien anhand ihrer Häufigkeiten oder berechneter Kennzahlen
- **Zweck:** Grössenunterschiede und Rangfolgen übersichtlich darstellen
- **Bsp.:** Städte nach Anzahl Ladestationen oder Marken nach Medianpreis vergleichen
- **Interpretation:** Die Balkenlänge zeigt die auf der Zahlenachse angegebene Grösse
- **Wichtig:** Prüfen, ob Anzahlen, Prozentwerte oder Kennzahlen dargestellt werden, eine Top-10-Auswahl blendet alle übrigen Kategorien aus

### Liniendiagramm

- **Definition:** Verbindet Werte entlang einer sinnvoll geordneten X-Achse
- **Zweck:** Entwicklungen und Verläufe zeigen, bspw. über die Zeit oder nach Zimmerzahl
- **Bsp.:** Der Medianpreis steigt von 2-Zimmer- zu 4-Zimmer-Wohnungen
- **Interpretation:** Steigende Abschnitte zeigen zunehmende Y-Werte, fallende Abschnitte abnehmende Y-Werte
- **Wichtig:** Die X-Werte müssen sinnvoll geordnet sein, Verbindungslinien beweisen keine gemessenen Zwischenwerte

### QQ-Plot

- **Definition:** Vergleicht die Quantile der beobachteten Daten mit denen einer theoretischen Verteilung, häufig der Normalverteilung
- **Zweck:** Prüfen, ob die Verteilungsform ungefähr zur gewählten Referenz passt
- **Bsp.:** Liegen besonders hohe Ladeleistungen deutlich über der Referenzlinie, passt der obere Datenbereich schlecht zur Normalverteilung
- **Interpretation:** Punkte ungefähr auf einer Geraden sprechen für eine passende Verteilungsform, systematische Krümmungen dagegen
- **Wichtig:** Die Referenzlinie muss weder durch 0 verlaufen noch eine Steigung von 1 besitzen, der Plot ist kein Beweis für Normalverteilung

### Scatterplot – Streudiagramm

- **Definition:** Stellt jede Beobachtung als Punkt mit zwei numerischen Werten dar
- **Zweck:** Zusammenhänge, Gruppen, Streuung und Ausreisser erkennen
- **Interpretation:** Eine steigende Punktwolke spricht für einen positiven Zusammenhang, eine fallende für einen negativen, Krümmungen für eine nichtlineare Beziehung
- **Bsp.:** Punkte weit abseits der übrigen Punktwolke können auffällige Beobachtungen sein
- **Gruppierung:** Farben können Kategorien unterscheiden, ein Gesamttrend kann innerhalb einzelner Gruppen anders aussehen

### Scatterplot-Matrix

- **Definition:** Zeigt paarweise Streudiagramme mehrerer numerischer Variablen
- **Zweck:** Mehrere mögliche Zusammenhänge gleichzeitig überblicken
- **Interpretation:** Ausserhalb der Diagonale zeigt der Spaltenname die X-Variable und der Zeilenname die Y-Variable
- **Diagonale:** Zeigt die Verteilung jeder einzelnen Variable, bspw. als Histogramm
- **Wichtig:** Gespiegelte Felder zeigen dasselbe Variablenpaar mit vertauschten Achsen, viele Variablen machen die Matrix schnell unübersichtlich

### Hexbin-Plot

- **Definition:** Fasst Beobachtungen zweier numerischer Variablen in sechseckigen Bereichen zusammen
- **Zweck:** Bei vielen überlappenden Punkten erkennen, wo sich Beobachtungen häufen
- **Bsp.:** Ein dunkles Sechseck bei 400 V und 32 A bedeutet bei einer entsprechend gewählten Farbskala viele Beobachtungen in diesem Bereich
- **Interpretation:** Die Farblegende zeigt standardmässig die Anzahl Beobachtungen je Sechseck
- **Wichtig:** Eine intensive Farbe bedeutet hohe Häufigkeit, nicht automatisch hohe X- oder Y-Werte

### Korrelationsmatrix und Heatmap

- **Definition:** Eine Korrelationsmatrix enthält die Korrelationen aller ausgewählten Variablenpaare, die Heatmap stellt sie zusätzlich farbig dar
- **Zweck:** Starke und schwache Zusammenhänge schnell erkennen
- **Bsp.:** Eine Zelle mit −0,85 zeigt einen starken negativen Zusammenhang zwischen der Zeilen- und Spaltenvariable
- **Interpretation:** Für den stärksten Zusammenhang den höchsten Betrag ausserhalb der Hauptdiagonale suchen
- **Wichtig:** Die Matrix ist symmetrisch, Selbstkorrelationen auf der Diagonale sind 1, sofern definiert
- **Farben:** Die Bedeutung hängt von der Farbskala ab, deshalb die Legende und die Zahlen lesen
