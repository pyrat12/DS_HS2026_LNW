## Datenquellen
Folien 4 bis 7

- Definition: Eine Datenquelle beschreibt den Ursprung bzw. Speicherort verwendeter Daten
- Zweck: Geeignete Informationen für eine konkrete Fragestellung beschaffen
- Bsp. Sensoren, Interviews, Webseiten, Publikationen und Datenbanken
- Bedeutung: Wachsende Datenmengen ermöglichen umfangreichere Analysen, erhöhen aber auch den Aufwand für Auswahl und Verarbeitung
- Erkenntnis: Viele verfügbare Daten sind nicht automatisch geeignete oder qualitativ gute Daten

### Primärdaten und Sekundärdaten

| Datenart | Bedeutung | Bsp. | Vorteil | Nachteil |
|---|---|---|---|---|
| Primärdaten | Daten für die eigene Untersuchung direkt erheben | Eigene Interviews, Befragungen und Messungen | Gezielt auf die Fragestellung abstimmbar | Erhebung benötigt Zeit und Ressourcen |
| Sekundärdaten | Bereits erhobene Daten erneut verwenden | Bestehende Statistiken, Webseiten und Unternehmensdaten | Häufig schneller und günstiger verfügbar | Ursprünglicher Zweck und Qualität können unpassend sein |

- Achtung: Die Einordnung hängt vom Untersuchungskontext ab
- Bsp. Eigene Sensormessungen sind Primärdaten, heruntergeladene Messdaten einer anderen Untersuchung sind Sekundärdaten
- Erkenntnis: Entscheidend ist, ob die Daten selbst erhoben oder bereits vorhandene Daten verwendet werden

## Dateiformate
Folien 33 bis 43

- Definition: Ein Dateiformat legt Aufbau und Bedeutung der gespeicherten Inhalte fest
- Zweck: Daten speichern, austauschen und mit geeigneten Werkzeugen verarbeiten

| Format | Aufbau | Verwendung bzw. Besonderheit |
|---|---|---|
| CSV | Texttabelle mit getrennten Feldern | Einfacher Austausch tabellarischer Daten |
| TXT | Reiner Text | Freie Texte und Protokolle |
| JSON | Schlüssel-Wert-Paare, Objekte und Arrays | Strukturierter Datenaustausch |
| XML | Hierarchische Elemente mit Tags und Attributen | Strukturierter Austausch verschachtelter Daten |
| HTML | Elemente zur Strukturierung von Webseiten | Darstellung im Browser und Quelle für Scraping |
| ZIP | Verlustfrei komprimiertes Archiv | Mehrere Dateien und Ordner bündeln |
| XLSX | Arbeitsmappe mit Tabellenblättern und Zellen | Tabellen, Formeln und mehrere Datenblätter |
| PDF | Dokument mit festgelegter Darstellung | Text, Bilder und Layout erhalten |
| Bilddateien | Pixelwerte und Metadaten | Bilder speichern und analysieren |

### CSV und TXT

- CSV: Jede Zeile beschreibt typischerweise einen Datensatz, Trennzeichen separieren die Felder
- Trennzeichen: Häufig Komma oder Semikolon
- Achtung: Trennzeichen, Zeichencodierung und Dezimalzeichen müssen zur Datei passen
- TXT: Besitzt keine automatisch vorgegebene Tabellenstruktur
- Erkenntnis: Eine Textdatei ist nicht automatisch eine korrekt eingelesene Tabelle

### JSON

- Aufbau: Daten werden als Schlüssel-Wert-Paare gespeichert
- Objekt: Geschweifte Klammern `{}`
- Array: Eckige Klammern `[]`
- Bsp. `{"brand": "Samsung", "price": 299, "available": true}`
- Achtung: JSON verwendet `true`, `false` und `null`, Python verwendet `True`, `False` und `None`
- Erkenntnis: Verschachtelte JSON-Daten müssen für eine Tabellenanalyse möglicherweise zuerst umgeformt werden

### XML und HTML

- XML: Besitzt genau ein Wurzelelement und korrekt verschachtelte Elemente
- XML-Tags: Gross-/Kleinschreibung beachten, Start- und Endtags müssen zusammenpassen
- XML-Attribute: Werte stehen in Anführungszeichen
- HTML: Beschreibt Struktur und Bedeutung der Inhalte einer Webseite
- CSS: Bestimmt die Darstellung
- JavaScript: Ermöglicht dynamisches Verhalten und nachgeladene Inhalte
- Bsp. HTML-Elemente sind `head`, `body`, `h1` und `p`
- Erkenntnis: HTML-Struktur, visuelle Darstellung und dynamisches Verhalten haben unterschiedliche Aufgaben

### Bilddateien

- Pixel: Beschreiben die Bildpunkte
- Kanäle: Beschreiben bspw. Rot, Grün und Blau
- Graustufenbild: Typischer Array-Aufbau `(Höhe, Breite)`
- RGB-Bild: Typischer Array-Aufbau `(Höhe, Breite, 3)`
- RGBA-Bild: Typischer Array-Aufbau `(Höhe, Breite, 4)`, Alpha beschreibt Transparenz
- Metadaten: Bspw. Bildformat und Bildabmessungen
- Achtung: Vier Kanäle bedeuten nicht automatisch ein vierdimensionales Array
- Erkenntnis: Bildformat, Array-Dimensionen und Anzahl Kanäle unterscheiden

## Parquet
Folien 44 bis 46

- Definition: Binäres, spaltenorientiertes Dateiformat für tabellarische Daten
- Zweck: Grosse Datenbestände effizient speichern und analysieren
- Aufbau: Werte werden spaltenweise gespeichert, Schema und Datentypen sind enthalten
- Kompression: Ähnliche Werte innerhalb einer Spalte lassen sich häufig gut komprimieren
- Spaltenauswahl: Nicht benötigte Spalten können beim Lesen übersprungen werden

| Merkmal | CSV | Parquet |
|---|---|---|
| Speicherung | Textbasiert | Binär und spaltenorientiert |
| Lesbarkeit | Direkt als Text lesbar | Geeignetes Werkzeug erforderlich |
| Datentypen | Beim Einlesen bestimmen | Im Schema gespeichert |
| Speicherbedarf | Häufig höher | Durch Kompression häufig geringer |
| Verwendung | Einfacher Datenaustausch | Effiziente Speicherung und Analyse |

- Bsp. `pd.read_parquet('daten.parquet', columns=['city', 'price'])`
- Achtung: Die Grössen- und Geschwindigkeitsunterschiede der Folien sind Beispielwerte
- Erkenntnis: CSV eignet sich für einfachen Austausch, Parquet besonders für grössere analytische Datensätze

## Funktionsweise eines Browsers

| Komponente | Aufgabe |
|---|---|
| Benutzeroberfläche | Adressleiste, Schaltflächen und Benutzereingaben |

## XPath

- Definition: Ausdruckssprache zur Auswahl von Knoten in einem XML- oder HTML-Baum

- Definition: Werkzeug zur automatisierten Steuerung eines Browsers

- Ablauf: `Python-Code → WebDriver → Browser`

## Data Preparation
PDF-Seiten 90 bis 95

- Definition: Rohdaten in eine geeignete Form für Analyse und Modellierung bringen
- Zweck: Daten bereinigen, vereinheitlichen, ergänzen und passend organisieren
- Aufgaben: Fehlende Werte behandeln, Duplikate prüfen, Datentypen korrigieren, Merkmale erstellen und Tabellen kombinieren

## Missing Values
PDF-Seiten 96 bis 98

- Definition: Ein benötigter Wert wurde für eine Beobachtung nicht erfasst
- Folgen: Weniger nutzbare Informationen, geringere statistische Aussagekraft und mögliche Verzerrungen
- Erkennen: `df.isna().sum()` zählt fehlende Werte je Spalte
- Anteil: `df.isna().mean() * 100` berechnet den Prozentanteil je Spalte
- Achtung: Leere Texte und Platzhalter wie unbekannt werden nicht automatisch als fehlend erkannt

### Entfernen und Imputation

| Vorgehen | Bedeutung | Vorteil | Nachteil |
|---|---|---|---|
| Entfernen | Betroffene Zeilen oder Spalten ausschliessen | Keine Ersatzwerte annehmen | Informationsverlust und mögliche Verzerrung |
| Imputation | Fehlende Werte durch geeignete Werte ersetzen | Mehr Beobachtungen bleiben nutzbar | Zusätzliche Annahmen und mögliche Verzerrung |

| Ersatzwert | Typischer Einsatz |
|---|---|
| Mittelwert | Numerische Daten mit ungefähr symmetrischer Verteilung |
| Median | Numerische Daten mit Schiefe oder Ausreissern |
| Modalwert | Häufigste Kategorie verwenden |
| Modellbasierter Wert | Andere Merkmale zur Schätzung berücksichtigen |

- Achtung: Einfache Imputation kann die Streuung unterschätzen und Zusammenhänge verändern
- Bsp. Bei rechtsschiefen Preisen kann der Median geeigneter sein als der Mittelwert
- Erkenntnis: Anzahl, Muster und fachliche Bedeutung der fehlenden Werte bestimmen das Vorgehen

## MCAR, MAR und MNAR
PDF-Seite 99

| Mechanismus | Bedeutung | Bsp. | Konsequenz |
|---|---|---|---|
| MCAR | Fehlen hängt weder vom fehlenden Wert noch von anderen Merkmalen ab | Zufälliger technischer Ausfall | Entfernen verursacht durch den Fehlmechanismus keine systematische Auswahlverzerrung |
| MAR | Fehlen kann von beobachteten Merkmalen abhängen | Ältere Personen überspringen eine Frage häufiger | Beobachtete Einflussgrössen bei der Behandlung berücksichtigen |
| MNAR | Fehlen hängt auch nach Berücksichtigung beobachteter Merkmale vom fehlenden Wert selbst ab | Personen mit tiefem Einkommen nennen dieses seltener | Einfaches Entfernen oder Ersetzen kann systematisch verzerren |

- MCAR: Missing Completely At Random
- MAR: Missing At Random
- MNAR: Missing Not At Random
- Achtung: MAR bedeutet nicht vollständig zufälliges Fehlen
- Achtung: Unter MAR bleiben beim Löschen nicht automatisch alle statistischen Beziehungen unverzerrt
- Erkenntnis: Die Ursache des Fehlens ist für die Behandlung wichtiger als die blosse Anzahl fehlender Werte

## Duplicated Values
PDF-Seite 100

- Definition: Datensätze oder relevante Merkmalskombinationen kommen mehrfach vor
- Problem: Unbeabsichtigte Wiederholungen können Beobachtungen mehrfach gewichten und Ergebnisse verzerren
- Erkennen: `df.duplicated().sum()` zählt Wiederholungen nach dem ersten Auftreten
- Untersuchen: `df[df.duplicated(keep=False)]` zeigt alle beteiligten Zeilen
- Entfernen: `df.drop_duplicates()` behält standardmässig die erste identische Zeile
- Schlüssel prüfen: Mit `subset` nur die relevanten Spalten vergleichen
- Achtung: Wiederholte Werte sind nicht automatisch unerwünschte Duplikate
- Bsp. Dasselbe Produkt darf bei unterschiedlichen Zeitpunkten oder Varianten mehrfach vorkommen
- Erkenntnis: Vor dem Entfernen klären, was eine einzelne Beobachtung eindeutig identifiziert

## Outliers und Measurement Errors
PDF-Seite 101

- Ausreisser: Beobachtung, die deutlich von anderen Beobachtungen abweicht
- Messfehler: Fehler bei der Messung bzw. Erfassung eines Werts
- Ursache: Ein Ausreisser kann ein echter seltener Fall oder ein Fehler sein
- Vorgehen: Auffällige Werte untersuchen und nur bei begründetem Fehler korrigieren oder entfernen
- Bsp. Preis von CHF 12'990 prüfen, wenn dasselbe Gerät in einer anderen Farbe CHF 1'299 kostet
- Achtung: Ein hoher Wert allein beweist keinen Fehler
- Erkenntnis: Auffälligkeit erkennen und Fehler nachweisen sind unterschiedliche Schritte

## Regular Expressions
PDF-Seiten 102 bis 105

- Definition: Zeichenfolge, die ein Suchmuster für Texte beschreibt
- Zweck: Inhalte erkennen, extrahieren, ersetzen und Eingaben prüfen
- Bsp. Zahlen aus Preisangaben oder Gewichtsangaben extrahieren

| Muster | Bedeutung |
|---|---|
| `[0-9]` bzw. `\d` | Eine Ziffer |
| `[0-9]+` | Eine oder mehrere Ziffern |
| `\D` | Ein Zeichen, das keine Ziffer ist |
| `\w` | Wortzeichen, bspw. Buchstabe, Ziffer oder Unterstrich |
| `\s` | Leerraum, bspw. Leerzeichen oder Tabulator |
| `.` | Beliebiges Zeichen, normalerweise ausser Zeilenumbruch |
| `\.` | Wörtlicher Punkt |
| `^dog` | Beginnt mit dog |
| `dog$` | Endet mit dog |
| `[abc]` | Eines der angegebenen Zeichen |
| `[^abc]` | Keines der angegebenen Zeichen |
| `*` | Null oder mehr Wiederholungen |
| `+` | Eine oder mehr Wiederholungen |
| `?` | Null oder eine Wiederholung |
| `{2,4}` | Zwei bis vier Wiederholungen |
| `(…)` | Gruppe bilden |
| `a\|b` | Alternative a oder b |

- Raw String: `r'…'` erleichtert die Schreibweise von Mustern mit Backslashes
- `str.contains()`: Prüft, ob ein Muster vorkommt
- `str.extract()`: Extrahiert passende Gruppen
- `str.replace()`: Ersetzt passende Inhalte
- `str.count()`: Zählt passende Muster
- Achtung: Dezimalstellen, Tausendertrennzeichen und Einheiten berücksichtigen
- Bsp. Ziffern aus 3.5 blind zusammenzufügen ergibt fälschlich 35
- Erkenntnis: Ein gefundenes Zahlenmuster muss noch korrekt interpretiert und umgewandelt werden

## String Manipulation
PDF-Seiten 106 bis 110

- Definition: Texte gezielt untersuchen, vereinheitlichen und verändern
- Zweck: Rohtexte bereinigen und zusätzliche Merkmale erstellen

| `.str.replace()` | Inhalte ersetzen |
| `.str.split()` | Text aufteilen |
| `.str.count()` | Vorkommen eines Musters zählen |

- Bsp. Unterschiedliche Schreibweisen einer Kategorie mit `lower()` und `strip()` vereinheitlichen
- Binäres Merkmal: Vorhandensein eines Begriffs als 1 bzw. 0 speichern
- Bsp. Wohnungsbeschreibung auf Luxusbegriffe prüfen
- Achtung: Gross-/Kleinschreibung vereinheitlichen korrigiert keine Tippfehler
- Achtung: Ein Begriff kann auch in einer Verneinung vorkommen
- Erkenntnis: Technisch korrekt gefundener Text ist nicht automatisch fachlich richtig interpretiert

## Date Formatting
PDF-Seiten 111 bis 113

- Zeitreihe: Beobachtungen in zeitlicher Reihenfolge
- Zweck: Datumswerte korrekt sortieren, filtern, vergleichen und weiterverarbeiten
- Parsen: Text in ein Datumsobjekt umwandeln
- Formatieren: Ein Datumsobjekt als Text darstellen
- Bsp. `pd.to_datetime(df['date'], format='%d.%m.%Y', errors='coerce')`
- Ungültige Werte: Werden bei `errors='coerce'` zu `NaT`
- Ableiten: Bspw. mit `.dt.year` das Jahr bestimmen

| Formatcode | Bedeutung |
|---|---|
| `%Y` | Jahr mit Jahrhundert |
| `%m` | Monat |
| `%d` | Tag |
| `%H` | Stunde |
| `%M` | Minute |
| `%S` | Sekunde |

- Achtung: Monat `%m` und Minute `%M` unterscheiden
- Achtung: `strftime()` erzeugt eine Textdarstellung
- Erkenntnis: Ein Datum als Text und ein echtes Datumsobjekt ermöglichen unterschiedliche Operationen

## Data Discretization
PDF-Seiten 114 bis 115

- Definition: Kontinuierliche Werte in eine endliche Anzahl von Intervallen bzw. Klassen einteilen
- Zweck: Daten vereinfachen und interpretierbare Gruppen erstellen
- Bsp. Alter in jung, mittel und alt einteilen
- Werkzeuge: `pd.cut()` und `KBinsDiscretizer`

| Vorgehen | Bedeutung |
|---|---|
| Vorgegebene Grenzen | Fachlich definierte Intervalle verwenden |
| Uniform | Intervalle mit gleicher Breite bilden |
| Quantile | Intervalle mit ungefähr gleicher Anzahl Beobachtungen bilden |
| Kmeans | Intervalle anhand von Clustern bestimmen |

- Vorteil: Kleine Schwankungen wirken weniger stark und Gruppen sind leichter verständlich
- Nachteil: Genaue Einzelwerte gehen verloren
- Achtung: Anzahl und Grenzen der Klassen beeinflussen das Ergebnis
- Achtung: Bei `pd.cut()` Intervallgrenzen und Werte ausserhalb der Grenzen prüfen
- Erkenntnis: Binning vereinfacht Daten, verursacht aber Informationsverlust

## One-Hot Encoding
PDF-Seiten 116 bis 117

- Definition: Kategoriale Merkmale in binäre Indikatorspalten umwandeln
- Zweck: Kategorien numerisch darstellen, ohne eine künstliche Rangfolge zu erzeugen
- Bsp. Farbe rot und blau als separate Spalten mit 1 bzw. 0 darstellen
- Werkzeuge: `pd.get_dummies()` und `OneHotEncoder()`

| Vorteil | Nachteil |
|---|---|
| Keine künstliche Reihenfolge zwischen nominalen Kategorien | Viele Kategorien erzeugen viele zusätzliche Spalten |
| Kategorien werden als separate Merkmale dargestellt | Viele Einträge sind 0 |
| Für zahlreiche numerisch arbeitende Modelle geeignet | Speicher- und Rechenaufwand können steigen |

- Referenzkategorie: Bei bestimmten Modellen kann eine Kategorie weggelassen werden
- Bsp. `drop_first=True` bei `pd.get_dummies()`
- Achtung: Eine Referenzkategorie wegzulassen ist keine allgemeine Pflicht für jedes Modell
- Erkenntnis: One-Hot Encoding verändert die Darstellung einer Kategorie, nicht deren fachliche Bedeutung

## Data Scaling
PDF-Seiten 118 bis 121

- Definition: Numerische Merkmale auf vergleichbare Grössenordnungen bringen
- Zweck: Verhindern, dass grosse Zahlenwerte bei skalensensitiven Verfahren übermässig wirken
- Bsp. Kilometerstand und Motorgrösse vor einer distanzbasierten Analyse skalieren

| Methode | Berechnung | Geeignet für | Grenze |
|---|---|---|---|
| Min-Max Scaling | `(x − min) / (max − min)` | Gemeinsamen Wertebereich, häufig 0 bis 1 | Empfindlich gegenüber Extremwerten |
| MaxAbs Scaling | `x / max(abs(x))` | Nullen und Vorzeichen erhalten | Empfindlich gegenüber Extremwerten |
| Robust Scaling | `(x − Median) / IQR` | Daten mit Ausreissern | Kein fest vorgegebener Wertebereich |

- IQR: Interquartilsabstand, `Q3 − Q1`
- Min-Max: Kleinster beobachteter Wert wird 0 und grösster 1
- MaxAbs: Verschiebt die Werte nicht und erhält vorhandene Nullen
- Robust Scaling: Verwendet Median und IQR statt Mittelwert und Standardabweichung
- Achtung: Spätere Werte ausserhalb der ursprünglichen Grenzen können ausserhalb des ursprünglichen Zielbereichs liegen
- Achtung: Skalierung entfernt keine Ausreisser
- Erkenntnis: Die passende Methode hängt von Datenverteilung, Ausreissern und Analyseverfahren ab

## Data Standardization
PDF-Seite 122

- Definition: Werte mithilfe von Mittelwert und Standardabweichung standardisieren
- Berechnung: `z = (x − Mittelwert) / Standardabweichung`
- Ergebnis: Auf den verwendeten Daten Mittelwert 0 und Standardabweichung 1 bei nicht konstanten Merkmalen
- Bsp. Ein Z-Wert von 2 liegt zwei Standardabweichungen über dem Mittelwert
- Wertebereich: Nicht auf 0 bis 1 begrenzt
- Werkzeug: `StandardScaler()`
- Achtung: Standardisierung erzeugt keine Normalverteilung
- Erkenntnis: Standardisierung verändert Lage und Streuung, die grundlegende Verteilungsform bleibt erhalten

### Scaling und Standardization unterscheiden

| Frage | Min-Max Scaling | Standardization |
|---|---|---|
| Was wird festgelegt? | Ein gemeinsamer Wertebereich | Mittelwert und Standardabweichung |
| Typisches Ergebnis | Werte zwischen 0 und 1 | Mittelwert 0 und Standardabweichung 1 |
| Fester Wertebereich | Für die Anpassungsdaten | Nein |
| Wird die Verteilung normal? | Nein | Nein |

## Data Transformation
PDF-Seiten 123 bis 126

- Definition: Mathematische Funktionen auf Werte anwenden
- Zweck: Zusammenhänge vereinfachen, Streuung stabilisieren und Modellannahmen besser erfüllen

| Transformation | Berechnung bzw. Bedingung | Möglicher Zweck |
|---|---|---|
| Logarithmus | `log(x)`, positive Werte erforderlich | Grosse Werte komprimieren und Rechtsschiefe reduzieren |
| Quadratwurzel | `sqrt(x)`, nicht negative Werte erforderlich | Grosse Werte moderater komprimieren |
| Potenz | Bspw. `x²` | Form eines Zusammenhangs verändern |
| Kehrwert | `1/x`, Wert darf nicht 0 sein | Nichtlinearen Zusammenhang umformen |
| Box-Cox | Positive Werte erforderlich, bei λ = 0 als Logarithmus definiert | Anpassbare Potenztransformation |

- Bsp. Eine Log-Transformation kann einen gekrümmten Zusammenhang vereinfachen
- Residuen: Eine Transformation der Zielvariable kann ungleichmässige Residuenstreuung reduzieren
- Achtung: Die gewünschte Verbesserung ist nicht garantiert und muss geprüft werden
- Achtung: Bei Regression beziehen sich wichtige Annahmen auf Fehler bzw. Residuen, nicht pauschal auf jede Eingangsvariable
- Erkenntnis: Nichtlineare Transformationen können die Verteilungsform verändern, reine Skalierung tut dies grundsätzlich nicht

## Combining Data
PDF-Seiten 127 bis 130

- Definition: Daten aus mehreren Quellen zusammenführen
- Zweck: Vorhandene Beobachtungen mit zusätzlichen Informationen anreichern
- Bsp. Wohnungen mit Gemeindedaten wie Einkommen, Bevölkerungsdichte und Bahnhofsdistanz verbinden
- Voraussetzung: Schlüssel, Datentypen und fachliche Bedeutung müssen zusammenpassen
- Räumliche Überlagerung: Daten anhand ihrer Lage zuordnen, bspw. Wohnung einer Gemeinde zuweisen

| Methode | Grundidee | Bsp. |
|---|---|---|
| `merge()` | Tabellen über Schlüsselspalten oder Indizes verbinden | Wohnung mit Gemeindedaten anreichern |
| `join()` | Spalten ergänzen, standardmässig über Indizes | Indexbasierte Zusatzinformationen übernehmen |
| `concat()` | Daten entlang von Zeilen oder Spalten aneinanderhängen | Mehrere Monatsdateien zusammenfügen |

- Erkenntnis: Schlüsselbasiertes Verknüpfen und blosses Aneinanderhängen sind unterschiedliche Operationen

## merge()
PDF-Seiten 131 bis 132

- Definition: DataFrames über passende Schlüsselwerte verbinden
- Zweck: SQL-artige Verknüpfungen zwischen Tabellen durchführen
- Bsp. `left.merge(right, on='key', how='left')`
- Standard: Ohne Angabe von `how` wird ein Inner Join verwendet

| Join-Typ | Welche Zeilen bleiben erhalten? |
|---|---|
| inner | Nur passende Kombinationen beider Seiten |
| left | Alle linken Zeilen und passende rechte Informationen |
| right | Alle rechten Zeilen und passende linke Informationen |
| outer | Passende Kombinationen und nicht zugeordnete Zeilen beider Seiten |

- Unterschiedliche Schlüsselnamen: Mit `left_on` und `right_on` angeben
- Fehlende Gegenseite: Wird bei entsprechenden Joins mit fehlenden Werten ergänzt
- Mehrere Treffer: Können die Zeilenanzahl vervielfachen
- Bsp. Zwei linke und drei rechte Zeilen zum gleichen Schlüssel ergeben sechs Kombinationen
- Achtung: Pandas kann fehlende Schlüssel miteinander verknüpfen, anders als typische SQL-NULL-Verknüpfungen
- Erkenntnis: Vor und nach dem Join Zeilenzahl, Schlüssel und Trefferlage prüfen

## join()
PDF-Seite 133

- Definition: Spalten verschiedener DataFrames bequem zusammenführen
- Standard: Verknüpfung über die Indizes mit einem Left Join
- Bsp. `left.join(right)`
- Mit `on`: Links wird die angegebene Spalte und rechts der Index verwendet
- Achtung: `left.join(right, on='key')` verbindet nicht automatisch zwei gleichnamige Spalten
- Erkenntnis: Bei `join()` besonders auf die Bedeutung der Indizes achten

## concat()
PDF-Seite 134

- Definition: Mehrere DataFrames entlang einer Achse aneinanderhängen
- Zeilenweise: `axis=0`
- Spaltenweise: `axis=1`
- Bsp. `pd.concat([df1, df2], ignore_index=True)`
- Neuer Index: `ignore_index=True` erzeugt einen fortlaufenden Index
- Unterschiedliche Spalten: Fehlende Angaben werden beim üblichen zeilenweisen Zusammenfügen ergänzt
- Achtung: Bei `axis=1` erfolgt die Ausrichtung anhand des Index
- Achtung: Duplikate werden nicht automatisch entfernt
- Erkenntnis: Vor dem Zusammenfügen Struktur, Indizes und mögliche Überschneidungen prüfen

## Sorting Data
PDF-Seiten 135 bis 136

- Definition: Zeilen anhand einer oder mehrerer Spalten ordnen
- Zweck: Reihenfolgen herstellen und höchste bzw. tiefste Werte erkennen
- Werkzeug: `sort_values()`
- Bsp. `df.sort_values(by=['brand', 'price'], ascending=[True, False])`
- Reihenfolge: Zuerst Marke aufsteigend, innerhalb derselben Marke Preis absteigend
- Fehlende Werte: Können mit `na_position` eingeordnet werden
- Achtung: Zahlen als Text können anders sortiert werden als echte Zahlen
- Erkenntnis: Sortieren verändert die Reihenfolge, nicht die Anzahl der Beobachtungen

## Aggregation mit groupby()
PDF-Seiten 137 bis 138

- Definition: Daten in Gruppen aufteilen und pro Gruppe Kennzahlen berechnen
- Zweck: Viele Einzelbeobachtungen zusammenfassen
- Ablauf: `Split → Apply → Combine`
- Bsp. `df.groupby('brand')['price'].mean()` berechnet den Durchschnittspreis je Marke
- Mehrere Schlüssel: Bspw. nach Marke und Farbe gemeinsam gruppieren

| Funktion | Bedeutung |
|---|---|
| mean | Mittelwert berechnen |
| median | Median bestimmen |
| min bzw. max | Kleinsten bzw. grössten Wert bestimmen |
| sum | Werte summieren |
| size | Anzahl Zeilen zählen |
| count | Anzahl vorhandener Werte zählen |
| agg | Mehrere bzw. unterschiedliche Aggregationen festlegen |

- Achtung: `size` und `count` können bei fehlenden Werten unterschiedliche Ergebnisse liefern
- Achtung: Fehlende Gruppenschlüssel werden standardmässig ausgeschlossen
- Interpretation: Gruppengrösse, Ausreisser und fehlende Werte berücksichtigen
- Erkenntnis: Aggregation verdichtet Daten und kann dadurch Details verdecken

## Reshaping Data
PDF-Seiten 139 bis 144

- Definition: Die Anordnung von Daten zwischen Zeilen, Spalten und Index verändern
- Zweck: Daten in eine für Analyse oder Darstellung geeignete Form bringen

| Methode | Bedeutung | Verwendung |
|---|---|---|
| stack | Spaltenebene in eine Indexebene überführen | Werte untereinander anordnen |
| unstack | Indexebene in eine Spaltenebene überführen | Werte nebeneinander anordnen |
| pivot | Werte einer Spalte zu neuen Spalten machen | Long nach Wide ohne Aggregation |
| melt | Mehrere Messspalten in Variablen- und Wertzeilen überführen | Wide nach Long |
| pivot_table | Umformen und gleichzeitig aggregieren | Mehrere Werte je Kombination zusammenfassen |

### pivot()

- Zweck: Eine Tabelle anhand von Index-, Spalten- und Wertangaben neu anordnen
- Bsp. `df.pivot(index='city', columns='year', values='price')`
- Ergebnis: Eine Zeile je Stadt und eine Spalte je Jahr
- Voraussetzung: Jede Kombination aus Stadt und Jahr besitzt höchstens einen zugehörigen Wert
- Achtung: Mehrere Werte je Kombination verursachen einen Fehler
- Erkenntnis: `pivot()` ordnet um, führt aber keine Aggregation durch

### melt()

- Zweck: Mehrere Messspalten untereinander in eine lange Darstellung bringen
- id_vars: Identifikationsspalten, die erhalten bleiben
- value_vars: Messspalten, die in Zeilen überführt werden
- Ergebnis: Zusätzliche Spalte für den bisherigen Variablennamen und eine für den Wert
- Bsp. Preis_2025 und Preis_2026 werden zu einer Jahres- und einer Preisspalte
- Erkenntnis: `melt()` eignet sich für die Umwandlung von Wide nach Long

### pivot_table()

- Zweck: Daten umformen und mehrere Werte je Kombination zusammenfassen
- Bsp. `df.pivot_table(index='city', columns='year', values='price', aggfunc='mean')`
- Ergebnis: Durchschnittspreis je Stadt und Jahr
- Aggregation: Bspw. mean, median, sum oder count
- Achtung: Die gewählte Aggregationsfunktion beeinflusst die fachliche Bedeutung
- Erkenntnis: `pivot_table()` kann im Gegensatz zu `pivot()` mehrere Werte je Kombination verarbeiten
