# AP02 – Spick Folien V5 EDA

## Exploratory Data Analysis
Folien 4 bis 6

- Definition: Daten systematisch untersuchen, um Verteilungen, Muster, Zusammenhänge und Auffälligkeiten zu verstehen
- Zweck: Daten kennenlernen und eine Grundlage für weiterführende Analysen schaffen
- Einordnung: EDA untersucht Daten zunächst ohne formale statistische Modellierung und Hypothesenprüfung
- Datenqualität: Fehlende Werte, Duplikate und unplausible Angaben erkennen
- Erkenntnis: Vor der Modellierung verstehen, welche Daten vorliegen und welche Einschränkungen sie besitzen

### Ziele der EDA

| Ziel | Bedeutung | Bsp. |
|---|---|---|
| Unexpected Discoveries | Unerwartete Muster entdecken | Auffällige Preisgruppen erkennen |
| Hypotheses | Mögliche Erklärungen und Hypothesen entwickeln | Grössere Wohnungen könnten teurer sein |
| Assess Assumptions | Voraussetzungen späterer Methoden prüfen | Verteilung und Ausreisser untersuchen |
| Selection of Methods | Geeignete Analyseverfahren auswählen | Median bei stark rechtsschiefen Preisen |
| Further Data Collection | Bedarf an zusätzlichen Informationen erkennen | Lageinformationen ergänzen |
| Data Quality | Qualität der vorhandenen Daten prüfen | Fehlende Preise und doppelte Angebote erkennen |

- Achtung: Eine durch EDA entwickelte Hypothese ist noch nicht bestätigt
- Erkenntnis: EDA unterstützt sowohl das Datenverständnis als auch die Planung der weiteren Analyse

## Einteilung der EDA-Methoden
Folien 7 bis 11

- Univariate EDA: Eine einzelne Variable untersuchen
- Multivariate EDA: Mehrere Variablen und ihre Beziehungen untersuchen
- Non-Graphical EDA: Kennzahlen und Tabellen verwenden
- Graphical EDA: Diagramme und Visualisierungen verwenden

| Einteilung | Non-Graphical | Graphical |
|---|---|---|
| Univariate | Mittelwert, Median, Standardabweichung und Schiefe | Histogramm, Boxplot und Balkendiagramm |
| Multivariate | Kreuztabelle, Kovarianz und Korrelation | Scatterplot, Scatterplot-Matrix und Korrelations-Heatmap |

- Bsp. Univariate Analyse: Verteilung der Mietpreise untersuchen
- Bsp. Multivariate Analyse: Zusammenhang zwischen Wohnfläche und Mietpreis untersuchen
- Numerische Daten: Bspw. Lage, Streuung und Verteilungsform betrachten
- Kategoriale Daten: Bspw. Häufigkeiten einzelner Kategorien betrachten
- Erkenntnis: Die passende Methode hängt von Fragestellung, Anzahl Variablen und Datentypen ab

## Anscombe's Quartet
Folie 12

- Definition: Vier Datensätze mit nahezu identischen Kennzahlen, aber deutlich unterschiedlichen grafischen Mustern
- Gemeinsame Kennzahlen: Bspw. Mittelwerte, Varianzen, Korrelationen und Regressionsgeraden
- Unterschiede: Bspw. lineare und nichtlineare Zusammenhänge sowie einzelne einflussreiche Ausreisser
- Datasaurus Dozen: Weitere Datensätze mit ähnlichen Kennzahlen und unterschiedlichen grafischen Formen
- Achtung: Gleiche Kennzahlen bedeuten nicht gleiche Datenstrukturen
- Erkenntnis: Statistische Kennzahlen immer durch passende Grafiken ergänzen

## Measures of Central Tendency
Folie 14

- Definition: Lagermasse beschreiben einen zentralen bzw. typischen Wert einer Verteilung
- Zweck: Viele Einzelwerte durch eine verständliche Kennzahl zusammenfassen

| Kennzahl | Bedeutung | Geeignet für | Grenze |
|---|---|---|---|
| Mean, Mittelwert | Summe aller Werte geteilt durch ihre Anzahl | Durchschnitt numerischer Werte | Empfindlich gegenüber Ausreissern |
| Median | Mittlerer Wert der sortierten Beobachtungen | Schiefe Verteilungen und Daten mit Ausreissern | Berücksichtigt die Grösse extremer Werte nur eingeschränkt |
| Mode, Modalwert | Häufigster Wert | Numerische und kategoriale Daten | Mehrere Modalwerte möglich |

- Median bei ungerader Anzahl: Mittleren Wert verwenden
- Median bei gerader Anzahl numerischer Werte: Mittelwert der beiden mittleren Werte verwenden
- Bsp. Werte `1, 2, 3, 4, 5, 5, 42` besitzen Mittelwert ca. 8.86, Median 4 und Modalwert 5
- Interpretation: Der hohe Wert 42 zieht den Mittelwert nach oben
- Python: `.mean()`, `.median()` und `.mode()`
- Erkenntnis: Der Mittelwert ist nicht bei jeder Verteilung der geeignetste typische Wert

## Quantile
Folien 15 bis 16

- Definition: Grenzwerte, die eine geordnete Verteilung nach bestimmten Anteilen aufteilen
- Zweck: Niedrige, mittlere und hohe Wertebereiche beschreiben
- Quartile: Teilen die Verteilung in vier Bereiche
- Dezile: Teilen die Verteilung in zehn Bereiche
- Perzentile: Teilen die Verteilung in hundert Bereiche

| Quantil | Bedeutung |
|---|---|
| 0.25 bzw. Q1 | Unteres Quartil |
| 0.50 bzw. Q2 | Median |
| 0.75 bzw. Q3 | Oberes Quartil |
| 0.95 | Grenze, bis zu der ungefähr 95 % der Werte liegen |

- Bsp. Ein 95%-Quantil von CHF 3'000 bedeutet, dass ungefähr 95 % der beobachteten Preise höchstens CHF 3'000 betragen
- IQR: Interquartilsabstand, `Q3 − Q1`
- Interpretation: Der Bereich zwischen Q1 und Q3 enthält die mittleren 50 % der Werte
- Python: `.quantile([0.25, 0.50, 0.75])`
- Achtung: Quantile können abhängig von der Berechnung zwischen beobachteten Werten liegen
- Erkenntnis: Quantile beschreiben die Verteilung differenzierter als ein einzelner Durchschnitt

## Measures of Dispersion
Folie 17

- Definition: Streuungsmasse beschreiben, wie stark Werte auseinanderliegen bzw. vom Mittelwert abweichen
- Zweck: Die Unterschiedlichkeit der Beobachtungen beurteilen

| Kennzahl | Bedeutung | Einheit |
|---|---|---|
| Varianz | Mittlere quadrierte Abweichung vom Mittelwert | Quadrierte Originaleinheit |
| Standardabweichung | Quadratwurzel der Varianz | Originaleinheit |

- Kleine Standardabweichung: Werte liegen vergleichsweise nahe beim Mittelwert
- Grosse Standardabweichung: Werte streuen stärker um den Mittelwert
- Grundgesamtheit: Bei der Varianz durch die Anzahl N teilen
- Stichprobe: Bei der üblichen Schätzung der Varianz durch n − 1 teilen
- Python: `.var()` und `.std()` verwenden standardmässig die Stichprobenberechnung mit `ddof=1`
- Achtung: Ausreisser können Varianz und Standardabweichung stark erhöhen
- Achtung: Ob eine Streuung gross ist, hängt von Einheit, Grössenordnung und Kontext ab
- Erkenntnis: Mittelwert und Streuung gemeinsam betrachten

## Skewness
Folien 18 bis 19

- Definition: Schiefe beschreibt die Asymmetrie einer Verteilung
- Zweck: Erkennen, auf welcher Seite ein längerer Ausläufer liegt
- Python: `.skew()`

| Schiefe | Bedeutung | Typisches Muster |
|---|---|---|
| Positiv | Rechtsschief | Viele kleinere Werte und wenige sehr grosse Werte |
| Negativ | Linksschief | Viele grössere Werte und wenige sehr kleine Werte |
| Nahe 0 | Geringe gemessene Schiefe | Häufig ungefähr symmetrische Verteilung |

- Bsp. Mietpreise sind häufig rechtsschief, weil wenige sehr teure Wohnungen einen langen rechten Ausläufer bilden
- Faustregel: Mittelwert über Median kann auf Rechtsschiefe hindeuten
- Faustregel: Mittelwert unter Median kann auf Linksschiefe hindeuten
- Achtung: Die Richtung der Schiefe beschreibt den Ausläufer, nicht die Seite mit den meisten Werten
- Achtung: Schiefe nahe 0 beweist weder Symmetrie noch Normalverteilung
- Erkenntnis: Schiefe mit Histogramm und Lagekennzahlen gemeinsam beurteilen

## Kurtosis
Folie 20

- Definition: Kennzahl für die Ausprägung der Verteilungsenden bzw. Tails
- Zweck: Die Neigung zu extremen Werten im Vergleich zu einer Normalverteilung beurteilen
- Python: `.kurtosis()` liefert die Excess Kurtosis
- Referenz: Eine Normalverteilung besitzt Excess Kurtosis 0

| Excess Kurtosis | Interpretation |
|---|---|
| Positiv | Stärker ausgeprägte Verteilungsenden als bei einer Normalverteilung |
| Negativ | Weniger stark ausgeprägte Verteilungsenden als bei einer Normalverteilung |
| 0 | Gleicher Kurtosis-Referenzwert wie die Normalverteilung |

- Achtung: Kurtosis beschreibt hauptsächlich die Verteilungsenden, nicht einfach die Höhe des Gipfels
- Achtung: Ein Wert von 0 beweist keine Normalverteilung
- Erkenntnis: Skewness und Kurtosis beschreiben unterschiedliche Eigenschaften der Verteilungsform

## Contingency Tables
Folien 21 bis 22

- Definition: Kreuztabelle mit Häufigkeiten für Kombinationen kategorialer Merkmale
- Zweck: Verteilungen und mögliche Zusammenhänge zwischen Kategorien untersuchen
- Aufbau: Kategorien einer Variable in den Zeilen und Kategorien einer anderen Variable in den Spalten
- Zellen: Anzahl Beobachtungen mit der jeweiligen Kombination
- Bsp. Anzahl Wohnungen nach Zimmerkategorie und Luxusmerkmal
- Python: `pd.crosstab(df['rooms'], df['luxurious'])`
- Achtung: Bei unterschiedlich grossen Gruppen zusätzlich relative Häufigkeiten betrachten
- Bsp. Zeilenprozente zeigen den Luxusanteil innerhalb einer Zimmerkategorie
- Erkenntnis: Absolute Anzahlen und relative Anteile beantworten unterschiedliche Fragen

## Covariance
Folien 23 bis 24

- Definition: Kovarianz beschreibt die gemeinsame Veränderung zweier numerischer Variablen
- Zweck: Die Richtung eines gemeinsamen linearen Veränderungsmusters beurteilen
- Wertebereich: Nicht auf −1 bis 1 begrenzt

| Kovarianz | Interpretation |
|---|---|
| Positiv | Hohe Werte der einen Variable treten tendenziell mit hohen Werten der anderen auf |
| Negativ | Hohe Werte der einen Variable treten tendenziell mit tiefen Werten der anderen auf |
| Nahe 0 | Kein deutliches gemeinsames lineares Veränderungsmuster |

- Bsp. Wohnfläche und Mietpreis besitzen häufig eine positive Kovarianz
- Abhängigkeit: Die Grösse der Kovarianz hängt von den Einheiten und Grössenordnungen ab
- Kovarianzmatrix: Zeigt die Kovarianzen mehrerer Variablenpaare
- Diagonale: Enthält die Varianz der jeweiligen Variable
- Python: `.cov()`
- Achtung: Ein grosser Betrag ist ohne Berücksichtigung der Einheiten kein allgemein vergleichbares Stärkemass
- Erkenntnis: Kovarianz eignet sich für die Richtung, Korrelation zusätzlich für eine standardisierte Stärke

## Correlation
Folien 25 bis 26

- Definition: Ein Korrelationskoeffizient beschreibt Richtung und Stärke eines bestimmten Zusammenhangs zwischen Variablen
- Pearson-Korrelation: Misst lineare Zusammenhänge zwischen numerischen Variablen
- Wertebereich: −1 bis +1

| Wert | Interpretation bei Pearson |
|---|---|
| Nahe +1 | Starker positiver linearer Zusammenhang |
| Nahe −1 | Starker negativer linearer Zusammenhang |
| Nahe 0 | Geringer linearer Zusammenhang |

- Stärke: Anhand des absoluten Werts beurteilen
- Bsp. −0.90 beschreibt einen stärkeren linearen Zusammenhang als +0.60
- Richtung: Das Vorzeichen zeigt gleichläufige bzw. gegenläufige Veränderung
- Achtung: Eine Korrelation nahe 0 schliesst einen nichtlinearen Zusammenhang nicht aus
- Achtung: Ausreisser können Pearson-Korrelationen stark beeinflussen
- Achtung: Korrelation beweist keine Kausalität
- Erkenntnis: Korrelationswert immer mit einem Scatterplot und fachlichem Wissen prüfen

### Pearson und Spearman unterscheiden

| Methode | Grundlage | Geeignet für |
|---|---|---|
| Pearson | Ursprüngliche numerische Werte | Lineare Zusammenhänge |
| Spearman | Rangplätze der Werte | Monotone Zusammenhänge und geeignete ordinale Daten |

- Monoton: Eine Variable steigt bzw. fällt tendenziell durchgehend mit der anderen, auch bei gekrümmtem Verlauf
- Bsp. Ein durchgehend steigender gekrümmter Zusammenhang kann eine hohe Spearman-Korrelation besitzen
- Python: `.corr(method='pearson', numeric_only=True)` bzw. `.corr(method='spearman', numeric_only=True)`
- Erkenntnis: Pearson untersucht Linearität, Spearman die gemeinsame Rangordnung

### Korrelationsmatrix

- Definition: Tabelle mit Korrelationskoeffizienten mehrerer Variablenpaare
- Zweck: Zusammenhänge zwischen vielen numerischen Merkmalen überblicken
- Symmetrie: Die Korrelation von A mit B entspricht der Korrelation von B mit A
- Diagonale: Bei nicht konstanten Variablen normalerweise 1
- Achtung: Für die Suche nach starken Beziehungen die Diagonale ausschliessen
- Achtung: Bei fehlenden Werten können verschiedene Paare auf unterschiedlichen Beobachtungen beruhen
- Erkenntnis: Auffällige Paare anschliessend gezielt grafisch untersuchen

## Line Chart
Folien 27 bis 28

- Definition: Datenpunkte werden durch Linien verbunden
- Zweck: Entwicklungen entlang einer geordneten Achse darstellen
- Verwendung: Besonders für Zeitreihen
- Bsp. Entwicklung eines Aktienkurses oder monatlicher Mietpreise
- X-Achse: Bspw. Zeit
- Y-Achse: Beobachteter Wert bzw. Kennzahl
- Achtung: Die Reihenfolge der X-Werte muss sinnvoll sein
- Achtung: Eine Verbindung ungeordneter Kategorien kann einen irreführenden Verlauf erzeugen
- Erkenntnis: Liniendiagramme zeigen zeitliche Entwicklungen und Veränderungen besonders gut

## Box Plot
Folien 29 bis 30

- Definition: Grafische Zusammenfassung numerischer Daten mithilfe von Median und Quartilen
- Zweck: Lage, Streuung und potenzielle Ausreisser erkennen
- Verwendung: Einzelne Verteilungen untersuchen oder mehrere Gruppen vergleichen

| Bestandteil | Bedeutung |
|---|---|
| Linie in der Box | Median |
| Untere Boxgrenze | Q1 |
| Obere Boxgrenze | Q3 |
| Box | Bereich der mittleren 50 % der Werte |
| Boxhöhe bzw. Boxbreite | IQR |
| Whiskers | Bereich bis zu den äussersten Werten innerhalb der verwendeten Grenzen |
| Einzelne Punkte | Werte ausserhalb der Whiskers |

- Häufige Regel: Untere Grenze `Q1 − 1.5 × IQR`, obere Grenze `Q3 + 1.5 × IQR`
- Whiskers: Enden bei den äussersten tatsächlich vorhandenen Werten innerhalb dieser Grenzen
- Interpretation: Eine grössere Box zeigt einen grösseren Abstand zwischen Q1 und Q3
- Achtung: Whiskers sind nicht grundsätzlich Minimum und Maximum
- Achtung: Punkte ausserhalb der Whiskers sind potenzielle Ausreisser, keine automatisch nachgewiesenen Fehler
- Erkenntnis: Ein Boxplot fasst die Verteilung kompakt zusammen, zeigt aber nicht alle Details ihrer Form

## Histogram
Folien 31 bis 32

- Definition: Numerische Werte werden in Intervalle eingeteilt und deren Häufigkeiten dargestellt
- Zweck: Verteilungsform, Schiefe, mehrere Gipfel und ungewöhnliche Werte erkennen
- Bins: Intervalle bzw. Klassen auf der X-Achse
- Bsp. Verteilung der Mietpreise pro Quadratmeter
- Rechtsschief: Viele moderate Werte und wenige sehr hohe Werte mit langem rechten Ausläufer

### Anzahl und Breite der Bins

- Wenige Bins: Grober Überblick, Details können verloren gehen
- Viele Bins: Mehr Details, zufällige Schwankungen können überbetont werden
- Gleiche Binbreite: Balkenhöhen können Häufigkeiten darstellen
- Unterschiedliche Binbreite: Für eine korrekte Flächendarstellung Häufigkeitsdichten verwenden
- Häufigkeitsdichte: Häufigkeit geteilt durch Intervallbreite
- Achtung: Achsenbeschriftung prüfen, dargestellt sein können Anzahl, Anteil oder Dichte
- Erkenntnis: Die Wahl der Bins beeinflusst den visuellen Eindruck der Verteilung

## Quantile-Quantile Plot
Folien 33 bis 34

- Definition: Quantile zweier Verteilungen gegeneinander darstellen
- Zweck: Prüfen, ob die Verteilungsformen ungefähr übereinstimmen
- Normal-QQ-Plot: Beobachtete Quantile mit theoretischen Quantilen einer Normalverteilung vergleichen

| Muster | Interpretation |
|---|---|
| Punkte ungefähr auf einer Geraden | Verteilung passt näherungsweise zur Vergleichsverteilung |
| Systematische Krümmung | Verteilungsformen unterscheiden sich |
| Starke Abweichungen an den Enden | Verteilungsenden passen nicht gut zur Vergleichsverteilung |
| Einzelne weit entfernte Punkte | Mögliche Extremwerte genauer untersuchen |

- Bsp. Sehr grosse Wohnflächen können am oberen Ende deutlich von der Referenzlinie abweichen
- Achtung: Ein QQ-Plot liefert eine grafische Beurteilung, keinen Beweis für Normalverteilung
- Erkenntnis: QQ-Plot und Histogramm ergänzen sich bei der Verteilungsprüfung

## Bar Chart
Folie 35

- Definition: Kategorien werden mit Balken dargestellt, deren Länge einer Kennzahl entspricht
- Zweck: Häufigkeiten oder Kennzahlen zwischen Kategorien vergleichen
- Darstellung: Vertikale oder horizontale Balken
- Bsp. Anzahl Fahrzeuge je Marke oder Medianpreis je Hersteller
- Achtung: Ein Balkendiagramm zeigt nicht automatisch Anzahlen, auch aggregierte Kennzahlen sind möglich
- Erkenntnis: Balkenlängen ermöglichen einen direkten Kategorienvergleich

### Histogramm und Balkendiagramm unterscheiden

| Merkmal | Histogramm | Balkendiagramm |
|---|---|---|
| Datengrundlage | Numerische Werte | Kategorien |
| Einteilung | Numerische Intervalle | Einzelne Kategorien |
| Reihenfolge | Durch Wertebereiche vorgegeben | Kann passend gewählt werden |
| Aussage | Verteilung einer numerischen Variable | Vergleich von Kategorien |

## Pie Chart
Folie 36

- Definition: Ein Kreis wird entsprechend den Anteilen in Segmente aufgeteilt
- Zweck: Zusammensetzung eines Ganzen darstellen
- Bsp. Anteile verschiedener Verbrauchsarten am gesamten Wasserverbrauch
- Vorteil: Wenige Anteile eines Ganzen sind schnell erkennbar
- Nachteil: Ähnliche Winkel und Flächen sind schwer vergleichbar
- Achtung: Bei vielen Kategorien ist ein Balkendiagramm häufig übersichtlicher
- Erkenntnis: Kreisdiagramme sparsam verwenden und auf wenige klar unterscheidbare Kategorien beschränken

## Scatter Plot
Folie 37

- Definition: Zusammengehörige Wertepaare werden als Punkte in einem Koordinatensystem dargestellt
- Zweck: Beziehungen zwischen zwei numerischen Variablen untersuchen
- Bsp. Wohnfläche auf der X-Achse und Mietpreis auf der Y-Achse

| Muster | Mögliche Interpretation |
|---|---|
| Steigender Verlauf | Positiver Zusammenhang |
| Fallender Verlauf | Negativer Zusammenhang |
| Gekrümmter Verlauf | Nichtlinearer Zusammenhang |
| Getrennte Punktgruppen | Mögliche Cluster oder unterschiedliche Gruppen |
| Einzelne isolierte Punkte | Potenzielle Ausreisser |
| Zunehmende vertikale Streuung | Streuung verändert sich mit den X-Werten |

- Achtung: Sichtbarer Zusammenhang beweist keine Ursache
- Achtung: Überlappende Punkte können die tatsächliche Anzahl Beobachtungen verdecken
- Erkenntnis: Scatterplots helfen, Korrelationswerte und Modellannahmen richtig einzuordnen

## Bubble Plot
Folie 38

- Definition: Scatterplot mit zusätzlichen Informationen über Punktgrösse und häufig Farbe
- Zweck: Mehrere Merkmale gleichzeitig darstellen
- Bsp. Wohnfläche und Preis als Achsen, Zimmerzahl als Punktgrösse und Bevölkerungsdichte als Farbe
- Vorteil: Zusätzliche Zusammenhänge können in derselben Grafik sichtbar werden
- Nachteil: Viele visuelle Merkmale können die Interpretation erschweren
- Achtung: Legenden und Grössenskalierung beachten
- Erkenntnis: Punktposition, Punktgrösse und Farbe können unterschiedliche Variablen darstellen

## Scatter Plot Matrix
Folie 39

- Definition: Raster aus Scatterplots verschiedener Variablenpaare
- Zweck: Viele numerische Zusammenhänge gleichzeitig überblicken
- Bsp. Beziehungen zwischen Preis, Fläche und Zimmerzahl untersuchen
- Leseregel: Spalte bestimmt gewöhnlich die X-Variable, Zeile die Y-Variable
- Diagonale: Zeigt je nach Umsetzung die Verteilung der einzelnen Variable
- Achtung: Bei vielen Variablen wird die Darstellung schnell unübersichtlich
- Erkenntnis: Die Matrix eignet sich für den Überblick, auffällige Paare anschliessend einzeln untersuchen

## Hexagonal Binning Plot
Folie 40

- Definition: Wertepaare werden in sechseckige Bereiche eingeteilt und deren Häufigkeiten farblich dargestellt
- Zweck: Dichte und Verteilung vieler Punkte sichtbar machen
- Verwendung: Grosse Datensätze mit vielen überlappenden Punkten
- Bsp. Häufige Kombinationen von Wohnfläche und Mietpreis erkennen
- Farbe: Zeigt gemäss Legende die Anzahl bzw. eine transformierte Darstellung der Anzahl Beobachtungen
- Achtung: Eine dunklere Fläche bedeutet nicht automatisch einen höheren Preis
- Erkenntnis: Hexbin zeigt Punktdichten besser als ein stark überlagerter Scatterplot

## Correlation Heat Map
Folie 41

- Definition: Farbige Darstellung einer Korrelationsmatrix
- Zweck: Richtung und Stärke vieler Zusammenhänge schnell erkennen
- Farbe: Entspricht dem Korrelationswert gemäss Farbskala
- Diagonale: Zeigt bei nicht konstanten Variablen die Selbstkorrelation von 1
- Stärke: Den absoluten Wert ausserhalb der Diagonale vergleichen
- Bsp. −0.90 ist stärker als +0.60, aber gegenläufig
- Achtung: Pearson-Heatmaps zeigen lineare Zusammenhänge
- Achtung: Korrelation bedeutet keine Kausalität
- Erkenntnis: Auffällige Korrelationen mit Scatterplots und Fachwissen prüfen

## Matplotlib
Folien 42 bis 44

- Definition: Python-Bibliothek zur Erstellung und Anpassung von Grafiken
- Zweck: Diagramme flexibel aufbauen und ihre Darstellung steuern
- pyplot: Sammlung von Funktionen zur Erstellung und Veränderung einer Grafik
- Figure: Gesamte Grafik
- Axes: Einzelner Zeichenbereich mit Achsen
- Bsp. `fig, ax = plt.subplots()` erzeugt eine Figure mit einem Zeichenbereich

| Diagramm | pyplot-Methode |
|---|---|
| Scatterplot | `plt.scatter()` |
| Liniendiagramm | `plt.plot()` |
| Vertikales Balkendiagramm | `plt.bar()` |
| Horizontales Balkendiagramm | `plt.barh()` |
| Histogramm | `plt.hist()` |
| Boxplot | `plt.boxplot()` |
| Kreisdiagramm | `plt.pie()` |

- Objektorientierte Verwendung: Bspw. `ax.hist()` oder `ax.scatter()`
- Beschriftung: Titel, Achsen, Einheiten und Legenden passend angeben
- Erkenntnis: Ein korrektes Diagramm benötigt neben Daten auch eine verständliche Beschriftung

## Seaborn
Folie 45

- Definition: Python-Bibliothek für statistische Visualisierungen auf Basis von Matplotlib
- Zweck: Statistische Grafiken mit weniger manuellem Aufwand erstellen
- Verwendung: Direkte Zuordnung von DataFrame-Spalten zu Achsen und weiteren Darstellungsmerkmalen
- Bsp. `sns.scatterplot(data=df, x='area', y='price', hue='category')`
- Zusätzliche Merkmale: Bspw. Farbe über `hue`, Grösse über `size` und Stil über `style`
- Erweiterung: Die Folien nennen auch die deklarative Schnittstelle `seaborn.objects`
- Erkenntnis: Seaborn vereinfacht statistische Darstellungen, Matplotlib ermöglicht zusätzliche Anpassungen

Folien 46 bis 47

## Automated EDA
Folien 58 bis 59

- Definition: Werkzeuge erzeugen automatisch Übersichten und Berichte über einen Datensatz
- Zweck: Einen schnellen ersten Überblick über Struktur und Qualität gewinnen
- Inhalte: Kennzahlen, Verteilungen, Korrelationen, fehlende Werte, Duplikate und Warnhinweise
- Werkzeuge: Bspw. Profiling-Werkzeuge, Sweetviz und D-Tale
- Ausgabe: Bspw. interaktiver HTML-Bericht

- Grenze: Automatische Hinweise kennen die fachliche Bedeutung der Daten nur eingeschränkt
- Achtung: Eine Warnung ist ein Prüfhinweis, kein automatisch nachgewiesener Fehler
- Erkenntnis: Automatisierte Berichte unterstützen die EDA, ersetzen aber keine gezielte Interpretation

## AI-assisted EDA
Folie 60

- Definition: KI unterstützt bei Planung, Programmierung und Interpretation der explorativen Analyse
- Zweck: Wiederkehrende Arbeiten beschleunigen und Erklärungen erhalten

| Möglichkeit | Bsp. |
|---|---|
| Code erzeugen | pandas- und Seaborn-Code aus einer Beschreibung erstellen |

- Risiko: KI kann Spalten, Funktionen und Ergebnisse erfinden
- Prüfung: Berechnungen und Interpretationen mit den tatsächlichen Daten kontrollieren

- Datenschutz: Personenbezogene oder vertrauliche Daten nicht unkontrolliert an externe Dienste übermitteln
- Achtung: Ein überzeugend formulierter Befund ist nicht automatisch korrekt
- Erkenntnis: KI kann EDA beschleunigen, die fachliche Beurteilung bleibt notwendig

## Zentrale Unterscheidungen für Verständnisfragen

| Frage | Kernantwort |
|---|---|
| EDA oder Hypothesenprüfung? | Daten erkunden und Hypothesen entwickeln oder eine formulierte Hypothese statistisch prüfen |
| Univariate oder multivariate EDA? | Eine Variable oder mehrere Variablen und ihre Beziehungen untersuchen |
| Non-Graphical oder Graphical EDA? | Kennzahlen und Tabellen oder Diagramme verwenden |
| Warum Kennzahlen und Grafiken kombinieren? | Gleiche Kennzahlen können unterschiedliche Datenmuster verdecken |
| Mittelwert oder Median? | Durchschnitt oder robuster mittlerer Wert bei Schiefe und Ausreissern |
| Median oder Modalwert? | Mittlerer sortierter Wert oder häufigster Wert |
| Varianz oder Standardabweichung? | Quadrierte Streuung oder Streuung in der Originaleinheit |
| Schiefe oder Kurtosis? | Asymmetrie oder Ausprägung der Verteilungsenden |
| Rechtsschief oder linksschief? | Langer rechter oder linker Ausläufer |
| Kreuztabelle oder Korrelation? | Kategorienkombinationen oder einen bestimmten numerischen Zusammenhang untersuchen |
| Kovarianz oder Korrelation? | Einheitenabhängige gemeinsame Veränderung oder standardisierte Richtung und Stärke |
| Pearson oder Spearman? | Linearer Zusammenhang der Werte oder monotoner Zusammenhang der Rangplätze |
| Bedeutet Korrelation 0 keinen Zusammenhang? | Nein, nichtlineare Zusammenhänge können trotzdem vorliegen |
| Bedeutet Korrelation Kausalität? | Nein, gemeinsame Ursachen und andere Erklärungen sind möglich |
| Histogramm oder Balkendiagramm? | Numerische Intervalle oder Kategorien vergleichen |
| Boxplot oder Histogramm? | Kompakte Zusammenfassung oder detailliertere Verteilungsform |
| QQ-Plot oder Histogramm? | Verteilungen anhand ihrer Quantile vergleichen oder Häufigkeiten darstellen |
| Scatterplot oder Hexbin? | Einzelne Wertepaare oder Häufigkeiten in räumlichen Bereichen zeigen |
| Scatterplot oder Bubble Plot? | Zwei Variablen oder zusätzliche Merkmale über Grösse und Farbe darstellen |
| Plotly oder Dash? | Interaktive Grafik oder vollständige interaktive Datenanwendung |
| Leaflet oder Folium? | JavaScript-Kartenbibliothek oder Python-Zugang zu Leaflet-Karten |
| Automatisierter Bericht oder fachliche Analyse? | Standardübersicht oder gezielte Beurteilung im Kontext der Fragestellung |
