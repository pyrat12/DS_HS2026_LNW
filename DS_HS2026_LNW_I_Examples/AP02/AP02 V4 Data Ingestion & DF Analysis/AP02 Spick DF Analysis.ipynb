# DataFrames analysieren – Daten untersuchen und verstehen
- Ein Pandas-DataFrame (`pd.DataFrame`) ist eine zweidimensionale Datenstruktur mit Zeilen und benannten Spalten (ähnlich einer Tabelle in SQL)
- Nach dem Einlesen von CSV-, Excel-, JSON-, HTML- oder anderen Datenquellen kann der DataFrame mit verschiedenen Pandas-Methoden untersucht werden

# Daten prüfen und bereinigen

### Datentypen in Pandas

| Datentyp | Beschreibung | Bsp. | Typische Operationen |
|---|---|---|---|
| `int64` / `Int64` | Ganze Zahlen. `Int64` erlaubt zusätzlich fehlende Werte | `3`, `75000`, `-5` | Rechnen mit `+`, `-`, `*`, `/`; `sum()`, `mean()`, `median()`, Zahlenvergleiche |
| `float64` / `Float64` | Zahlen mit Dezimalstellen. Auch ganze Zahlen können so gespeichert sein | `3.5`, `1249.90` | Rechnen, `mean()`, `median()`, `std()`, `round(2)`, `between()` |
| `str` / `string` | Text oder Zeichenfolgen. Zahlen darin gelten zunächst als Text | `"Samsung"`, `"CHF 999.–"`, `"123"` | `.str.lower()`, `.str.upper()`, `.str.strip()`, `.str.replace()`, `.str.contains()`, `.str.extract()` |
| `object` | Allgemeiner Datentyp für Python-Objekte. Häufig Text, aber auch gemischte Inhalte | `"Samsung"` oder Mischung aus Text und Zahlen | Bei Textinhalten `.str`-Methoden; vor Zahlenoperationen Inhalte prüfen und gegebenenfalls konvertieren |
| `bool` / `boolean` | Wahrheitswerte. `boolean` erlaubt zusätzlich fehlende Werte | `True`, `False` | Filtern, UND `&`, ODER `\|`, NICHT `~`, `sum()` zählt wahre Werte, `mean()` berechnet deren Anteil |
| `datetime64[...]` | Datum und Uhrzeit. Die genaue Zeitauflösung kann variieren | `2026-10-30 14:30:00` | Chronologisch sortieren, Datumsvergleiche, `.dt.year`, `.dt.month`, `.dt.day`; Datumsdifferenzen berechnen |
| `timedelta64[...]` | Zeitdauer oder Differenz zwischen zwei Zeitpunkten | `3 days`, `02:30:00` | Zeitdauern addieren, vergleichen und mitteln, `.dt.total_seconds()` |
| `category` | Werte aus einer begrenzten Menge von Kategorien. Optional mit festgelegter Reihenfolge | `"Benzin"`, `"Diesel"`, `"Elektro"` | `value_counts()`, Gruppieren, Filtern. Bei geordneten Kategorien auch Reihenfolgevergleiche |

### Datentypen prüfen und umwandeln

| Befehl | Erklärung |
|---|---|
| `display(df.dtypes)` | Datentyp jeder Spalte anzeigen |
| `df.info()` | Spalten, Datentypen und Anzahl vorhandener Werte anzeigen |
| `df['spalte'] = pd.to_numeric(df['spalte'], errors='coerce')` | Zahlentexte numerisch umwandeln. Ungültige Werte werden fehlend |
| `df['spalte'] = df['spalte'].astype('Int64')` | Geeignete Werte als ganze Zahlen mit Unterstützung fehlender Werte speichern |
| `df['spalte'] = df['spalte'].astype('float64')` | Geeignete Werte als Dezimalzahlen speichern |
| `df['spalte'] = df['spalte'].astype('string')` | Werte als Text speichern. Fehlende Werte bleiben fehlend |
| `df['spalte'] = df['spalte'].astype('category')` | Werte als Kategorien speichern |
| `df['datum'] = pd.to_datetime(df['datum'], format='%d.%m.%Y', errors='coerce')` | Datumstexte umwandeln, bspw. `"30.10.2026"`. Ungültige Werte werden `NaT` |

**Wichtig:**
- `"123"` ist Text, `123` eine ganze Zahl und `123.0` eine Dezimalzahl. Numerische Filter und Berechnungen benötigen numerische Datentypen
- `int64` erlaubt keine fehlenden Werte; `Int64` erlaubt `<NA>`. `float64` verwendet für fehlende Werte üblicherweise `NaN`
- `NaN`, `<NA>` und `NaT` kennzeichnen fehlende Werte, mit `isna()` prüfen
- Vor `pd.to_numeric()` Währungen, Einheiten und Tausendertrennzeichen entfernen sowie Dezimalkommas durch Punkte ersetzen
- `astype()` kann bei ungeeigneten Werten einen Fehler auslösen
- `astype('bool')` interpretiert nicht leere Texte als `True` – auch `"False"`. Solche Textwerte gezielt zuordnen, bspw. mit `.map({'True': True, 'False': False}).astype('boolean')`

```python
#Einfache Zahlentexte wie "256" in Ganzzahlen umwandeln; fehlende Werte bleiben erlaubt
df_dfaspick_smartphones['storage_integer'] = pd.to_numeric(
    df_dfaspick_smartphones['storage_gb'], errors='coerce'                      #Zahlentexte umwandeln; ungültige Angaben => fehlend
).astype('Int64')                                                               #"256" => 256; fehlend => <NA>

display(df_dfaspick_smartphones[['storage_gb', 'storage_integer']].head())      #Ausgangswerte und Ergebnis vergleichen
display(df_dfaspick_smartphones['storage_integer'].dtype)                       #Datentyp kontrollieren: Int64

#ANPASSEN: DataFrame, Ausgangsspalte 'storage_gb' und Zielspalte 'storage_integer'
#ACHTUNG: Gültige Werte müssen ganzzahlig sein; "256GB" muss vorher bereinigt werden


#Mehrere Spalten mit einfachen Zahlentexten in Ganzzahlen umwandeln
for spalte in ['storage_gb', 'battery_mah']:                                    #Ausgewählte Spalten einzeln durchlaufen
    df_dfaspick_smartphones[spalte] = pd.to_numeric(
        df_dfaspick_smartphones[spalte], errors='coerce'
    ).astype('Int64')                                                           #"256" => 256; ungültige Angaben => <NA>

display(df_dfaspick_smartphones[['storage_gb', 'battery_mah']].head())          #Umgewandelte Werte kontrollieren
display(df_dfaspick_smartphones[['storage_gb', 'battery_mah']].dtypes)          #Beide Spalten sollten Int64 sein

#ANPASSEN: DataFrame und Spaltenliste in Schleife und Anzeigen
#ACHTUNG: Diese Variante überschreibt die ausgewählten Spalten; nur ganzzahlige Werte verwenden


#Einfache Zahlentexte wie "849.50" in Dezimalzahlen umwandeln
df_dfaspick_smartphones['price_float'] = pd.to_numeric(
    df_dfaspick_smartphones['price_chf'], errors='coerce'                       #Zahlentexte umwandeln; ungültige Angaben => NaN
).astype('float64')                                                             #"849.50" => 849.5; "849" => 849.0

display(df_dfaspick_smartphones[['price_chf', 'price_float']].head())           #Ausgangswerte und Ergebnis vergleichen
display(df_dfaspick_smartphones['price_float'].dtype)                           #Datentyp kontrollieren: float64

#ANPASSEN: DataFrame, Ausgangsspalte 'price_chf' und Zielspalte 'price_float'
#ACHTUNG: Dezimalstellen müssen mit einem Punkt geschrieben sein; Währungen und Einheiten vorher entfernen


#Werte ausdrücklich als Text speichern, beispielsweise Produkt-IDs, Modellnamen oder Postleitzahlen
df_dfaspick_smartphones['product_id_text'] = (
    df_dfaspick_smartphones['product_id'].astype('string')
)                                                                               #"PM-10001" bleibt Text; fehlende Werte bleiben <NA>

display(df_dfaspick_smartphones[['product_id', 'product_id_text']].head())      #Ausgangswerte und Textspalte vergleichen

#ANPASSEN: DataFrame, Ausgangs- und Zielspalte, bspw. 'model' und 'model_text'
#ACHTUNG: Führende Nullen, die beim Einlesen bereits verloren gingen, werden dadurch nicht wiederhergestellt


#Preisangaben mit CHF, Tausenderapostrophen und Preisendungen bereinigen und in Zahlen umwandeln
df_dfaspick_smartphonesraw['price_numeric'] = pd.to_numeric(
    df_dfaspick_smartphonesraw['price_raw']
        .astype('string')                                                       #Textdatentyp sicherstellen; fehlende Werte bleiben erhalten
        .str.replace('CHF', '', regex=False)                                    #Währungszeichen entfernen
        .str.replace(r"['’]", '', regex=True)                                   #Beide Tausenderapostrophe entfernen: "1'249" => "1249"
        .str.replace(r'\.[–—-]', '', regex=True)                                #Preisendungen .–, .— und .- entfernen
        .str.replace(',', '.', regex=False)                                     #Dezimalkomma durch Punkt ersetzen
        .str.strip(),                                                           #Äusseren Leerraum entfernen
    errors='coerce'                                                             #"Preis auf Anfrage" und andere ungültige Angaben => NaN
)

display(
    df_dfaspick_smartphonesraw[['price_raw', 'price_numeric']].head()
)                                                                               #"CHF 1'249.–" => 1249.0; Rohtext bleibt erhalten

#ANPASSEN: DataFrame, Rohtextspalte 'price_raw' und Zielspalte 'price_numeric'
#ANPASSEN: Währung, Einheiten und Trennzeichen an die tatsächlichen Rohtexte anpassen
#ACHTUNG: Kommas nur durch Punkte ersetzen, wenn sie Dezimalzeichen und keine Tausendertrennzeichen sind


#Nach der Rohpreisumwandlung alle Geräte mit einem numerischen Preis über 1000 CHF anzeigen
display(
    df_dfaspick_smartphonesraw[
        df_dfaspick_smartphonesraw['price_numeric'] > 1000
    ]
)                                                                               #Genau 1000 ist ausgeschlossen; fehlende Preise werden nicht ausgewählt

#ANPASSEN: DataFrame und numerische Spalte; hier zuvor den Rohpreisumwandlungs-Block ausführen
#ANPASSEN: > 1000 an die verlangte Bedingung anpassen, bspw. <= 500 oder .between(500, 1000)


#Nach der Rohpreisumwandlung alle Geräte anzeigen, deren Preis fehlt oder nicht konvertiert werden konnte
display(
    df_dfaspick_smartphonesraw[
        df_dfaspick_smartphonesraw['price_numeric'].isna()
    ]
)                                                                               #Betroffene Zeilen inklusive ursprünglicher Preisangabe anzeigen

#ANPASSEN: DataFrame und numerische Spalte; hier zuvor den Rohpreisumwandlungs-Block ausführen
#ACHTUNG: NaN kann durch ursprünglich fehlende Angaben oder eine unpassende Bereinigung entstanden sein


#Bei einer bereits numerischen Preisspalte direkt alle Geräte über 1000 CHF anzeigen
display(
    df_dfaspick_smartphones[
        df_dfaspick_smartphones['price_chf'] > 1000
    ]
)                                                                               #Keine vorherige Textbereinigung oder Umwandlung erforderlich

#ANPASSEN: DataFrame, numerische Spalte und Filterbedingung


#Bei einer bereits numerischen Preisspalte direkt alle Geräte ohne Preis anzeigen
display(
    df_dfaspick_smartphones[
        df_dfaspick_smartphones['price_chf'].isna()
    ]
)                                                                               #Nur Datensätze mit fehlendem Preis auswählen

#ANPASSEN: DataFrame und numerische Spalte


#Wiederkehrende Markennamen als Kategorien speichern, ohne die ursprüngliche Markenspalte zu überschreiben
df_dfaspick_smartphones['brand_category'] = (
    df_dfaspick_smartphones['brand'].astype('category')
)                                                                               #"Samsung", "Apple", ... => Kategorien ohne festgelegte Rangfolge

display(df_dfaspick_smartphones[['brand', 'brand_category']].head())            #Ausgangswerte und Kategoriespalte vergleichen
display(df_dfaspick_smartphones['brand_category'].value_counts())               #Anzahl Geräte je Marke anzeigen

#ANPASSEN: DataFrame, Ausgangs- und Zielspalte
#ACHTUNG: Textspalten funktionieren auch ohne astype('category') bei groupby(), Kreuztabellen und One-Hot Encoding


#Verfügbarkeit auf einen bestimmten Text prüfen und als True, False oder fehlenden Wahrheitswert speichern
df_dfaspick_smartphones['available_bool'] = (
    df_dfaspick_smartphones['availability']
        .astype('string')                                                       #Textdatentyp mit erlaubten fehlenden Werten
        .eq('Sofort lieferbar')                                                 #Exakte Übereinstimmung prüfen
        .astype('boolean')
)                                                                               #Passender Text => True; anderer Text => False; fehlend => <NA>

display(df_dfaspick_smartphones[['availability', 'available_bool']].head())      #Ausgangstext und Wahrheitswert vergleichen

display(
    df_dfaspick_smartphones[df_dfaspick_smartphones['available_bool']]
)                                                                               #Nur Zeilen mit True anzeigen; False und <NA> werden ausgeschlossen

#ANPASSEN: DataFrame, Ausgangs- und Zielspalte sowie Vergleichstext
#ACHTUNG: astype(bool) allein interpretiert den nicht leeren Text "False" ebenfalls als True


#Vorhandene Wahrheitswerte in eine binäre Zahlenspalte mit 0 und 1 umwandeln
df_dfaspick_smartphones['available_binary'] = (
    df_dfaspick_smartphones['available_bool'].astype('Int64')
)                                                                               #True => 1; False => 0; fehlend => <NA>

display(df_dfaspick_smartphones[['available_bool', 'available_binary']].head()) #Wahrheitswerte und binäre Zahlen vergleichen

#ANPASSEN: DataFrame, Ausgangs- und Zielspalte
#VORAUSSETZUNG: Die Wahrheitswertspalte muss existieren; hier zuvor den Verfügbarkeits-Block ausführen


#Ein Beispieldatum als Text anlegen, weil die Smartphone-Daten ursprünglich keine Datumsspalte enthalten
df_dfaspick_smartphones['date_text'] = '01.11.2026'                             #Reines Umwandlungsbeispiel; kein echtes Erfassungsdatum
display(df_dfaspick_smartphones[['date_text']].head())

#ANPASSEN: Diesen Beispielblock bei einer bereits vorhandenen Datumsspalte weglassen


#Datumstexte in echte Datumswerte umwandeln, damit zeitliche Berechnungen und .dt-Methoden möglich sind
df_dfaspick_smartphones['date_parsed'] = pd.to_datetime(
    df_dfaspick_smartphones['date_text'],
    format='%d.%m.%Y',                                                          #Eingabeformat: Tag.Monat.Jahr
    errors='coerce'                                                             #Ungültige Angaben => NaT
)

display(df_dfaspick_smartphones[['date_text', 'date_parsed']].head())           #Datumstext und echten Datumswert vergleichen

#ANPASSEN: DataFrame, Ausgangs- und Zielspalte sowie das tatsächliche Eingabeformat
#VORAUSSETZUNG: Datumstextspalte vorhanden; hier zuvor den Beispielblock ausführen


#Echte Datumswerte für die Darstellung wieder in formatierten Text umwandeln
df_dfaspick_smartphones['date_formatted'] = (
    df_dfaspick_smartphones['date_parsed'].dt.strftime('%Y-%m-%d')
)                                                                               #Datum => Text wie "2026-11-01"

display(df_dfaspick_smartphones[['date_parsed', 'date_formatted']].head())      #Datumswert und formatierten Text vergleichen

#ANPASSEN: DataFrame, Ausgangs- und Zielspalte sowie das gewünschte Ausgabeformat
#ACHTUNG: Das Ergebnis ist Text; für Datumsberechnungen weiterhin die echte Datumsspalte verwenden


#Datentypen und Beispielzeilen nach den Umwandlungen kontrollieren
display(df_dfaspick_fountains.dtypes)                                           #Datentypen der Trinkbrunnen-Spalten
display(df_dfaspick_smartphones.dtypes)                                         #Datentypen der Smartphone-Spalten inklusive neuer Spalten
display(df_dfaspick_smartphones.head())                                         #Erste fünf Smartphone-Datensätze
display(df_dfaspick_smartphonesraw[['price_raw', 'price_numeric']].dtypes)      #Rohpreis sollte Text, price_numeric numerisch sein

#ANPASSEN: DataFrames und ausgewählte Spalten an die tatsächlich ausgeführten Umwandlungen anpassen


#Bereits numerische Trinkbrunnen-Spalten auswählen und ihre statistischen Kennzahlen anzeigen
display(
    df_dfaspick_fountains.select_dtypes(include='number').describe()
)                                                                               #Anzahl, Mittelwert, Streuung, Quartile und Extremwerte je numerischer Spalte

#ANPASSEN: DataFrame; numerisch gespeicherte IDs sind nicht automatisch sinnvolle Messwerte


#Den Median der vorhandenen numerischen Smartphone-Preise berechnen und mit Einheit ausgeben
print(
    f"Der Median ist {df_dfaspick_smartphones['price_chf'].median()} CHF"
)                                                                               #Fehlende Preise werden ignoriert; bei gerader Anzahl Mittelwert der beiden mittleren Werte

#ANPASSEN: DataFrame, numerische Preisspalte und Einheit im Ausgabetext
```

### Missing Values – Fehlende Werte erkennen

| Befehl | Erklärung |
|---|---|
| `df.isna()` | Prüft jede Zelle auf fehlende Werte. `True` bedeutet fehlend |
| `df.isna().sum()` | Anzahl fehlender Werte pro Spalte |
| `df.isna().sum().sum()` | Gesamtanzahl fehlender Werte im gesamten DataFrame |
| `df.isna().mean() * 100` | Prozentualer Anteil fehlender Werte pro Spalte |
| `df.isna().any(axis=1)` | Prüft für jede Zeile, ob mind. ein Wert fehlt |
| `df[df.isna().any(axis=1)]` | Zeigt alle Datensätze mit mind. einem fehlenden Wert |
| `df.notna().sum()` | Anzahl vorhandener Werte pro Spalte |
| `df.count()` | Anzahl nicht fehlender Werte pro Spalte |
| `df.isna().sum().loc[lambda x: x > 0]` |	Zeigt nur Spalten mit fehlenden Werten und deren Anzahl

**Wichtig:** `isna()` erkennt bspw. `NaN`, `None` und `NaT`. Leere Strings (`''`) oder reine Leerzeichen gelten nicht automatisch als fehlende Werte

**Unterschied:** `count()` zählt nur vorhandene Werte pro Spalte, während `len(df)` sämtliche Zeilen zählt

```python
display(df_dfaspick_fountains.isna().sum())                                     #Anzahl fehlender Werte pro Spalte anzeigen

display(df_dfaspick_smartphones[df_dfaspick_smartphones.isna().any(axis=1)])    #Ganze Zeilen anzeigen, in denen mind. ein Wert fehlt

display(df_dfaspick_smartphones[['price_chf', 'brand']].isna().sum())           #Nur bestimmte Spalten prüfen

#ANPASSEN: Beim Zeilenfilter den DataFrame-Namen ausserhalb und innerhalb der eckigen Klammern gleich anpassen
#OPTIONAL: Nur bestimmte Spalten prüfen → df[['spalte1', 'spalte2']].isna().sum()
#OPTIONAL: Zeilen mit fehlendem Wert in einer bestimmten Spalte anzeigen → df[df['spalte'].isna()]
#ACHTUNG: isna() erkennt NaN, None und NaT; leere Strings und reine Leerzeichen gelten nicht automatisch als fehlend
```

### Statistische Kennzahlen untersuchen

| Befehl | Erklärung |
|---|---|
| `df.describe()` | Deskriptive Statistik der numerischen Spalten: Anzahl, Mittelwert, Standardabweichung, Minimum, Quartile und Maximum |
| `df.describe(include='all')` | Statistische Übersicht über alle Spalten, einschliesslich kategorialer Spalten |
| `df['price'].min()` | Kleinsten Wert einer numerischen Spalte ermitteln |
| `df['price'].max()` | Grössten Wert einer numerischen Spalte ermitteln |
| `df['price'].mean()` | Arithmetischen Mittelwert berechnen |
| `df['price'].median()` | Median berechnen |
| `df['price'].sum()` | Alle vorhandenen numerischen Werte aufsummieren |

**Interpretation von `describe()`:**
- **`count`:** Anzahl nicht fehlender Werte (fehlende Werte werden nicht mitgezählt)
- **`mean`:** Arithmetischer Mittelwert (Summe aller Werte geteilt durch deren Anzahl)
- **`std`:** Standardabweichung (wie stark die Werte um den Durchschnitt streuen; grösser bedeutet stärker verstreut)
- **`min` / `max`:** Kleinster bzw. grösster Wert
- **`25%` / `50%` / `75%`:** Erstes Quartil, Median und drittes Quartil (25 %, 50 % bzw. 75 % der Werte liegen auf oder unter dem jeweiligen Wert)
- **`unique`:** Anzahl unterschiedlicher Kategorien (bei geeigneten nicht numerischen Spalten, jede Kategorie wird nur einmal gezählt)
- **`top`:** Häufigste Kategorie
- **`freq`:** Häufigkeit der häufigsten Kategorie

### DataFrame gezielt filtern

| Befehl | Erklärung |
|---|---|
| `df[['brand', 'price_chf']]` | Nur ausgewählte Spalten anzeigen |

| `df[df['brand'] == 'Samsung']` | Nur Samsung auswählen |
| `df[df['brand'] != 'Samsung']` | Samsung ausschliessen |
| `df[(df['brand'] == 'Samsung') & (df['price_chf'] > 1000)]` | UND: Samsung mit einem Preis über 1000 auswählen |
| `df[(df['brand'] == 'Samsung') \| (df['brand'] == 'Apple')]` | ODER: Samsung oder Apple auswählen |

| `df[~df['brand'].isin(['Samsung', 'Apple'])]` | Marken aus einer Liste ausschliessen |

**Wichtig:** `==` bedeutet gleich, `!=` ungleich, `>` grösser, `<` kleiner, `>=` grösser oder gleich und `<=` kleiner oder gleich. Mehrere Bedingungen jeweils einklammern und mit `&` (UND) oder `|` (ODER) verbinden; `~` kehrt eine Bedingung um. Die Auswahl verändert den ursprünglichen DataFrame nicht

# DataFrames analysieren – Textsuche und numerische Werte filtern

**Erkenntnis:** Mit Pandas können DataFrames gezielt nach Texten, Zahlen, Wertebereichen und kombinierten Bedingungen durchsucht werden. Dabei werden nur die Datensätze angezeigt, die bestimmte Suchkriterien erfüllen. Dadurch können auch sehr grosse Datensätze effizient analysiert werden

**Zweck:** Auffällige Werte finden, bestimmte Produkte oder Kategorien identifizieren, Preise vergleichen und Datensätze für weitere Analysen auswählen

**Wichtige Regeln für die Prüfung:**
- Textwerte werden in Anführungszeichen geschrieben: `'Samsung'`
- Numerische Werte werden ohne Anführungszeichen geschrieben: `1000`
- Für kombinierte Bedingungen `&`, `|` und `~` verwenden, nicht die Python-Operatoren `and`, `or` und `not`
- Bei kombinierten Pandas-Filtern jede Bedingung einzeln einklammern
- Vor numerischen Vergleichen sicherstellen, dass die Spalte numerische Werte enthält
- Fehlende Werte können Filterergebnisse beeinflussen und sollten separat untersucht werden
- Bei grossen DataFrames Ergebnisse mit `.head(n)` begrenzen


### Textsuche: Textwerte und Zeichenfolgen finden

| Befehl | Erklärung |
|---|---|
| `df[df['price_raw'] == 'Preis auf Anfrage']` | Alle Datensätze anzeigen, deren Preistext exakt «Preis auf Anfrage» lautet |
| `df[df['price_raw'] != 'Preis auf Anfrage']` | Alle Datensätze anzeigen, deren Preistext nicht exakt «Preis auf Anfrage» lautet |
| `df[df['price_raw'].str.contains('Anfrage', case=False, na=False)]` | Alle Datensätze anzeigen, deren Preistext das Wort «Anfrage» enthält |
| `df[df['price_raw'].str.contains('Preis auf Anfrage', case=False, na=False, regex=False)]` | Einen bestimmten Text unabhängig von der Gross-/Kleinschreibung suchen und wörtlich interpretieren |
| `df[df['brand'].str.startswith('Sam', na=False)]` | Alle Datensätze anzeigen, deren Markenname mit «Sam» beginnt |
| `df[df['brand'].str.endswith('ung', na=False)]` | Alle Datensätze anzeigen, deren Markenname auf «ung» endet |
| `df[df['brand'].isin(['Samsung', 'Apple'])]` | Datensätze anzeigen, deren Marke Samsung oder Apple ist |
| `df[df['brand'].str.contains('Samsung', case=False, na=False)]` | Alle Datensätze anzeigen, deren Markenname «Samsung» enthält |
| `df['brand'].value_counts()` | Häufigkeit der einzelnen Marken ermitteln |
| `df['brand'].unique()` | Alle unterschiedlichen Marken anzeigen |
| `df['brand'].nunique()` | Anzahl unterschiedlicher Marken ermitteln |

**Wichtige Funktionen und Parameter:**
- **`==`:** Exakter Vergleich eines Werts (Gross-/Kleinschreibung beachten)
- **`!=`:** Alle Werte ausser dem angegebenen Wert auswählen
- **`.str.contains('Text')`:** Prüft, ob eine bestimmte Zeichenfolge innerhalb eines Textwerts vorkommt
- **`case=False`:** Ignoriert die Gross-/Kleinschreibung bei `str.contains()`
- **`na=False`:** Behandelt fehlende Werte als Nicht-Treffer
- **`regex=False`:** Interpretiert den Suchtext wörtlich statt als regulären Ausdruck
- **`.str.startswith('Text')`:** Sucht Werte, die mit einem bestimmten Text beginnen
- **`.str.endswith('Text')`:** Sucht Werte, die mit einem bestimmten Text enden
- **`.isin([...])`:** Prüft, ob ein Wert in einer Liste vorgegebener Werte enthalten ist
- **`.astype('string')`:** Konvertiert eine Spalte in den Pandas-String-Datentyp, damit Textoperationen auch bei gemischten Ausgangsdaten möglich sind
- **`.value_counts()`:** Zählt die Häufigkeit einzelner Werte
- **`.unique()`:** Gibt die unterschiedlichen Werte zurück
- **`.nunique()`:** Zählt die unterschiedlichen Werte

**Textsuche über den gesamten DataFrame:**
- Mit `df.astype('string')` lassen sich alle Spalten als Text betrachten
- Mit `.apply()` kann jede Spalte nach einer Zeichenfolge durchsucht werden
- Mit `.any(axis=1)` werden jene Zeilen identifiziert, in denen mindestens eine Spalte den Suchtext enthält

**Wichtig:**
- `==` findet nur exakte Übereinstimmungen, während `str.contains()` auch Teiltexte findet
- `str.contains()` unterstützt reguläre Ausdrücke. Für eine gewöhnliche Textsuche empfiehlt sich `regex=False`
- Bei numerischen oder gemischten Spalten vor Textoperationen gegebenenfalls `.astype('string')` verwenden
- Fehlende Werte sollten bei Textsuchen mit `na=False` berücksichtigt werden
- Die Suche über sämtliche Spalten kann bei grossen DataFrames aufwendig sein

```python
#Denke daran head(2)) bzw. [:2] rauszulöschen. Habe sie hier nur drin, um den Output zu beschränken

#Exakten Textwert suchen
#Alle Datensätze anzeigen, deren Preistext exakt 'Preis auf Anfrage' lautet
#==: Exakte Übereinstimmung inklusive Gross-/Kleinschreibung
display(df_dfaspick_smartphonesraw[df_dfaspick_smartphonesraw['price_raw'] == 'Preis auf Anfrage'])

#Bestimmten Textwert ausschliessen
#Alle Datensätze anzeigen (limitiert auf 5), deren Preistext nicht exakt 'Preis auf Anfrage' lautet
#!=: Ungleich-Vergleich durchführen
display(df_dfaspick_smartphonesraw[df_dfaspick_smartphonesraw['price_raw'] != 'Preis auf Anfrage'].head(2))

#Nach einem Text innerhalb einer Spalte suchen
#astype('string'): Spaltenwerte als Text behandeln
#str.contains(): Text innerhalb eines Werts suchen
#case=False: Gross-/Kleinschreibung ignorieren
#na=False: Fehlende Werte als Nicht-Treffer behandeln
#regex=False: Suchtext wörtlich interpretieren
display(df_dfaspick_smartphonesraw[df_dfaspick_smartphonesraw['price_raw'].astype('string').str.contains('Preis auf Anfrage', case=False, na=False, regex=False)])

#Nach einem Teilwort suchen
#Alle Datensätze anzeigen, deren Preistext 'Anfrage' enthält
display(df_dfaspick_smartphonesraw[df_dfaspick_smartphonesraw['price_raw'].astype('string').str.contains('Anfrage', case=False, na=False, regex=False)])

#Text suchen, der mit einer bestimmten Zeichenfolge beginnt
#Marken suchen, deren Name mit 'Sam' beginnt
#str.startswith(): Anfang einer Zeichenfolge überprüfen
display(df_dfaspick_smartphonesraw[df_dfaspick_smartphonesraw['brand'].astype('string').str.startswith('Sam', na=False)].head(2))

#Text suchen, der mit einer bestimmten Zeichenfolge endet
#Marken suchen, deren Name auf 'ung' endet
#str.endswith(): Ende einer Zeichenfolge überprüfen
display(df_dfaspick_smartphonesraw[df_dfaspick_smartphonesraw['brand'].astype('string').str.endswith('ung', na=False)].head(2))

#Mehrere Textwerte gleichzeitig suchen
#Alle Datensätze der Marken Samsung oder Apple anzeigen
#isin(): Prüfen, ob ein Wert in der angegebenen Liste enthalten ist
display(df_dfaspick_smartphonesraw[df_dfaspick_smartphonesraw['brand'].isin(['Samsung', 'Apple'])].head(2))

#Alle Datensätze der Marke Samsung anzeigen
display(df[df['brand'] == 'Samsung'].head(2))

#Alle unterschiedlichen Textwerte ermitteln
#Alle vorhandenen Marken anzeigen
#unique(): Gibt jeden unterschiedlichen Wert nur einmal zurück
print(df_dfaspick_smartphonesraw['brand'].unique()[:2])                         #unique() liefert ein Array; deshalb [:2] statt .head(2)
#Anzahl unterschiedlicher Textwerte ermitteln
#Anzahl unterschiedlicher Marken ermitteln
#nunique(): Fehlende Werte standardmässig nicht mitzählen
print(df_dfaspick_smartphonesraw['brand'].nunique())

#Häufigkeiten der Textwerte ermitteln
#Anzahl Datensätze pro Marke anzeigen
#value_counts(): Häufigkeiten standardmässig absteigend sortieren
print(df_dfaspick_smartphonesraw['brand'].value_counts())

#Die zehn häufigsten Textwerte (Marken) anzeigen
print(df_dfaspick_smartphonesraw['brand'].value_counts().head(10))       

#Alle Spalten nach einem bestimmten Text durchsuchen
#astype('string'): Sämtliche Spalten als Text behandeln
#apply(): Suchfunktion auf jede Spalte anwenden
#lambda: Funktion für jede einzelne Spalte definieren
#any(axis=1): True, wenn mindestens eine Spalte pro Zeile den Suchtext enthält
#Alle Datensätze anzeigen, in denen der Suchtext irgendwo vorkommt
treffer = df_dfaspick_smartphonesraw.astype('string').apply(
    lambda spalte: spalte.str.contains('Preis auf Anfrage', case=False, na=False, regex=False)
).any(axis=1)
display(df_dfaspick_smartphonesraw[treffer])
```

### Numerische Werte: Zahlen und Wertebereiche finden

| Befehl | Erklärung |
|---|---|
| `df[df['price'] == 999]` | Alle Datensätze mit dem exakten Preis 999 anzeigen |
| `df[df['price'] != 999]` | Alle Datensätze ausser denen mit dem Preis 999 anzeigen |
| `df[df['price'] > 1000]` | Alle Datensätze mit einem Preis über 1000 anzeigen |
| `df[df['price'] >= 1000]` | Alle Datensätze mit einem Preis von mindestens 1000 anzeigen |
| `df[df['price'] < 500]` | Alle Datensätze mit einem Preis unter 500 anzeigen |
| `df[df['price'] <= 500]` | Alle Datensätze mit einem Preis von höchstens 500 anzeigen |
| `df[df['price'].between(500, 1000)]` | Alle Datensätze mit Preisen zwischen 500 und 1000 anzeigen, inklusive Grenzwerte |
| `df[df['price'].isin([499, 799, 999])]` | Alle Datensätze anzeigen, deren Preis einem der angegebenen Werte entspricht |
| `df[(df['price'] > 500) & (df['price'] < 1000)]` | Preise zwischen 500 und 1000 suchen; beide Bedingungen müssen erfüllt sein |
| `df[(df['price'] < 500) \| (df['price'] > 1000)]` | Preise unter 500 oder über 1000 suchen; mindestens eine Bedingung muss erfüllt sein |
| `df[~df['price'].between(500, 1000)]` | Alle Datensätze ausserhalb des angegebenen Preisintervalls anzeigen |
| `df[df['price'].notna()]` | Datensätze mit vorhandenem Preis anzeigen |
| `df[df['price'].isna()]` | Datensätze mit fehlendem Preis anzeigen |
| `df.loc[df['price'] > 1000, ['brand', 'price']]` | Datensätze über 1000 filtern und nur Marke und Preis anzeigen |

**Vergleichsoperatoren:**
- **`==`:** Gleich
- **`!=`:** Ungleich
- **`>`:** Grösser als
- **`>=`:** Grösser oder gleich
- **`<`:** Kleiner als
- **`<=`:** Kleiner oder gleich
- **`&`:** Logisches UND – beide Bedingungen müssen erfüllt sein
- **`|`:** Logisches ODER – mindestens eine Bedingung muss erfüllt sein
- **`~`:** Logisches NICHT – eine boolesche Bedingung umkehren

**Weitere wichtige Methoden:**
- **`.between(min, max)`:** Prüft, ob ein Wert innerhalb eines Intervalls liegt; standardmässig inklusive beider Grenzen
- **`.isin([wert1, wert2])`:** Sucht mehrere bestimmte Zahlen gleichzeitig
- **`.loc[bedingung, spalten]`:** Filtert Zeilen und wählt gleichzeitig bestimmte Spalten aus
- **`.isna()`:** Erkennt fehlende Werte
- **`.notna()`:** Erkennt vorhandene Werte

```python
#Exakten numerischen Wert suchen
#Alle Datensätze mit dem exakten Preis 999 anzeigen
#==: Exakten numerischen Vergleich durchführen
display(df_dfaspick_smartphones[df_dfaspick_smartphones['price_chf'] == 999])

#Bestimmten numerischen Wert ausschliessen
#Alle Datensätze anzeigen, deren Preis nicht 999 beträgt
#!=: Ungleich-Vergleich durchführen
display(df_dfaspick_smartphones[df_dfaspick_smartphones['price_chf'] != 999])

#Grösser als einen bestimmten Wert
#Alle Datensätze mit einem Preis über 1000 anzeigen
#>: Grösser als, ohne Grenzwert
display(df_dfaspick_smartphones[df_dfaspick_smartphones['price_chf'] > 1000])

#Grösser oder gleich einem bestimmten Wert
#Alle Datensätze mit einem Preis von mindestens 1000 anzeigen
#>=: Grösser oder gleich, inklusive Grenzwert
display(df_dfaspick_smartphones[df_dfaspick_smartphones['price_chf'] >= 1000]) 

#Kleiner als einen bestimmten Wert
#Alle Datensätze mit einem Preis unter 500 anzeigen
#<: Kleiner als, ohne Grenzwert
display(df_dfaspick_smartphones[df_dfaspick_smartphones['price_chf'] < 500])

#Kleiner oder gleich einem bestimmten Wert
#Alle Datensätze mit einem Preis von höchstens 500 anzeigen
#<=: Kleiner oder gleich, inklusive Grenzwert
display(df_dfaspick_smartphones[df_dfaspick_smartphones['price_chf'] <= 500]) 

#Numerischen Wertebereich durchsuchen
#Alle Datensätze mit Preisen zwischen 500 und 1000 anzeigen
#between(): Prüfen, ob Werte innerhalb eines Intervalls liegen
#Standardmässig beide Grenzwerte eingeschlossen
display(df_dfaspick_smartphones[df_dfaspick_smartphones['price_chf'].between(500, 1000)]) 

#Mehrere bestimmte Zahlen gleichzeitig suchen
#Alle Datensätze mit einem der angegebenen Preise anzeigen
#isin(): Prüfen, ob ein Wert in einer Liste enthalten ist
display(df_dfaspick_smartphones[df_dfaspick_smartphones['price_chf'].isin([499, 799, 999])])

#Mehrere Bedingungen mit UND kombinieren
#Preise zwischen 500 und 1000 anzeigen, ohne Grenzwerte
#&: Beide Bedingungen müssen erfüllt sein
#(): Jede Bedingung einzeln einklammern
display(df_dfaspick_smartphones[(df_dfaspick_smartphones['price_chf'] > 500) & (df_dfaspick_smartphones['price_chf'] < 1000)])

#Mehrere Bedingungen mit ODER kombinieren
#Preise unter 500 oder über 1000 anzeigen
#|: Mindestens eine Bedingung muss erfüllt sein
display(df_dfaspick_smartphones[(df_dfaspick_smartphones['price_chf'] < 500) | (df_dfaspick_smartphones['price_chf'] > 1000)])

#Eine Bedingung umkehren
#Alle Datensätze ausserhalb des Preisintervalls anzeigen
#~: Boolesche Bedingung umkehren
#ACHTUNG: Fehlende Werte können dabei ebenfalls ausgewählt werden
display(df_dfaspick_smartphones[~df_dfaspick_smartphones['price_chf'].between(500, 1000)]) 

#Nur Datensätze mit vorhandenem Preis anzeigen
#Datensätze mit nicht fehlendem Preis anzeigen
#notna(): Vorhandene Werte erkennen
display(df_dfaspick_smartphones[df_dfaspick_smartphones['price_chf'].notna()]) 

#Nur Datensätze mit fehlendem Preis anzeigen
#Datensätze mit fehlendem Preis anzeigen
#isna(): Fehlende Werte wie NaN erkennen
display(df_dfaspick_smartphones[df_dfaspick_smartphones['price_chf'].isna()]) 

#Numerischen Filter anwenden und bestimmte Spalten auswählen
#Nur Marke und Preis von Smartphones über 1000 anzeigen
#loc[Zeilenbedingung, Spaltenliste]: Zeilen und Spalten gezielt auswählen
display(df_dfaspick_smartphones.loc[df_dfaspick_smartphones['price_chf'] > 1000, ['brand', 'price_chf']])
```

### Grösste und kleinste Werte ermitteln

| Befehl | Erklärung |
|---|---|

| `df['price'].mean()` | Durchschnittlichen Preis berechnen |

**Unterschied:**
- **`max()` / `min()`:** Liefern einzelne numerische Werte
- **`nlargest()` / `nsmallest()`:** Liefern vollständige Datensätze mit den grössten bzw. kleinsten Werten
- **`sort_values()`:** Sortiert den DataFrame nach bestimmten Spalten
- **`head(n)`:** Begrenzt die Ausgabe auf die ersten `n` Zeilen


### Textsuche und numerische Bedingungen kombinieren

Bsp.: Alle Samsung-Smartphones mit einem Preis über 1000 suchen

`df[(df['brand'] == 'Samsung') & (df['price'] > 1000)]`

**Interpretation:**
- Die erste Bedingung sucht nach Samsung
- Die zweite Bedingung sucht Preise über 1000
- Mit `&` müssen beide Bedingungen gleichzeitig erfüllt sein
- Jede einzelne Bedingung muss in Klammern stehen

```python
#Samsung-Smartphones mit einem Preis über 1000 suchen
#Nur Samsung-Smartphones mit einem Preis über 1000 anzeigen
#&: Beide Bedingungen müssen gleichzeitig erfüllt sein
display(df_dfaspick_smartphones[(df_dfaspick_smartphones['brand'] == 'Samsung') & (df_dfaspick_smartphones['price_chf'] > 1000)])

#Samsung- oder Apple-Smartphones mit Preisen zwischen 500 und 1000 suchen
#Nur Samsung- oder Apple-Smartphones innerhalb der angegebenen Preisspanne anzeigen
#isin(): Mehrere Marken berücksichtigen
#between(): Numerischen Wertebereich einschränken
#&: Beide Bedingungen kombinieren
display(df_dfaspick_smartphones[
    (df_dfaspick_smartphones['brand'].isin(['Samsung', 'Apple'])) &
    (df_dfaspick_smartphones['price_chf'].between(500, 1000))
])
```
