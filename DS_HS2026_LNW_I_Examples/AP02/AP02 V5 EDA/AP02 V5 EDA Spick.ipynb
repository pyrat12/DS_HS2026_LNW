# Exploratory Data Analysis (EDA) (V5) – Einführung

### 1. Was ist Exploratory Data Analysis (EDA)?

- **Exploratory Data Analysis (EDA / Explorative Datenanalyse):** Erster Schritt der Datenanalyse, bei dem Daten systematisch untersucht werden, um ihre Eigenschaften, Verteilungen, Muster und Zusammenhänge zu verstehen – ohne bereits formale statistische Modelle oder Hypothesentests durchzuführen.
- **Ziel:** Die Daten kennenlernen, ihre Qualität beurteilen und eine Grundlage für weiterführende statistische Analysen schaffen.

**Objectives of EDA (Ziele):**
- **Unexpected discoveries:** Unerwartete Muster, Trends und Auffälligkeiten in den Daten entdecken.
- **Hypotheses:** Mögliche Erklärungen für beobachtete Zusammenhänge finden und daraus Hypothesen entwickeln.
- **Assess assumptions:** Voraussetzungen für spätere statistische Verfahren überprüfen (z.B. Verteilung der Daten).
- **Selection of methods:** Geeignete statistische Methoden und Analyseverfahren auswählen.
- **Further data collection:** Erkennen, welche zusätzlichen Daten oder Untersuchungen notwendig sein könnten.
- **Data Quality:** Fehlende Werte (Missing Values), Duplikate (Duplicates), Ausreisser (Outliers) und unplausible Werte identifizieren.

### 2. Systematics of EDA Methods – Einteilung der Methoden

EDA-Methoden werden anhand zweier Dimensionen unterschieden:

**Anzahl untersuchter Variablen:**
- **Univariate EDA:** Untersuchung einer einzelnen Variable (z.B. Verteilung der Mietpreise).
- **Multivariate EDA:** Gemeinsame Untersuchung von zwei oder mehr Variablen, insbesondere deren Zusammenhänge (z.B. Zusammenhang zwischen Wohnfläche und Mietpreis).

**Art der Untersuchung:**
- **Non-Graphical EDA:** Analyse mithilfe statistischer Kennzahlen und Tabellen.
- **Graphical EDA:** Analyse mithilfe von Diagrammen und Visualisierungen.

| EDA-Methode | Non-Graphical (nicht-grafisch) | Graphical (grafisch) |
|---|---|---|
| **Univariate** | Mean, Median, Standard Deviation, Skewness | Histogram, Box Plot, Bar Chart |
| **Multivariate** | Correlation, Covariance, Contingency Table | Scatter Plot, Scatter Plot Matrix, Correlation Heatmap |

### 3. Univariate Non-Graphical EDA

- **Definition:** Untersucht die Verteilung und Eigenschaften einer einzelnen Variable mithilfe statistischer Kennzahlen. Ermöglicht unter anderem die Erkennung von Ausreissern.
- **Numerische Daten:** Verwendung von statistischen Kennzahlen:
  - **Measures of Central Tendency (Lagermasse):** Mean (Mittelwert), Median, Mode (Modalwert).
  - **Measures of Dispersion (Streuungsmasse):** Standard Deviation (Standardabweichung), Variance (Varianz), Range (Spannweite).
  - **Measures of Shape (Formmasse):** Skewness (Schiefe), Kurtosis (Wölbung).
- **Kategoriale Daten:** Untersuchung der Häufigkeiten einzelner Kategorien (z.B. Anzahl Fahrzeuge je Treibstoffart).

### 4. Multivariate Non-Graphical EDA

- **Definition:** Untersucht Beziehungen zwischen zwei oder mehr Variablen mithilfe statistischer Kennzahlen oder Tabellen.
- **Contingency Table (Kreuztabelle):** Zeigt die Häufigkeitsverteilung von Kombinationen kategorialer Variablen.
  - Beispiel: Anzahl Personen nach Geschlecht und bevorzugtem Dessert.
  - Ermöglicht, mögliche Zusammenhänge zwischen kategorialen Merkmalen zu erkennen.
- **Correlation (Korrelation):** Beschreibt die Stärke und Richtung eines statistischen Zusammenhangs zwischen Variablen (z.B. Wohnfläche und Mietpreis).
- **Covariance (Kovarianz):** Beschreibt die Richtung der gemeinsamen Veränderung zweier numerischer Variablen.

### 5. Univariate Graphical EDA

- **Definition:** Untersucht eine einzelne Variable grafisch und ergänzt damit die Erkenntnisse aus der nicht-grafischen Analyse.
- **Typische Methoden:** Histogram, Box Plot und Bar Chart.
- **Beispiel:** Verteilung der weltweiten Einkommen:
  - Die Einkommensverteilung ist stark rechtsschief (Right-Skewed).
  - Die meisten Menschen haben ein vergleichsweise tiefes Einkommen, während wenige sehr hohe Einkommen erzielen.
  - Eine logarithmische X-Achse kann helfen, eine stark rechtsschiefe Verteilung übersichtlicher darzustellen.
- **Erkenntnis:** Grafiken machen Verteilungen, Ausreisser und Auffälligkeiten sichtbar, die durch einzelne Kennzahlen nicht ausreichend erkennbar sind.

### 6. Multivariate Graphical EDA

- **Definition:** Untersucht Beziehungen zwischen zwei oder mehr Variablen mithilfe grafischer Darstellungen.
- **Typische Methoden:** Scatter Plot (Streudiagramm), Scatter Plot Matrix und Correlation Heatmap.
- **Beispiel:** Bevölkerungspyramide mit Altersverteilung nach Geschlecht für verschiedene Jahre.
  - Ermöglicht, Unterschiede zwischen Bevölkerungsgruppen und Veränderungen über die Zeit zu erkennen.

### 7. Warum Graphical EDA wichtig ist – Anscombe's Quartet

- **Anscombe's Quartet:** Vier unterschiedliche Datensätze mit nahezu identischen statistischen Kennzahlen:
  - Mean (Mittelwert)
  - Variance (Varianz)
  - Correlation (Korrelation)
  - Regression Line (Regressionsgerade)
- **Besonderheit:** Obwohl diese Kennzahlen nahezu identisch sind, zeigen die vier Datensätze in Streudiagrammen völlig unterschiedliche Muster, darunter nichtlineare Zusammenhänge und Ausreisser.
- **Datasaurus Dozen:** Moderne Erweiterung dieses Prinzips mit zwölf Datensätzen, die nahezu identische statistische Kennzahlen, aber völlig unterschiedliche grafische Formen aufweisen (z.B. die Form eines Dinosauriers).

**Zentrale Erkenntnis:**
- Statistische Kennzahlen allein können irreführend sein, weil unterschiedliche Datenverteilungen dieselben Kennzahlen aufweisen können.
- Deshalb sollten Non-Graphical EDA und Graphical EDA immer kombiniert werden.

### 8. Basic Data Quality Checks (Introduction-Notebook)

Vor einer eigentlichen Analyse sollte die Datenqualität überprüft werden:

- **`df.shape`:** Anzahl Zeilen und Spalten des DataFrames anzeigen.
- **`df.head(3)`:** Erste drei Datensätze zur ersten Sichtprüfung anzeigen.
- **`df.dtypes`:** Datentypen der einzelnen Spalten überprüfen.
- **`df.isnull().sum()`:** Anzahl fehlender Werte pro Spalte ermitteln.
- **`df.duplicated().sum()`:** Anzahl vollständig identischer (doppelter) Zeilen ermitteln.

**Erkenntnis:** Eine systematische EDA kombiniert statistische Kennzahlen, grafische Darstellungen und Datenqualitätsprüfungen, um Daten zu verstehen und eine zuverlässige Grundlage für weitere Analysen zu schaffen.

## Non-Graphical Exploratory Data Analysis (EDA)

Die **Non-Graphical EDA** untersucht Daten mithilfe statistischer Kennzahlen und Tabellen, ohne grafische Visualisierungen zu verwenden. Ziel ist es, **Verteilungen, Streuungen und Zusammenhänge** zwischen Variablen zu erkennen

### Univariate Non-Graphical EDA – Eine Variable untersuchen
- **Measures of Central Tendency:** Mittelwert (Mean), Median und Modus (Mode) beschreiben die zentrale Lage einer Verteilung
- **Quantiles:** Zeigen, unterhalb welcher Werte bestimmte Anteile der Beobachtungen liegen (z.B. Quartile und Perzentile)
- **Measures of Dispersion:** Varianz (Variance) und Standardabweichung (Standard Deviation) messen die Streuung der Werte um den Mittelwert
- **Skewness & Kurtosis:** Beschreiben die Asymmetrie (Schiefe) und die Ausprägung der Verteilungsenden (Tails)

### Multivariate Non-Graphical EDA – Mehrere Variablen untersuchen
- **Contingency Tables:** Kreuztabellen zeigen die gemeinsamen Häufigkeiten kategorialer Variablen
- **Correlation:** Misst Richtung und Stärke eines Zusammenhangs zwischen numerischen Variablen (z.B. Pearson-Korrelation von −1 bis +1)
- **Covariance:** Misst, ob zwei numerische Variablen tendenziell gemeinsam steigen oder sich gegenläufig verändern

**Erkenntnis:** Univariate EDA beschreibt einzelne Variablen und ihre Verteilung. Multivariate EDA untersucht Zusammenhänge zwischen mehreren Variablen

```python
# Libraries
import pandas as pd                                                             #pandas: Daten einlesen, bearbeiten und statistisch auswerten
import numpy as np                                                              #NumPy: Numerische Berechnungen und mathematische Funktionen
import warnings                                                                 #Python-Modul zur Verwaltung von Warnmeldungen
warnings.filterwarnings('ignore')                                               #Warnmeldungen unterdrücken (wie in der Dozentenlösung)

# Load the dataset
df_evcharging = pd.read_csv('ev_charging_stations.csv')                         #CSV-Datei einlesen und als DataFrame df speichern
                                                                                #read_csv(): CSV-Datei in einen pandas-DataFrame laden

print('Shape:', df_evcharging.shape)                                            #Anzahl Zeilen und Spalten anzeigen
df_evcharging.head(3)                                                           #Erste drei Datensätze anzeigen

#ANPASSEN: 'ev_charging_stations.csv' an den tatsächlichen Dateinamen und Dateipfad anpassen
#ANPASSEN: df an den gewünschten DataFrame-Namen anpassen
#NICHTS ANPASSEN: pandas, numpy und warnings sind die Bibliotheken aus der Musterlösung
#VORAUSSETZUNG: CSV-Datei muss im angegebenen Verzeichnis vorhanden sein
#ACHTUNG: warnings.filterwarnings('ignore') unterdrückt sämtliche Warnmeldungen, auch potenziell wichtige
#ERKENNTNIS: Der Datensatz wird geladen und steht anschliessend für alle EDA-Methoden zur Verfügung
```

### Deskriptive Statistik – Statistische Kennzahlen mit describe()

**Zweck:** Numerische Daten zusammenfassen, ihre Verteilung untersuchen und mögliche Ausreisser erkennen.

**Methode:** `describe()` berechnet automatisch die wichtigsten statistischen Kennzahlen.

| Kennzahl | Bedeutung |
|---|---|
| `count` | Anzahl gültiger Werte (ohne NaN) |
| `mean` | Arithmetischer Mittelwert (ausreisserempfindlich) |
| `std` | Standardabweichung (Streuung der Werte) |
| `min` | Kleinster Wert |
| `25%` | Erstes Quartil (Q1): 25 % der Werte liegen darunter oder sind gleich |
| `50%` | Median (Q2): Mittlerer Wert, robust gegenüber Ausreissern |
| `75%` | Drittes Quartil (Q3): 75 % der Werte liegen darunter oder sind gleich |
| `max` | Grösster Wert |

**Varianten:**
- `df['spalte'].describe()` → Eine numerische Spalte analysieren.
- `df.describe()` → Alle numerischen Spalten analysieren.
- `df.groupby('kategorie')['wert'].describe()` → Kennzahlen für jede Kategorie separat berechnen.
- `.round(2)` → Ergebnisse auf zwei Dezimalstellen runden.

**Wichtig:**
- `mean` kann durch extreme Werte stark beeinflusst werden; `median` ist robuster.
- Grosse Unterschiede zwischen `mean` und `median` können auf eine schiefe Verteilung hindeuten.
- `describe()` entfernt oder korrigiert keine Ausreisser.

**Erkenntnis:** Deskriptive Statistik beschreibt die Verteilung vorhandener Daten, ohne daraus Ursache-Wirkungs-Beziehungen abzuleiten.

```python
# DESKRIPTIVE STATISTIK – STATISTISCHE KENNZAHLEN

# 1. Eine numerische Spalte analysieren
print(df_evcharging['power_kw'].describe().round(2))                            #Anzahl, Mittelwert, Standardabweichung, Minimum, Quartile und Maximum berechnen

# 2. Alle numerischen Spalten analysieren
display(df_evcharging.describe().round(2))                                      #Statistische Kennzahlen aller numerischen Spalten anzeigen

# 3. Kennzahlen nach Kategorien vergleichen
display(
    df_evcharging.groupby('power_type')['power_kw'].describe().round(2)
)                                                                               #Statistische Kennzahlen der Ladeleistung für jede Ladeart separat berechnen

#ANPASSEN: df_evcharging an den verwendeten DataFrame anpassen
#ANPASSEN: 'power_kw' an die untersuchte numerische Spalte anpassen
#ANPASSEN: 'power_type' an die kategoriale Spalte für die Gruppierung anpassen
#OPTIONAL: round(2) an die gewünschte Anzahl Dezimalstellen anpassen
#NICHTS ANPASSEN: describe(), groupby() und display() als Methoden
#VORAUSSETZUNG: Numerische Daten für die statistische Auswertung
#ACHTUNG: describe() ignoriert fehlende Werte bei der Berechnung numerischer Kennzahlen
#ERKENNTNIS: mean ist ausreisserempfindlich; median beschreibt die zentrale Lage robuster
```

### Measures of Central Tendency – Mean, Median und Mode (Lagermasse)

- **Definition:** Lagermasse beschreiben die zentrale bzw. typische Lage der Werte einer numerischen Variable. Sie gehören zur **Univariate Non-Graphical EDA**
- **Mean (Mittelwert) – `.mean()`:** Arithmetischer Durchschnitt aller Werte (Summe / Anzahl). Empfindlich gegenüber Ausreissern
- **Median (Zentralwert) – `.median()`:** Mittlerer Wert der sortierten Daten (50%-Quantil). Robust gegenüber Ausreissern
- **Mode (Modalwert) – `.mode()[0]`:** Am häufigsten vorkommender Wert. `.mode()` gibt eine Series zurück; `[0]` wählt den ersten Modalwert aus, falls mehrere existieren
- **`f"...{wert:.2f}..."`:** f-String zur formatierten Ausgabe von Variablen und Berechnungen. `:.2f` zeigt eine Dezimalzahl mit genau zwei Nachkommastellen an

**Interpretation der Verteilung:**
- **Mean > Median:** Hinweis auf eine rechtsschiefe Verteilung (Right-Skewed). Wenige hohe Werte ziehen den Mittelwert nach oben
- **Mean < Median:** Hinweis auf eine linksschiefe Verteilung (Left-Skewed). Wenige tiefe Werte ziehen den Mittelwert nach unten
- **Mean ≈ Median:** Kann auf eine ungefähr symmetrische Verteilung hindeuten
- **Wichtig:** Der Vergleich von Mean und Median liefert nur Hinweise auf die Schiefe. Für eine zuverlässigere Beurteilung zusätzlich Skewness und die grafische Verteilung untersuchen

**Beispiel:** EV-Ladestationen mit Mean = 50.5 kW und Median = 22 kW → Hinweis auf eine stark rechtsschiefe Verteilung, da wenige Schnellladestationen (150–400 kW) den Mittelwert erhöhen

**Erkenntnis:** Bei stark schiefen Verteilungen ist der Median häufig repräsentativer als der Mean, weil er weniger durch Extremwerte beeinflusst wird

```python
# Measures of Central Tendency – Mean, Median and Mode

print(f"Mean:   {df_evcharging['power_kw'].mean():.2f} kW")                     #Arithmetischen Mittelwert der Spalte berechnen
                                                                                #mean(): Summe aller gültigen Werte / Anzahl gültiger Werte
                                                                                #:.2f: Ergebnis mit zwei Nachkommastellen ausgeben

print(f"Median: {df_evcharging['power_kw'].median():.2f} kW")                   #Median (Zentralwert) der Spalte berechnen
                                                                                #median(): Mittlerer Wert der sortierten Daten; robust gegenüber Ausreissern

print(f"Mode:   {df_evcharging['power_kw'].mode()[0]:.2f} kW")                  #Häufigsten Wert (Modalwert) der Spalte bestimmen
                                                                                #mode(): Gibt den/die häufigsten Wert(e) als Series zurück
                                                                                #[0]: Ersten Modalwert auswählen, falls mehrere existieren

#ANPASSEN: df an den verwendeten DataFrame anpassen
#ANPASSEN: 'power_kw' an den tatsächlichen numerischen Spaltennamen anpassen
#ANPASSEN: 'kW' an die Einheit der untersuchten Variable anpassen (z.B. CHF, m², km)
#ANPASSEN: :.2f an die gewünschte Anzahl Nachkommastellen anpassen (z.B. :.1f für eine)
#NICHTS ANPASSEN: .mean(), .median() und .mode() sind die pandas-Methoden zur Berechnung der Lagermasse
#VORAUSSETZUNG: df ist bereits geladen und die untersuchte Spalte enthält numerische Werte
#ACHTUNG: .mode()[0] gibt nur den ersten Modalwert zurück, auch wenn mehrere Werte gleich häufig vorkommen
#ERKENNTNIS: Mean > Median deutet häufig auf eine rechtsschiefe Verteilung hin; der Median ist robuster gegenüber Ausreissern
```

### Quantiles – Verteilung numerischer Variablen untersuchen

- **Definition:** Quantile sind Grenzwerte, die eine sortierte Datenverteilung in bestimmte Anteile unterteilen. Sie zeigen, unterhalb welcher Werte ein bestimmter Prozentsatz der Beobachtungen liegt
- **Zweck:** Verteilung, Streuung und mögliche Extremwerte numerischer Variablen untersuchen, ohne sich ausschliesslich auf den Mittelwert zu verlassen
- **`.quantile([0.05, 0.25, 0.50, 0.75, 0.95])`:** Berechnet die angegebenen Quantile für alle ausgewählten numerischen Spalten:
  - **0.05 (5%-Quantil):** Rund 5% der Werte liegen unterhalb oder auf diesem Wert
  - **0.25 (Q1 / unteres Quartil):** Rund 25% der Werte liegen unterhalb oder auf diesem Wert
  - **0.50 (Q2 / Median):** Mittlerer Wert der Verteilung
  - **0.75 (Q3 / oberes Quartil):** Rund 75% der Werte liegen unterhalb oder auf diesem Wert
  - **0.95 (95%-Quantil):** Rund 95% der Werte liegen unterhalb oder auf diesem Wert
- **Quartile:** Teilen die Daten anhand von Q1, Q2 und Q3 in vier gleich grosse Anteile
- **Perzentile:** Quantile, die als Prozentwerte angegeben werden (z.B. 95. Perzentil = 0.95)
- **Interquartilsabstand (IQR):** Q3 − Q1; beschreibt die Streuung der mittleren 50% der Werte
- **`.round(1)`:** Rundet die berechneten Ergebnisse auf eine Nachkommastelle
- **Hinweis:** pandas berechnet Quantile bei Bedarf durch Interpolation zwischen benachbarten Werten. Fehlende Werte (NaN) werden ignoriert

**Bsp.:** Das 95%-Quantil der Ladeleistung (`power_kw`) zeigt, bis zu welcher Ladeleistung ungefähr 95% der Ladestationen liegen

**Erkenntnis:** Quantile ermöglichen eine differenzierte Beurteilung der Datenverteilung und helfen, besonders hohe oder tiefe Werte zu identifizieren

```python
# Quantiles of numerical columns

num_cols = ['power_kw', 'amperage', 'voltage']                                  #Liste der numerischen Spalten erstellen, deren Quantile berechnet werden sollen

df_evcharging[num_cols].quantile([0.05, 0.25, 0.50, 0.75, 0.95]).round(1)       #Quantile für jede ausgewählte Spalte berechnen
                                                                                #df_evcharging[num_cols]: Nur die definierten Spalten auswählen
                                                                                #quantile(): Berechnet die angegebenen Quantile
                                                                                #0.05: 5%-Quantil
                                                                                #0.25: 25%-Quantil (Q1 / unteres Quartil)
                                                                                #0.50: 50%-Quantil (Q2 / Median)
                                                                                #0.75: 75%-Quantil (Q3 / oberes Quartil)
                                                                                #0.95: 95%-Quantil
                                                                                #round(1): Ergebnisse auf eine Nachkommastelle runden

#ANPASSEN: df_evcharging an den verwendeten DataFrame anpassen
#ANPASSEN: 'power_kw', 'amperage', 'voltage' an die vorhandenen numerischen Spaltennamen anpassen
#ANPASSEN: [0.05, 0.25, 0.50, 0.75, 0.95] an die gewünschten Quantile anpassen
#ANPASSEN: round(1) an die gewünschte Anzahl Nachkommastellen anpassen
#NICHTS ANPASSEN: .quantile() ist die pandas-Methode zur Berechnung von Quantilen
#VORAUSSETZUNG: df_evcharging ist geladen und die ausgewählten Spalten enthalten numerische Werte
#ERKENNTNIS: Quantile zeigen, unterhalb welcher Grenzwerte bestimmte Anteile der Beobachtungen liegen
```

### Measures of Dispersion – Variance & Standard Deviation (Streuungsmasse)

- **Definition:** Streuungsmasse beschreiben, wie stark die Werte einer numerischen Variable um ihren Mittelwert streuen. Sie gehören zur **Univariate Non-Graphical EDA**.
- **Variance (Varianz) – `.var()`:** Misst die Streuung anhand der quadrierten Abweichungen vom Mittelwert. Je höher die Varianz, desto stärker streuen die Werte.
  - Die Einheit wird quadriert (z.B. kW²).
  - pandas berechnet standardmässig die **Sample Variance (Stichprobenvarianz)** mit `ddof=1` (Division durch n−1).
- **Standard Deviation (Standardabweichung) – `.std()`:** Quadratwurzel der Varianz. Beschreibt die Streuung der Werte um den Mittelwert.
  - Besitzt dieselbe Einheit wie die ursprünglichen Werte (z.B. kW).
  - pandas berechnet standardmässig die **Sample Standard Deviation (Stichprobenstandardabweichung)** mit `ddof=1`.
- **`df[num_cols]`:** Wählt die numerischen Spalten aus, deren Streuung untersucht werden soll.
- **`.round(2)`:** Rundet die berechneten Ergebnisse auf zwei Nachkommastellen.
- **Hinweis:** Fehlende Werte (`NaN`) werden bei beiden Berechnungen standardmässig ignoriert.

**Interpretation:**
- **Hohe Varianz / Standardabweichung:** Werte streuen stark um den Mittelwert → grosse Unterschiede zwischen den Beobachtungen.
- **Tiefe Varianz / Standardabweichung:** Werte liegen näher am Mittelwert → geringe Streuung.
- **Standardabweichung = 0:** Alle gültigen Werte sind identisch.
- **Beispiel:** Die Ladeleistung (`power_kw`) besitzt eine Standardabweichung von rund 77.3 kW. Dies zeigt die grosse Streuung zwischen langsamen AC-Ladestationen und leistungsstarken DC-Schnellladestationen.

**Erkenntnis:** Die Standardabweichung ist häufig leichter interpretierbar als die Varianz, weil sie dieselbe Einheit wie die ursprünglichen Daten besitzt. Beide Masse sind empfindlich gegenüber Ausreissern.

```python
# Variance & Standard Deviation
num_cols = ['power_kw', 'amperage', 'voltage']              #Numerische Spalten auswählen, deren Streuung berechnet werden soll

print('Sample variance:')                                   #Überschrift für die Stichprobenvarianz ausgeben
print(df_evcharging[num_cols].var().round(2))               #Varianz für jede ausgewählte numerische Spalte berechnen
                                                            #df_evcharging[num_cols]: Nur die definierten Spalten auswählen
                                                            #var(): Stichprobenvarianz berechnen (ddof=1 → Division durch n−1)
                                                            #round(2): Ergebnisse auf zwei Nachkommastellen runden
print()                                                     #Leerzeile zur besseren Lesbarkeit

print('Sample standard deviation:')                         #Überschrift für die Stichprobenstandardabweichung ausgeben
print(df_evcharging[num_cols].std().round(2))               #Standardabweichung für jede ausgewählte numerische Spalte berechnen
                                                            #std(): Quadratwurzel der Stichprobenvarianz berechnen
                                                            #round(2): Ergebnisse auf zwei Nachkommastellen runden
print()                                                     #Leerzeile zur besseren Lesbarkeit

#ANPASSEN: df_evcharging an den verwendeten DataFrame anpassen
#ANPASSEN: 'power_kw', 'amperage', 'voltage' an die vorhandenen numerischen Spaltennamen anpassen
#ANPASSEN: round(2) an die gewünschte Anzahl Nachkommastellen anpassen
#NICHTS ANPASSEN: .var() und .std() sind die pandas-Methoden zur Berechnung der Streuungsmasse
#VORAUSSETZUNG: df_evcharging ist geladen und die ausgewählten Spalten enthalten numerische Werte
#OPTIONAL: .var(ddof=0) und .std(ddof=0) für die Berechnung der Populationsvarianz bzw. Populationsstandardabweichung (Division durch n)
#ACHTUNG: Varianz und Standardabweichung sind empfindlich gegenüber Ausreissern
#ERKENNTNIS: Hohe Werte bedeuten eine starke Streuung um den Mittelwert; tiefe Werte eine geringe Streuung
```

### Covariance – Kovarianz und Kovarianzmatrix

- **Definition:** Die Kovarianz (Covariance) misst, wie zwei numerische Variablen gemeinsam variieren. Sie zeigt, ob hohe Werte einer Variable tendenziell mit hohen oder niedrigen Werten einer anderen Variable auftreten.
- **Zweck:** Richtung eines linearen Zusammenhangs zwischen zwei numerischen Variablen untersuchen.
- **EDA:** Multivariate Non-Graphical EDA (Zusammenhang zwischen mindestens zwei Variablen).

**Interpretation der Kovarianz:**
- **Positive Kovarianz (> 0):** Steigt eine Variable, steigt tendenziell auch die andere → positiver linearer Zusammenhang.
- **Negative Kovarianz (< 0):** Steigt eine Variable, sinkt tendenziell die andere → negativer linearer Zusammenhang.
- **Kovarianz nahe 0:** Kein oder nur ein geringer linearer Zusammenhang erkennbar.
- **Wertebereich:** −∞ bis +∞ (keine festen Grenzen).
- **Wichtig:** Die Kovarianz zeigt die Richtung des Zusammenhangs, ihre absolute Grösse ist jedoch von den Masseinheiten und der Skalierung der Variablen abhängig.

**Pandas – `.cov()`:**
- **`df_evcharging['power_kw'].cov(df_evcharging['amperage'])`:** Berechnet die Kovarianz zwischen genau zwei numerischen Variablen.
- **`df_evcharging[num_cols].cov()`:** Berechnet eine Kovarianzmatrix für mehrere ausgewählte numerische Variablen.
- **`num_cols`:** Liste der numerischen Spalten, deren Kovarianzen untersucht werden sollen.
- **Kovarianzmatrix:** Tabelle, die für jedes Variablenpaar die jeweilige Kovarianz enthält.
- **Hauptdiagonale:** Enthält die Varianzen der einzelnen Variablen (Kovarianz einer Variable mit sich selbst = Varianz).
- **Symmetrie:** Die Kovarianz zwischen A und B ist identisch mit der Kovarianz zwischen B und A.
- **Berechnung:** Pandas verwendet standardmässig die Stichprobenkovarianz mit dem Nenner n − 1.

**Unterschied zwischen Covariance und Correlation:**
- **Covariance (`.cov()`):** Misst die gemeinsame Veränderung zweier Variablen. Wertebereich −∞ bis +∞, abhängig von Masseinheiten und Skalierung.
- **Correlation (`.corr(method='pearson')`):** Misst Richtung und Stärke des linearen Zusammenhangs. Standardisierter Wertebereich von −1 bis +1.
- **Zusammenhang:** Die Pearson-Korrelation entspricht der Kovarianz geteilt durch das Produkt der beiden Standardabweichungen.
- **Praxis:** Kovarianz eignet sich zur Untersuchung gemeinsamer Veränderungen; Korrelation ist besser geeignet, um die Stärke unterschiedlicher linearer Zusammenhänge zu vergleichen.

**Beispiel EV Charging Stations:**
- **`power_kw` und `amperage`:** Eine positive Kovarianz bedeutet, dass höhere Ladeleistungen tendenziell mit höheren Stromstärken einhergehen.
- **`power_kw` und `voltage`:** Eine positive Kovarianz bedeutet, dass höhere Ladeleistungen tendenziell mit höheren Spannungen einhergehen.
- Die Kovarianzwerte verschiedener Variablenpaare sollten nicht allein anhand ihrer absoluten Grösse verglichen werden, da unterschiedliche Einheiten und Skalierungen das Ergebnis beeinflussen.

**Wichtig:**
- Die Kovarianz berücksichtigt lineare Zusammenhänge; nichtlineare Beziehungen können trotz geringer Kovarianz bestehen.
- Eine Kovarianz von 0 bedeutet nicht automatisch, dass die Variablen unabhängig sind.
- Die Kovarianz ist empfindlich gegenüber Ausreissern.
- Fehlende Werte (NaN) werden von Pandas bei der paarweisen Kovarianzberechnung standardmässig ausgeschlossen.
- Kovarianz bedeutet nicht Kausalität.

**Erkenntnis:** Die Kovarianz zeigt, ob sich zwei numerische Variablen tendenziell gemeinsam oder gegenläufig verändern. Im Gegensatz zur Pearson-Korrelation ist sie nicht standardisiert und eignet sich deshalb weniger gut zum Vergleich der Zusammenhangsstärke verschiedener Variablenpaare.

```python
# COVARIANCE – KOVARIANZ UND KOVARIANZMATRIX
num_cols = ['power_kw', 'amperage', 'voltage']                                  #Numerische Spalten für die Kovarianzanalyse festlegen

# Covariance between two variables
cov_pair = df_evcharging['power_kw'].cov(df_evcharging['amperage'])             #Kovarianz zwischen Ladeleistung und Stromstärke berechnen
print('Covariance (power_kw, amperage):', round(cov_pair, 2))                   #Ergebnis auf zwei Nachkommastellen ausgeben

# Covariance matrix
cov = df_evcharging[num_cols].cov()                                             #Kovarianzmatrix aller ausgewählten numerischen Variablen berechnen
                                                                                #Berechnet die Kovarianz für jedes Variablenpaar
                                                                                #Hauptdiagonale enthält die Varianzen der einzelnen Variablen
                                                                                #Standardmässig Stichprobenkovarianz mit n-1

display(cov.round(2))                                                           #Kovarianzmatrix als Tabelle mit zwei Nachkommastellen anzeigen

#ANPASSEN: df_evcharging an den verwendeten DataFrame anpassen
#ANPASSEN: 'power_kw' und 'amperage' an die beiden zu untersuchenden numerischen Spalten anpassen
#ANPASSEN: num_cols an die gewünschten numerischen Spalten anpassen
#ANPASSEN: round(..., 2) an die gewünschte Anzahl Nachkommastellen anpassen
#NICHTS ANPASSEN: .cov() berechnet die Kovarianz; die Methode bleibt gleich
#VORAUSSETZUNG: df_evcharging ist geladen und die ausgewählten Spalten sind numerisch
#ACHTUNG: Kovarianz ist nicht standardisiert und hängt von den Masseinheiten und der Skalierung ab
#ACHTUNG: NaN-Werte werden bei der Berechnung standardmässig paarweise ausgeschlossen
#ACHTUNG: Kovarianz bedeutet nicht Kausalität
#ERKENNTNIS: Positive Kovarianz = Variablen steigen oder sinken tendenziell gemeinsam
#ERKENNTNIS: Negative Kovarianz = Variablen verändern sich tendenziell gegenläufig
#ERKENNTNIS: Kovarianz nahe 0 = Kein oder nur geringer linearer Zusammenhang
#ERKENNTNIS: Für den Vergleich der Zusammenhangsstärke eignet sich die standardisierte Pearson-Korrelation besser
```

### Skewness & Kurtosis – Schiefe und Wölbung einer Verteilung

- **Definition:** Skewness und Kurtosis sind statistische Kennzahlen zur Beschreibung der Form einer numerischen Datenverteilung. Sie gehören zur **Univariate Non-Graphical EDA**.
- **Skewness (Schiefe) – `.skew()`:** Misst die Asymmetrie einer Verteilung, also ob die Werte stärker nach links oder rechts auslaufen
  - **Skewness > 0:** Rechtsschiefe Verteilung (Right-Skewed) → Langer Ausläufer nach rechts; wenige sehr hohe Werte
  - **Skewness < 0:** Linksschiefe Verteilung (Left-Skewed) → Langer Ausläufer nach links; wenige sehr tiefe Werte
  - **Skewness ≈ 0:** Geringe gemessene Schiefe; kann auf eine ungefähr symmetrische Verteilung hindeuten
- **Kurtosis (Wölbung) – `.kurtosis()`:** Misst insbesondere die Ausprägung der Verteilungsenden (Tails) und die Neigung zu Extremwerten
  - **Kurtosis > 0:** Schwerere Tails als bei einer Normalverteilung → stärkere Neigung zu Extremwerten
  - **Kurtosis < 0:** Leichtere Tails als bei einer Normalverteilung → geringere Neigung zu Extremwerten
  - **Kurtosis = 0:** Referenzwert der Normalverteilung
- **Excess Kurtosis (Exzess-Kurtosis):** pandas berechnet standardmässig die Exzess-Kurtosis (Kurtosis − 3). Deshalb besitzt die Normalverteilung den Referenzwert 0 statt 3
- **`f"...{wert:.3f}..."`:** Gibt das Ergebnis mit genau drei Nachkommastellen aus

**Interpretation:**
- **Positive Skewness:** Wenige hohe Werte können den Mittelwert nach oben ziehen (Mean > Median)
- **Negative Skewness:** Wenige tiefe Werte können den Mittelwert nach unten ziehen (Mean < Median)
- **Hohe positive Excess Kurtosis:** Hinweis auf ausgeprägte Verteilungsenden und mögliche Extremwerte
- **Wichtig:** Skewness und Kurtosis allein reichen nicht aus, um die vollständige Verteilung oder Normalverteilung zu beurteilen. Deshalb zusätzlich grafische Methoden wie Histogramme oder QQ-Plots verwenden

**Bsp.:** Bei EV-Ladestationen können wenige besonders leistungsstarke DC-Schnellladestationen zu einer positiven Skewness und einer erhöhten Kurtosis der Ladeleistung führen

**Erkenntnis:** Skewness beschreibt die Asymmetrie einer Verteilung, während Kurtosis insbesondere die Ausprägung der Verteilungsenden und die Neigung zu Extremwerten beschreibt

```python
# Skewness & Kurtosis

skew = df_evcharging['power_kw'].skew()                                         #Schiefe der numerischen Spalte berechnen und in skew speichern
                                                                                #skew(): Misst die Asymmetrie der Verteilung
                                                                                #Positiv: rechtsschief; negativ: linksschief; nahe 0: geringe Schiefe

kurt = df_evcharging['power_kw'].kurtosis()                                     #Exzess-Kurtosis der numerischen Spalte berechnen und in kurt speichern
                                                                                #kurtosis(): Misst insbesondere die Ausprägung der Verteilungsenden
                                                                                #Positiv: schwerere Tails; negativ: leichtere Tails als bei der Normalverteilung
                                                                                #Normalverteilung: Exzess-Kurtosis = 0

print(f'Skewness: {skew:.3f}')                                                  #Schiefe mit drei Nachkommastellen ausgeben
                                                                                #f-String: Variable skew in die Textausgabe einsetzen
                                                                                #:.3f: Zahl mit genau drei Nachkommastellen formatieren

print(f'Excess kurtosis: {kurt:.3f}')                                           #Exzess-Kurtosis mit drei Nachkommastellen ausgeben

#ANPASSEN: df_evcharging an den verwendeten DataFrame anpassen
#ANPASSEN: 'power_kw' an den tatsächlichen numerischen Spaltennamen anpassen
#ANPASSEN: :.3f an die gewünschte Anzahl Nachkommastellen anpassen
#NICHTS ANPASSEN: .skew() und .kurtosis() sind die pandas-Methoden zur Berechnung der Schiefe und Exzess-Kurtosis
#VORAUSSETZUNG: df_evcharging ist geladen und die untersuchte Spalte enthält numerische Werte
#ACHTUNG: Skewness nahe 0 garantiert keine symmetrische Verteilung; Exzess-Kurtosis nahe 0 garantiert keine Normalverteilung
#ERKENNTNIS: Skewness > 0 → rechtsschief; Skewness < 0 → linksschief
#ERKENNTNIS: Excess Kurtosis > 0 → schwerere Tails; < 0 → leichtere Tails als bei der Normalverteilung
```

### Contingency Table – Kontingenztabelle / Kreuztabelle

- **Definition:** Eine Contingency Table (Kontingenztabelle / Kreuztabelle) zeigt die gemeinsame Häufigkeitsverteilung von zwei oder mehr kategorialen Variablen. Sie gehört zur **Multivariate Non-Graphical EDA**.
- **Zweck:** Häufigkeiten verschiedener Merkmalskombinationen vergleichen und mögliche Zusammenhänge zwischen kategorialen Variablen erkennen.
- **`pd.crosstab()`:** Erstellt eine Kreuztabelle, indem die Häufigkeiten der verschiedenen Kategorienkombinationen gezählt werden.
  - **Erster Parameter (`index`):** Bestimmt die Kategorien in den Zeilen der Tabelle.
  - **Zweiter Parameter (`columns`):** Bestimmt die Kategorien in den Spalten der Tabelle.
  - **`margins=True`:** Fügt eine zusätzliche Zeile und Spalte mit den Gesamtsummen hinzu.
  - **`margins_name='Total'`:** Legt die Bezeichnung der Gesamtsummen fest.
- **Ergebnis:** Jede Zelle enthält die Anzahl der Beobachtungen, die zur jeweiligen Kombination aus Zeilen- und Spaltenkategorie gehören.
- **Hinweis:** Standardmässig werden absolute Häufigkeiten und keine Prozentwerte berechnet. Fehlende Kategorienkombinationen werden als 0 angezeigt; fehlende Werte (`NaN`) werden standardmässig nicht mitgezählt.

**Interpretation:**
- **Zeilen:** Ausprägungen der ersten kategorialen Variable.
- **Spalten:** Ausprägungen der zweiten kategorialen Variable.
- **Zellen:** Anzahl Beobachtungen mit der entsprechenden Merkmalskombination.
- **Total:** Gesamthäufigkeiten pro Kategorie sowie die Gesamtanzahl der berücksichtigten Beobachtungen.
- **Beispiel:** Anzahl EV-Ladestationen nach Ladeleistungsart (`power_type`: AC/DC) und Steckertyp (`plug_type`). Die Tabelle zeigt, wie häufig bestimmte Kombinationen vorkommen.

**Erkenntnis:** Kreuztabellen ermöglichen einen strukturierten Vergleich kategorialer Variablen und können Hinweise auf Zusammenhänge liefern. Sie beweisen jedoch keinen kausalen Zusammenhang.

```python
# Contingency Table

ct = pd.crosstab(                   #Kreuztabelle erstellen und in ct speichern
    df_evcharging['power_type'],    #Erste kategoriale Variable: Bestimmt die Zeilen der Tabelle
    df_evcharging['plug_type'],     #Zweite kategoriale Variable: Bestimmt die Spalten der Tabelle
    margins=True,                   #Zusätzliche Zeile und Spalte mit den Gesamtsummen hinzufügen
    margins_name='Total'            #Bezeichnung der Gesamtsummen auf 'Total' setzen
)
ct                                  #Kreuztabelle anzeigen

#ANPASSEN: df_evcharging an den verwendeten DataFrame anpassen
#ANPASSEN: 'power_type' an die erste kategoriale Spalte (Zeilen) anpassen
#ANPASSEN: 'plug_type' an die zweite kategoriale Spalte (Spalten) anpassen
#ANPASSEN: margins=True auf False setzen, wenn keine Gesamtsummen angezeigt werden sollen
#ANPASSEN: margins_name='Total' an die gewünschte Bezeichnung der Gesamtsummen anpassen
#NICHTS ANPASSEN: pd.crosstab() ist die pandas-Funktion zur Erstellung einer Kreuztabelle
#VORAUSSETZUNG: pandas wurde als pd importiert und df_evcharging ist bereits geladen
#VORAUSSETZUNG: Die ausgewählten Spalten existieren und enthalten kategorial auswertbare Werte
#ACHTUNG: Standardmässig werden absolute Häufigkeiten berechnet und fehlende Werte (NaN) nicht mitgezählt
#ERKENNTNIS: Jede Zelle zeigt, wie häufig die jeweilige Kombination der beiden Merkmale vorkommt
```

### Pearson Correlation Matrix – Lineare Zusammenhänge zwischen numerischen Variablen

- **Definition:** Die Pearson-Korrelation misst die **Richtung und Stärke des linearen Zusammenhangs** zwischen zwei numerischen Variablen. Eine Korrelationsmatrix zeigt diese Zusammenhänge für mehrere Variablen gleichzeitig. Sie gehört zur **Multivariate Non-Graphical EDA**.
- **Zweck:** Erkennen, welche numerischen Variablen miteinander zusammenhängen, wie stark dieser Zusammenhang ist und ob er positiv oder negativ verläuft.
- **`.corr(method='pearson')`:** Berechnet die Pearson-Korrelationskoeffizienten für alle möglichen Spaltenpaare.
  - **`df[num_cols]`:** Wählt die numerischen Spalten aus, die untersucht werden sollen.
  - **`.corr()`:** Erstellt die Korrelationsmatrix; `method='pearson'` legt die Pearson-Methode explizit fest (Standard).
  - **`.round(3)`:** Rundet die Korrelationskoeffizienten auf drei Nachkommastellen.
- **Pearson-Korrelationskoeffizient (r):** Liegt zwischen **−1 und +1**:
  - **r = +1:** Perfekter positiver linearer Zusammenhang → beide Variablen steigen gemeinsam.
  - **r > 0:** Positiver Zusammenhang → höhere Werte einer Variable gehen tendenziell mit höheren Werten der anderen einher.
  - **r ≈ 0:** Kein oder kaum linearer Zusammenhang.
  - **r < 0:** Negativer Zusammenhang → höhere Werte einer Variable gehen tendenziell mit tieferen Werten der anderen einher.
  - **r = −1:** Perfekter negativer linearer Zusammenhang.
  - **Stärke:** Je näher |r| an 1 liegt, desto stärker ist der lineare Zusammenhang; je näher an 0, desto schwächer.
- **Korrelationsmatrix interpretieren:**
  - **Zeilen und Spalten:** Enthalten dieselben numerischen Variablen.
  - **Zellen:** Zeigen den Korrelationskoeffizienten zwischen den jeweiligen Variablen.
  - **Hauptdiagonale:** Enthält 1.0, sofern die Korrelation definiert ist (Variable mit sich selbst).
  - **Symmetrie:** Die Korrelation zwischen A und B ist identisch zur Korrelation zwischen B und A.

**Interpretation:**
- **Starke positive Korrelation:** Zwei Variablen nehmen tendenziell gemeinsam zu.
- **Starke negative Korrelation:** Nimmt eine Variable zu, nimmt die andere tendenziell ab.
- **Schwache Korrelation:** Es besteht nur ein geringer linearer Zusammenhang.
- **Stärkste Korrelation identifizieren:** Den höchsten absoluten Korrelationskoeffizienten ausserhalb der Hauptdiagonale suchen.
- **Beispiel:** Bei EV-Ladestationen besteht ein starker positiver Zusammenhang zwischen `power_kw` und `amperage`: Höhere Stromstärken gehen mit höheren Ladeleistungen einher.

**Wichtig:**
- **Korrelation ≠ Kausalität:** Ein statistischer Zusammenhang beweist keinen Ursache-Wirkungs-Zusammenhang.
- **Pearson misst nur lineare Zusammenhänge:** Eine nichtlineare Beziehung kann trotz r ≈ 0 bestehen.
- **Ausreisser:** Können die Pearson-Korrelation stark beeinflussen.
- **Fehlende Werte (NaN):** pandas berücksichtigt für jedes Spaltenpaar nur Beobachtungen, bei denen beide Werte vorhanden sind.
- **Konstante Spalten:** Besitzen eine undefinierte Pearson-Korrelation (NaN).

**Erkenntnis:** Die Pearson-Korrelationsmatrix zeigt auf einen Blick, welche numerischen Variablen stark oder schwach und positiv oder negativ linear zusammenhängen.

```python
# Pearson Correlation Matrix

num_cols = ['power_kw', 'amperage', 'voltage']                      #Liste der numerischen Spalten erstellen, deren Korrelationen untersucht werden sollen

corr = df_evcharging[num_cols].corr(method='pearson').round(3)      #Pearson-Korrelationsmatrix für alle ausgewählten numerischen Spalten berechnen
                                                                    #df_evcharging[num_cols]: Nur die definierten numerischen Spalten auswählen
                                                                    #corr(): Berechnet die paarweisen Korrelationskoeffizienten
                                                                    #method='pearson': Pearson-Korrelation für lineare Zusammenhänge verwenden
                                                                    #round(3): Ergebnisse auf drei Nachkommastellen runden
display(corr)                                                       #Korrelationsmatrix anzeigen

# Interpretation:
#Positive Korrelation (r > 0): Beide Variablen steigen tendenziell gemeinsam
#Negative Korrelation (r < 0): Steigt eine Variable, sinkt die andere tendenziell
#Schwache Korrelation (r nahe 0): Geringer linearer Zusammenhang
#Starke Korrelation (r nahe +1 oder -1): Starker linearer Zusammenhang
#Stärkste Korrelation: Höchsten absoluten r-Wert ausserhalb der Hauptdiagonale identifizieren

#ANPASSEN: df_evcharging an den verwendeten DataFrame anpassen
#ANPASSEN: 'power_kw', 'amperage', 'voltage' an die vorhandenen numerischen Spaltennamen anpassen
#ANPASSEN: round(3) an die gewünschte Anzahl Nachkommastellen anpassen
#NICHTS ANPASSEN: .corr(method='pearson') berechnet die Pearson-Korrelationsmatrix
#VORAUSSETZUNG: df_evcharging ist geladen und die ausgewählten Spalten enthalten numerische Werte
#ACHTUNG: Die Hauptdiagonale enthält Selbstkorrelationen (r = 1) und wird beim Vergleich verschiedener Variablen ignoriert
#ACHTUNG: Pearson misst nur lineare Zusammenhänge und ist empfindlich gegenüber Ausreissern
#ACHTUNG: Korrelation bedeutet nicht automatisch Kausalität
#ERKENNTNIS: Je näher r an +1 oder -1 liegt, desto stärker ist der lineare Zusammenhang zwischen zwei Variablen
```

### Spearman-Rangkorrelation

**Zweck:** Stärke und Richtung eines monotonen Zusammenhangs zwischen zwei Variablen untersuchen

**Unterschied zu Pearson:**
- **Pearson:** Misst lineare Zusammenhänge anhand der ursprünglichen Zahlenwerte; empfindlich gegenüber Ausreissern
- **Spearman:** Misst monotone Zusammenhänge anhand der Rangplätze; weniger empfindlich gegenüber Ausreissern
- **Monoton:** Mit zunehmendem X nimmt Y tendenziell zu oder ab, nicht zwingend auf einer Geraden

**Interpretation des Korrelationskoeffizienten (r bzw. ρ):**
- **+1:** Perfekter positiver Zusammenhang
- **0:** Kein monotoner Zusammenhang bei Spearman
- **−1:** Perfekter negativer Zusammenhang

**Methoden:**
- `df['x'].corr(df['y'], method='spearman')` → Spearman zwischen zwei Spalten berechnen
- `df[['x', 'y']].corr(method='spearman')` → Spearman-Korrelationsmatrix berechnen
- `.round(2)` → Ergebnis auf zwei Dezimalstellen runden

**Voraussetzung:** Beide numerischen Variablen müssen im selben DataFrame vorhanden sein. Falls sie aus verschiedenen Dateien stammen, zuerst über einen gemeinsamen Schlüssel zusammenführen (join machen mit .merge())

**Achtung:** Korrelation bedeutet nicht Kausalität. Ein Scatter Plot hilft, die Beziehung und mögliche Ausreisser zu untersuchen

```python
# SPEARMAN-RANGKORRELATION
num_cols = ['power_kw', 'amperage', 'voltage']                                  #Numerische Spalten auswählen

corr = df_evcharging[num_cols].corr(method='spearman').round(2)                 #Spearman-Korrelationsmatrix berechnen und auf 2 Stellen runden
display(corr)                                                                   #Korrelationsmatrix anzeigen

# Einzelne Spearman-Korrelation zwischen zwei Spalten
r = df_evcharging['power_kw'].corr(df_evcharging['amperage'], method='spearman')
print('Spearman-Korrelation:', round(r, 2))                                     #Ergebnis auf 2 Stellen runden

#ANPASSEN: df_evcharging an den verwendeten DataFrame anpassen
#ANPASSEN: 'power_kw', 'amperage', 'voltage' an die vorhandenen numerischen Spalten anpassen
#ANPASSEN: round(2) an die verlangte Anzahl Nachkommastellen anpassen
#NICHTS ANPASSEN: method='spearman' für die Spearman-Rangkorrelation
#ACHTUNG: Spearman misst monotone, nicht ausschliesslich lineare Zusammenhänge
#ERKENNTNIS: Die Berechnung funktioniert wie bei Pearson, nur mit method='spearman'
```

# V5 EDA Graphical

## Graphical Exploratory Data Analysis (EDA)

Die **Graphical EDA** untersucht Daten mithilfe grafischer Visualisierungen. Ziel ist es, **Verteilungen, Trends, Ausreisser, Muster und Zusammenhänge** zwischen Variablen visuell zu erkennen

### Univariate Graphical EDA – Eine Variable untersuchen
- **Line Chart (Liniendiagramm):** Zeigt die Entwicklung numerischer Werte über eine geordnete Abfolge, insbesondere über die Zeit
- **Box Plot (Boxplot):** Visualisiert Median, Quartile, Streuung und potenzielle Ausreisser einer numerischen Variable
- **Histogram:** Zeigt die Häufigkeitsverteilung numerischer Werte, indem diese in Intervalle (Bins) eingeteilt werden
- **Quantile-Quantile Plot (QQ-Plot):** Vergleicht die Quantile einer Datenverteilung mit einer theoretischen Verteilung (z.B. Normalverteilung), um Abweichungen zu erkennen
- **Bar Chart (Balkendiagramm):** Vergleicht die Häufigkeiten oder Werte verschiedener Kategorien
- **Pie Chart (Kreisdiagramm):** Visualisiert die prozentualen Anteile einzelner Kategorien an einem Gesamtwert

### Multivariate Graphical EDA – Mehrere Variablen untersuchen
- **Scatter Plot (Streudiagramm):** Zeigt den Zusammenhang zwischen zwei numerischen Variablen und macht Korrelationen, Muster und Ausreisser sichtbar
- **Bubble Plot (Blasendiagramm):** Erweitert den Scatter Plot um eine dritte numerische Variable, die durch die Grösse der Punkte dargestellt wird
- **Scatter Plot Matrix:** Stellt paarweise Streudiagramme mehrerer numerischer Variablen dar, um deren Zusammenhänge gleichzeitig zu untersuchen
- **Hexagonal Binning Plot:** Gruppiert Datenpunkte in sechseckige Bereiche und visualisiert deren Häufigkeit bzw. Dichte; besonders geeignet für grosse Datensätze
- **Heat Map (Heatmap):** Stellt numerische Werte mithilfe von Farben dar, beispielsweise in einer Korrelationsmatrix, um Muster und Zusammenhänge sichtbar zu machen

**Erkenntnis:** Univariate Graphical EDA visualisiert die Verteilung und Eigenschaften einzelner Variablen. Multivariate Graphical EDA visualisiert Beziehungen, Muster und Zusammenhänge zwischen mehreren Variablen

```python
# Libraries
import pandas as pd                 #Pandas: Daten einlesen und analysieren
import numpy as np                  #NumPy: Numerische Berechnungen
import matplotlib.pyplot as plt     #Matplotlib: Diagramme erstellen; plt wird für Histogramme und andere Plots benötigt
import seaborn as sns               #Seaborn: Statistische Diagramme wie Boxplots erstellen
import scipy.stats as stats         #SciPy: Statistische Methoden und Verteilungen
import warnings                     #Python-Modul zur Verwaltung von Warnmeldungen
warnings.filterwarnings('ignore')   #Warnmeldungen unterdrücken

sns.set_theme(style='whitegrid')    #Einheitliches Diagrammdesign mit weissem Hintergrund und Gitternetzlinien
```

### Histogram – Häufigkeitsverteilung numerischer Variablen

- **Definition:** Ein Histogramm stellt die Häufigkeitsverteilung einer numerischen Variable grafisch dar. Die Werte werden in Intervalle (Bins) unterteilt, deren Balkenhöhe die Anzahl der Beobachtungen im jeweiligen Intervall zeigt. Es gehört zur **Univariate Graphical EDA**.
- **Zweck:** Verteilungen, Häufigkeiten, Streuung, Schiefe und mögliche Ausreisser numerischer Variablen erkennen.
- **`plt.subplots(figsize=(8, 4))`:** Erstellt eine Abbildung (`fig`) und eine Zeichenfläche (`ax`).
  - **`fig`:** Gesamte Abbildung (Figure).
  - **`ax`:** Einzelnes Diagramm bzw. Zeichenfläche (Axes).
  - **`figsize=(8, 4)`:** Breite und Höhe der Abbildung in Zoll.
- **`ax.hist()`:** Erstellt das Histogramm:
  - **`df['spalte']`:** Numerische Variable, deren Verteilung dargestellt werden soll.
  - **`bins=40`:** Unterteilt den Wertebereich in 40 Intervalle. Mehr Bins zeigen feinere Details, weniger Bins eine gröbere Verteilung.
  - **`color='steelblue'`:** Legt die Farbe der Balken fest.
  - **`edgecolor='white'`:** Legt die Farbe der Balkenumrandungen fest.
- **`ax.set_xlabel()`:** Beschriftet die X-Achse (numerische Werte bzw. Intervalle).
- **`ax.set_ylabel()`:** Beschriftet die Y-Achse (standardmässig Anzahl Beobachtungen).
- **`ax.set_title()`:** Legt den Titel des Diagramms fest.
- **`plt.tight_layout()`:** Optimiert die Abstände zwischen Diagramm und Beschriftungen.
- **`plt.show()`:** Zeigt das fertige Diagramm an.

**Interpretation:**
- **Symmetrische Verteilung:** Werte verteilen sich ungefähr gleichmässig um die Mitte.
- **Rechtsschiefe Verteilung (Right-Skewed):** Viele tiefe Werte und wenige sehr hohe Werte → langer Ausläufer nach rechts.
- **Linksschiefe Verteilung (Left-Skewed):** Viele hohe Werte und wenige sehr tiefe Werte → langer Ausläufer nach links.
- **Mehrere Häufigkeitsgipfel:** Können auf unterschiedliche Gruppen innerhalb des Datensatzes hinweisen.
- **Ausreisser:** Einzelne Beobachtungen liegen deutlich ausserhalb des Hauptbereichs.
- **Beispiel:** Bei EV-Ladestationen liefern die meisten Stationen 22 kW. Wenige DC-Schnellladestationen mit 150–400 kW erzeugen einen langen rechten Ausläufer.

**Wichtig:**
- **Histogramm ≠ Balkendiagramm:** Ein Histogramm stellt numerische Wertebereiche dar, ein Balkendiagramm vergleicht Kategorien.
- **Bins beeinflussen die Darstellung:** Unterschiedliche Bin-Anzahlen können die wahrgenommene Verteilung verändern.
- **Standardmässig absolute Häufigkeiten:** Bei `ax.hist()` zeigt die Y-Achse normalerweise die Anzahl Beobachtungen pro Bin.

**Erkenntnis:** Histogramme visualisieren die Verteilung einer numerischen Variable und ermöglichen es, deren Form, häufige Werte, Schiefe und mögliche Ausreisser schnell zu erkennen.

```python
# Histogram
fig, ax = plt.subplots(figsize=(8, 4))                                              #Abbildung (fig) und Zeichenfläche (ax) erstellen
                                                                                    #figsize=(8, 4): Breite 8 Zoll, Höhe 4 Zoll

ax.hist(df_evcharging['power_kw'], bins=40, color='steelblue', edgecolor='white')   #Histogramm der numerischen Variable erstellen
                                                                                    #df_evcharging['power_kw']: Numerische Spalte auswählen
                                                                                    #bins=40: Wertebereich in 40 Intervalle unterteilen
                                                                                    #color='steelblue': Balkenfarbe festlegen
                                                                                    #edgecolor='white': Weisse Balkenumrandung festlegen
                                                                                    #Balkenhöhe: Anzahl Beobachtungen im jeweiligen Intervall

ax.set_xlabel('Power (kW)')                                                         #X-Achse beschriften (numerische Werte)
ax.set_ylabel('Count')                                                              #Y-Achse beschriften (Anzahl Beobachtungen pro Intervall)
ax.set_title('Distribution of Charging Power (kW)')                                 #Diagrammtitel festlegen

plt.tight_layout()                                                                  #Abstände automatisch optimieren
plt.show()                                                                          #Histogramm anzeigen

#ANPASSEN: df_evcharging an den verwendeten DataFrame anpassen
#ANPASSEN: 'power_kw' an die gewünschte numerische Spalte anpassen
#ANPASSEN: bins=40 an die gewünschte Anzahl Intervalle anpassen
#ANPASSEN: figsize=(8, 4) an die gewünschte Diagrammgrösse anpassen
#ANPASSEN: color='steelblue' und edgecolor='white' an die gewünschten Farben anpassen
#ANPASSEN: ax.set_xlabel(), ax.set_ylabel() und ax.set_title() an die untersuchte Variable anpassen
#NICHTS ANPASSEN: plt.subplots(), ax.hist(), plt.tight_layout() und plt.show() sind die benötigten Matplotlib-Methoden
#VORAUSSETZUNG: matplotlib.pyplot wurde als plt importiert und df_evcharging ist geladen
#VORAUSSETZUNG: Die untersuchte Spalte enthält numerische Werte
#ACHTUNG: Die Anzahl Bins beeinflusst das Erscheinungsbild der Verteilung
#ERKENNTNIS: Rechtsschief → langer Ausläufer nach rechts; linksschief → langer Ausläufer nach links
#ERKENNTNIS: Histogramme zeigen Häufigkeitsverteilungen numerischer Variablen und helfen, mögliche Ausreisser zu erkennen
```

### Box Plot – Verteilung, Streuung und Ausreisser vergleichen

- **Definition:** Ein Box Plot (Box-and-Whisker Plot) visualisiert die Verteilung einer numerischen Variable anhand von Median, Quartilen, Streuung und möglichen Ausreissern. Mehrere Boxplots ermöglichen den Vergleich verschiedener Kategorien.
- **Zweck:** Unterschiede zwischen Gruppen erkennen, insbesondere bezüglich zentraler Lage (Median), Streuung (IQR), Schiefe und Ausreissern.
- **EDA:** Univariate Verteilungsanalyse; beim Vergleich einer numerischen Variable zwischen mehreren Kategorien eine bivariate Analyse.

**Bestandteile eines Box Plots:**
- **Q1 (25%-Quantil):** 25 % der Beobachtungen liegen bei oder unter diesem Wert; untere Boxkante.
- **Median / Q2 (50%-Quantil):** 50 % der Beobachtungen liegen bei oder unter diesem Wert; Linie innerhalb der Box.
- **Q3 (75%-Quantil):** 75 % der Beobachtungen liegen bei oder unter diesem Wert; obere Boxkante.
- **IQR (Interquartile Range):** Q3 − Q1; beschreibt die Streuung der mittleren 50 % der Beobachtungen.
- **Whiskers (Antennen):** Reichen standardmässig bis zu den äussersten tatsächlich beobachteten Werten innerhalb der Grenzen Q1 − 1.5 × IQR und Q3 + 1.5 × IQR.
- **Outliers (Ausreisser):** Werte ausserhalb dieser Grenzen werden standardmässig als einzelne Punkte dargestellt.

**Variante 1 – Seaborn: `sns.boxplot()`**
- **`data=df_evcharging`:** Verwendeter DataFrame.
- **`x='power_type'`:** Kategoriale Variable auf der X-Achse (Gruppen).
- **`y='power_kw'`:** Numerische Variable auf der Y-Achse (untersuchte Verteilung).
- **`order=[...]`:** Bestimmt die Reihenfolge der Kategorien auf der X-Achse.
- **`ax=ax`:** Zeichnet den Boxplot auf der zuvor erstellten Zeichenfläche.
- **Vorteil:** Seaborn übernimmt die Gruppierung der Daten automatisch. Keine separate Datenaufbereitung erforderlich.

**Variante 2 – Matplotlib: `ax.boxplot()`**
- **`order=[...]`:** Definiert die gewünschten Kategorien und ihre Reihenfolge.
- **`data = [df_evcharging[df_evcharging['power_type'] == t]['power_kw'].values for t in order]`:** List Comprehension, welche die numerischen Werte für jede Kategorie separat sammelt.
- **`ax.boxplot(data)`:** Erstellt für jede Werteliste einen eigenen Boxplot.
- **`tick_labels=order`:** Beschriftet die Boxplots mit den Kategorienamen (Matplotlib ≥ 3.9).
- **`patch_artist=True`:** Ermöglicht farbig gefüllte Boxen.
- **`boxprops=dict(facecolor='steelblue', alpha=0.6)`:** Bestimmt die Füllfarbe und Transparenz der Boxen.
- **Unterschied:** Matplotlib erwartet bereits gruppierte numerische Werte; Seaborn übernimmt diese Gruppierung automatisch.

**Interpretation:**
- **Höherer Median:** Die Gruppe weist tendenziell höhere Werte auf.
- **Grössere Box (IQR):** Die mittleren 50 % der Werte sind stärker gestreut.
- **Längere Whiskers:** Grössere Spannweite der nicht als Ausreisser dargestellten Werte.
- **Punkte ausserhalb der Whiskers:** Potenzielle Ausreisser.
- **Median ausserhalb der Boxmitte oder ungleiche Whiskers:** Können auf eine asymmetrische Verteilung hinweisen.

**Beispiel EV Charging Stations:**
- **AC_3_PHASE:** Ladeleistungen sind stark um 22 kW konzentriert → geringe Streuung.
- **DC:** Tendenziell höhere Ladeleistungen mit grösserer Streuung.
- Der Boxplot ermöglicht den direkten Vergleich der Ladeleistungsverteilungen zwischen verschiedenen Ladearten.

**Wichtig:**
- Ein Boxplot zeigt eine statistische Zusammenfassung, aber nicht die genaue Verteilungsform oder mögliche mehrere Häufigkeitsgipfel.
- Für die detaillierte Verteilungsform eignet sich ein Violin Plot.
- Die 1.5-IQR-Regel kennzeichnet potenzielle Ausreisser; solche Werte sind nicht automatisch fehlerhaft.
- Die Boxhöhe stellt den IQR dar, nicht die absolute Anzahl Beobachtungen.
- Die Position der Whiskers hängt von der tatsächlichen Datenverteilung ab.

**Erkenntnis:** Boxplots ermöglichen den schnellen Vergleich von Median, Quartilen, Streuung und möglichen Ausreissern zwischen mehreren Gruppen. Für die einfache Erstellung mit kategorialen Daten eignet sich `sns.boxplot()`, während `ax.boxplot()` eine manuelle Gruppierung der Werte erfordert.

```python
# BOX PLOT – VARIANTE 1: SEABORN (EMPFOHLEN)

fig, ax = plt.subplots(figsize=(7, 5))                                          #Abbildung und Zeichenfläche erstellen; Grösse in Zoll

sns.boxplot(data=df_evcharging, x='power_type', y='power_kw',                   #Boxplots: Ladeleistung nach Ladeart; Gruppierung erfolgt automatisch
            order=['AC_1_PHASE', 'AC_3_PHASE', 'DC'], ax=ax)                    #Kategorienreihenfolge festlegen; ax = Zeichenfläche

ax.set_xlabel('Power Type')                                                     #X-Achse: Ladearten
ax.set_ylabel('Charging Power (kW)')                                            #Y-Achse: Ladeleistung
ax.set_title('Charging Power by Power Type')                                    #Diagrammtitel festlegen

plt.tight_layout()                                                              #Abstände optimieren
plt.show()                                                                      #Boxplot anzeigen


# BOX PLOT – VARIANTE 2: MATPLOTLIB (ALTERNATIVE)

order = ['AC_1_PHASE', 'AC_3_PHASE', 'DC']                                                  #Reihenfolge der Kategorien festlegen

data = [df_evcharging[df_evcharging['power_type'] == t]['power_kw'].values for t in order]  #Numerische Werte jeder Kategorie separat sammeln
                                                                                            #for t in order: Kategorien durchlaufen
                                                                                            #df_evcharging['power_type'] == t: Zeilen der Kategorie auswählen
                                                                                            #['power_kw'].values: Numerische Werte als Array übernehmen
                                                                                            #Ergebnis: Liste mit einem Array pro Kategorie

fig, ax = plt.subplots(figsize=(7, 5))                                                      #Abbildung und Zeichenfläche erstellen; Grösse in Zoll

ax.boxplot(data, tick_labels=order, patch_artist=True,                                      #Boxplot aus gruppierten Werten erstellen; Kategorien beschriften; Füllung aktivieren
           boxprops=dict(facecolor='steelblue', alpha=0.6))                                 #Füllfarbe und Transparenz der Boxen festlegen

ax.set_xlabel('Power Type')                                                                 #X-Achse: Ladearten
ax.set_ylabel('Charging Power (kW)')                                                        #Y-Achse: Ladeleistung
ax.set_title('Charging Power by Power Type')                                                #Diagrammtitel festlegen

plt.tight_layout()                                                                          #Abstände optimieren
plt.show()                                                                                  #Boxplot anzeigen


#ANPASSEN: df_evcharging an den verwendeten DataFrame anpassen
#ANPASSEN: 'power_type' an die kategoriale Spalte (Gruppierung) anpassen
#ANPASSEN: 'power_kw' an die numerische Spalte (untersuchte Werte) anpassen
#ANPASSEN: order bzw. order=[...] an die vorhandenen Kategorien und ihre gewünschte Reihenfolge anpassen
#ANPASSEN: 'AC_1_PHASE', 'AC_3_PHASE', 'DC' sind BEISPIELWERTE und müssen durch die tatsächlichen Kategorien ersetzt werden
#NICHTS ANPASSEN: sns.boxplot() übernimmt die Gruppierung automatisch; ax.boxplot() verwendet die zuvor gruppierten Werte
#VORAUSSETZUNG: x bzw. order enthält Kategorien; y bzw. data enthält numerische Werte
#ACHTUNG: tick_labels=order erfordert Matplotlib 3.9 oder neuer; bei älteren Versionen labels=order verwenden
#ACHTUNG: Whiskers reichen standardmässig bis zu den äussersten Datenpunkten innerhalb von Q1 - 1.5 × IQR und Q3 + 1.5 × IQR
#ACHTUNG: Ausreisser sind nicht automatisch fehlerhafte Daten
#ERKENNTNIS: Median = Linie in der Box; Q1/Q3 = Boxgrenzen; IQR = Boxhöhe; Whiskers = äussere Werte innerhalb der Ausreissergrenzen
#ERKENNTNIS: Boxplots vergleichen die Verteilung numerischer Werte zwischen verschiedenen Kategorien
```

### Bar Chart – Häufigkeiten kategorialer Variablen visualisieren

- **Definition:** Ein Bar Chart (Balkendiagramm) stellt die Häufigkeiten oder Werte verschiedener Kategorien mithilfe von Balken dar. Es gehört zur **Univariate Graphical EDA**, wenn die Häufigkeitsverteilung einer einzelnen kategorialen Variable untersucht wird.
- **Zweck:** Kategorien vergleichen, häufigste oder seltenste Ausprägungen identifizieren und Rangfolgen grafisch darstellen.
- **`.value_counts()`:** Zählt die Häufigkeiten einer kategorialen Variable und sortiert diese standardmässig absteigend.
  - **`.head(10)`:** Wählt die zehn häufigsten Kategorien aus (Top 10).
  - **`.index`:** Enthält die Kategorienamen (z.B. Städtenamen).
  - **`.values`:** Enthält die dazugehörigen Häufigkeiten.
  - **`[::-1]`:** Kehrt die Reihenfolge der Elemente um, damit die Kategorie mit der höchsten Häufigkeit im horizontalen Diagramm oben steht.
- **`plt.subplots(figsize=(9, 4))`:** Erstellt eine Abbildung (`fig`) und eine Zeichenfläche (`ax`) mit einer Breite von 9 und einer Höhe von 4 Zoll.
- **`ax.barh()`:** Erstellt ein horizontales Balkendiagramm.
  - **Erster Parameter:** Kategorienamen auf der Y-Achse.
  - **Zweiter Parameter:** Werte bzw. Häufigkeiten auf der X-Achse.
  - **`color='steelblue'`:** Legt die Balkenfarbe fest.
  - **`edgecolor='white'`:** Legt die Farbe der Balkenumrandung fest.
- **`ax.bar()`:** Alternative für ein vertikales Balkendiagramm.
- **`ax.set_xlabel()`:** Beschriftet die X-Achse.
- **`ax.set_title()`:** Legt den Diagrammtitel fest.
- **`plt.tight_layout()`:** Optimiert die Abstände und verhindert überlappende Beschriftungen.
- **`plt.show()`:** Zeigt das fertige Diagramm an.

**Interpretation:**
- **Lange Balken:** Hohe Häufigkeit der entsprechenden Kategorie.
- **Kurze Balken:** Tiefe Häufigkeit der entsprechenden Kategorie.
- **Top-N-Analyse:** Beschränkt die Darstellung auf die N häufigsten Kategorien und ermöglicht einen übersichtlichen Vergleich.
- **Beispiel:** Die zehn Städte mit den meisten EV-Ladestationen werden anhand ihrer Stationsanzahl absteigend verglichen.

**Unterschied zum Histogramm:**
- **Bar Chart:** Kategoriale Variable → Jede Kategorie erhält einen eigenen Balken; zwischen den Balken bestehen normalerweise Abstände.
- **Histogram:** Numerische Variable → Werte werden in Intervalle (Bins) unterteilt; die Balken liegen normalerweise direkt nebeneinander.

**Wichtig:**
- `value_counts()` ignoriert fehlende Werte (`NaN`) standardmässig.
- Bei horizontalen Balkendiagrammen ist `[::-1]` hilfreich, um die häufigste Kategorie oben anzuzeigen.
- `head(10)` begrenzt die Darstellung auf die zehn häufigsten Kategorien; andere Kategorien werden nicht angezeigt.

**Erkenntnis:** Mit `.value_counts().head(10)` und `ax.barh()` lassen sich die häufigsten Kategorien eines Datensatzes ermitteln und übersichtlich als horizontales Balkendiagramm visualisieren.

```python
# Bar Chart – Top 10 Categories by Frequency

top10 = df_evcharging['city'].value_counts().head(10) #Die zehn häufigsten Kategorien ermitteln und in top10 speichern, evtl. .sort_values() anhängend für aufsteigend
                                                      #df_evcharging['city']: Kategoriale Spalte auswählen
                                                      #value_counts(): Häufigkeiten zählen und absteigend sortieren
                                                      #head(10): Die zehn häufigsten Kategorien auswählen

fig, ax = plt.subplots(figsize=(9, 4))                #Neue Abbildung (fig) und Zeichenfläche (ax) erstellen
                                                      #figsize=(9, 4): Breite 9 Zoll, Höhe 4 Zoll

ax.barh(top10.index[::-1], top10.values[::-1], color='steelblue', edgecolor='white') #Horizontales Balkendiagramm erstellen
                                                                                     #barh(): Horizontale Balken zeichnen
                                                                                     #top10.index: Kategorienamen
                                                                                     #top10.values: Häufigkeiten der Kategorien
                                                                                     #[::-1]: Reihenfolge umkehren, damit die häufigste Kategorie oben steht
                                                                                     #color='steelblue': Balkenfarbe festlegen
                                                                                     #edgecolor='white': Balkenumrandung festlegen

ax.set_xlabel('Number of Charging Stations')                                         #X-Achse beschriften (Häufigkeit)
ax.set_title('Top 10 Cities by Number of EV Charging Stations')                      #Diagrammtitel festlegen

plt.tight_layout()                                                                  #Abstände automatisch optimieren
plt.show()                                                                          #Balkendiagramm anzeigen

#ANPASSEN: df_evcharging an den verwendeten DataFrame anpassen
#ANPASSEN: 'city' an die gewünschte kategoriale Spalte anpassen
#ANPASSEN: head(10) an die gewünschte Anzahl der häufigsten Kategorien anpassen
#ANPASSEN: figsize=(9, 4) an die gewünschte Diagrammgrösse anpassen
#ANPASSEN: color='steelblue' und edgecolor='white' bei Bedarf anpassen
#ANPASSEN: ax.set_xlabel() und ax.set_title() passend zum untersuchten Datensatz beschriften
#NICHTS ANPASSEN: value_counts(), index[::-1], values[::-1], barh(), tight_layout() und show() bleiben bei dieser Vorgehensweise gleich
#VORAUSSETZUNG: matplotlib.pyplot wurde als plt importiert und df_evcharging ist geladen
#VORAUSSETZUNG: Die ausgewählte Spalte enthält kategorial auswertbare Werte
#ACHTUNG: value_counts() ignoriert fehlende Werte (NaN) standardmässig
#ACHTUNG: head(10) zeigt nur die zehn häufigsten Kategorien und blendet alle übrigen aus
#ERKENNTNIS: Die Balkenlänge entspricht der Häufigkeit einer Kategorie; die häufigste Kategorie erscheint oben
```

### QQ Plot (Quantile-Quantile Plot) – Normalverteilung grafisch überprüfen

- **Definition:** Ein QQ-Plot vergleicht die Quantile einer numerischen Variable mit den theoretischen Quantilen einer bestimmten Wahrscheinlichkeitsverteilung (z.B. Normalverteilung). Er gehört zur **Univariate Graphical EDA**.
- **Zweck:** Überprüfen, ob eine numerische Variable annähernd einer bestimmten theoretischen Verteilung folgt, und Abweichungen wie Schiefe oder ausgeprägte Verteilungsenden erkennen.
- **`stats.probplot()`:** Erstellt einen QQ-Plot mithilfe von SciPy.
  - **`df['spalte']`:** Numerische Variable, deren Verteilung überprüft werden soll.
  - **`dist='norm'`:** Verwendet die Normalverteilung als theoretische Vergleichsverteilung.
  - **`plot=ax`:** Zeichnet den QQ-Plot auf der angegebenen Matplotlib-Zeichenfläche.
- **`plt.subplots(figsize=(6, 6))`:** Erstellt eine Abbildung (`fig`) und eine Zeichenfläche (`ax`) mit einer Breite und Höhe von jeweils 6 Zoll.
- **`ax.set_title()`:** Legt den Diagrammtitel fest.
- **`plt.tight_layout()`:** Optimiert die Abstände zwischen Diagramm und Beschriftungen.
- **`plt.show()`:** Zeigt den fertigen QQ-Plot an.

**Aufbau des QQ-Plots:**
- **X-Achse:** Theoretische Quantile der gewählten Vergleichsverteilung (z.B. Normalverteilung).
- **Y-Achse:** Tatsächliche Quantile der untersuchten numerischen Variable.
- **Referenzlinie:** Zeigt den erwarteten linearen Verlauf, wenn die Daten gut zur theoretischen Verteilung passen.
- **Datenpunkte:** Repräsentieren die beobachteten Quantile der untersuchten Variable.

**Interpretation:**
- **Punkte liegen ungefähr auf der Referenzlinie:** Daten folgen näherungsweise der gewählten theoretischen Verteilung.
- **Punkte weichen systematisch von der Referenzlinie ab:** Datenverteilung unterscheidet sich von der theoretischen Verteilung.
- **Starke Abweichungen im oberen Bereich (Upper Tail):** Hohe Werte unterscheiden sich deutlich von den theoretisch erwarteten Werten.
- **Starke Abweichungen im unteren Bereich (Lower Tail):** Tiefe Werte unterscheiden sich deutlich von den theoretisch erwarteten Werten.
- **Rechter Bereich krümmt sich deutlich nach oben:** Kann auf eine rechtsschiefe Verteilung mit besonders hohen Werten hinweisen.
- **Beispiel:** Bei EV-Ladestationen weicht `power_kw` stark von der Normalverteilung ab, insbesondere durch wenige leistungsstarke DC-Schnellladestationen.

**Wichtig:**
- **QQ-Plot ≠ Histogramm:** Ein Histogramm zeigt die Häufigkeitsverteilung; ein QQ-Plot vergleicht die beobachtete Verteilung mit einer theoretischen Verteilung.
- **QQ-Plot ≠ statistischer Normalitätstest:** Er ermöglicht eine visuelle Beurteilung, liefert aber keinen p-Wert.
- **Referenzlinie:** Muss nicht im 45°-Winkel verlaufen, da theoretische und beobachtete Quantile unterschiedliche Skalen besitzen können.
- **Fehlende Werte (NaN):** Vor der Berechnung gegebenenfalls entfernen, z.B. mit `.dropna()`.

**Erkenntnis:** Mit `stats.probplot(..., dist='norm', plot=ax)` lässt sich grafisch beurteilen, ob eine numerische Variable annähernd normalverteilt ist. Je stärker und systematischer die Punkte von der Referenzlinie abweichen, desto schlechter passt die theoretische Normalverteilung.

```python
# QQ Plot

fig, ax = plt.subplots(figsize=(6, 6))                                          #Neue Abbildung (fig) und Zeichenfläche (ax) erstellen
                                                                                #figsize=(6, 6): Breite und Höhe jeweils 6 Zoll

stats.probplot(df_evcharging['power_kw'], dist='norm', plot=ax) #QQ-Plot erstellen und mit einer theoretischen Normalverteilung vergleichen
                                                                #df_evcharging['power_kw']: Numerische Spalte auswählen
                                                                #stats.probplot(): Beobachtete und theoretische Quantile vergleichen
                                                                #dist='norm': Normalverteilung als Referenz verwenden
                                                                #plot=ax: QQ-Plot auf der bestehenden Zeichenfläche darstellen
                                                                #X-Achse: Theoretische Quantile der Normalverteilung
                                                                #Y-Achse: Beobachtete Quantile der numerischen Variable
                                                                #Referenzlinie: Orientierung zur Beurteilung der Verteilungsform

ax.set_title('QQ Plot — Charging Power (kW)')                                   #Diagrammtitel festlegen

plt.tight_layout()                                                              #Abstände automatisch optimieren
plt.show()                                                                      #QQ-Plot anzeigen

#ANPASSEN: df_evcharging an den verwendeten DataFrame anpassen
#ANPASSEN: 'power_kw' an die gewünschte numerische Spalte anpassen
#ANPASSEN: dist='norm' anpassen, falls eine andere theoretische Verteilung untersucht werden soll
#ANPASSEN: figsize=(6, 6) an die gewünschte Diagrammgrösse anpassen
#ANPASSEN: ax.set_title() passend zur untersuchten Variable beschriften
#NICHTS ANPASSEN: stats.probplot(), plot=ax, plt.tight_layout() und plt.show() bleiben bei dieser Vorgehensweise gleich
#VORAUSSETZUNG: scipy.stats wurde als stats und matplotlib.pyplot als plt importiert
#VORAUSSETZUNG: df_evcharging ist geladen und die untersuchte Spalte enthält numerische Werte
#ACHTUNG: Falls NaN-Werte vorhanden sind, vor der Berechnung mit .dropna() entfernen
#ACHTUNG: Ein QQ-Plot ermöglicht eine visuelle Beurteilung, ersetzt aber keinen statistischen Normalitätstest
#ERKENNTNIS: Punkte ungefähr auf der Referenzlinie → Daten folgen näherungsweise der theoretischen Verteilung
#ERKENNTNIS: Starke systematische Abweichungen → Daten folgen nicht gut der theoretischen Verteilung
#ERKENNTNIS: Starke Abweichungen im Upper Tail → Auffällig hohe Werte bzw. ausgeprägter rechter Verteilungsbereich
```

### Scatter Plot – Zusammenhänge zwischen numerischen Variablen visualisieren

- **Definition:** Ein Scatter Plot (Streudiagramm) stellt den Zusammenhang zwischen zwei numerischen Variablen grafisch dar. Jeder Punkt repräsentiert eine Beobachtung. Durch unterschiedliche Farben können zusätzlich Kategorien unterschieden werden. Er gehört zur **Multivariate Graphical EDA**.
- **Zweck:** Lineare und nichtlineare Zusammenhänge, Korrelationen, Cluster, Muster und mögliche Ausreisser zwischen numerischen Variablen erkennen.
- **`plt.subplots(figsize=(8, 5))`:** Erstellt eine Abbildung (`fig`) und eine Zeichenfläche (`ax`).
  - **`figsize=(8, 5)`:** Breite und Höhe der Abbildung in Zoll.
- **`df.groupby('spalte')`:** Gruppiert die Daten nach einer kategorialen Variable, damit jede Kategorie separat dargestellt werden kann.
  - **`for ptype, grp in ...`:** Durchläuft die einzelnen Gruppen mithilfe einer Schleife.
  - **`ptype`:** Name der aktuellen Kategorie (z.B. AC_1_PHASE, AC_3_PHASE oder DC).
  - **`grp`:** Teil-DataFrame mit den Beobachtungen der aktuellen Kategorie.
- **`ax.scatter()`:** Erstellt das Streudiagramm:
  - **Erster Parameter (`grp['voltage']`):** Numerische Variable auf der X-Achse.
  - **Zweiter Parameter (`grp['amperage']`):** Numerische Variable auf der Y-Achse.
  - **`label=ptype`:** Verwendet den aktuellen Gruppennamen als Beschriftung für die Legende.
  - **`alpha=0.5`:** Transparenz der Punkte (0 = unsichtbar, 1 = vollständig sichtbar).
  - **`s=30`:** Grösse der Datenpunkte.
  - **Farben:** Matplotlib verwendet bei mehreren Aufrufen von `ax.scatter()` automatisch unterschiedliche Farben.
- **`ax.set_xlabel()` / `ax.set_ylabel()`:** Beschriften die X- und Y-Achse.
- **`ax.set_title()`:** Legt den Diagrammtitel fest.
- **`ax.legend()`:** Zeigt die Legende mit den Kategorien und ihren Farben an.
- **`plt.tight_layout()`:** Optimiert die Abstände zwischen Diagramm und Beschriftungen.
- **`plt.show()`:** Zeigt das fertige Streudiagramm an.

**Interpretation:**
- **Positiver Zusammenhang:** Punkte verlaufen tendenziell von links unten nach rechts oben → Höhere X-Werte gehen mit höheren Y-Werten einher.
- **Negativer Zusammenhang:** Punkte verlaufen tendenziell von links oben nach rechts unten → Höhere X-Werte gehen mit tieferen Y-Werten einher.
- **Kein erkennbarer Zusammenhang:** Punkte sind ungeordnet verteilt → Kein deutlicher Zusammenhang sichtbar.
- **Cluster:** Mehrere Punkte bilden erkennbare Gruppen mit ähnlichen Eigenschaften.
- **Ausreisser:** Einzelne Punkte liegen deutlich ausserhalb der übrigen Beobachtungen.
- **Farbliche Gruppierung:** Ermöglicht den Vergleich der Verteilungen verschiedener Kategorien innerhalb desselben Diagramms.
- **Beispiel:** Bei EV-Ladestationen werden Spannung (`voltage`) und Stromstärke (`amperage`) untersucht. Die Farben unterscheiden zwischen AC- und DC-Ladearten.

**Wichtig:**
- **Scatter Plot ≠ Korrelationsmatrix:** Ein Scatter Plot zeigt den Zusammenhang grafisch; die Pearson-Korrelation quantifiziert einen linearen Zusammenhang anhand eines Koeffizienten.
- **Korrelation ≠ Kausalität:** Ein erkennbarer Zusammenhang beweist keine Ursache-Wirkungs-Beziehung.
- **Überlagerungen:** Bei vielen identischen oder ähnlichen Werten können sich Punkte überdecken. `alpha=0.5` macht solche Überlagerungen besser sichtbar.
- **Ohne Gruppierung:** Ein einfacher Scatter Plot kann auch ohne `groupby()` erstellt werden, wenn keine kategoriale Unterscheidung benötigt wird.

**Erkenntnis:** Scatter Plots visualisieren Zusammenhänge zwischen zwei numerischen Variablen. Mithilfe von `groupby()` und einer `for`-Schleife können zusätzliche kategoriale Merkmale farblich unterschieden werden, wodurch Cluster, Muster und Gruppenunterschiede sichtbar werden.

```python
# SCATTER PLOT – ZWEI NUMERISCHE VARIABLEN (STANDARD)

fig, ax = plt.subplots(figsize=(8, 5))                                          #Abbildung erstellen

ax.scatter(df_evcharging['voltage'], df_evcharging['amperage'])                 #X- und Y-Werte als Punkte darstellen

ax.set_xlabel('Voltage (V)')                                                    #X-Achse beschriften
ax.set_ylabel('Amperage (A)')                                                   #Y-Achse beschriften
ax.set_title('Voltage vs. Amperage')                                            #Diagrammtitel festlegen

plt.tight_layout()                                                              #Abstände optimieren
plt.show()                                                                      #Diagramm anzeigen

#ANPASSEN: df_evcharging an den verwendeten DataFrame anpassen
#ANPASSEN: 'voltage' und 'amperage' an die beiden numerischen Spalten anpassen
#ANPASSEN: Achsenbeschriftungen und Diagrammtitel anpassen
#ERKENNTNIS: Für einen einfachen Scatter Plot braucht es KEIN groupby() und KEINE for-Schleife
```

```python
# SCATTER PLOT – NACH KATEGORIEN EINGEFÄRBT

fig, ax = plt.subplots(figsize=(8, 5))                                          #Abbildung erstellen

for ptype, grp in df_evcharging.groupby('power_type'):                          #Daten nach Kategorien gruppieren
    ax.scatter(grp['voltage'], grp['amperage'],
               label=ptype, alpha=0.5, s=30)                                    #Jede Kategorie separat darstellen

ax.set_xlabel('Voltage (V)')                                                    #X-Achse beschriften
ax.set_ylabel('Amperage (A)')                                                   #Y-Achse beschriften
ax.set_title('Voltage vs. Amperage by Power Type')                              #Diagrammtitel festlegen

ax.legend()                                                                     #Legende für Kategorien anzeigen
plt.tight_layout()
plt.show()

#ANPASSEN: df_evcharging an den verwendeten DataFrame anpassen
#ANPASSEN: 'power_type' an die kategoriale Gruppierungsspalte anpassen
#ANPASSEN: 'voltage' und 'amperage' an die numerischen Spalten anpassen
#ANPASSEN: Achsenbeschriftungen und Diagrammtitel anpassen
#ACHTUNG: groupby() nur verwenden, wenn Kategorien getrennt dargestellt werden sollen
```

### Scatter Plot Matrix – Paarweise Zusammenhänge mehrerer numerischer Variablen visualisieren

- **Definition:** Eine Scatter Plot Matrix (Streudiagramm-Matrix) stellt die paarweisen Zusammenhänge zwischen mehreren numerischen Variablen in einer Matrix aus Streudiagrammen dar. Auf der Hauptdiagonale werden die Verteilungen der einzelnen Variablen visualisiert. Sie gehört zur **Multivariate Graphical EDA**.
- **Zweck:** Zusammenhänge, Korrelationen, Cluster, Muster und mögliche Ausreisser zwischen mehreren numerischen Variablen gleichzeitig erkennen.
- **`sns.pairplot()`:** Erstellt automatisch eine Matrix aus Streudiagrammen und Verteilungsdiagrammen.
  - **`df[num_cols + ['power_type']]`:** Wählt die numerischen Variablen sowie die kategoriale Gruppierungsvariable aus.
  - **`num_cols`:** Liste der numerischen Spalten, deren Zusammenhänge untersucht werden sollen.
  - **`+ ['power_type']`:** Ergänzt die Spaltenliste um die kategoriale Variable für die farbliche Gruppierung.
  - **`hue='power_type'`:** Unterscheidet die Kategorien durch unterschiedliche Farben und erstellt eine Legende.
  - **`diag_kind='hist'`:** Zeigt auf der Hauptdiagonale Histogramme der einzelnen numerischen Variablen.
  - **`diag_kind='kde'`:** Alternative für geglättete Dichtekurven statt Histogrammen.
  - **`plot_kws={'alpha': 0.4, 's': 15}`:** Übergibt zusätzliche Einstellungen an die Streudiagramme:
    - **`alpha=0.4`:** Transparenz der Punkte (0 = unsichtbar, 1 = vollständig sichtbar).
    - **`s=15`:** Grösse der Datenpunkte.
- **`g = sns.pairplot(...)`:** Speichert die erstellte Diagrammmatrix als Seaborn-PairGrid in der Variable `g`.
- **`g.fig.suptitle()`:** Erstellt einen gemeinsamen Titel für die gesamte Diagrammmatrix.
  - **`y=1.02`:** Positioniert den Titel etwas oberhalb der Diagrammmatrix.
- **`plt.show()`:** Zeigt die vollständige Diagrammmatrix an.

**Aufbau einer Scatter Plot Matrix:**
- **Hauptdiagonale:** Zeigt die Verteilung jeder einzelnen numerischen Variable (z.B. Histogramme).
- **Ausserhalb der Hauptdiagonale:** Zeigt die paarweisen Zusammenhänge der numerischen Variablen als Scatter Plots.
- **Farben (`hue`):** Unterscheiden die Kategorien innerhalb der einzelnen Diagramme.
- **Anzahl Teilgrafiken:** Bei n numerischen Variablen entsteht standardmässig eine n × n-Matrix.
- **Beispiel:** Drei numerische Variablen ergeben eine 3 × 3-Matrix mit neun Teilgrafiken (drei Histogramme und sechs Scatter Plots).

**Interpretation:**
- **Positiver Zusammenhang:** Punkte steigen tendenziell von links unten nach rechts oben.
- **Negativer Zusammenhang:** Punkte fallen tendenziell von links oben nach rechts unten.
- **Kein erkennbarer Zusammenhang:** Punkte bilden kein deutliches Muster.
- **Cluster:** Datenpunkte bilden erkennbare Gruppen mit ähnlichen Eigenschaften.
- **Ausreisser:** Einzelne Punkte liegen deutlich ausserhalb der übrigen Beobachtungen.
- **Hauptdiagonale:** Ermöglicht die Beurteilung der Verteilungen einzelner Variablen.
- **Beispiel:** Bei EV-Ladestationen werden `power_kw`, `amperage` und `voltage` paarweise verglichen. Die Farben unterscheiden zwischen den verschiedenen Ladearten (`power_type`).

**Unterschied zum einfachen Scatter Plot:**
- **Scatter Plot – `ax.scatter()`:** Untersucht gezielt zwei numerische Variablen in einem einzelnen Diagramm.
- **Scatter Plot Matrix – `sns.pairplot()`:** Untersucht automatisch alle paarweisen Kombinationen mehrerer numerischer Variablen.
- **`hue`:** Ermöglicht die farbliche Gruppierung, ohne die Kategorien manuell mit `groupby()` und einer Schleife durchlaufen zu müssen.

**Wichtig:**
- Je mehr numerische Variablen ausgewählt werden, desto mehr Teilgrafiken entstehen. Bei vielen Variablen kann die Matrix unübersichtlich werden.
- Die dargestellten Zusammenhänge sind nicht automatisch kausal.
- Durch `alpha` können überlagerte Datenpunkte besser sichtbar gemacht werden.
- Die Variable unter `hue` muss auch im übergebenen DataFrame enthalten sein.

**Erkenntnis:** Mit `sns.pairplot()` lassen sich mehrere numerische Variablen gleichzeitig untersuchen. Die Kombination aus Scatter Plots, Histogrammen und farblicher Gruppierung ermöglicht es, Verteilungen, Zusammenhänge und Cluster effizient zu erkennen.

```python
# Scatter Plot Matrix

num_cols = ['power_kw', 'amperage', 'voltage']      #Liste der numerischen Spalten erstellen, die miteinander verglichen werden sollen

g = sns.pairplot(                                   #Scatter Plot Matrix erstellen und in g speichern
    df_evcharging[num_cols + ['power_type']],       #Numerische Spalten und kategoriale Gruppierungsvariable auswählen
                                                    #num_cols: Liste mit den numerischen Variablen
                                                    #+ ['power_type']: Kategoriale Spalte zur Liste hinzufügen
    hue='power_type',                               #Datenpunkte anhand der Kategorien farblich unterscheiden
                                                    #hue: Erstellt unterschiedliche Farben und eine Legende für die Gruppen
    diag_kind='hist',                               #Histogramme auf der Hauptdiagonale anzeigen
                                                    #hist: Zeigt die Häufigkeitsverteilung jeder einzelnen numerischen Variable
    plot_kws={'alpha': 0.4, 's': 15}                #Zusätzliche Einstellungen für die Scatter Plots festlegen
                                                    #plot_kws: Dictionary mit Parametern für die Streudiagramme
                                                    #alpha=0.4: Transparenz der Datenpunkte festlegen (0 bis 1)
                                                     #s=15: Grösse der Datenpunkte festlegen
)

g.fig.suptitle('Scatter Plot Matrix — EV Charging Stations', y=1.02)    #Gemeinsamen Titel für die gesamte Diagrammmatrix erstellen
                                                                        #g.fig: Zugriff auf die gesamte Abbildung
                                                                        #suptitle(): Titel oberhalb der gesamten Matrix setzen
                                                                        #y=1.02: Vertikale Position des Titels leicht nach oben verschieben

plt.show()                                                              #Vollständige Scatter Plot Matrix anzeigen

#ANPASSEN: df_evcharging an den verwendeten DataFrame anpassen
#ANPASSEN: 'power_kw', 'amperage', 'voltage' an die gewünschten numerischen Spalten anpassen
#ANPASSEN: 'power_type' in der Spaltenauswahl und bei hue an die gewünschte kategoriale Gruppierungsvariable anpassen
#ANPASSEN: diag_kind='hist' bei Bedarf auf 'kde' ändern, wenn Dichtekurven statt Histogrammen gewünscht sind
#ANPASSEN: alpha=0.4 an die gewünschte Transparenz der Datenpunkte anpassen
#ANPASSEN: s=15 an die gewünschte Grösse der Datenpunkte anpassen
#ANPASSEN: g.fig.suptitle() an den gewünschten Diagrammtitel anpassen
#ANPASSEN: y=1.02 bei Bedarf anpassen, um die Titelposition zu verändern
#NICHTS ANPASSEN: sns.pairplot(), plot_kws, g.fig.suptitle() und plt.show() bleiben bei dieser Vorgehensweise gleich
#VORAUSSETZUNG: seaborn wurde als sns und matplotlib.pyplot als plt importiert
#VORAUSSETZUNG: df_evcharging ist geladen und num_cols enthält numerische Spalten
#VORAUSSETZUNG: Die bei hue angegebene kategoriale Spalte ist ebenfalls im ausgewählten DataFrame enthalten
#ACHTUNG: Bei n numerischen Variablen entsteht standardmässig eine n × n-Matrix mit n² Teilgrafiken
#ACHTUNG: Viele numerische Variablen können die Diagrammmatrix unübersichtlich machen
#ERKENNTNIS: Die Hauptdiagonale zeigt die Verteilungen der einzelnen Variablen
#ERKENNTNIS: Die übrigen Diagramme zeigen paarweise Zusammenhänge, Cluster und mögliche Ausreisser
#ERKENNTNIS: hue ermöglicht den farblichen Vergleich verschiedener Kategorien innerhalb der Diagrammmatrix
```

### Hexagonal Binning Plot (Hexbin Plot) – Häufigkeitsverteilung zweier numerischer Variablen

- **Definition:** Ein Hexagonal Binning Plot (Hexbin Plot) visualisiert die gemeinsame Häufigkeitsverteilung zweier numerischer Variablen. Der Wertebereich wird in sechseckige Zellen (Hexagons) unterteilt, deren Farbe die Anzahl der Beobachtungen innerhalb der jeweiligen Zelle darstellt. Er gehört zur **Multivariate Graphical EDA**.
- **Zweck:** Zusammenhänge, Cluster, Häufigkeitsmuster und Bereiche mit hoher oder tiefer Beobachtungsdichte erkennen. Besonders geeignet für grosse Datensätze mit vielen überlappenden Datenpunkten.
- **`plt.subplots(figsize=(8, 5))`:** Erstellt eine Abbildung (`fig`) und eine Zeichenfläche (`ax`).
  - **`figsize=(8, 5)`:** Breite und Höhe der Abbildung in Zoll.
- **`ax.hexbin()`:** Erstellt einen Hexbin Plot:
  - **`df['voltage']`:** Erste numerische Variable auf der X-Achse.
  - **`df['amperage']`:** Zweite numerische Variable auf der Y-Achse.
  - **`gridsize=25`:** Legt die Auflösung des sechseckigen Gitters fest (ungefähre Anzahl Hexagons in X-Richtung).
  - **`cmap='Blues'`:** Verwendet eine blaue Farbskala zur Darstellung der Häufigkeiten.
  - **`mincnt=1`:** Zeigt nur Hexagons mit mindestens einer Beobachtung.
- **`hb = ax.hexbin(...)`:** Speichert das erstellte Hexbin-Objekt in `hb`, damit es für die Farblegende verwendet werden kann.
- **`plt.colorbar(hb, ax=ax, label='Count')`:** Fügt eine Farblegende hinzu, welche die Häufigkeiten der Hexagons erklärt.
  - **`hb`:** Verknüpft die Farblegende mit dem Hexbin Plot.
  - **`ax=ax`:** Ordnet die Farblegende der bestehenden Zeichenfläche zu.
  - **`label='Count'`:** Beschriftet die Farblegende.
- **`ax.set_xlabel()` / `ax.set_ylabel()`:** Beschriften die X- und Y-Achse.
- **`ax.set_title()`:** Legt den Diagrammtitel fest.
- **`plt.tight_layout()`:** Optimiert die Abstände zwischen Diagramm und Beschriftungen.
- **`plt.show()`:** Zeigt das fertige Diagramm an.

**Interpretation:**
- **Dunkle Hexagons:** Viele Beobachtungen innerhalb des jeweiligen Wertebereichs (bei `cmap='Blues'`).
- **Helle Hexagons:** Wenige Beobachtungen innerhalb des jeweiligen Wertebereichs.
- **Cluster:** Mehrere benachbarte Hexagons mit hoher Häufigkeit bilden einen erkennbaren Schwerpunkt.
- **Isolierte Hexagons:** Können auf seltene Wertekombinationen oder mögliche Ausreisser hinweisen.
- **Beispiel:** Bei EV-Ladestationen zeigen sich Häufungen bei typischen AC-Spannungen (230/400 V) mit niedrigeren Stromstärken sowie bei höheren Spannungen und Stromstärken, die auf DC-Schnellladestationen hindeuten.

**Unterschied zum Scatter Plot:**
- **Scatter Plot – `ax.scatter()`:** Zeichnet jede Beobachtung als einzelnen Punkt. Bei vielen Beobachtungen können sich Punkte überlagern.
- **Hexbin Plot – `ax.hexbin()`:** Fasst Beobachtungen innerhalb sechseckiger Bereiche zusammen und stellt deren Häufigkeit durch Farben dar.
- **Scatter Plot:** Eignet sich besonders, um einzelne Beobachtungen und Ausreisser zu erkennen.
- **Hexbin Plot:** Eignet sich besonders, um Dichtemuster und Cluster bei grossen Datensätzen sichtbar zu machen.

**Wichtig:**
- **`gridsize` beeinflusst die Darstellung:** Ein grösserer Wert erzeugt kleinere Hexagons und zeigt feinere Details; ein kleinerer Wert erzeugt grössere Hexagons und eine gröbere Darstellung.
- **Farben zeigen Häufigkeiten:** Die Farbintensität repräsentiert standardmässig die Anzahl der Beobachtungen pro Hexagon, nicht deren ursprünglichen X- oder Y-Wert.
- **Keine direkte kategoriale Gruppierung:** Im Gegensatz zum Scatter Plot mit `groupby()` werden die Kategorien nicht automatisch durch unterschiedliche Farben dargestellt.
- **Farbskala beachten:** Welche Farben hohe oder tiefe Häufigkeiten repräsentieren, hängt von der gewählten `cmap` ab.

**Erkenntnis:** Mit `ax.hexbin()` lassen sich die Häufigkeiten zweier numerischer Variablen als zweidimensionale Dichteverteilung visualisieren. Besonders bei grossen Datensätzen werden Cluster und Häufungen deutlicher sichtbar als bei einem gewöhnlichen Scatter Plot.

```python
# Hexagonal Binning Plot

fig, ax = plt.subplots(figsize=(8, 5)) #Neue Abbildung (fig) und Zeichenfläche (ax) erstellen
                                        #figsize=(8, 5): Breite 8 Zoll, Höhe 5 Zoll

hb = ax.hexbin(df_evcharging['voltage'], df_evcharging['amperage'], gridsize=25, cmap='Blues', mincnt=1) #Hexbin Plot erstellen und in hb speichern
                                                                                                        #hexbin(): Datenpunkte in sechseckige Bereiche unterteilen
                                                                                                        #df_evcharging['voltage']: Numerische Variable auf der X-Achse
                                                                                                        #df_evcharging['amperage']: Numerische Variable auf der Y-Achse
                                                                                                        #gridsize=25: Auflösung des sechseckigen Gitters festlegen
                                                                                                        #cmap='Blues': Blaue Farbskala für die Häufigkeiten verwenden
                                                                                                        #mincnt=1: Nur Hexagons mit mindestens einer Beobachtung anzeigen
                                                                                                        #hb: Speichert das Hexbin-Objekt für die Farblegende

plt.colorbar(hb, ax=ax, label='Count') #Farblegende mit den Häufigkeiten der Hexagons anzeigen
                                       #hb: Verknüpft die Farblegende mit dem Hexbin Plot
                                       #ax=ax: Ordnet die Farblegende der Zeichenfläche zu
                                       #label='Count': Beschriftung der Farblegende festlegen

ax.set_xlabel('Voltage (V)')    #X-Achse beschriften
ax.set_ylabel('Amperage (A)')   #Y-Achse beschriften
ax.set_title('Hexbin: Voltage vs. Amperage') #Diagrammtitel festlegen

plt.tight_layout()  #Abstände automatisch optimieren
plt.show()          #Hexbin Plot anzeigen

#ANPASSEN: df_evcharging an den verwendeten DataFrame anpassen
#ANPASSEN: 'voltage' an die gewünschte numerische Spalte für die X-Achse anpassen
#ANPASSEN: 'amperage' an die gewünschte numerische Spalte für die Y-Achse anpassen
#ANPASSEN: gridsize=25 an die gewünschte Auflösung des sechseckigen Gitters anpassen
#ANPASSEN: cmap='Blues' an die gewünschte Farbskala anpassen
#ANPASSEN: mincnt=1 an die gewünschte Mindesthäufigkeit pro Hexagon anpassen
#ANPASSEN: figsize=(8, 5) an die gewünschte Diagrammgrösse anpassen
#ANPASSEN: label='Count' an die gewünschte Beschriftung der Farblegende anpassen
#ANPASSEN: ax.set_xlabel(), ax.set_ylabel() und ax.set_title() passend zu den untersuchten Variablen beschriften
#NICHTS ANPASSEN: ax.hexbin(), plt.colorbar(), plt.tight_layout() und plt.show() bleiben bei dieser Vorgehensweise gleich
#VORAUSSETZUNG: matplotlib.pyplot wurde als plt importiert und df_evcharging ist geladen
#VORAUSSETZUNG: Die ausgewählten X- und Y-Spalten enthalten numerische Werte
#ACHTUNG: Die Farbintensität zeigt standardmässig die Anzahl Beobachtungen pro Hexagon
#ACHTUNG: gridsize beeinflusst die Auflösung und damit die Sichtbarkeit von Clustern
#ERKENNTNIS: Dunklere Hexagons bedeuten bei cmap='Blues' höhere Häufigkeiten
#ERKENNTNIS: Hexbin Plots eignen sich besonders für grosse Datensätze mit vielen überlappenden Datenpunkten
```

### Correlation Heatmap – Lineare Zusammenhänge zwischen numerischen Variablen visualisieren

- **Definition:** Eine Correlation Heatmap (Korrelations-Heatmap) visualisiert die Korrelationsmatrix mehrerer numerischer Variablen mithilfe von Farben. Sie gehört zur **Multivariate Graphical EDA**.
- **Zweck:** Positive und negative Korrelationen sowie deren Stärke zwischen mehreren numerischen Variablen schnell erkennen und vergleichen.
- **`.corr(method='pearson')`:** Berechnet die paarweisen Pearson-Korrelationskoeffizienten:
  - **`df[num_cols]`:** Wählt die numerischen Spalten aus, deren Korrelationen untersucht werden sollen.
  - **`method='pearson'`:** Misst Richtung und Stärke linearer Zusammenhänge (r zwischen −1 und +1).
- **`plt.subplots(figsize=(6, 5))`:** Erstellt eine Abbildung (`fig`) und eine Zeichenfläche (`ax`).
- **`sns.heatmap()`:** Visualisiert die Korrelationsmatrix als farbige Tabelle:
  - **`corr`:** Die zuvor berechnete Korrelationsmatrix.
  - **`annot=True`:** Zeigt die Korrelationskoeffizienten zusätzlich als Zahlen in den Zellen an.
  - **`fmt='.2f'`:** Formatiert die angezeigten Zahlen auf zwei Nachkommastellen.
  - **`cmap='RdBu_r'`:** Verwendet eine Farbskala von Blau (negativ) über Weiss (neutral) nach Rot (positiv).
  - **`center=0`:** Zentriert die Farbskala beim Korrelationswert 0.
  - **`vmin=-1`:** Legt −1 als untere Grenze der Farbskala fest.
  - **`vmax=1`:** Legt +1 als obere Grenze der Farbskala fest.
  - **`linewidths=0.5`:** Fügt Trennlinien zwischen den Zellen hinzu.
  - **`ax=ax`:** Zeichnet die Heatmap auf der bestehenden Zeichenfläche.
- **`ax.set_title()`:** Legt den Diagrammtitel fest.
- **`plt.tight_layout()`:** Optimiert die Abstände zwischen Diagramm und Beschriftungen.
- **`plt.show()`:** Zeigt die Heatmap an.

**Interpretation bei `cmap='RdBu_r'`:**
- **Dunkelrot (r nahe +1):** Starker positiver linearer Zusammenhang → Beide Variablen steigen tendenziell gemeinsam.
- **Weiss (r nahe 0):** Schwacher oder kein linearer Zusammenhang.
- **Dunkelblau (r nahe −1):** Starker negativer linearer Zusammenhang → Steigt eine Variable, sinkt die andere tendenziell.
- **Hauptdiagonale:** Enthält Selbstkorrelationen (r = 1), sofern die Korrelation definiert ist.
- **Symmetrie:** Die Korrelationsmatrix ist gespiegelt, da die Korrelation zwischen A und B gleich der Korrelation zwischen B und A ist.
- **Beispiel:** Bei EV-Ladestationen besteht ein starker positiver Zusammenhang zwischen `power_kw` und `amperage`, der durch dunkelrote Zellen sichtbar wird.

**Unterschied zur Pearson Correlation Matrix:**
- **Correlation Matrix – `.corr(method='pearson')`:** Berechnet die Korrelationskoeffizienten und gibt diese als numerische Tabelle aus.
- **Correlation Heatmap – `sns.heatmap()`:** Visualisiert dieselben Korrelationskoeffizienten zusätzlich mithilfe von Farben.
- **Wichtig:** Die Heatmap führt keine neue Korrelationsberechnung durch, sondern visualisiert die zuvor berechnete Matrix.

**Wichtig:**
- **Korrelation ≠ Kausalität:** Ein statistischer Zusammenhang beweist keine Ursache-Wirkungs-Beziehung.
- **Pearson misst nur lineare Zusammenhänge:** Nichtlineare Beziehungen können trotz geringer Pearson-Korrelation bestehen.
- **Ausreisser:** Können die Korrelationskoeffizienten stark beeinflussen.
- **`fmt='.2f'` ≠ `.round(2)`:** `fmt` beeinflusst nur die angezeigten Nachkommastellen innerhalb der Heatmap. Die ursprünglichen Korrelationswerte werden nicht verändert.
- **Farbskala:** Bei einer anderen `cmap` können andere Farben für positive und negative Korrelationen stehen.

**Erkenntnis:** Mit `.corr(method='pearson')` und `sns.heatmap()` lassen sich die linearen Zusammenhänge zwischen mehreren numerischen Variablen berechnen und grafisch darstellen. Farben zeigen die Richtung und Stärke der Korrelationen auf einen Blick.

```python
# Pearson Correlation Heatmap

num_cols = ['power_kw', 'amperage', 'voltage'] #Liste der numerischen Spalten erstellen, deren Korrelationen visualisiert werden sollen

corr = df_evcharging[num_cols].corr(method='pearson') #Pearson-Korrelationsmatrix für die ausgewählten numerischen Spalten berechnen
                                                      #df_evcharging[num_cols]: Numerische Spalten auswählen
                                                      #corr(): Berechnet die paarweisen Korrelationskoeffizienten
                                                      #method='pearson': Misst Stärke und Richtung linearer Zusammenhänge
                                                      #Ergebnis: Korrelationsmatrix mit Werten zwischen -1 und +1

fig, ax = plt.subplots(figsize=(6, 5))   #Neue Abbildung (fig) und Zeichenfläche (ax) erstellen
                                         #figsize=(6, 5): Breite 6 Zoll, Höhe 5 Zoll

sns.heatmap(                             #Korrelationsmatrix als farbige Heatmap visualisieren
    corr,                                #Zuvor berechnete Korrelationsmatrix verwenden
    annot=True,                          #Korrelationskoeffizienten als Zahlen in den Zellen anzeigen
    fmt='.2f',                           #Zahlen mit zwei Nachkommastellen anzeigen
    cmap='RdBu_r',                       #Farbskala: Blau = negativ, Weiss = neutral, Rot = positiv
    center=0,                            #Farbskala beim Korrelationswert 0 zentrieren
    vmin=-1,                             #Untere Grenze der Farbskala auf -1 setzen
    vmax=1,                              #Obere Grenze der Farbskala auf +1 setzen
    linewidths=0.5,                      #Breite der Trennlinien zwischen den Zellen festlegen
    ax=ax                                #Heatmap auf der bestehenden Zeichenfläche darstellen
)

ax.set_title('Pearson Correlation Heatmap — EV Charging Stations') #Diagrammtitel festlegen
plt.tight_layout()                      #Abstände automatisch optimieren
plt.show()                              #Korrelations-Heatmap anzeigen

#ANPASSEN: df_evcharging an den verwendeten DataFrame anpassen
#ANPASSEN: 'power_kw', 'amperage', 'voltage' an die gewünschten numerischen Spalten anpassen
#ANPASSEN: figsize=(6, 5) an die gewünschte Diagrammgrösse anpassen
#ANPASSEN: annot=True auf False setzen, falls keine Zahlen in den Zellen angezeigt werden sollen
#ANPASSEN: fmt='.2f' an die gewünschte Anzahl Nachkommastellen anpassen
#ANPASSEN: cmap='RdBu_r' an die gewünschte Farbskala anpassen
#ANPASSEN: linewidths=0.5 an die gewünschte Breite der Zelltrennlinien anpassen
#ANPASSEN: ax.set_title() passend zum untersuchten Datensatz beschriften
#NICHTS ANPASSEN: .corr(method='pearson') und sns.heatmap() sind die Methoden zur Berechnung und Visualisierung der Korrelationen
#NICHTS ANPASSEN: center=0, vmin=-1 und vmax=1 sind für die Darstellung von Pearson-Korrelationen sinnvoll
#VORAUSSETZUNG: seaborn wurde als sns und matplotlib.pyplot als plt importiert
#VORAUSSETZUNG: df_evcharging ist geladen und num_cols enthält numerische Spalten
#ACHTUNG: fmt='.2f' rundet nur die angezeigten Werte, nicht die berechnete Korrelationsmatrix
#ACHTUNG: Die Hauptdiagonale enthält Selbstkorrelationen (r = 1), sofern diese definiert sind
#ACHTUNG: Korrelation bedeutet nicht automatisch Kausalität
#ERKENNTNIS: Dunkelrot → Starke positive Korrelation (r nahe +1)
#ERKENNTNIS: Dunkelblau → Starke negative Korrelation (r nahe -1)
#ERKENNTNIS: Weiss → Schwacher oder kein linearer Zusammenhang (r nahe 0)
```

### Line Chart – Entwicklungen und Trends visualisieren

- **Definition:** Ein Line Chart (Liniendiagramm) stellt Datenpunkte dar, die durch gerade Linien miteinander verbunden werden.
- **Zweck:** Entwicklungen, Trends, Veränderungen und Schwankungen entlang einer geordneten X-Achse erkennen.
- **EDA:** Univariate Graphical EDA bei der Darstellung einer numerischen Variable über die Zeit; grundsätzlich auch für Zusammenhänge zwischen zwei geordneten numerischen Variablen geeignet.

**Aufbau und Interpretation:**
- **X-Achse:** Enthält eine geordnete Variable, häufig Zeitpunkte (z.B. Tage, Monate, Jahre).
- **Y-Achse:** Enthält die untersuchte numerische Variable (z.B. Umsatz, Temperatur, Anzahl Ladestationen).
- **Datenpunkte (Markers):** Stellen die einzelnen Beobachtungen dar.
- **Verbindungslinien:** Verbinden die Datenpunkte in ihrer Reihenfolge und machen Entwicklungen sichtbar.
- **Steigende Linie:** Die dargestellten Werte nehmen zu.
- **Fallende Linie:** Die dargestellten Werte nehmen ab.
- **Horizontale Linie:** Die Werte bleiben unverändert.
- **Schwankende Linie:** Die Werte verändern sich wiederholt nach oben und unten.

**Matplotlib – `ax.plot()`:**
- **`fig, ax = plt.subplots(figsize=(8, 4))`:** Erstellt eine Abbildung und eine Zeichenfläche.
- **`ax.plot(x, y)`:** Erstellt ein Liniendiagramm mit X- und Y-Werten.
  - **`x`:** Werte auf der X-Achse, z.B. Zeitpunkte oder aufsteigend sortierte numerische Werte.
  - **`y`:** Numerische Werte auf der Y-Achse.
  - **`color='steelblue'`:** Bestimmt die Linienfarbe.
  - **`marker='o'`:** Markiert die einzelnen Datenpunkte mit Kreisen.
  - **`linestyle='-'`:** Zeichnet eine durchgezogene Linie.
  - **`linewidth=2`:** Bestimmt die Liniendicke.
  - **`label='...'`:** Definiert eine Beschriftung für die Legende.
- **`ax.set_xlabel()` / `ax.set_ylabel()`:** Beschriften die Achsen.
- **`ax.set_title()`:** Legt den Diagrammtitel fest.
- **`ax.grid()`:** Fügt Gitternetzlinien hinzu.
- **`ax.legend()`:** Zeigt die Legende an, falls Datenreihen mit `label` beschriftet wurden.
- **`plt.tight_layout()`:** Optimiert die Abstände.
- **`plt.show()`:** Zeigt das Diagramm an.

**Beispiel EV Charging Stations:**
- Die Ladeleistungen (`power_kw`) werden aufsteigend sortiert.
- Für jede Leistungsstufe wird die kumulierte Anzahl Ladestationen berechnet.
- Das Liniendiagramm zeigt, wie viele Ladestationen insgesamt höchstens eine bestimmte Ladeleistung besitzen.
- Starke Anstiege zeigen Leistungsbereiche mit vielen Ladestationen.

**Unterschied zum Scatter Plot:**
- **Line Chart – `ax.plot()`:** Verbindet geordnete Datenpunkte mit Linien; besonders geeignet für Entwicklungen und Trends.
- **Scatter Plot – `ax.scatter()`:** Stellt Datenpunkte einzeln dar; besonders geeignet zur Untersuchung von Zusammenhängen zwischen zwei numerischen Variablen.
- **Wichtig:** Ein Line Chart setzt eine sinnvolle Reihenfolge der X-Werte voraus. Ein Scatter Plot benötigt diese nicht.

**Wichtig:**
- Liniendiagramme eignen sich besonders für Zeitreihen, aber auch für andere sinnvoll geordnete numerische Daten.
- Bei Zeitreihen müssen die Daten chronologisch sortiert sein.
- Ungeordnete Kategorien sollten normalerweise mit einem Bar Chart dargestellt werden.
- Linien zwischen Messpunkten können Zwischenwerte suggerieren, obwohl diese nicht beobachtet wurden.
- Mehrere Datenreihen können mit mehreren `ax.plot()`-Aufrufen in derselben Abbildung dargestellt werden.

**Erkenntnis:** Ein Line Chart verbindet geordnete Datenpunkte und eignet sich besonders zur Visualisierung von zeitlichen Entwicklungen, Trends und Schwankungen. Mit `ax.plot()` lassen sich Linienfarbe, Marker, Linienstil und Liniendicke individuell konfigurieren.

```python
# LINE CHART – MATPLOTLIB

# Prepare data
cumulative = df_evcharging['power_kw'].value_counts().sort_index().cumsum()     #Kumulierte Anzahl Ladestationen nach aufsteigender Ladeleistung berechnen
                                                                                #value_counts(): Häufigkeit jeder Ladeleistung zählen
                                                                                #sort_index(): Ladeleistungen aufsteigend sortieren
                                                                                #cumsum(): Häufigkeiten fortlaufend aufsummieren

# Create line chart
fig, ax = plt.subplots(figsize=(8, 4))                                          #Abbildung und Zeichenfläche erstellen; Grösse in Zoll

ax.plot(cumulative.index, cumulative.values,                                    #Liniendiagramm: Ladeleistung (X) gegen kumulierte Anzahl Stationen (Y)
        color='steelblue', marker='o',                                          #Linienfarbe festlegen; einzelne Datenpunkte als Kreise markieren
        linestyle='-', linewidth=2)                                             #Durchgezogene Linie mit Liniendicke 2

ax.set_xlabel('Charging Power (kW)')                                            #X-Achse: Ladeleistung
ax.set_ylabel('Cumulative Number of Stations')                                  #Y-Achse: Kumulierte Anzahl Ladestationen
ax.set_title('Cumulative Charging Stations by Charging Power')                  #Diagrammtitel festlegen
ax.grid(True, alpha=0.3)                                                        #Gitternetzlinien mit geringer Deckkraft anzeigen

plt.tight_layout()                                                              #Abstände und Beschriftungen optimieren
plt.show()                                                                      #Liniendiagramm anzeigen

#ANPASSEN: df_evcharging an den verwendeten DataFrame anpassen
#ANPASSEN: 'power_kw' an die gewünschte numerische Spalte anpassen
#ANPASSEN: .value_counts().sort_index().cumsum() nur verwenden, wenn eine kumulierte Häufigkeitsverteilung dargestellt werden soll
#ANPASSEN: cumulative.index und cumulative.values durch andere geordnete X- und zugehörige Y-Werte ersetzen, z.B. bei Zeitreihen
#ANPASSEN: figsize, color, marker, linestyle und linewidth bei Bedarf anpassen
#ANPASSEN: Achsenbeschriftungen und Diagrammtitel passend wählen
#NICHTS ANPASSEN: ax.plot(), plt.subplots(), plt.tight_layout() und plt.show() bleiben bei dieser Vorgehensweise gleich
#VORAUSSETZUNG: df_evcharging ist geladen und matplotlib.pyplot wurde als plt importiert
#VORAUSSETZUNG: Die X-Werte besitzen eine sinnvolle Reihenfolge
#ACHTUNG: Bei Zeitreihen zuerst chronologisch sortieren, damit die Punkte korrekt verbunden werden
#ACHTUNG: Ein Line Chart ist normalerweise nicht für ungeordnete kategoriale Daten geeignet
#ERKENNTNIS: Steigende Linie = zunehmende Werte; fallende Linie = abnehmende Werte; horizontale Linie = konstante Werte
#ERKENNTNIS: Im vorliegenden kumulativen Beispiel kann die Linie nur steigen oder konstant bleiben
```

### Violin Plot (Ergänzung)– Verteilungsdichte, Quartile und Multimodalität visualisieren

- **Definition:** Ein Violin Plot visualisiert die Verteilung einer numerischen Variable mithilfe einer geglätteten Dichteschätzung (Kernel Density Estimation, KDE). Die charakteristische Violinform entsteht durch die gespiegelte Darstellung dieser Dichte.
- **Zweck:** Verteilungen zwischen verschiedenen Gruppen vergleichen und dabei insbesondere Median, Quartile, Streuung, Datendichte und mögliche Häufigkeitsgipfel erkennen.
- **EDA:** Univariate Verteilungsanalyse; beim Vergleich einer numerischen Variable zwischen mehreren Kategorien eine bivariate Analyse.

**Aufbau und Interpretation eines Violin Plots:**
- **Breite Bereiche:** Hohe geschätzte Datendichte → Viele Beobachtungen konzentrieren sich in diesem Wertebereich.
- **Schmale Bereiche:** Geringe geschätzte Datendichte → Weniger Beobachtungen in diesem Wertebereich.
- **Symmetrische Violinform:** Die Dichte wird links und rechts gespiegelt. Dies bedeutet nicht, dass die tatsächliche Datenverteilung symmetrisch ist.
- **Quartile (bei `inner='quartile'`):**
  - **Q1 (25%-Quantil):** 25 % der Werte liegen bei oder unter diesem Wert.
  - **Median / Q2 (50%-Quantil):** 50 % der Werte liegen bei oder unter diesem Wert.
  - **Q3 (75%-Quantil):** 75 % der Werte liegen bei oder unter diesem Wert.
- **Mehrere breite Bereiche:** Können auf eine multimodale Verteilung mit mehreren Häufigkeitsgipfeln hinweisen.

**Seaborn – `sns.violinplot()`:**
- **`data=df_evcharging`:** Verwendeter DataFrame.
- **`x='power_type'`:** Kategoriale Variable auf der X-Achse; bestimmt die Gruppen.
- **`y='power_kw'`:** Numerische Variable auf der Y-Achse; deren Verteilung wird untersucht.
- **`order=['AC_1_PHASE', 'AC_3_PHASE', 'DC']`:** Legt die Reihenfolge der Kategorien fest.
- **`inner='quartile'`:** Zeigt Median, Q1 und Q3 als Linien innerhalb der Violine.
  - **`inner='box'`:** Zeigt alternativ eine vereinfachte Boxplot-Darstellung innerhalb der Violine.
  - **`inner=None`:** Zeigt keine zusätzlichen statistischen Kennzahlen innerhalb der Violine.
- **`ax=ax`:** Zeichnet den Violin Plot auf der zuvor erstellten Zeichenfläche.

**Kernel Density Estimation (KDE):**
- **Definition:** Statistisches Verfahren zur Schätzung einer kontinuierlichen Wahrscheinlichkeitsdichte anhand beobachteter Daten.
- **Funktionsweise:** Die beobachteten Werte werden zu einer geglätteten Dichtekurve zusammengefasst. Beim Violin Plot wird diese Kurve gespiegelt.
- **Interpretation:** Hohe Dichte führt zu breiten Stellen, niedrige Dichte zu schmalen Stellen.
- **Wichtig:** Die Breite zeigt die geschätzte Dichte und nicht unmittelbar die absolute Anzahl der Beobachtungen.
- **Einschränkung:** Die Glättung kann Häufigkeitsgipfel abschwächen, künstlich erzeugen oder die Dichte über tatsächlich beobachtete Werte hinaus fortsetzen.

**Unterschied zum Box Plot – Übungsfrage des Dozenten:**

**Question: What additional information does the violin plot provide compared to the box plot?**

**Answer:**
- The violin plot shows the full estimated distribution shape and data density, not just summary statistics such as the median and quartiles.
- Wider sections indicate higher estimated data density, while narrower sections indicate lower density.
- It can reveal multiple peaks (multimodal distributions), e.g. DC charging clusters around 50 kW, 150 kW and 350+ kW.
- A box plot summarises the distribution using the median, quartiles, whiskers and potential outliers, but does not reveal these density patterns.

**Vergleich Box Plot vs. Violin Plot:**
- **Box Plot:** Zeigt Median, Quartile, IQR, Whiskers und mögliche Ausreisser → Gut für kompakte statistische Vergleiche.
- **Violin Plot:** Zeigt die geschätzte Verteilungsdichte und optional Median/Quartile → Gut zum Erkennen der Verteilungsform und mehrerer Häufigkeitsgipfel.
- **Gemeinsamkeit:** Beide eignen sich zum Vergleich numerischer Verteilungen zwischen verschiedenen Kategorien.
- **Unterschied:** Ein Violin Plot zeigt die Dichtestruktur, die im Box Plot nicht sichtbar ist. Ausreisser werden dagegen im Box Plot standardmässig explizit als Punkte dargestellt.

**Beispiel EV Charging Stations:**
- **DC:** Die Violine zeigt mehrere Konzentrationen bei unterschiedlichen Ladeleistungen (gemäss Musterlösung etwa 50, 150 und über 350 kW). Diese multimodale Struktur ist im Box Plot nicht direkt erkennbar.
- **AC_3_PHASE:** Eine sehr schmale Verteilung rund um 22 kW zeigt, dass die meisten Ladestationen eine ähnliche Ladeleistung besitzen.
- **Interpretation:** Der Box Plot fasst beide Verteilungen statistisch zusammen; der Violin Plot macht zusätzlich Unterschiede in der geschätzten Verteilungsform sichtbar.

**Wichtig:**
- Ein Violin Plot basiert auf einer geschätzten Dichte und zeigt nicht die exakten Häufigkeiten einzelner Werte.
- Eine breitere Violine bedeutet nicht automatisch mehr Beobachtungen als in einer anderen Gruppe; dies hängt auch von der Dichteskalierung ab.
- Insbesondere bei diskreten Werten und kleinen Stichproben kann die KDE-Darstellung irreführend sein.
- Für die genaue Identifikation potenzieller Ausreisser eignet sich ein Box Plot besser.

**Erkenntnis:** Ein Violin Plot kombiniert eine geglättete Darstellung der Verteilungsdichte mit optionalen statistischen Kennzahlen. Gegenüber dem Box Plot liefert er insbesondere zusätzliche Informationen über die Verteilungsform, die Konzentration der Beobachtungen und mögliche Multimodalität.

```python
# VIOLIN PLOT – SEABORN

fig, ax = plt.subplots(figsize=(7, 5))                                          #Abbildung und Zeichenfläche erstellen; Grösse in Zoll

sns.violinplot(data=df_evcharging, x='power_type', y='power_kw',                #Violin Plot: Geschätzte Verteilungsdichte der Ladeleistung je Ladeart
               order=['AC_1_PHASE', 'AC_3_PHASE', 'DC'],                        #Reihenfolge der Kategorien auf der X-Achse festlegen
               inner='quartile', ax=ax)                                         #Q1, Median und Q3 innerhalb der Violine anzeigen; ax = Zeichenfläche

ax.set_xlabel('Power Type')                                                     #X-Achse: Ladearten
ax.set_ylabel('Charging Power (kW)')                                            #Y-Achse: Ladeleistung
ax.set_title('Charging Power by Power Type (Violin Plot)')                      #Diagrammtitel festlegen
plt.tight_layout()                                                              #Abstände optimieren
plt.show()                                                                      #Violin Plot anzeigen

print('Interpretation:')
print('The violin plot shows the full shape of the distribution (not just quartiles).')         #Zeigt zusätzlich die geschätzte Verteilungsform und Dichte
print('For DC: the wide body at higher kW values reveals a multi-modal distribution')           #Breite Bereiche deuten auf mehrere Häufigkeitsgipfel bei DC hin
print('(clusters at 50 kW, 150 kW and 350+ kW), which the box plot hides.')                     #Mehrere Konzentrationen gemäss Musterlösung, im Boxplot nicht direkt sichtbar
print('For AC_3_PHASE: the very narrow violin confirms that nearly all stations are at 22 kW.') #Starke Konzentration der Ladeleistungen bei 22 kW

#ANPASSEN: df_evcharging an den verwendeten DataFrame anpassen
#ANPASSEN: x='power_type' an die kategoriale Gruppierungsvariable anpassen
#ANPASSEN: y='power_kw' an die numerische Variable anpassen
#ANPASSEN: order=[...] an die vorhandenen Kategorien und ihre gewünschte Reihenfolge anpassen
#ANPASSEN: inner='quartile' bei Bedarf durch 'box' oder None ersetzen
#ANPASSEN: figsize=(7, 5), Achsenbeschriftungen und Diagrammtitel bei Bedarf anpassen
#ANPASSEN: print()-Interpretationen an die tatsächlich beobachteten Verteilungen anpassen
#NICHTS ANPASSEN: sns.violinplot(), plt.subplots(), plt.tight_layout() und plt.show() bleiben bei dieser Vorgehensweise gleich
#VORAUSSETZUNG: 'power_type' enthält Kategorien und 'power_kw' numerische Werte
#ACHTUNG: Der Violin Plot basiert auf einer geglätteten Dichteschätzung (KDE), nicht auf exakten Häufigkeiten
#ACHTUNG: Breite Stellen zeigen eine hohe geschätzte Dichte, nicht zwingend eine höhere absolute Beobachtungszahl
#ACHTUNG: KDE kann bei diskreten Werten oder kleinen Stichproben die tatsächliche Verteilung verzerren
#ERKENNTNIS: Breite Bereiche = hohe Datendichte; schmale Bereiche = geringe Datendichte
#ERKENNTNIS: Mehrere breite Bereiche können auf Multimodalität (mehrere Häufigkeitsgipfel) hinweisen
#ERKENNTNIS: Violin Plot zeigt die geschätzte Verteilungsform; Box Plot zeigt vor allem statistische Kennzahlen und mögliche Ausreisser
```
